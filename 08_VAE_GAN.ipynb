CONFRONTO TRA VAE E GAN: DUE FILOSOFIE GENERATIVE

Conosciamo la GAN con la loro competizoine da falsario e poliziotto
e la VAE con la loro eleganza probabilità e lo spazio latente continuo

Ma quale scegliere?

Sono due filosofia diverse
Quand'estetica deve cedere il passo alla precisoine statitica.

- Qualità contro Copertura: l estetica delle GAN e la diversità dei VAE
- Stabilità del Training: la convergenza dell'ELBO contro il gioco minmax
- Invertibiltà: etrarre lo  spazio latente in modo efficiente

Cosa verchiamo davvere quando generiamo

La differenza tra le GAN e le VAE non riside solo nell'architettura, ma dall'obbiettivo  probabilistico che cercano di ottimizzare.
Mentre le GAN sono ossessionate dal realismo dei singoli campioni
cercadurre immagini che il discriminatore
VAE si concentrano sulla rappresentazione dell'intero dataset senza trascurare alcuna variante.
