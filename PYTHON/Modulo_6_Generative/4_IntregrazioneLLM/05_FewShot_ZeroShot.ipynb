FEW-SHOT E ZERO-SHOT LEARNING

Immagina di insegnare un compito nuovo ad una macchina, solo parlandole, senza toccare neanche un singolo neurone.
Zero-Shot e Few-Shot learning, passiamo dall'ai che deve essere istruita per un compito all'ai che capisce il compito al volo.

- In-context learning: imparare senza aggiornare i pesi
- selezione degli esempi: strategie di prompt rappresentativi
- limiti della context window: gestire spazio e complessità

In-context Learning
L'apprendimento per osmosi contestuale
L'in-context learning (ICL) è la capacità di un modello di attarsii a un nuovo compito semplicemente leggendoo esempi forniti nel prompt di input
A differenza del fine-tuning classico, l'ICL non richiede il calcolo dei gradienti nè l modifica dei parametri interni del Transformer. Il modello è in sola lettura, eppure impara. Tutto ciò avviene nel layer di attenzione

Zero-Shot significa: chiedi al modello di svolgere un compito senza mostrargli esempi
Il modello deve capire da solo come applicare le regole.
Zero-Shot = istruzione + nuovo input -> risposta

Few-Shot significa: prima mostri al modello alcuni esempi di input e output corretti, poi gli chiedi di trattare un nuovo caso allo stesso modo
Il modello osserva gli esempi e capisce meglio il criterio che vuoi usare.
Few-Shot = istruzione + pochi esempi + nuovo input -> risposta
Con il Few-Shot prompting non cambi i pesi del modello, ma semplicemente metti esempi nel contesto. In questo caso gli esempi stanno nel prompt

Diverso rispetto al fine-tuning dove gli esempi vengono utilizzati per modificare i pesi del modello

Few-shot -> insegno temporaneamente con esempi (il modello impara ma non nel senso tradizionale del Machine Learning)
Fine-tuning -> modifico il modello

Un problema del Few-Shot è il costo
Più esempi metti, più token hai, prompt più grande, più memoria/context window -> maggiore costo computazionale.

Possiamo pertanto prevedere una progressione:
1. Provo Zero-shot -> 
2. Se non è abbastanza stabile ->
3. Aggiungo Few-shot ->
4. Se serve conoscenza aziendale ->
5. RAG ->
6. Se il comportamento deve cambiare in modo stabile ->
7. Fine-tuning

Few-shot e Zero-shot sono importanti perché mostrano una differenza profonda rispetto al Machine Learning classico: spesso non devi riaddestrare un modello per ogni nuovo compito. Puoi modificare il comportamento semplicemente cambiando il contesto e gli esempi forniti nel prompt.

Con One-Shot e con Few-Shot stiamo modificando la probabilità gli esempi modificano la probabilità delle soluzioni possibili.

Il segreto risiede nella self-attention. Il meccanismo di self-attention identifica le relazioni strutturali tra gli esempi forniti e la nuova richiesta dell'utente (analisi dei pattern).
Il modello rimane in modalità 'sola lettura', l'apprendimento è temporaneo confinato alla singola sessione di generazione (zero-grad inference).
Gli LLM possono passare da compiti di traduzione a compiti di coding nello stesso istante semplicemente cambiando il prefisso del prompt (generalizzazione immediata).

L'Origine dell'ICL
L'In-Context learning è una proprietà emergente, non è stato progettato esplicitamente, ma è emerso come proprietà secondaria dell'addestramento su dataset massivi.
Più il modello è grande, più diventa capace di estrapolare regole complesse da un numero esiguo di dimostrazioni. La sua capacità di minimizzare la perdita durante l'inferenza diventa raffinata più il modello è grande.
Non copiano solo il testo, questo modelli estrapolano la funzione matematica che legano gli esempi alla soluzione.

Ma se gli esempi sono così potenti, sceglierli male può essere un disastro.

Selezione degli Esempi
L'arte di scegliere i campioni giusti
Nel few-shot learning, la qualità e la varietà degli esempi sono più importanti della loro quantità numerica. La qualità batte la quantità.
Esempi mal scelti possono introdurre bias nel modello o portare a un'imitazione superficiale che ignora la logica sottostante.
Se gli esempi sono troppo facili, il modello non impare i casi complessi.
Se sono troppo difficili o rumorosi la rete potrebbe trovare una logica dove non c'è.

Strategia di Selezione degli Esempi
Ottimizzare il segnale negli esempi
- Diversità: includere casi limite ed esempi variegati per coprire diverse sfaccettature del compito richiesto
- Rappresentatività: scegliere esempi che siano semanticamente vicini alla query reale che il modello dovrà processare
- Ordine degli esempi: il modello tende ad essere influenzato maggiormente dall'ultimo esempio visto nel prompt (recency bias)

Una buona selezione degli esempi agisce come un compressore di informazione per la distribuzione di output

Trappole più comuni da evitare
- Se gli esempi few-shot mostrano sempre gli stessi esempi, mostrando sempre la stessa classe di output, il modello tenderà a rispondere sempre con quella classe (label bias).
- Mantenere uno schema identico tra gli esempi aiuta il modello a isolare la logica dal rumore della informazione. Mantieni una formattazione coerente, usa sempre gli stessi simoboli (esempio input: e output:), per aiutare l'attenzione ad isolare la logica dal rumore.
- Il numero ottimale di esempi (k) varia solitamente tra 1 e 10; oltre questa soglia, il guadagno di performance diminuisce e stiamo solo mangiando spazio prezioso.

E se avessimo migliaia di esempi a disposizione, come scegliamo i migliori in tempo reale?
Qui la tecnica si evolve
Retrieval for Few-Shot
Esempi dinamici
I sistemi moderni utilizzano database vettoriali per pescare gli esempi più pertinenti per ogni specifica richiesta dell'utente.
Quando un utente fa una domanda, cerchiamo istantaneamente gli esempi nel nostro archivio, facendo un confronto vettoriale. Calcoliamo la distanza coseno tra vettori  e costruiamo un prompt dinamico. 
E' come dare al modello un manuale di istruzioin che si riscrive ogni volta per mostrare solo i casi più pertinenti.
Questo apprroccio trasforma il prompt statico in un'entità dinamica che si adatta al contesto in tempo reale.

Ma tutti questi esempi si scontano con un limite fisico invalicabile, lo spazio.

Limiti della Context Window
Gestire lo spazio sul tavolo da lavoro
Ogni LLM ha un limite fisico al numero di token che può processare simultaneamente, noto come finestra di contesto.
Ogni parola, ogni spazio, ogni virgola del prompot occupa un token (a volte una parola anche più token).
Esiste un trade-off critico tra il numero di esempi few-shot che possiamo fornire e lo spazio rimanente per la risposta del modello.
Più esempi mettiamo meno spazio rimane per i ragionamenti complessi.

Gestire questa economia è il cuore dell'ingegneria del prompt
La lunghezza influenza non solo lo spazio ma anche la qualità del pensiero dell'ai

Spazio e Complessità
- Token Limit: se il prompt accede la finestra di contesto, il modello inizierà a dimenticare le istruzioni iniziali o gli esempi.
- Attenzione Quadratica: più lungo è il prompt, più lenta e costosa diventa l'elaborazione dei pesi di attenzione
- Truncation Strategy: decidere quali parti del contesto eliminare per far spazio alla generazione dell'output senza far in modo che sia l'ai che decida di tagliere le parti iniziali

La gestione efficiente della context window è la sfida ingegneristica principale per i prompt di grandi dimensioni.

Ottimizzazione del Volume
Fortunatamente la ricerca ci sta dando strumenti per ottimizzare questo numero
Usiamo inanzitutto la prompt compression: rimuovere parole non essenziali e utilizzare una tokenizzazione efficiente per massimizzare l'informazione per token.
Esistono modelli a contesto lungo finestra da 128k a 1M di token, ma hanno un difetto: l'attenzione tende a diluirsi nelle parti centrali (lost in the middle). Se il prompt è troppo lungo il modello tende a dimenticari le parti centrali
Le API moderne permettono di riutilizzare il contesto degli esempi few-show per risparmiare tempo e calcolo tra chiamate diverse.

Lost in the Middle
La topologia dell'attenzione
Le ricerche dimostrano che gli LLM ricordano meglio le informazioni poste all'inizio (le istruzioni) e alla fine (la query attuale) del prompt.
Posizionare gli esempi più importanti o l'istruzione finale nelle zone ad alta attenzione è una strategia di prompting avanzata.
Conoscere la topologia dell'attenzione, permette di posizionare le informazioni dove il modello è più sveglio.
