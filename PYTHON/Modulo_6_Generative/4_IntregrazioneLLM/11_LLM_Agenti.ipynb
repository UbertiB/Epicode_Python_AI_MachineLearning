LLM COME MOTORE DI REASONING (AGENTI)

Finora abbiamo trattato gli LLM come dei completatori di parole, dei geni della statistica capaci di prevedere la parola successiva.
Ma cosa succede se smettiamo di chiedere loro di scrivere e chiediamo a loro di pensare.

Come trasformare un modello linguistico in un agente
Un entità capace di ragionare, pianificare, e agire nel mondo reale, per risolvere problemi che non potrebbe risolvere con la sua sola memoria.
E' il passaggio dal sapere al saper fare

Dobbiamo prima insegnare all'ai come riflettere

3 pilastri fondamentali dell'intelligenza operativa:
- Chain of Thought (CoT): scomporre problemi complessi in passaggi logici. tecnica che permette alle macchine di pensare a voce alta.
- Framework ReAct (Reason + Act): unisce il ragionamento all'azione in  un ciclo continuo
- Tool Use: come dare all'AI le braccia per usare calcolatrici, motori di ricerca, 
API, ecc.

LLm normale:
inpu -> llm -> output
Agente:
Input -> llm -> decisione -> tool -> risultato
Quindi un agente non è un llm più intelligente. E' un sistema software nel quale l'LLM viene usato come motore decisionale, mentre altri componenti eseguono concretamente le azioni.
Il framework ReAct ha fornalizzato proprio l'idea di alternare decisione e azione, usando i risultati delle azioni successive per aggiornare il piano.

Esempio:
Chiedi a Qwen:
"quanto fa 127x583?"
Un LLM può cercare di produrre direttamente: 74041 ma potrebbe sbagliare.
Un agente può decidere:
- Devo fare un calcolo
- uso Calculator
- 127x583
- 74041
- Rispondo
Il ruole dell'LLM non è quindi necessariamente fare il calcolo
E' capire: "per questa richiesta quale strumento devo utilizzare?"
Toolformer, per esempio, ha studiato proprio modelli capaci di imparare quando chiamare strumenti esterni, quali argomenti utilizzare e come incorporare il risultato nella generazione successiva.

Cos'è quindi un Agente?
Un agente è un sistema nel quale un modello riceve un obbiettivo, osserva lo stato corrente, sceglie un'azione tra quelle disponibili, esegue eventualmente uno strumento, osserva il risultato e decide cosa fare successivamente.
Quindi:
obbiettivo -> osservazione -> llm -> decisione -> azione -> risultato -> nuova osservazione -> llm -> ...
Fino a quando non raggiungiamo l'obbiettivo, oppure errore, oppure serve approvazione umana.

"LLM come motore di reasoning" non significa necessarriamente "LLM ragiona esattamente come una persona"
Operativamente significa che il modello viene usato per attività come:
- interpretare l'obbiettivo - scomporre il problema - scegliere la prossima azione -  scegliere uno strumento - preparare i parametri - interpretare il risultato - decidere se continuare
Per esempio:
"controlla questo ordine cliente"
LLm può determinare:
- devo capire chi è il cliente - devo estrarre le righe - devo verificare gli articoli - devo controllare i prezzi - devo verificare eventuali anomalie - devo preparare il risultato

Non significa che il modello debba eseguire personalmente ognuna di queste operazioni.

Supponiamo che l'agente disponga di:
- get_customer() - get_item() - get_customer_price() - get_stock() - get_open_orders() - calculate_total() - creare_order_draft() - send_for_validation()
LLM deve conoscere il nome dello strumento, la descrizione, i parametri richiesti, il tipo di risultato.

L'agente riceve:
"il cliente A ordina 500 pezzi dell'articolo ABC come l'ultima volta"
identifico il cliente, identifico l'articolo, 'come l'ultima volta' richiede lo storico dei prezzi, chiamo get_customer_prica()
Tool get_customer_price(A, ABC, 500, 28/09/2026)
Ora LLM continua usanto il dato recuperato, invece di inventarlo.

Possiamo pertanto utilizzare Qweb, LoRA, RAG, SQL, Pydantic, l'agente li può coordinare.

L'agente decide che per sapere il prezzo deve usare SQL, per capire una procedura tecnica, deve usare RAG documentale, per calcolare il totale deve usare codice Python o una funzione, per classificare l'email deve usare LMM, per creare l'oridne, deve usare API ERP

LoRA specializza il comportamento di un LLM generico, esempio specializza il comportamente di Qwen

Perchè l'AI a volte sbaglia calcoli elementari?
Perchè tenta di invovinare la risposta in un colpo solo, senza un ragionamento o una strategia.
Chain of Throught (CoT)
Il ragionamento sequenziale è una proprietà emergente dei modelli di grande scala, che permette di superare i limiti della predizione diretta del token finale.
Attraverso la Chaing of Thought, forziamo il modello a generare una traccia esplicita della logica, prima di arrivare al risultato, migliorando drasticamnte la precisione su compiti aritmetici e simbolici.

Vediamo matematicamente come questa memoria di lavoro testuale cambi l'output
In termini probabilistici, introduce una variabile latente linguistica z, invece di mappare direttamente l'input x all'output y, il modello genera una sequenza di passaggi logici.
La scomposizoine permette di identificare i sotto-problemi necessari per arrivare alla soluzione finale di un quesito complesso.
I passaggi latenti generati nel prompt agiscono come una memoria di lavoro per il modello.
La capacità del modello di rilevare incoerenze nei passiggi precedenti durante la generazione a catena, è detto Self Correction.
L'output finale è condizionato non solo dall'input originale, ma dall'intera sequenza di ragionamento prodotta precedentemente.

Esistono due modi principali per attivare questa capacità:
1) Zero-shot Cot: attivazione del ragionamento tramite l'aggiunta della frase "pensiamo passo dopo passo" nel prompt
2) Few-shot CoT: fornire esempi che mostrano esplicitamente come risolvere un problema tramite passaggi logici intermedi (esempi di problema, ragionamento, soluzione)
Error Compounding: riconoscere che un errore in un passaggio iniziale della catena porta inevitabilmente a una soluzione finale errata.

Ma il solo pensiero non basta se il modello non ha accesso a dati freschi.
Il CoT non è per tutti
Diventa un arma potente solo quando superiamo una cerca scale, tipicamente sopra i 10 miliardi di parametri.
Nei modelli piccoli, forzare il ragionamento può essere controproducente, la rete non ha abbastanza coerenza interna e finisce per inventare logiche assurde aumentando le allucinazioni.
Il ragionamento è un lusso che solo le reti più grandi possono permettersi

E se volessimo che questo ragionamento portasse ad un'azione concreta nel mondo reale?

Il Framework ReAct
Reason + Act: l'unione tra logica e interazione
Mentre il CoT è un processo puramente interno, ReAct permette al modello di interagire con l'ambiente esterno per recuperare informazioni mancanti (è un dialogo con il mondo).
Il modello alterna una fase di ragionamento (Thought) a una di azione (Action), osservando poi il risultato (Observation) per aggiornare il proprio piano.
Immagina di dover organizzare un viaggio, non puoi farlo solo pensando a dove prenotare i voli, guardare il meteo e prenotare l'hotel, bisogna agire concretamente.
ReAct permette di alternare momenti di riflessione a momenti di azione esterna.
Esempio: non so che ore sono, userò lo strumento orologio
Questa è l'unione tra la logica del cervello e l'interazione dei sensi.

Il Ciclo dell'Agente
il ciclo dell'agente si divide in 3 fasi
Pensiero, Azione, Osservazione
1) Thought: il modello scrive un breve paragrafo per decidere quale sia il passo successivo più logico. Il modello decide cosa fare, per rispondere alla domanda devo prima torvare l'altezza del monumento
2) Action: la scelta di uno strumento specifico, come una ricerca web o un calcolo matematico. Il modello scrive un comando speciale per usare un tools
3) Observation: il feedback reale ricevuto dallo strumento che viene pre-iniettato nel contesto del prompt. L'ambiente risponde, il monumento è alto 300 mt.
Lo stato dell'agente si aggiorna con questa nuova informazione reale
Questo loop permette all'agente di correggere i propri errori basandosi su dati oggettivi e non solo sulla propria memoria statistica.

Questa architettura conferisce agli agente capacità incredibili.
Possono fare pianificazione dinamica, se l'azione di cercare fallisce, il pensiere successivo da un altra soluzione
Possono recuperare informazioni che non esistevano quando sono stati addestarti, superando il problema quando i lmodello non può imparare eventi che sono avvenuti dopo il suo addestramento
Il limite qui è solo il context management, ogni ciclo pensiere, azione osservazione, riempie la memoria del modello, dobbiamo essere bravi a riassumere le scoperte fatte.

Ma come fa un modello a parlare con una calcolatrice o con un database?

Decision Making
Mappare lo spazio delle azioni
Il modello deve imparare a mappare le intenzioni espresse in linguaggio naturale su una sintassi di comando specifica. Capacità di mappare un intenzione su uno spazio di azioni discrete.
L'agente agisce come un controllore logico che arbitra tra la propria conoscenza interna e le risorse computazionali esterne.

Non è un comando rigido, come in un software tradizionale.

Entriamo nel dettaglio di queste braccia dell'ai

Tool Use e Function Calling
Estendere le braccia dell'ai
Un LLM non può eseguire codice Python o consultare un database direttamente; ha bisogno di un'interfaccia di comunicazione strutturata.
Il Tool Use (o function calling) permette di definire una lista di funzioni disponibili che il modello può invocare fornendo i parametri necessari in formato JSON
Il modello non esegue la funzione, ma è incaricato di scrivere i parametri giusti per farla girare, quello lo facciamo noi, come se l'ai fosse l ingegnere che progetta il pezzo e poi lo gira all'operaio.

Vediamo come definiamo questo contratto tra linguaggio e codice

Usiamo definizioni in formato JSON, diciamo al modello:
- Function Definition: forniamo al modello lo schema (nome, descrizione, parametri) degli strumenti a sua disposizione
- Parameter Extraction: la capacità del modello di estrarre dal testo dell'utente i valori corretti per popolare la chiamata alla funzione (valorizzando i parametri)
- Execution Sandbox: l'ambiente protetto dove il codice o API vengono eseguiti prima di restituire il risultato al modello
Il modello funge da interfaccia semantica che traduce un'intenzione umana in una serie di parametri tecnici validi.

Workflow Operativo deve poi essere sicuro, eseguiamo il codice in una execution sandbox, una zona dove l'ai non può fare danni al sistema (esempio non può cancellare il databse).
Se poi va in errore passiamo il codice all'ai, l'agente legge il codice, ragione su ciò che è andato male e rigenere il codice chiamante la funzione risolutiva.

Precisione dei parametri
Il modello utilizza la propria comprensione del mondo per decidere quale strumento sia più adatto alla richiesta corrente.
Il modello confronta la richiesta dell'utente con le descrizioni testuali dei vari tools. Se abbiamo descritto bene una funzione, la rete riconoscere la similitudine semantica con la domanda dell'utente.
Descrivere gli strumenti all'ai, è diventata una compentenza fondamentale.
Più la descrizione è accurata più l'agente sarà preciso.
Questo processo di selezione si basa sulla massimizzazione della similitudine semantica tra il compito e la descrizione dello strumento.
