PROMPT ENGINEERING: STRUTTURA BASE

Il Prompt Engineering è imparare a guidare i Transformer, per estrarre il massimo del valore da modelli che hanno miliardi di parametri.

Il Prompt Engineering è il modo in cui formuliamo istruzioni chiare e strutturate per ottenere da un LM una risposta il più possibile utile, coerente e prevedibile.
prompt poco chiaro  -> risposta più variabile
prompt ben strutturato -> risposta più controllabile

Un prompt non è solo 'una domanda' può contenere il ruole, l'obbiettivo, contesto, dati, regole, formato output.

- Anatomia del prompt: istruzione, contesto e dati di input
- Prompting pattern: tecnicìhe di persona e template e delimitatori
- Mitigazione delle allucinazioni, imparando le strategie per tenere il modello ancorato alla realtà dei fatti.

Anatomia del Prompt
Un prompt efficace non è una semplice domanda, ma una struttura informativa complessa che guida il modello verso lo spazio latente desiderato.
Non stiamo facendo solo una domanda, ma stiamo cercando di dominare una zona specifica dello spazio latente del modello. Dobbiamo fornire coordinate precise affinchè LLM non si perda nelle infinite varietà

Scomporre il prompt in istruzione (cosa deve fare il modello), contesto (informazioni di supporto per delimitare il campo di gioco), dati in input (materiale grezzo su cuiil modello deve lavorare) e indicatori di output (come vogliamo sia l'output) permette di massimizzare la precisoine della risposta del modello.

Ricordare sempre che il modello sta calcolando una probabilità condizoinata. Ogni ingrediente si aggiunge, stringe il cerchio attorno alla risposta corretta.

RUOLO:
Il ruolo serve a orientare il tipo di risposta.
puoi dire: "agisci come un esperto di logistica e pianificazione della produzione"
oppure "sei un assistente specializzato nell'analisi degli ordini clienti"
Il roulo non rende il modello magicamente esperto, serve soprattutto, a impostare il contesto e il comportamento atteso

CONTESTO:
Qui spieghi la situazione. Aiuta il modello ad interpretare correttamente ciò che leggerà

COMPITO:
Probabilmente la parte più importante. Il compito deve essere specifico, bisogna dire chiaramente cosa deve fare il modello, evitando istruzioni ambigue.
Esempio male:
"guarda questa mail"
meglio
"classifica questa mail in una sola delle categorie indicate, poi estrai i dati dell'ordine"

INPUT

VINCOLI

Esempio: non inventare dati mancanti, noon calcolare il prezzo se non esplicitamente indicato, ecc

Come usare pattern collaudati
Schemi ricorsivi per risultati prevedibili
I pattern sono modelli di progettazione che standarizzano il modo in cui forniamo istruzioni agli LLM per compiti ripetibili. Sono come gli schemi di gioco di una squadra di calcio. sono strategia pronte all'uso. 
Significa standarizzare la nostra comunicazione per ottenere risultati prevedibili, senza improvvisare ogni volta.
Tecniche come l'assegnazione di una persona o l'uso di delimitatori permettono di isolare semanticamente le varie parti del prompt.
Trasformando il linguaggio naturale in un vero e proprio linguaggio di programmazione semantica.

Vediamo i 3 patterng più potenti che ogni ingeniere deve conoscere.

Techinche Standard.
Persona, Template, Delimitatori.
- Persona Pattern: definire il ruolo del modello, ad esempio comportati come un esperto di programmazione python. Chiedi al modello di mettersi nei panni di un esperto. Questo sposta la distribuzione probabilistica verso termini tecnici e stili previsi.
- Delimitatori: utilizzo di caratteri speciali come triple virgolette o tag XML per separare chiaramente i dati dalle istruzioni. Serve per dire alla rete, quello che c'è qui dentro è solo testo non istruzioni
- Template Pattern: fornire una struttura fissa con segnaposto che il modello deve compilare seguendo regole precise

Questi pattern riducono l'ambiguità del linguaggio naturale, rendendo l'interazione più simile a una chiamata di fuzione programmabile.

Vediamo come questi pattern migliorano l'integrazione con i sistemi software reali

- Isolamento dell'input: separare il testo dell'utente con simboli chiari previene attacchi di prompt injection (ovvero quando i dati dell'utente tentano di prendere il controllo dell'ai) e interpretazioni errate di comandi
- Controllo del formato: definire template di output come JSON o Markdown facilita l'integrazione del modello in pipeline softweare automatizzate.
- Placeholders: l'uso di variabili simboliche nel prompt aiuta il modello a comprendere quali parti del testo sono dinamiche

Ma c'è un modo ancora più potente di guidare la rete, mostrare non solo dire

In-Context Learning
Insegnare attraverso gli esempi
Fornitre esempi nel prompt trasforma lo schema da zero-shot a few-shot, migliorando drasticamente la comprensione del compito
Forniamo 2 o 3 esempi di input ed output desiderato direttamente all'interno del prompt
Il modello non apprende nuovi pesi (non impare nulla di nuovo in modo permanente), ma sfrutta la propria capacità di pattern matching immediato all'interno della finestra di contesto.

Purtroppo però anche il migliore prompt può scontrasi con l'allucinazione

Mitigrazione delle Allucinazioni
Garantire l'affidabilità delle risposte
Le allucinazioni sono il peccato originale dei modelli probabilistici. L'Ai non sta mentendo, sta solo calcolando la parola più probabile.
Le allucinazioni sono risposte in cui il modello genera informazioni false con alta confidenza statistica.
Comprendere perchè avvengono e come strutturare i prompt per prevenirle è vitale per applicazioni critiche.
Succede perchè il modello è addestrato a completare le sequenze, non a verificare i fatti.
Se la sequenza più probabile è un dato falso ma grammaticalmente perfetto, l'ai lo sceglie senza esitare.

E' possibile usare il prompt per tenere il modello ancorato ai fatti.

Strategie Anti-Allucinazione
Ridurre il rumore probabilistico
Chidi al modello di rispondere basandosi esclusivamente sul testo fornito nel contesto (grounding). Obbligo il modello a rispondere usando solo i dati fornito nel contesto.
Istruire il modello a pensare passo dopo passo per verificare la logica interna del ragionamento (chain of thought). 
Fornire al modello la possibiità di rispondere non so se l'informazione non è presente (opzione di incertezza).
Le allucinazioni derivano spesso dalla natura autoregressiva dei modelli che tentano di completare la sequenza più probabile a ogni costo.

Se un modello sbaglia comunque? possiamo insegnarli a correggersi

Verifica della Coerenza
Chiedere al modello di rileggere la propria risposta e identificare eventuali errori fattuali prima di consegnare l'output (self-correction). Rileggi la tua risposta, controlla i fatti rispetto alla prompt originale e correggi i tuoi errori.
E' possibile ridurre la temperatura di campionamento per rendere l'output più deterministico e focalizzato sui token a probabilità più alta (controllo della temperatura). Rendiamo il modello meno creativo forzandolo a scegliere sempre la parola con la probabilità più alta, riducendo lo spazio per le risposte fantasiose
Un'altra strategia è obbligare il modello a citare le fonti, per supportare ogni affermazione generata.

Dobbiamo però sempre ricordare che stiamo parlando con un modello statistico, non con un database.

Il Limite della Verità
Probabilità contro Realtà
Un LLM non possiede una nozione di verità assoluta, ma solo di plausibilità statistica basata sui dati di addestramento.
Il prompt engineering agisce come un vincolo che forza la plausibilità statistica a coincidere con la verità dei dati forniti.
Il prompt agisce come una lente che focalizza la luce, se la lente è sporca o mal posizionata, l'immagine della verità sarà distorta.

Più il contesto è cristallino, più la statistica coinciderà con la realtà.
