Esplorazione dei Parametri di Generazione
1) utilizza la funzione pipeline di hugging face con il modello gpt2
3) genera un testo partendo dalla frase "il futuro dell'intelligenza artificiale è" usando una temperature di 0.2
3) Ripeti la generazione con una temperatura di 1.5 e un top-p di 0.9. Confronta la coerenza logica e la varietà dei due risultati

Suggerimento: osserva come la temperatura alta introduca parole meno comuni ma potenzialmente fuori contesto

```python
from transformers import pipeline, set_seed
import torch

# Seed per rendere riproducibile il sampling
set_seed(42)

# Usa la prima GPU CUDA se disponibile, altrimenti CPU
device = 0 if torch.cuda.is_available() else "cpu"

generator = pipeline(
    "text-generation",
    model="gpt2",
    device=device
)

prompt = "Il futuro dell'intelligenza artificiale è"

# 1. Generazione conservativa
output_safe = generator(
    prompt,
    max_new_tokens=40,
    temperature=0.2,
    do_sample=True,
    truncation=True
)[0]["generated_text"]

# 2. Generazione più creativa
output_creative = generator(
    prompt,
    max_new_tokens=40,
    temperature=1.5,
    top_p=0.9,
    do_sample=True,
    truncation=True
)[0]["generated_text"]

# 3. Intermedio
output_intermedio = generator(
    prompt,
    max_new_tokens=40,
    temperature=0.6,
    top_p=0.9,
    top_k=50,
    do_sample=True,
    truncation=True,
    repetition_penalty=1.15,
    no_repeat_ngram_size=3    
)[0]["generated_text"]

print(f"--- OUTPUT CONSERVATIVO (T=0.2) ---\n{output_safe}\n")
print(f"--- OUTPUT CREATIVO (T=1.5, Top-p=0.9) ---\n{output_creative}")
print(f"--- OUTPUT INTERMEDIO (T=0.6, Top-p=0.9) ---\n{output_intermedio}")
```
