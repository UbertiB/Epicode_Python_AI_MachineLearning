DIFFERENZA TRA LLM E MODELLI CLASSICI NLP

Siamo passati da modelli che leggono parole come etichette a modelli che comprendono il linguaggio
Se NLP classico è un dizionario statico, LLM è un enciclopedia che scrive se stessa mentre la consultate

Un LLM è comunque un modello di NLP. Quando diciamo 'LLM vs NLP classico? stiamo confrontando due generazioni molto diverse di approccio al linguaggio.
Nei modelli NLP classici, in genere costruiamo una pipeline specifica per un compito. Un esempio potrebbe essere classificare un e-mail, potremmo trasformare le parole in numeri con TF-IDF e poi usare Logistic Regression, SVM, Naive Bayes o Random Forest. Il modello impara esattamente quel compito.
Con un LLM invece, lo stesso modello può essere utilizzato per moltissimi compiti diversi semplicemente cambiando le istruzioni.
Non dobbiamo necessariamente bisogno di creare un modello differente per ogni operazione.
La differenza fondamentale è:
NPL classico:
    modello specializzato
    normalmente per un compito specifico
    necessita feature engineering
    TF-IDF, Bow, embeddings semplici
    spesso richiede dataset etichettato specifico
    output spesso classe o numero
    relativamente leggero
    più semplice da controllare
    generalmente economico
    non 'comprende' molto il contesto
LLM:
    modello generalista
    molti compiti diversi
    apprende rappresentazioni linguistiche
    embeddings contestuali profondi
    può lavorare anche zero-shot/few-shot
    può generare testo strutturato o libero
    molto più pesante
    più flessibile ma meno prevedibile
    maggior costo computazionale
    gestisce contesto linguistico complesso

Un punto fondamentale riguarda le rappresentazioni delle parole
Con Bag of Words o TF-IDF, le parole sono trattate in base alla loro presenza frequenza.
esempio
'ordine cliente urgente'
potrebbe diventare
ordine = 0.72
cliente = 0.41
urgente = 0.58
Il significato delle varie parole dipende poco dal contesto.

Con un LLM invece, la rappresentazione della parola dipende dalle parole che la circondano. Il modello comprende il significato differente di una stessa parola in base al suo contesto.
Questo deriva dall'architettura Transformer e soprattutto dal meccanismo Attention

Un altro grande cambiamente è il modo in cui addestriamo il modello

Nel Machine Learning NLP tradizoinale potremmo avere:
10.000 email -> etichettata manualmente -> TF-IDF -> SVM -> classificatore ordini/offerte/reclami
Se domani vogliamo estrarre anche 'codice articolo' e 'quanti' dobbiamo costruire un altro sistema o modificare significativamente quello esistente.

Un LLM, invece, è stato già pre-addestrato su enormi quantità di tesot. Possiede già una conoscenza generale della struttura del linguaggio.
Possiamo dirgli:
"analizza questa mail, restituisci: "tipo documento", "cliente", "numero ordine"

C'è però un errore che spesso si fa: pensare che gli LLM abbiano sostituito completamente l'NLP classico.
Non è così
Per molti problemi NLP classico può essere addirittura migliore.

Supponiamo di avere già 100.000 email già perfettamente etichettate e che dobbiamo soltanto stabilire spam/non spam. Una Logistic Regression o SVM potrebbe essere più veloce, più economico, più semplice, più deterministico di un LLM.
Usare un LLM enorme solo per dire spam/non spam potrebbe essere inutile.

C'è poi una differenza ancora più profonda.
Un modello NLP classico normalmente esegue una funzione: input -> output
Un LLM può invece eseguire una sequenza di ragionamento operativo: email -> capisce il tipo di documento -> estrai i dati -> controlla se mancano informazioni -> produce JSON -> segnala anomalie -> propone un'azione successiva.
Ed  è proprio qui che nascono gli agenti AI

NLP classico e LLM non devono necessariamente essere concorrenti. Possono lavorare insieme 
esempio nlp classico stabile se un'email è un ordine, una richiesta offerta, un reclamo e se debba essere scartata. Con LLm estraggo i dati in modo strutturato generando un JSON

Un'altra differenze è il tipo di addestramento
Nel NLP classico prevale il training, cioè insegnamo esplicitamente un compito al modello.
Con gli LLM abbiamo diverse possibilità:
- Prompting
- Few-shot propting
- RAG
- Fine-tuning

Prompt
Prompting significa semplicemente dare istruzioni

Zero-Shot:
"classifica questa mail come ordine, offerta, reclamo"

Few-Shot significa dare alcuni esempi:
email: "vorrei 500 pezzi..."
risposta: ordine
email: "potete quotarmi ..."
risposta: offerta
email: "il prodotto ricevuto è difettoso..."
risposta: reclamo
Poi gli diamo una nuoav mail.

RAG
con il RAG aggiungiamo conoscenza esterna
LLM + documenti aziendali + database + manuali + listini + erp
Il modello non deve avere tutto memorizzato nei propri pesi

Fine-tuning
adattiamo il comportamento del modello mediante ulteriori esempi

Il NLP classico costruisce generalmente un modello specializzato per un compito; un LLM costruisce una rappresentazione generale del linguaggio che può essere adattata a molti compiti tramite istruzioni, contesto o fine-tuning.

Versatilità Operativa
Un LLm applica la logica appresa dai testi di programmazione per risolvere problemi di logica di linguaggio naturale (trasferimento di conoscenza).
Usare un unico modello per più task riduce la complessità infrastrutturale pur richiedendo hardware più potente per l'inferenza.
Modificare il comportamento di un sistema richiede solo un cambio di prompt invece di una nuova sessione di training di giorni (adattamento rapido)

La Potenza dei Foundation Models
Base comune per infinite vette.
I modelli moderni fungono da fondamenta su cui costruire applicazioni verticali tramite piccoli adattamenti leggeri.
Abbandoniamo il feature engineering (ovvero il lavoro manuale sui dati) per il prompt enginerring
Il passaggio fondamentale è la transisione da "feature engineering" manuale a "prompt engineering" strategico.
L'obbiettivo della rete rimene sempre lo stesso: massimizzare la probabilità del token successivo.

Per essere così bravi questi modelli devono avere una memoria incredibile.

Gestione del Contesto
Sconfiggere la memoria a breve termine.
Il tallone di Achille del NLP classico (come LSTM) erano le frasi lunghe, faticano a ricordare l'inizio di una frase lunga quando arrivano alla fine della frase.
Gli LLM utilizzano il meccanismo di attenzione (cancellando il concetto di scorrimento del tempo) per guardare l'intero documento simultaneamente, annullando il problema della dimenticanza progressiva.

Recurrenza (ricorrenza) vas Attenzione
Dal processamente seriale a quello parallelo.
RNN e LSTM: elaborano una parola alla volta mantenendo uno stato interno che tende a degradare con il tempo. Un passo dopo l'altro, e lente.
Self-Attention: permette a ogni parola di 'pesare' istantaneamente l'importanza di tutte le altre parole nel contesto in un solo passaggio
Context Windows: è la capacità degli LLM di mantenere in memoria migliaia di token per generare risposte coerenti su lunghi testi.
Questa architettura permete di collegare concetti distasnti pagine intere all'interno di un unico ragionamento fluido.

Ma questa potenza ha un costo fisico da considerare
Efficiente Temporale.
l'attenzione globale ci regalo un'efficienza temporale.
Non dobbiamo più aspettare la parola precedente per processare la successiva.
A differenza delle reti ricorrenti, i Transformer possono essere addestrati su hardware massivamente parallelo.
La rete può capire che il soggetto citato a inizio capitolo è lo stesso a cui si riferisce un pronome alla fine (relazioni lontane).
La complessità quadratica dell'attenzione limita ancora la lunghezza massima del testo gestibile senza ottimizzazioni (limiti fisici)

L'attenzione ha una complessità quadratica, raddoppiare la lunghezza del testo quadruplica la complessità del calcolo.

Usiamo i positional encoding, sono segnali matematici, che marchiano ogni parola con la sua posizione precisa.
