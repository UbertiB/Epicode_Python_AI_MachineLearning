L'Universo di CLIP
Contrastive Languade-Image Pre-training
I modelli di visione e quelli di linguaggio, fino a pochi anni fa (2026) erano mondi separati. CLIP ha creato un ponte, insegnando alla rete a collegare un'immagine alla sua descrizione testuale. Invece di usare classi fisse, ha usato iil linguggio naturale preso dal web.
E' come se avessimo dato alla rete milioni di fotografie con le relative didascalie.
CLIP è il ponte semantico che permette di guardare la foto di un tramonto ed ascoltare la frase ... il sole che scende all'orrizzonte... parlano della stessa cose.
L'apprendimento contrastivo permette di mappare testi e immagini in uno spazio latente condiviso, dove la distanza tra vettori riflette la coerenza semantica.

Usiamo un dueal encoder, un encoder per i pixel (per le immagini) ed un encoder per il testo. Addestrando simultaneamente i due encoder su milioni di coppie prese dal web
L'obbiettivo è massimizzare il prodotto scalare tra immagini e testi corrispondenti, minimizzando quello tra coppie errate.
Grazie alla comprensione del testo, il modello può classificare oggetti mai visti durante il training usando semplici prompt (zero-shot transfer).
Questa architettura trasforma la classificazione da un problema di etichette discrete a un problema di ricerca semantica.
Immagina una grande scacchiera dove cerchiamo di far combaciare ogni pezzo con la sua descrizione, l'obbiettivo della rete è massimizzare la simpatia tra l immagine corretta ed il suo testo, allontanando tutto il resto. La funzione softmax crea questa gerarchia di importanza.

Dai Modellli Avversari alla Diffuzione
L'evoluzione della sintesi guidata.
Il prompt testuale viene convertito in un vettore che agisce come guida per il generatore o il processo di denoising.
La Cross-Attenzione è il meccanismo che permette ai pixel di 'prestare attenzione' a specifiche parole del prompt durante la creazione. Ogni pixel che viene creato, alza lo sguardo e legge una parola del prompt, se il prompt dice 'cappello rosso' i pixel della zona superiore sapranno che devono assumere quella tonalità. Non è più una creazione alla cieca, pixel dopo pixel.
Oltre le GAN i modelli di diffusione hanno sostituito le GAN grazie alla maggiore stabilità e alla capacità di gestire prompt testuali complessi.
Il modello impara a invertire il processo di degradazione del rumore seguendo la direzione indicata dal linguaggio.

Questa tecnologia ha trovato la sua massima espressione in architetture come Stable Diffusion

Architetture Moderne
Stable Diffusion: opera in modo intelligente, non lavorano direttamente sui pixel ma nello spazio latente dei VAE. 
Possiamo usare la Classifier-Free Guidance per decidere quanto l'ai debba essere obbediente al nostro testo, è pertanto una tecnica per amplificare l'influenza del prompt, permettendo all'utente di decidere quanto essere federe la testo (prompt).
L'integrazione con gli LLM permette di modificare parti di un'immagine semplicemente chiedendo di 'aggiungere' un cappello (esempio), questo, dato grazie alla multimodaltà, è chiamato Inpainting e Editing.

Il Processo di Reverse Diffusion
Denoising condizionato
Il processo di reverse diffusion è come osservare uno scultore che estrae una statua da un blocco di marmo grezzo. Il blocco iniziale è il rumore casuale, a ogni step il modello stima quanto rumore deve rimuovere per avvicinarsi all'immagine descritta del prompt.
A ogni step temporale, il modello stima il rumore da rimuovere per avvicinarsi a un'immagine reale coerente con il contesto fornito.
Il prompt testuale funge da ancora semantica, impedendo al modello di collassare su immagini casuali prive di significato.
Il prompt spinge il rumore a colassare verso la forma desiderata.
E' la vittoria dell'informazione sul caos.

Ma cosa succede se facciamo il passo inverso, se chiediamo all'ai di raccontarci ciò che vede.

Image Captioning
Dall'immagine alla storia
Qui l'ai smette di essere pittrice e diventa scrittrice.
L'obbiettivo è estrarre la narrazione dai pixel. Non si tratta solo di riconoscere un gatto ma di descrivere un gatto persiano che dorme sul divano.
L'operazione inversa della generazione è la descrizione: estrarre il significato di un'immagine e tradurlo in linguaggio naturale attraverso un LLM.
Un encoder visivo (ViT) può dialogare con un decoder linguistico (GPT) per generare didascalie accurate e contestualizzate.

Qual'è l'architettura che permette questo dialogo tra occhi e bocca

Vision-Language Models
Il dialogo tra Encoder e Decoder
- Encoder Visivo: solitamente un Vision Transformer (ViT) che scompone l'immagine in patch (o piccolo quadrati) e ne estrae le feature globali (tramite vettori).
- Projection Layer: uno strato che allinea le dimensioni delle feature visive (vettori) con quelle degli embedding del linguaggio. Adatta il segnale visivo affinchè LLM possa intrarpretarlo come fossero parole.
- Autoregressive Decoding: l'LLM riceve i token visivi come prefisso e genera la descrizione una parola alla volta.
- Il sistema impara a mappare la struttura spaziale degli oggetti nela struttura sequenziale della sintassi.
E' un dialogo tra un modulo che vede e uno che parla.

Applicazioni Avanzate
- Visual Question Answering: permette all'utente di porre domande sull'immagine, come 'quante persone ci sono nel parco?', unendo visione e ragionamento.
Puoi caricare la foto di un frigor e chiedere, cosa posso cucinare questa sera?
l'ai devi prima vedere e tradurre gli ingredienti, poi ragionare sulle ricette, ed infine rispondere. 
- Spiegabilità: il modello può descrivere il motivo per cui ha classificato un'immagine in un certo modo, rendendo l'AI meno scatola nera.

L'Allineamento Multimodale
Superare il collo di bottiglia
La sfida principale è far capire al decoder che un certo vettore latente corrisponde a un conceettto visivo concreto. Un immagine contiene troppi dettagli, molti dei quali sono inutili per una descrizione testuale.
L'uso di connettori leggeri (come Q-Former) permette di estrarre solo le informazioni visive rilevanti per il compito testuale (agendo come filtri intelligenti) scartando il rumore di fondo.
Questo trasforma un ammasso di pixel in un concetto astratto che LLm può pensare e calcolare.

CLIP è stato addestrato affinche immagini e testi semanticamente coerenti, finiscano vicini nello stesso spazio vettoriale.

```python
import torch
from PIL import Image
import requests
from transformers import CLIPProcessor, CLIPModel

# CONFIGURAZIONE DEL MODELLO
# Carichiamo CLIP (Contrastive Language-Image Pre-training), un modello in grado di 
# mappare testo e immagini in uno spazio vettoriale condiviso.
model_id = "openai/clip-vit-base-patch32"

# CLIPModel: La classe principale che contiene i pesi del modello neurale.
# CLIPProcessor: Gestisce la preparazione dei dati (tokenizzazione testo e pre-processamento immagini).
model = CLIPModel.from_pretrained(model_id)
processor = CLIPProcessor.from_pretrained(model_id)

# PREPARAZIONE DEL DATABASE DI IMMAGINI
# Definiamo una lista di URL che rappresentano le immagini nelle quali vogliamo effettuare la ricerca.
urls = [
    "http://images.cocodataset.org/val2017/000000039769.jpg", # Rappresenta gatti
    "http://images.cocodataset.org/val2017/000000281759.jpg", 
]

# Scarichiamo le immagini tramite requests e le convertiamo in oggetti PIL (Python Imaging Library).
# Questa conversione è necessaria perché il CLIPProcessor accetta oggetti PIL o array NumPy.
images = [Image.open(requests.get(url, stream=True).raw) for url in urls]

# Definizione della query di ricerca testuale.
query_testuale = "una foto accogliente di piccoli felini domestici"

# ESTRAZIONE DELLE CARATTERISTICHE (EMBEDDING)
# Il CLIPProcessor trasforma il testo "grezzo" in token e le immagini in tensori normalizzati.
# return_tensors="pt" indica che vogliamo ottenere tensori PyTorch come output.
inputs = processor(text=[query_testuale], images=images, return_tensors="pt", padding=True)

# Disabilitiamo il calcolo dei gradienti (torch.no_grad) per risparmiare memoria durante l'inferenza.
with torch.no_grad():
    # Passiamo gli input processati al modello CLIP. 
    # L'operatore ** esegue lo scompattamento del dizionario 'inputs' negli argomenti corretti del modello.
    outputs = model(**inputs)
    
    # Il modello produce due vettori (embedding) fondamentali:
    # image_embeds: Rappresentazione numerica delle caratteristiche visive [NumeroImmagini, 512]
    # text_embeds: Rappresentazione numerica del significato semantico del testo [1, 512]
    image_embeds = outputs.image_embeds 
    text_embeds = outputs.text_embeds   

# CALCOLO DELLA SIMILITUDINE SEMANTICA
# Per confrontare i vettori, dobbiamo prima normalizzarli (portare la loro lunghezza a 1).
# Questo permette di utilizzare il prodotto scalare come misura di Cosine Similarity.
image_embeds = image_embeds / image_embeds.norm(p=2, dim=-1, keepdim=True)
text_embeds = text_embeds / text_embeds.norm(p=2, dim=-1, keepdim=True)

# Calcolo dell'affinità tramite prodotto matriciale (@). 
# Moltiplichiamo il vettore di testo per la trasposta del vettore delle immagini (.T).
# Risultato: un punteggio che indica quanto il testo è "vicino" a ciascuna immagine nello spazio semantico.
scores = (text_embeds @ image_embeds.T) * 100 

# OUTPUT DEI RISULTATI
print(f"Query: '{query_testuale}'")
for i, score in enumerate(scores[0]):
    # Mostriamo il punteggio di affinità per ogni immagine caricata.
    print(f"Immagine {i+1} - Score di affinità: {score.item():.2f}")
```
