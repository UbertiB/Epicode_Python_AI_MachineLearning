llm, immagina di non dover più costruire un attrezzo per ogni lavoro, ma di avere a disposizione una materia prima così duttile da potersi trasformare in qualsiasi cosa, un poeta, un programmatore, un trduttore o un analista.
Passiamo dal concetto di AI addestrata per un compito, a quello di AI che apprende la struttura stessa del pensiere attraverso il linguaggio.

Ma come siamo arrivati a queste dimensioni colossali?
La risposta risiede nella fisica delle informazioni

- Scaling Laws: la fisica dietro la potenza computazionale.
- Foundation Models: un unico modello  base per mille applicazioni.

Scaling Laws
La relazione tra parametri, dati e calcolo
L'era dei Large Language Models non è nata solo da nuove intuizioni architettoniche, ma dalla scoperta che la performance segue leggi di potenza prevedibili.
Capire le Scaling Laws significa prevedere quanto un modello diventerà intelligente aumentando il numero di parametri, la mole di dati e la potenza di calcolo utilizzata (leggi di potenza prevedibili).
Oggi siamo in grado di rpevedere quanto un modello diventerà intelligente ancora prima di accendere i server basandoci su: 1) quanti parametri ha 2) quanti dati legge 3) quante energia gli diamo.

La Legge di Chinchilla
Ottimizzare le risorse di addestramento
Per anni abbiamo pensato che bastasse aumentare i parametri per migliorare. In realtà è stato dimostrato che l'intelligenza è un equilibrio. E' come nutrire un atleta, non serve a nulla avere un corpo gigantesco (tanti parametri) se non gli abbastanza calorie (dati in ingresso). Se raddoppio la dimensione del modello, devi raddopiare anche la quantità di testo che gli fai leggere.
La performance di un modello non dipende solo dalla sua dimensione, ma dall'qeuilibrio tra parametri e token di addestramento.
Aumentare il numero di parametri senza aumentare proporzionalmente i dati porta a un rendimento decrescente e spreco di calcolo.
Il calcolo totale per l'addestramento viene approssimato come prodotto tr ai parametri, dati e una costante di costo per operazione.
L'obbiettivo è minimizzare la loss finale distribuendo il budget computazionale in modo ottimale tra la dimensione della rete e il dataset.
Non vince chi ha il modello più grande ma chi ha il modello più equilibrato.
Modello come LLaMA sono stati addestrati su molti più token rispetto ai loro predecessori per massimizzare l'efficienza durante l'inferenza.
Scalare i modelli richiede tecniche di inizializzazione e normalizzazione specifiche per evitare divergenze dal gradiente in virgola mobile.

Se la line a+ dritta, investnedo 100 volte tanto, otteniamo un risulato normalmente migliore.

Tutto questo ci porta al concetto di Foundation Models
L'unico modello per domarli tutti
In passato, ogni task NLP richiedeva un modello specifico.
Oggi i Foundation Model offrono una base universale pre-addestrata su dati massivi.
Questi modelli venono poi adattati a compiti specifici attraverso tecniche di fine-tuning, ma conservando una conoscenza del mondo vasta e trasversale.

Non dobbiamo re-inventare la ruota ogni volta, dobbiamo solo imparare a guidare questo veicolo universale.
Questa versatilità trasforma il modo di programmare.
Un foundation model non impara a memoria, impara la statistica del linguaggio. Impara le strutture grammaticali e semantiche della lingua attraverso l'auto-supervisione su internet.
L'adattabilità permette di usare lo stesso modello per scrivere codice, riassumere testo o tradurre inglese.
Il paradigma passa dalla costruzione di modelli alla progettazione di istruzioni e contesti per modelli esistenti.
L'assenza del modello risiede nella sua capacità di fungere da piattaforma per innumerevoli.
Leggendo l'intero internet ha appreso non solo la grammatica, ma il senso comune, la logica ed anche i bias umani.
Diventa pertanto un esperto di legge o un assistente medico.
il nostro codice non è più quello di scrivere codice, ma progettare istruzionoi che estraggono la conoscenza già presenza

Emergent Properties
Quando la quantità diventa qualità
Uno dei fenomeni più misteriosi del Deep Learning è l'emergere improvviso di capacità non previste nei modelli più piccoli.
Proprietà come il ragionamento logico, la risoluzione di enirgmi o la capacità di tradurre lingue rare appaiono solo sopra certe soglie di parametri.
I modelli pertanto iniziano a fare cose per cui non sono mai stati addestrati.
Non è un miglioarmento graduale è un salto di qualità che assomiglia molto ad un'inuizione.

Il Salto di Capacità
Il salto di capacità rompe ogni linearità
Le proprietà emergenti non sono visibili nei modelli con pochi parametri, rendendo difficile prevedere l'intelligenza finale.
Esempi tipici includono il few-shot learning e la capacità di seguire istruzioni complesse senza addestramento specifico
Questi fenomeni suggeriscono che l'LLM stia costruendo rappresentazioni interne del mondo e della logica, non solo riproducendo testo. Sta costruendo modelli mentali interni di pura logica.
Il passaggio da 7 miliardi a 70 miliardi di parametri spesso sblocca intere categorie di risoluzione di problemi.

Esempi di Emergenza (proprietà emergenti)
Modelli piccoli sbagliano operazioni semplici, mentre quelli grandi applicando regole matematiche generali per risolvere calcoli mai visti.
La capacità di mantente la coerenza in dialoghi lunghi o testi complessi è una proprietà che emerge con la scala
Cogliere sfumature linguistiche sottili come il sarcasmo richiede un'astrazione che i modelli compatti non riescono a raggiungere.

Ma attenzione, essere grandi non significa essere infallibili

Il Limite dell'Emergenza
Allucinazioni e Ragionamento, mantenere sempre uno sguardo critico
Nonostante le proprietà emergenti, gli LLM rimangono motori probabilistici soggetti a errori fattuali e allucinazioni.
Studiare dove l'emergenza fallisce, dove la logica si rompe, è il campo di ricerca più attivo per garantire l'affidabilità dei sitemi futuri.
