Integrazione Sicura

1) Crea un file .env contenente una chiave fittizia 'HF_TOKEN'
2) Scrivi uno script python che utilizzi python-dotenv per leggere la chiave e configurare un'istanza di InferenceClient di Hugging-Face
3) Effettua una chiamata al modello 'gpt2' (o un altro modello leggero) per completare la frase 'L'integrazione di sistemi AI richiede" inserendo degli errori (nella chiave o nel modello)
4) Gestisci eventuali errori con un blocco try-except

Suggerimento: non caricare mai il file .env su GitHub!

```python
import os
from dotenv import load_dotenv
from huggingface_hub import InferenceClient

# 1. Caricamento variabili d'ambiente
script_dir = os.getcwd()

env_path = os.path.join(script_dir, ".env")

load_dotenv(
    dotenv_path=env_path,
    override=True
)

api_token = os.getenv("HF_TOKEN")

if not api_token:
    raise ValueError(
        "Errore: variabile HF_TOKEN mancante nel file .env"
    )


# 2. Configurazione client
model_id = "openai/gpt-oss-120b"

client = InferenceClient(
    provider="auto",
    api_key=api_token
)


def esegui_completamento():

    prompt = "L'integrazione di sistemi IA richiede"

    print(
        f"Tentativo di chiamata API con modello: "
        f"{model_id}..."
    )

    try:

        # 3. Chat Completion
        response = client.chat.completions.create(
            model=model_id,
            messages=[
                {
                    "role": "user",
                    "content": (
                        "Completa brevemente questa frase: "
                        f"{prompt}"
                    )
                }
            ],
            max_tokens=400,
            temperature=0.7
        )

        testo_generato = (
            response.choices[0].message.content
        )

        print("-" * 30)
        print(
            f"Risultato:\n{testo_generato}"
        )
        print("-" * 30)

    except Exception as e:

        print("\n[!] ERRORE RILEVATO")
        print("Tipo:", type(e).__name__)
        print("Dettaglio:", repr(e))


if __name__ == "__main__":
    esegui_completamento()
```
