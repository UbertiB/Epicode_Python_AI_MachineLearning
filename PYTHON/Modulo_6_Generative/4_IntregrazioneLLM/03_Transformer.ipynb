ARCHITETTURA TRANSFORMER: IL MOTORE DEGLI LLM

Il Transformer è stata definita la rivoluzione industriale del linguaggio naturale.

Il Transformer è il motore degli LLM perchè è l'architettura che permette al modello di capire quali parole di un testo sono importanti rispetto alle altre, anche quando sono lontane tra loro.

RNN/LSTM (testo elaborato in sequenza)
parola1 -> parola2 -> parola3 -> ...
Transformer (ogni token confrontato con gli altri token)
parola1 <-> parola2 <-> parola3 <-> ...

Attention
Attention è il concetto più importanto dei Transformer
Esempio: 
partiamo dalla frase: "la banca ha approvato il prestito dopo aver verificato il reddito"
Quando il Transformer leggere la parola "prestito" può attribuire molta imporanza alle parole: "banca", "reddito", "approvato" e poca importanza a tutte le altre meno informative.
Non legge ogni parola isolatamente, costruisce il significato di una parola in funzione del suo contesto.

testo -> tokenizzazione -> token ID -> embedding -> informazione sulla posizione -> transformer blocks -> self-attention -> Feed Forward Network -> ripetuto molte volte -> probabilità token successivo -> generazione

Esempio
partiamo dalla frase: "il cliente invia un ordine"

Il modello non lavora direttamente sulle parole, prima il tokenizer divide il testo in token
potremmo avere questa suddivisione:
il -> 154
cliente ->8241
invia -> 3720
un -> 91
ordine -> 6254
questo sono token ID

Il Transformer poi trasforma i token id in vettori
Qui entra in gioco Embedding layer
6254 (ordine) -> embedding -> [0.21,-0.74,0.13,0.88,...]
Quindi ogni token diventa un vettore composta da centinaia o migliaia di numeri 
Ogni vettore (che rappresenta un token che rappresenta una parola e parte di parola) rappresenta caratteristiche apprese dal modello
L'embedding iniziale non è ancora la rappresentazione contestuale finale.
La vera trasformazione avviene pasasndo attraverso i blocchi del transformer.

C'è però un problema
Il Transfrmer elabora i token parallelamente, quindi non saprebbe distinguere tra "il cane morde l'uomo" e "l'uomo morde il cane"
Gli stessi token sono presenti, ma l'ordine cambia il significato.
Per questo bisogna aggiungere informazione sulla posizone
Embeddiing parola + Embedding posizione = rappresentazione iniziale
Il -> posizione1
cliente -> posizione2
invia -> posizione3
un -> posizione4
ordine -> posizione5

I Transformer possono usare tecniche diverse per rappresentare le posizioni, per esempio il Rotary Positional Embeddings, molto comune negli LLM moderni (anno 2026)

Ora una parte importante: Self Attention
La Self Attention permette a ogni parola in una sequenza di interagire con tutte le altre, decidendo autonomamente a quali prestare maggiore importanza
Come la mente, filtra il rumore e si focalizza solo sulle parole importanti. Permette ad ogni parola in una frase di guardare tutte le altre e decidere. Il contesto è sempre tutto disponibile.
Dalla frase: il cliente invia un ordine
Quando il transformer analizza la parola "ordine" non tratta tutte le parole allo stesso modo
Potrebbe assegnare maggiore attenzione a: "cliente" e "invia"
e meno attenzione a: "il", "un"
Quindi cerca una rappresentazione contestuale della parole "ordine"

Ma come fa, matematicamente, una parola a cercare la sue compagne?
Il cuore matematico della Self-Attention utilizza 3 mettori per ogni token:
Q=Query
K=Key
V=Value
Immagina che ad ogni parola:
- Q Query: "quali altre parole sono importanti per capire me?" - "cosa sto cercando", rappresenta la parola corrente che cerca informazioni rilevanti nel resto della frase 
K Key: "che tipo di informazione possiedo" - "cosa posso offrire". funge da etichetta o indici per tutte le altre parole presentin nella sequenza.
V Value: "questa è l'informazione che posso fornire" - "informazione effettiva". contiene il contenuto informativo che viene estratto una volta stabilita la pertinenza tra Query e Key
Il modello confronta Query x Key per capire quanto due token siano collegati.
Il prodotto scalare tra Query e Key determina lo score di attenzione, che viene poi normalizzato tramite una funzione Softmax
Immagina una biblioteca, la query è il foglio con scritto cosa stai cercando, la key è l'etichetta posta su ogni libro nello scaffale ed il modello calcola quanto la query assomiglia alla key 
Esempio
frase: il cliente invia un ordine
quando elaboro la parola ordine
ordine -> cliente 0.3
ordine -> invia 0.25
ordine -> il 0.02
ordine -> un 0.03
Questa è la nuova rappresentazione di "ordine" diventa una combinazione ponderata delle informazioni contenute negli altri token.

Multi-Head Attention
non usiamo solo una attenzione, ne usiamo diverse in parallelo per permettere al modello di focalizzarsi contemporaneamente su aspetti sintattici e semantici diversi.
Multi-Head Attentione = più modi contemporanei di interprestasre il contesto.
Una testa di attenzione potrebbe specializzari nel capire la grammatica, un altra nel risolvere nomi propri, un altra nel cogliere il tono emotivo.

L'unico neo la Complessità quadratica. Il calcolo dell'attenzione cresce con il quadrato della lunghezza della sequenza, rappresentando il limite principale per contesti molto lunghi.

L'output della Self-Attention è che l'embedding di una parola non è più una statua di marmo immobile. lo score di attenzione trasforma un vettore di una parola, in un vettore specifico di quella parola per il contesto in questione.
E' una somma pesata, prendiamo un pezzetto di significato di tutte le parole vicine per arrichire la parola corrente, in questo modo il paragrafo non è più una lista di termini, ma una rete densa di relazioni semantiche che vibrano all'unisono.

Il Transformer originale del 2017 aveva un Encoder per leggere l'input ed un Decoder per scrivere l'output.
tutta via, con il tempo, abbiamo capito che potevamo separare questi due mondi.
Se vuoi capire cosa dice il testo utilizza solo metà dell'architettura, se vuoi che l'ai scriva una poesia, usa l'altra metà dell'architettura.
La ricerca ha portato a varianti ottimizzate per task specifici.
Comprendere queste differenze è importante per scegliere il modello giusto tra comprensione del testo e generazione creativa.
Questa biforcazione  ha creato le due più grande famiglie di modelli che dominano il mercato oggi.

Chi sono i protagonisti che queste famiglie?

Abbiamo qundi 3 filosofie
- BERT è l'Encoder-only: è un lettore, guarda avanti e indietro per estrarre il senso profondo, Imbattibile per classificare o estrarre nomi
- GPT è il Decoder-only: è lo scrittore puro. Il futuro, per lui, non esiste ancora, vede solo quello che ha scritto e cerca di indovinare la parola successiva. Ottimizzato per la generazione di testo prevedendo un token alla volta in modo auto-regressivo
- T5/BART Encoder-Decoder: è il traduttore perfetto, legge in una lingua e scrive in un altra. legge un lungo testo e lo riassume in poche righe. Eccellente dove un input deve essere trasformato in un output diverso.

Ognuno ha il suo carattere archiettonico
Internamente la differenza sta nel modo in cui mascheriamo l'informazione

Encoder è bidirezionale, è come uno studente che legge l'intera pagina prima di rispondere. guarda l'intera frase contemporaneamente permettendo una comprensione olista che i modelli unidirezionali non hanno

Il Decoder usa il Causal Masking, una maschera triangolare che gli impedisce di sbirciare le parole successive, assicura che la predizione della parole corrente dipenda solo dalle precedenti.

Nei modelli ibridi esiste la cross-Attention, il decoder consulta l'output dell'encoder (consulta gli appunti di chi ha letto) per mantenere la coerenza con il testo sorgente durante la generazione, assicurandosi di non andare fuori traccia rispetto al testo originale.

Tutto questo funziona perchè il Transformer è una struttura modulare.

Modularità del Transformer
Blocchi ripetibili e scalalbilità
Ogni blocco del Transformer contiene layer di attenzione seguiti da reti Feed-Forward e connessioni residue con layer normalization.
Questa struttura modulare permette di scalare i modelli aggiungendo layer, arrivando alle centinaia di miliardi di parametri dei modelli moderni.
Vuoi un modello più intelligente? aggiungi più layer.

Ma c'è un problema, come facciamo a sapere qual'è il primo mattone e quale l'ultimo?

Positionale Encoding
Restituire l'ordine al caos
Poichè il Transformer non ha ciclo o convoluzioni (e vede tutto il blocco insieme), è intrinsecamente indifferente all'ordine delle parol (bag of words).
La frase, "il cane morde l'uomo" e "l'uomo morde il cane" sono identiche (sono solo una sacchetto di parole).
Per insegnare alla rete che la posizione di una parola conta (senza tornare alla lentezza delle reti riccorrenti), dobbiamo iniettare informazioni geometriche direttamente negli embedding.

Seno e Coseno
Codifica trigonometrica della posizione
Uun timbro speciali basato sulla posizione
Per la prima parola usiamo un onda velocce, per la seconda una leggermente più lenta e così via. Ogni posizione della sequenza riceve una firma numerica unica e irripetibile.
Queste onde permettono di capire la distanza relativa.
- Utilizziamo funzioni sinusoidali a diverse frequenze per creare un vettore unico per ogni posizione nella sequenza.
- Questi vettori di posizione vengono sommati agli embedding delle parole prima di entrare nel primo layer dle Transformer.
- La natura periodica delle funzioni permette al modello di generalizzare anche su lunghezze di testo mai viste durante il training.
- Il positional encoding garantisce che la distanza relativva tra due parole possa essere appresa facilmente dai layer di attenzione.

Per ogni tecnica esistono diverse varianti di Posizionamento

Varianti di Posizionamento
- Learned Encodings: invece di funzioni fisse, la rete impara da sola i vettori di posizione durante il training.
- Relative Positioning: tecniche più moderne (come RoPE o ALiBi) codificano la distanza tra parole invece della loro posizione assoluta nel testo.

Invarianza e Ordine: senza questa componente, la frase 'il gatto mangia il topo' sarebbe identica a 'il topo mangia il gatto' per la rete.
E'ciò che permette alla rete di distinguere tra soggetto ed oggetto, rendendo il linguaggio una struttura ordinata.

Come questa informazione (di posizione) si fonde con il significato delle parole
Sinergia tra Spazio e Significato
La somma tra embedding (vettore di significato) e encoding (cooridnata temporale) non distrugge l'informazione semantica, ma aggiunge una coordinata spazio-temporale utilizzabile.

Questa intuizione è ciò che ha permesso di abbandonare definitivamente la lentezza delle rete neurali ricorrenti (RNN)
