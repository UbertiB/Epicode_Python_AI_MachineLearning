CREAZIONE API FASTAPI PER AI (SCALABILE)

Con Flask ci permette di accedere ad un servizio AI, ma se volessimo costruire un sistema capace di gestire migliaia di utenti simultanei, qui entra in gioco Fast API.
Immagina di passare da una piccola bottega artigianale, ad una febbria automatizzata ad alta precisione.

FastAPI permette, ad un programma, di inviare dati a Python e ricevere una risposta.
- Il programma client invia una richeista, per esempio un POST con un JSON
- FastAPI riceve e controlla i dati
- La funzione python personalizzata esegue il modello AI
- FastAPI restituisce il risultato.
Il vantaggio è quello di poter dichiarare quali dati accettare.
Esempio: la richiesta deve contenere un campo 'features' con esattamente 4 numeri. 
FastAPI usa la libreria Pydantic per verificare questa struttura prima di chiamare la funzione python personalizzata.
La validazione dei dati di FastAPI permette di rifiutare le richieste che non rispettano le regole del chiamante.
Le risposte, in formato JSON, restituiscono dati utilizzabili da altri programmi.
FastAPI, con la gestione delle richieste concorrenti, aiuta a organizzare richieste provenienti da più client.

- Async/Await: gestione nativa della conceorrenza
- Pydantic: validazione automatica, veriifca ogni singolo bit in entrata
- Swagger UI: strumento che genera documentazione mentre si scrive il codice.

Come possiamo rendere Python davvero veloce?
Mentra Flask (sistemi sincroni) gestisce le richieste in modo sincrono, FastAPI sfrutta le moderne capacità di Python per gestire migliaia di connessioni simultanee (concorrenza nativa)  senza attendere il completamento seriale di ogni opearzione.

Come come fa Python a gestire tutte queste operazioni senza fare consufusione
Il segreto è l'evento Loop
Definiamo le nostre funzioni come Coroutine (usando la parola chiave async), quando incontriamo un'operazione lenta, usiamo await, come se dicessimo al server: mentre aspetti che questa matrice venga scaricata, vai pure avanti. Sono pertanto operazioni che possono essere sospese per permettere l'esecuzione di altri compiti (Async Def)
Await è la parola chiave per sospendere la coroutine corrente in attesa di un'operazione non bloccante, da utilizzarsi pertanto con un'operazione lenta.
E' come dire al server: mentre aspetti che questa funzione termina, vai pure avanti con altro.
FastAPI eccelle nella concorrenza I/O, permettendo alla CPU di cambiare contesto invece di restare inattiva.
L'architettura asincrona trasforma il server in un sistema reattivo capace di servire più richieste AI con la stessa impronta di memoria.
La produttività totale aumenta non tanto perchè la CPU è più potente, ma perchè la latenza media non blocca più l'intero sistema.
Stiamo trasformando il server in un organismo reattivo.

Ma per far girare questo motore asincrono, abbiamo bisogno di un carburante speciale.
Ecosistema ASGI
Se Flask usa WSGI FastAPI usa ASGI, che è lo standard per le comunicazioni asincrone.
Il nostro pilota si chiama Uvicorn, un'application server ultra leggera che permette di gestire migliaia di richieste simultanee, con un impronta di memoria ridottissima.
Questo approccio è una manna dal cielo per i sistemi RAG o pipeline, dove il tempo speso a recuperare documenti supera il tempo di calcolo puro (I/O Bound AI).
Mentra aspettiamo che il database vettoriale trovi il contesto giusto, il server può già iniziare a processare la query per l'utente sucessivo.
FastAPI eredita la potenza di Starlette per la gestione dei websocket e del routing ad alte prestazioni.

In un APP AI tradizionale, un utente con una connessione lenta, sequestra un intero worker del server, in Fast API la rete viene nascosta dall'evento loop e il tempo di risposta medio diventa più stabile, non importa se il traffico fluttua, il sistema è progettato per gestire le attese in modo intelligente, garantendo che la GPU sia saturata da calcoli e non satura da attese innutili.

Ma la velocità è nulla senza il controllo, ed ecco che entra in scena Pydantic

Validazione con Pydantic
Schemi dati robusti per l'input AI
Uno dei problemi critici nelle API di Machine Learning è la ricezione di dati malformati che causano il crash del modello (es. stringhe al posto di float).
FastAPI utilizza Pydantic per definire classi di dati con tipizzazione statica, garantendo che ogni input venga validato prima ancora di toccare la funzione di inferenza.
E'  una libreria che ci permette di definire degli schemi rigidi.
Se il dato che arriva dal web non è perfetto, FastAPI lo blocca al cancello di ingresso, restituendo un errore specifico all'utente, prima ancora che quel dato sporco, arrivi a toccare il modello.

Come definiamo queste regole di frontiera?
Integrità dei Dati e Schemi
- BaseModel: classe fondamentale per definire la struttura del payload JSON richiesto dal modello.
- Type Hinting (nativo di Python): utilizzo della sintassi standard di Python per dichiarare tipi (list, float, int) che Pydantic verificherà a runtime. Vincoli fisici, esempio se un immagine deve avere un valore di luminosità positivo.
- Field Constraints: possibilità di definire limiti numerici (es. ge=0) per prevenire input fisicamente impossibili nel Deep Learning.

La validazione automatica  diventa deterministica. O il dato è coerente con lo schema è viene generato un errore 422, niente più sorprese a metà dell'inferenza.

Vantaggio del Parsing Rigido
Pydantic V2 è stato quasi interamente scritto in Rust, questo significa che è estremamente veloce.
Inoltre Pydantic è estremamente gentile con gli sviluppatori che usano la vostra applicazione, restituendo un JSON dettagliato che spiega quale campo ha fallito e perchè, e non un generico errore di sistema

Possiamo anche inserire esempi nel codice.

Usare questa inferfaccia operativa per simulare dati estremi permette di rilevare comportamente imprevisti del modello.

1. Il client invia un JSON strutturato (esempio le 4 misure del fiore)
2. Fast API legge il JSON
3. Pydantic controlla i dati
4. La funzione creata prepara l'array ed esegue model.predict()
5. FastAPI restituisce il risultato come JSON
La lettura e la validazione del corpo della richiesta, insieme alla documentazione interattiva, sono gestite automaticamente a partire dagli schemi dichiarati

Nell'esempio la richiesta contiene:
{
  "sepal_length": 5.1,
  "sepal_width": 3.5,
  "petal_length": 1.4,
  "petal_width": 0.2
}

POST significa che stai inviando dati da elaborare (dal client). L'operazione richiesta è una classificazione (potrebbe anche essere una scrittura nel database ERP)

Il Client, tramite il browser con la pagina /docs, invia le misure del fiore
il Server, il programma Python con FastAPI e Uvicorn, riceve le misure, esegue il modello e restituisce il risultato.

La pagina /docs, viene generata da FastAPI e permette di provare l'API dal browser

Questa istruzione:
@app.post("/predict_species")
significa: quando arriva una richiesta POST all'indirizzo /predict_species, esegui la funzione scritto sotto

Nell'esempio di questa lezione vediamo SOLO come ricevere dati, come scriverlo non è ancora argomento di lezione.

```python
import asyncio
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel, Field, ConfigDict
from sklearn.datasets import load_iris
from sklearn.linear_model import LogisticRegression
import numpy as np

# --- CONFIGURAZIONE DEL MODELLO DI MACHINE LEARNING ---
# In questa fase prepariamo il modello che verrà utilizzato dall'API.
# Il dataset Iris contiene misurazioni di fiori (lunghezza/larghezza sepali e petali).
iris = load_iris()
X, y = iris.data, iris.target

# Inizializziamo e addestriamo un modello di regressione logistica per la classificazione.
# Questo oggetto (model) sarà condiviso tra le diverse richieste gestite da FastAPI.
model = LogisticRegression(max_iter=200).fit(X, y)

# --- DEFINIZIONE DELLO SCHEMA DEI DATI (PYDANTIC) ---
# La classe IrisSpecs eredita da BaseModel di Pydantic.
# Il suo compito è validare i dati JSON in ingresso prima che raggiungano la logica di predizione.
class IrisSpecs(BaseModel):
    # Ogni campo è definito con un tipo (float) e vincoli di validazione (gt=0: deve essere maggiore di zero).
    # Field permette di aggiungere metadati e regole di validazione automatica.
    sepal_length: float = Field(..., gt=0, description="Lunghezza del sepalo in centimetri")
    sepal_width: float = Field(..., gt=0, description="Larghezza del sepalo in centimetri")
    petal_length: float = Field(..., gt=0, description="Lunghezza del petalo in centimetri")
    petal_width: float = Field(..., gt=0, description="Larghezza del petalo in centimetri")

    # Utilizziamo model_config (ConfigDict) per definire i comportamenti del modello.
    # json_schema_extra fornisce un esempio pratico per la documentazione interattiva.
    model_config = ConfigDict(
        json_schema_extra={
            "example": {
                "sepal_length": 5.1,
                "sepal_width": 3.5,
                "petal_length": 1.4,
                "petal_width": 0.2
            }
        }
    )

# --- INIZIALIZZAZIONE DELL'APPLICAZIONE WEB (FASTAPI) ---
# L'oggetto 'app' è il punto centrale che coordina il routing e la gestione delle richieste.
app = FastAPI(title="Iris Predictor API")

@app.post("/predict_species")
async def predict_species(specs: IrisSpecs):
    """
    Endpoint per la predizione della specie.
    Riceve un oggetto IrisSpecs, esegue una simulazione di asincronia e restituisce il risultato.
    L'interazione avviene tramite il passaggio dell'oggetto 'specs' validato da Pydantic.
    """
    
    # Simuliamo un'operazione asincrona (ad esempio un log su database o una chiamata esterna).
    # L'uso di 'await' garantisce che il server non rimanga bloccato durante l'attesa,
    # permettendo di gestire altre richieste simultaneamente.
    await asyncio.sleep(0.8) 
    
    try:
        # Trasformiamo l'oggetto Pydantic in un array NumPy compatibile con Scikit-Learn.
        # Estraiamo i valori singoli dall'istanza della classe IrisSpecs.
        input_data = np.array([[
            specs.sepal_length, 
            specs.sepal_width, 
            specs.petal_length, 
            specs.petal_width
        ]])
        
        # Eseguiamo l'inferenza utilizzando il modello addestrato all'avvio dell'app.
        # Il metodo predict restituisce la classe predetta (0, 1 o 2).
        prediction = model.predict(input_data)[0]
        
        # Recuperiamo il nome leggibile della specie dal dataset Iris.
        # iris.target_names contiene ['setosa', 'versicolor', 'virginica'].
        species_name = iris.target_names[prediction]
        
        # Restituiamo un dizionario che FastAPI convertirà automaticamente in formato JSON.
        return {
            "status": "successo",
            "prediction": {
                "class_index": int(prediction),
                "species_name": species_name
            },
            "model_version" : "iris-demo-v1",
            "note": "Classificazione effettuata tramite modello di regressione logistica"
        }
        
    except Exception as e:
        # Se si verifica un errore durante il calcolo, solleviamo un'eccezione HTTP 500.
        # Questo assicura che il client riceva una risposta strutturata invece di un crash generico.
        raise HTTPException(status_code=500, detail=f"Errore tecnico durante l'inferenza: {str(e)}")

# Istruzioni per l'esecuzione in locale:
# Digitare nel terminale: uvicorn nome_file:app --reload
```

```python
import uvicorn

# Configuriamo il server usando l'oggetto app
# già creato nel notebook.
config = uvicorn.Config(
    app=app,
    host="127.0.0.1",  # Accessibile dal tuo PC
    port=8081,         # Proviamo una porta diversa dalla 8000
    reload=False,     # Ricaricamento automatico disabilitato
    log_level="info")

# Creiamo e avviamo il server.
server = uvicorn.Server(config)
await server.serve()
```
