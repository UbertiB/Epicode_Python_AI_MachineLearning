INTRODUZIONE ALLE API AI: REST, gRPC

Come facciamo a portare la nostra intelligenza fuori dal nostro pc o dalla nostra rete.

Le API pèermettono ai modelli di parlare con il resto del mondo software

1) Architettura client-server
2) REST full principi: per esporre inferenza in modo universale
3) gRPC: per chi ha bisogno di performance estreme

Perchè separiamo chi chiede, da chi risponde?
Architettura client-server in AI
Separazione delle responsabilità computazionali
L'esecuzione di modelli di Deep Learning richiede risorse hardware che spesso superano le capacità dei dispositivi utente.
L'architettura client-server risolve questo problema.
Il server funge da custode del modello (cervello) e dei pesi, mentre il client invia solo i dati minimi necessari per l'inferenza, ricevendo il risultato elaborato.
Il server ha a disposizione strumenti pesanti e costosi (le GPU).

Questo disaccopiamento non è solo una scelta hardware
I vantaggi di questa divisione sono:
- Astrazione Hardware: l'utente non deve possedere una GPU, tutto il calcolo intensivo è centralizzato nel backend. l'utente può avere anche un vecchio pc
- Sicurezza dei Pesi: la proprietà intelletuale del modello rimane protetta sul server e non viene mai distribuita al client. La proprietà intellettuale rimene sul server.
- Aggiornamenti Centralizzati: è possibile migliorare il modello sul server senza richiedere all'utente di scaricare nuovi file o installare aggiornamenti. 
- L'isolamento permette di scalare il numero di istanze indipendentemente dal numero di interfacce utente attive.

Ma come gestiamo il traffico quando arriviamo a migliaia di ordini contemporaneamente?
Gestione del Carico
Gestire il carico è un arte, usiamo strumenti come API Gateway che è un vigile urgano digitale che smistea le richieste
Un punto di ingresso unico che smista le richieste tra diversi modelli o versioni, gestendo anche autenticazione e logging (API Gateway).
Ogni richiesta di inferenza deve idealmente essere indipendente dalle altre, facilitando la distribuzione su cluster di calcolo (Statelessness).
L'hardware Pooling è l'ottimizzazione dell'uso della GPU tramite code di richieste che massimizzano il parallelismo del chip.

Tutto questo ha un costo, la latenza.
Trasferire tutto sul cloud introduce un ritardo fisico dovuto al trasferimento dei dati.
Questo trade-off deve essere bilanciato in base alla complessità del modello.
La latenza totale è la somma dal tempo di rete e del tempo di inferenza puro sulla GPU del server (tempo di calcolo del modello).
Se il modello è piccolo, probabilmente conviene farlo girare in locale, ma per modelli grandi, il tempo di caricamente in rete è nettamente inferiore al calcolo.

Le REST sono il linguaggio universale del web. E' come l'inglese per il commercio internazionale, tutto lo parlano.
RESTful Principles per AI
Utilizzo standard dei verbi HTTP
REST è lo standard universale per le API web. Utilizzare correttamente i suoi principi rende il servizio AI integrabile con qualsiasi linguaggio o piattaforma moderna.
Mappare le operazioni di Machine Learning sui verbi HTTP permette di creare interfacce intuitive per sviluppatori e sistemi automatici.
Non stiamo inventando un nuovo modo di comunicare, stiamo mappando la complessità del ML sulle azioni standard del protocollo HTTP.

Nel mondo REST usiamo principalmente due strumenti:
1) POST per inferenza. Poichè l'inferenza non è idempotente e richiede spesso payload JSON complessi, si usa sempre il metodo POST
2) GET per Stato: utilizzato per controllare se il server è attivo o per monitorare lo stato di job di training lunghi in background.
L'interfaccia REST è quindi come  un contratto, io ti do questo payload, tu mi restituisci questa funzione del dato
Status Code: Uso di 200 per successo, 429 per Rate Limiting e 503 se la GPU è momentaneamente sovracarica.
L'interfaccia REST agisce come un contratto semantico tra chi fornisce il dato e chi fornisce la predizione.

Ma cosa mettiamo dentro questi messaggi REST
Payload e Headers
Il contenuto è tipicamente in formato JSON, formato standard per scambiare tensori o input testuali, facile da debuggare ma con un overhead di parsing non trascurabile. 
Ma non dimentichiamo gli Headers, qui inseriamo i token di autenticazione, uso di token Bearer per proteggere l'accesso alle risorse computazionali ed evitare abusi economici. Dobbiamo assicurarci che, chi bussa alla nostra porta, abbia il permesso, ed il credito, per farlo.
Usare sempre il versioning dell'urll

REST è fantastico ma ha un tallone d'Achille, la sua natura testuale.
Quando inviamo immagini ad alta risoluzione o video, la conversione in Base64 (per trasformare i pixel in testo) può aumentare le dimensioni dei file del 33%
Per un'applicazione di chat va bene, ma se stai costruendo un sistema di guida autonoma o un sistema di traduzione istantanea, questo overhead diventa critico per applicazioni real-time, spingendo la ricerca verso protocolli più efficienti.

Per superare questo limite dobbiamo guardare a tecnologie nate per la velocità
gRPC è Performance
gRPC è un framework sviluppato da Google che utilizza HTTP/2 e Protocolo Buffers per garantire una velocità di comunicazione superiore a REST. gRPC non usa il testo per comunicare ma i dati binari, questo è ideali per il Deep Learning, dove muoviamo i tensori (ovvero numeri)
In ambito AI, gRPC è ideale per lo streaming di audio, video o il trasferimento massimo di tensori numerici grezzi.
Se REST è una lettera raccomandata.
gRPC è una telefonata su linea dedicata.

In molti sistemi AI il tempo che ci metti ad impachettare e spachettare i dati è superiore al tempo che il modello ci mette a pensare.
Passando a gRPC eliminiamo questo collo di bottiglia.
