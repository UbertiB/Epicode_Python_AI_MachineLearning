OPENAI API: PANORAMICA ENDPOINTS

Le API di OpenAI trasformano l'astrazione del linguaggio in geometrica operativa.

- Chat Completions: come manipolare non solo le parole ma le probabilità stesse con cui il modello risponde.
- Embedding: rappresentazione vettoriale per ricerca e clustering
- Fine-tuning API: specializzare modelli GPT su dataset personalizzati. Fare un lavaggio del cervello ad un modello per insegnarli uno stile specifico

Chat Completions Endpoing
L'evoluzione dell'interfaccia di generazione
L'endopoint 'v1/chat/completions' è il cuore pulsante dell'ecosistema OpenAI. Non serve solo per dialogare, ma agisce come un orchestratore di istruzioni complesse attraverso una struttura a messaggi.
Immaginalo com un direttore di orchestra, non li diamo solo un testo ma una struttura gerarchica
Invece di un singolo blocco di testo, utilizziamo array di oggetti con ruoli definiti: 'system' per direttive che il modello non deve mai dimenticare; 'user' per gli input; e 'assistant' per la cronologia delle risposte (è la memoria storica).
Questo architettura ha permesso di creare agenti che non perdono il filo del discorso agendo come una vera memoria di lavoro distribuita.

Ma come facciamo a controllare la creatività o la precisione di questo direttore di orchiestra?
Spesso si sente parlare di temperatura.
Ma i veri professionisti guardano anche a tools e logic bias.
- Tools e Functions: permette al modello di generare argomenti strutturati in JSON per chiamare funzioni esterne, trasformando l'LLM in un agente operativo. Il modello non si limita a parlare ma agisce, generando JSON pronti per chiamare software esterni.
- Logit Bias: parametro che modifica la probabilità di apparizione di specifici token. Un valore di 100 forza un token, mentre -100 lo eslcude totalmente.
Aggiungiamo un valore b ai logit prima della softmax, forziamo il modello a usare specifiche parole 

Ma cosa accade ai logit prima che diventino parole?
Meccanismi di Steering
Il logit_bias agisce direttamente sui valori pre-softmax del transformer. Questo controllo è fondamentale per applicazioni dove certi termini sono vietati per motivi di sicurezza o branding.
I logit sono come l'energia assegnata a ogni parola del vocabolario prima della normalizzazione probabilistica.
Il nostro bias interviene qui, modificando il valore l, stiamo spostando l'attenzione del trasformatore.

Lasciando la generazione, passiamo alla comprensione

Embeddings Endpoint
Il ponte tra testo e geometria vettoriale.
L'endpoint 'v1/embeddings' è il ponte tra il mondo umano e quello matematico, converte il testo in vettori densi di numeri reali. Questi vettori catturano il significato semantico permettendo confronti matematici tra frasi diverse.
Se chat competions ('v1/chat/completions') è la bocca del modello, gli embedding ('v1/embeddings') sono i suoi occhi.
Trasformano una frase in una lista di numeri. In questo spazio le frasi simili di significato, si trovnao vicine anche se non condividono nemmeno una parola
Modelli come 'text-embedding-3-small' permettono di gestire fino a 8192 token di contesto, producendo vettori di 1536 dimensioni ottimizzati per il recupero dati.

Ma cosa possiamo fare con questi vettori?

Ricerca Semantica e Clustering
Le applicazioni pratiche sui vettori
- RAG (Retrieval-Augmented Generation): utilizzo degli embeddings per trovare i documenti più rilevanti in una base di conoscenza prima di interrogare GPT-4
- Clustering: raggruppare migliaia di documenti simili in modo non supervisionato applicando algoritmi come K-Means sui vettori ottenuti. Es. posso raggruppare le mail per argomento senza aver bisogno di leggerle
- Anomaly Detecion: identificare input o log fuori norma calcolando la distanza vettoriale rispetto alla distribuzione dei dati standard.
- La vicinanza semantica tra due testi si misura tramite il calolo della similitudine tra i loro rispettivi vettori. Quanto due pensieri puntano nella stessa direzione all'interno dello spazio semantico.

Esempi di Utilizzo Vettoriale
- Esempio Ricerca: data una query 'come si cuicina la pasta' otteniamo il suo embedding e lo confrontiamo con gli embeddings di un database di ricette  per trovare il match più vicino.
- Esempio Classificazione: possiamo usare gli embeddings come feature input per un classificatore lineare (es. Logistic Regression) molto più veloce di un LLM intero.
- Visualizzazione: attraverso algoritmi di riduzione dimensionale come t-SNE o UMAP, possiamo proiettare i vettori in 2D per osservare come il modello raggruppa i concetti.

Ma come ottimizziamo lo spazio occupato da questi milioni di numeri?
Normalizzazione e Dimensioni
Ottimizzazione dello storage
OpenAI fornisce embeddings già normalizzati sulla sfera unitaria, semplificando il calcolo della similitudine al solo prodotto scalare (dot product). La forza di questi vettori normalizzati è sempre 1, conta solo l'orientamento.
Nei nuovi modelli è possibile ridurre le dimensioni del vettore (es. da 1536 a 512) tramite il parametro 'dimensions' senza una perdita drastica di performance. Risparmio il 30% sulla dimensione del databse perdnedo meno dell'1% di precisione.
E' come avere una foto che rimane nitida anche se riduco le dimensioni del file.

Se tutto questo non basta, dobbiamo passare alla chirurgia pensate, il fine-tuning.

Fine-Tuning API
Specializzazione per compiti verticali
Mentre il prompt engineering lavora sul contesto, il fine-tuning modifica i parametri del modello (solitamente gpt 3.5-turbo o gpt 4o-mini) per apprendere pattern specifici.
Il fine-tuning non serve per dare nuove informazioni al modello, per quello usiamo le RAG, ma serve per cambiargli la personalità o insegnarli un formato rigoroso.
Questo endpoint è ideale quando abbiamo bisogno di un formato di output estremamente rigido o quando vogliamo istruire il modello su uno stile comunicativo proprietario.
Se vuoi che il modello risponda come un avvocato o se vuoi che emetta solo codice di programmazione, il fine-tuning è la strada, stiamo modificando i pesi interni, non solo il contesto temporaneo.

Ma come prepariamo di dati per questo addestramento.
il formato JSONL
- Struttura a conversazione: ogni riga del file deve contenere un oggetto JSON con il campo 'messages', replicando la struttura del Chat Completions.
- Dataset Quality: sono sufficienti 50-100 esempi di alta qualità per vedere miglioramenti significativi nel comportamento del modello. Qualità batte quantità
- Validation Set: OpenAI permette di caricare un secondo file per monitorare la loss durante l'addestramento ed evitare l'overfitting. il modello impara a minimizzare la distanza tra la sua risposta e quella del dataset maestro.
- Il processo di fine-tuning minimizza la cross-entropy tra le risposte del modello e quelle fornite nel dataset di esempio
E' una forma di apprendimento per immitazione profonda.

Come si presenta una riga di questo dataset:
Non va gestita la GPU o parametri complessi.
 - Esempio Riga JSON: {'messages':[{'role':'system','content':'Sei un bot sarcastico'},{'role':'user','content':'ciao'},{'role':'assistant','content':'Ancora tu?'}]}
 - Hyperparameters: è possibile configurare il numero di epoche, il moltiplicatore del learning rate e la dimensione del batch, anche se i valori di default sono spesso ottimali.
 - Modle Deployment: una volta terminato il job, Open AI genera un ID modello unico (es. 'ft:gpt-3.5-turbo:org:custom-name') utilizzabile immediatamente nell'endpoint chat. E' un modlelo che esiste solo per noi.

 Ma conviene sempre fare fine-tuning?

 Quando usare il Fine-Tuning
 Analisi dei costi e benefici
 Il fine tunint ha un costo iniziale di addestramento, tuttavia puoò far risparmiare in inferenza, perchè se il modello è già istruito non devi mandarmi prompot lunghi ed ogni singola chiamata costa meno. Paghi oggi per addestrare per pagare meno in futuro.
 Ma se i dati cambiano ogni ora, resta sulla RAG
 Il fine-tuning riduce la necessità di prompt lunghi ed esempi 'few-shot' risparmiando token in inferenza a scapito di un costo iniziale di addestramento.
 Tuttavia, non è adatto per inietare nuove conoscenze fattuali (per quello si usa il RAG), ma eccelle nel definire lo stile e il comportamento.

```python
import torch
import json
#1
# Definiamo lo schema per il modello (JSON Schema)
# Questo istruisce l'LLM a rispondere solo con i campi richiesti
ticket_tool = {
    "type": "function",
    "function": {
        "name": "classify_ticket",
        "description": "Estrae categoria e urgenza da un ticket tecnico",
        "parameters": {
            "type": "object",
            "properties": {
                "topic": {"type": "string", "enum": ["hardware", "software", "network"]},
                "priority": {"type": "integer", "description": "Scala 1-5"},
                "is_urgent": {"type": "boolean"}
            },
            "required": ["topic", "priority", "is_urgent"]
        }
    }
}

# Esempio di messaggio di sistema per definire il "carattere" 
system_prompt = "Sei un analista tecnico senior. Rispondi sempre in modo conciso e strutturato."

# 2
def check_similarity(query_vector, historical_vectors):
    """
    Confronta un nuovo ticket con lo storico usando PyTorch.
    query_vector: tensor [1, 1536]
    historical_vectors: tensor [N, 1536]
    """
    # 1. Convertiamo in tensori PyTorch (se non lo sono già)
    q = torch.tensor(query_vector)
    h = torch.tensor(historical_vectors)

    # 2. Calcolo della similitudine cosina (vettorializzato)
    # Finiamo per fare un prodotto scalare su vettori già normalizzati
    similarities = torch.nn.functional.cosine_similarity(q, h)
    
    # 3. Troviamo il match più vicino
    best_match_idx = torch.argmax(similarities)
    return best_match_idx.item(), similarities[best_match_idx].item()

# Esempio d'uso fittizio
current_ticket = [0.12, -0.05, 0.88] # Immaginiamo sia un embedding ridotto
history = [[0.11, -0.04, 0.85], [0.9, 0.1, -0.2]]
idx, score = check_similarity(current_ticket, history)

print(f"Match trovato nel database storico all'indice {idx} con score: {score:.4f}")

```
