```python
from fastapi import FastAPI
from pydantic import BaseModel, Field, FiniteFloat

app = FastAPI(title="API di prova")


# Descrive la struttura dei dati che il client deve inviare.
class Richiesta(BaseModel):
    features: list[FiniteFloat] = Field(
        min_length=4,
        max_length=4,
    )


# Riceve una richiesta POST all'indirizzo /verifica.
@app.post("/verifica")
def verifica(dati: Richiesta):
    return {
        "features_ricevute": dati.features,
        "numero_features": len(dati.features),
    }
```
