CYCLEGAN: TRADUZIONE DI IMMAGINI TRA DOMINI NON ACCOPPIATI

CycleGAN è una GAN pensata per fare image-to-image translation quando non hai coppie perfettamente allineate di immagini.
Il problema che risolve è questo_
Dominio A -> Dominio B
senza bisogno di avere:
immagine A1 <-> esatta immagine B1 corrispondente
Esempio 
cavallo -> zebra
zebre -> cavallo
Non serve avere la foto dello stesso cavallo prima 'normale' e poi trasformato in zebra. Basta avere un insieme di foto di cavalli e un insieme separato di foto di zebre.

Le GAN classiche sono sempre state addestrate come se avessimo a disposizione un insegnante perfetto per ogni immagine A abbiamo una corrispettiva immagine B.
Ma nel mondo reale la natura non è così generosa.

La CycleGAN introduce la capacità di traddurre immagini tra due mondi che non si sono mai incontrati.
Una macchina può imparare a trasformare un cavallo in una zebra senza aver mai visto una foto dello stesso identico animale in entrambe le versioni.

Dobbiamo cambiare il modo di pensare ai dati.

- Dati non accoppiati: superare il limite dei dataset speculari
- Cuore matematico la cycle consistency loss, il vincolo del viaggio di ritorno, che impedisce alla rete di inventare realtà assurde.
- Identity loss: per assicurarci che la rete non cambi ciò che deve restare immuatato (come i colori di base o la struttura dell'orizzonte)

Cosa facciamo quando non abbiamo una mappa punto per punto tra i nostri dati, appartenenti a due mondi diversi?

Dati non Accoppiati
Tradurre senza una mappa punto apunto
Fino ad ora, per tradurre un'immagine da un dominio A a un dominio B, avevamo bisogno di coppie perfette: ad esempio, la stessa foto scattata in estate e in inverno, questo se voglio tradurre una foto estiva in una foto invernale.
CycleGAN risolve il problema fondamentale dell'impossibilità di reperire dati accoppiati per molti task reali, permettendo di apprendere il mapping tra due insiemi di immagini completamente slegati.

Per gestire questa libertà non basta un solo modello, ne servono 4 che lavorano in armonia.

Mapping tra Domini Disgiunti
Imparare lo stile senza la corrispondenza.
- Dominio X ed Y: la rete apprende le caratteristiche collettive di due gruppi di immagini senza esempi individuali corrispondenti.
- Due Generatori: G mappa da X a Y, mentre F mappa da Y a X (percorso inverso) per garantire la bidirezionalità del processo.
- Due Discriminatori: per ogni modulo abbiamo un guardiano, Dx valuta se le immagini generate da F sembrano appartenere al dominio X, Dy fa lo stesso per il dominio Y
- L'obbiettivo è imparare una distribuzione condizionata senza avere una guida esplicita per ogni pixel dell'input.
G: X->Y; F:Y->X
I generatori cercnao di imparare l'essenza del stile avversario
Mentre i Discriminatori affinano il loro gusto per il realismo.
E' un sistema che si autobilancia vicendevolmente
Ma senza un vincolo forte potremmo produrre immagini bellissime ma svincolate dall'input.

Il Probelma dell'Unpaired Data
Con i dati disaccopiati non abbiamo una verità assoluta di riferimento.
Possiamo usare migliaia di foto di cavalli e migliaia foto di zebre prese da fonti diverse per addestrare un traduttore automatico.
Poiche non esiste una risposta corretta unica, il rischio è che la rete inventi immagini realistiche ma totalmente diverse dall'input originale.
Senza coppie di dati, abbiamo bisogno di una funzione di perdita innovativa che forzi il contenuto a rimanere costante mentre lo stile cambia (vincoli strutturali).
E' la sfida della coerenza strutturale

Vediamo come questa capacità di astrazione ci permette di superare il limiti dei modelli precedenti.
Oltre Pix2Pix
Dalla supervisione all'intuizione
Mentre i modelli supervisionati combattevano una battaglia pixel per pixel (Pix2Pix calcola una distanza pixel per pixe), le CycleGAN combattano una battaglia di concetti, deve capire le caratteristiche di alto livello, come la trama delle strice di una zebra. Non ci interessa più che il pixel in posizone 10x10 sia identico, ci interessa che la rete capisca cosè una striscia o come si muove una pennellata.
Questa astrazione permette di applicare la traduzione a contesti artistici o fotografici dove la corrispondenza esatta è fisicamente impossibile da ottenere
Questa indipendenza dai dati accoppiati è ciò che l'ha resa famosa.
L'intuizione dello stile prende il posto della pura memoria matematica

Ma come facciamo a garantire matematicamente che l'intuizione non diventi pura invenzione?

Cycle Consistency Loss
Il concetto di coerenza ciclica
Serve a impedire al generatore di fare trasformazioni completamente arbitrarie.
Senza cycle consistency loss il modello potrebbe imparare: qualsiasi cavallo -> qualsiasi zebra plausibile e perdere completamente l'identità dell'immagine originale
Con il ciclo, invece, deve essere possibile tornare indietro (dalla zebra al cavallo)
input -> trasformazione -> trasformazione inversa -> input originale circa ricostruito.

Usiamo l'analogia della traduzione linguistica.
Se traduciamo una frase dall'italiano all'inglese e poi la ritraduciamo dall'inglese all'italiano,  dovremmo ottenere la frase di partenza (quasi esattamente).
Se la frase finale è diversa, significa che la traduzione originale era imprecisa o ha perso il senso profondo.
La CycleGAN applica questo principio alle immagini. 
E' il patto di fedeltà tra due mondi.
CycleGAN applica questo principio alle immagini: se trasformiamo un cavallo in zebra e poi la zebra in cavallo, l'immagine finale deve essere identica (o quasi) a quella inziale.

La Cycle Consistency Loss ha tra componenti importanti:
1) Adversarial loss: il generatore G deve produrre immagini che sembrino appartenere al dominio B
2) Loss inversa: il generatore F deve produrre immagini credibili appartenenti al dominio A
3) Cycle consistency loss: immagine originale vs immagine ricostruita dopo il cliclo. La distanza viene spesso calcolata con L1
Questa capacità si chiama unpaired image-to-image translation
E' ciò che la distingue da Pix2Pix

Vediamo come si traduce questo concetto in un ciclo bidirezionale di andata e ritorno.

Viaggio di Andata e Ritorno
Forzare la conservazione del contenuto
- Forward Cycle Consistency: trasformare x in G(x) e poi di nuovo in F(F(x)) deve restituire x.  Trasforma x in y e poi di nuovo in x
- Backward Cycle Consistency: trasformare y in F(y) e poi di nuovo in G(F(y)) deve restituire y.
trasfomra y in x e x in y
- Uso della norma L1: si preferisce L1 per minimizzare le differenze pixel-pixel tra l'originale e il ricostruito (scarto tra origine e ricostruito), garantendo nitidezza 
- Questa perdita agisce come un regolarizzatore che impedisce ai generatori di ignorare l'input originale.

Logica della Perdita Ciclica
La perdita ciclica limita drasticamente l'insieme delle funzioni di mapping possibili (riduce lo spazio delle soluzioni possibili), rendendo il training convergente (rendendo il problema trattabile)
Grazie al ciclo, la forma e la posizione degli oggetti nell'immagine originale vengono preservate durante la trasformazione stilistica (mantenimento dell'identità), Con i ciclo, la rete è costratta a trovare un legame biunivoco.
I due generatori lavorano in simbiosi: G impara a creare, F impare a invertire (smonta lo stile di G per ritrovare l'origine), correggendosi a vicenda continuamente

Ma cosa accade effettivamente al sengale informativo durante questo doppio passaggio^

Il Flusso Informativo
Controllare il segnale.
Senza la cycle loss, il generatore potrebbe produrre immagini casuali del dominio target che soddisfano il discriminatore ma perdono il legame con l'input
La ricostruzione ciclica asicura che ogni pixel dell'output sia una funzione deterministica e reversibile dei pixel in ingresso. Se x è cerica uguale al risultato del doppio passaggio, abbiamo creato un collo di bottiglia informativo perfetto.
L'informazione neccessaria necessaria a ricostruire x originale, deve essere nascosta dentro l'immagine di y generata
La zebra deve trasportare l'anima del cavallo per permettere il ritorno all'immagine del cavallo.

Identity Loss
Immagina di chiedere al traduttore di volgere un testo dall'italiano all'italiano. Non dovrrebbe cambiare una virgola, nelle immagini è lo stesso. Se diamo la foto di una zebra, il generatore deve riconoscsre di essere già a destinazione (zembra) e non alterare l'immagine.

Autoconservazione del Dominio
Garantire la federltà dei colori
- Se passimao un'immagine del dominio Y al generatore G (che dovrebbe mappare da X a Y), l'output deve essere l'immagine stessa.
- Questa perdita è fondamentale per task come la traduzione giorno-notte, dove non vogliamo che la notte diventi blu se l'input è già notturno
- L'identity Loss agisce come un vincolo di regolarizzazione che stabilizza ulteriormente il mapping spaziale.
- Aiuta la rete a distinguere tra ciò che deve essere cambiato (lo stile) e ciò che deve restare immutato (la struttura globale.)
Identity Loss vincola il modello alla realtà cromatica del dominio di destinazione.

Applicazioni della Identity Loss
L'identity Loss impedisce alla rete di inventare colori assurdi per oggetti che hanno già un colore naturale (correzione cromatica).
In ambito medico, assicura che le strutture degli organi non vengano deformate durante la trasposizione tra diverse modalità di scansione (preservazione anatomica). Se traduciamo una TAC in una risonanza magnetica l'identity loss garantisce che la struttura degli organi non vengano trasformati o spostati.
La Traduzione Artistica permette di mantenere le palette dell'autore originale quando si applica un filtro stilistico moderno.

Ma come mettiamo insieme tutti questi pezzi diversi all'interno del nostro algoritmo totale.

Equilibrio tra le Loss
Pesare il contributo
In una CycleGAN completa, la funzione di costo finale è una combinazione pesata di GAN loss, Cycle loss e Identy loss.
Tipicamente si assegna un peso molto più alto alla cycle loss rispetto alla GAN loss per dare priorità alla ricostruzione. Questo perchè vogliamo che la coerenza del contenuto è più importante dell'inganno del discriminatore.
Vogliamo una traduzione che sia onesta, non solo visivamente gradevole.

La flessibilità delle CycleGAN ha un prezzo: non 'capisce' semanticamente l'immagine.
Può imparare correlazioni indesiderate
Esempio
dominio A: cavalli quasi sempre in prati
dominio B: zebre quasi sempre nella savana
La rete potrebbe imparare non solo cavallo->zebra ma anche prato->savana, perchè statisticamente questi elementi sono correlati.

Altro limie: la cycle consistency non garantisce che la trasformazione da A a B e poi da B a A sia semanticamente corretta secondo il significato umano.
