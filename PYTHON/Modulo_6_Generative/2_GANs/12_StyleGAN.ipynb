STYLEGAN: LA RIVOLUZIONE DEL CONTROLLO STILISTICO.

Le StyleGAN sono l'evoludzione delle GAN o di PGGAN
Con PGGAN il problema era: "come genero immagini sempre piu grandi senza destabilizzare il training?
Con le StyelGAN diventa: "come posso controllare separatamente quali caratteristiche dell'immagine vengono generate alle diverse scale?"
Presentata da INVIDIA ha introdotto il concetto di controllo stilistico. La macchina non impara solo a copiare ma a scomporre la realtà in livelli indipendenti (tratti somaticici ma perfino la grana delle pelle)

Prima potevamo generare un volto ma non decidere, per esempio, il colore degli occhi, senza cambiare anche la forma del naso.

- Mapping network
- AdaIN inietta lo stile ad ogni risoluzione
- Ruolo del rumore stocastico, la casualità gestisce i dettagli microscopici che rendeno l'immagine vera all'occhio umano

In una GAN normale z= rumore casuale -> Generator -> immagine
z -> posa -> forma del volto -> occhi -> capelli -> colore -> texture -> microdettagli
Ma tutte queste iformazioni possono risultare molto intrecciate.
Nelle GAN il rumore z entra direttamente nel generatore. 

Style gan introduce una mapping network dedicata, è un livello intermedio che permette di controllare molto meglio 'come' viene generata l'immagine. 
Invece di usare il rumore grezzo, la rete lo elabora preventivamente. Il rumore viene trasformato in uno spazio intermedio (omega) capace di separare le caratteristiche visive in modo indipendnete e controllabile.
z casuale -> mappint Network -> w=vettore di stile -> Generator -> Immagine
La vera innovazoine è il vettore w
Il Mapping Network trasforma lo spazio letente originale z in uno spazio intermedio chiamato w
In Z possono essere mischiate informazioni come colore capelli, forma del viso, orientamente, illuminazione ,ecc. Nella GAN tradizionale queste caratteristiche possono essere fortemente intrecciate.
StyelGAN cerca di ottenere uno spazio più caratteristico, se deve modificare una caratteristiche tende a non influenzare le altre.
Es. w1=forma del viso, w2= capelli, w3=colore pelle, w4=illuminazione, w5=dettagli fini.
Non è realmente così netto matematicamente, ma è il comportamente che si cerca.

Nello spazio latente Z le caratteristiche (feature) sono come fili di lana intrecciati (annodate): cambiare il colore dei capelli potrebbe alterare involontariamente la forma del viso.
La Mapping Network (composta di 8 layer densi) proietta Z in uno spazio intermedio W che non è vincolato a seguire la distribuzione di addestramento originale. In questo spazio W questi fili intrecciati sono srotolati e ogni dimensione può rappresentare un concetto isolato.
Lo spazio W permette un controllo semantico superiore, dove ogni dimensione può rappresentare un concetto isolato.
Questa trasformazione non lineare rende lo spazio latente più flessibile e facile da navigare per l'utente
E' come passare da un telecomando che fa tutto, ad un mixer dove ogni cursore controlla un singolo strumento.

Vediamo perchè questa nuova geometria è così preziosa per noi

I Vantaggi dello Spazio W
La Mapping Network modella una distribuzione che riflette meglio la varietà dei dati reali senza essere obbligato a seguire la distribuzione gaussiana iniziale (flessibilità geometrica), che per sua natura tende a schiacciare i dati verso il centro.
W è una spazio ilbero che riflette la vera distribuzione dei dati reali.
W diventa la sorgente centrale che alimenterà ogni singolo blocco della rete di sintesi successiva attraverso segnali specifici.
Senza questo passaggio, l'architettura non potrebbe separare gli attributi globali (come la posa) da quelli locali (come il colore degli occhi) in modo così netto (supporto al controllo).

La rete genera l'immagine progressivamente, passando da caratteristiche grossolane e caratteristiche sempre più fini.
4x4       -> struttura generale del volto
8x8       -> posizione testa/forma
16x16     -> occhi/naso/capelli
64x64     -> texture
128x128   -> dettagli pelle
256x256   -> dettagli fini
512x512
1024x1024
Gli stili possono quindi influenzare livelli differenti


La prova della superiorità di W sta nel modo con cui ci muoviamo al suo interno.

Perchè W è superiore?
Linearità dello spazio dei concetti.
Nello spazio W l'interpolazione tra due punti produce transizioni molto più naturali e coerenti rispetto allo spazio Z originale. Nello spazio W l'interpolazione è lineare, se ci muoviamo dal punto A  (es uomo) al punto B (es. donna) passeremmo attraverso una serie di volti intermedi perfettamente realistici e coerenti.
Questa transazione ci dice che la rete ha capito le rogole del mondo reale ed ha eliminato le zone di buio informativo.
Questo accade perchè la rete ha imparato a srotolare la varietà dei dati, eliminando le zone di bassa probabilità

Come fa questo vettore W a trasformarsi effettivamente in pixel
Qui entra in gioco AdaIN

Il Meccanismo AdaIN
Iniezione dello stile ad ogni scala.
La differenza con le vecchie GAN il generatore di StyleGAN non inizia con rumore, inizia con una costante apprendibile che viene modificata strato dopo strato.
Immagina un blocco di marmo grezzo che non cambia mai, per scolpirlo usiamo AdaIN, Adaptive Instasnce Normalization. 
AdaIN traduce il vettore W in parametri visivi concreti per ogni layer.
Prende lo stile derivato da W e lo usa per vestire questa costante, layer dopo layer, risoluzione dopo risoluzione.

Entriamo nella meccanica di questa vestizione digitale

Modulazione e Demodulazione
Il processo è un binomio di modulazione e demodulazione
AdaIN prima normalizza la feature maps per togliere lo stile precedente (normalizza a media zero e varianza unitaria).
Poi applica una scala e un offset presi dal vettore latente W
E' come regolare il volume ed il tono di ogni canale radio
Lo stile controlla quanto ogni feature deve essere presente nell'immagine finale
Poichè questo avviene a diverse sscale, possiamo controllare l'immagine con una precisione chirurgica.
Questa precisione ci permette di gestire una vera gerarchia della visione.

poi non tutti i layer che costruiscono questa struttura sono uguali

La Gerarchia degli Stili
Gestisce gli stili su 3 livelli
Nei layer a bassa risoluzione (4x4, 8x8), AdaIN controlla la posa, la rotazione della testa e la forma generale del volto (Stili grossolani)
Nei layer intermedi vengono definiti i tratti somatici più specifici, la pettinatura e l'espressione degli occhi (stili medi)
Negli ultimi strati ad alta risoluzione, lo stile influenza solo le palette cromatiche  i micro-contrasti della pelle (stili fini).
Questa gerarchia è cio che rende il modello bilanciato e strutturato

E se provassimo a mescolare quest modelli tra persone diverse?

E' qui che nasce il potere dello Style Mixing
Il Poter della Gerarchia
La vera magia di AdaIN è che ci permette di fare 'Style Mixing': possiamo prendere la posa di una persona (layer più grossolani) e applicare i colori di un'altra (layer fini).
Questa modularità è possibile perchè ogni risoluzione è diventata un canale indipendente di informazione visiva, controllato da W. E' una rivoluzione per design ed editing.

Puoi usare livelli iniziali della persona A con i livelli fini della persona B

Ma per quanto W sia potente c'è un dettaglio che non possiamo controllare
Il caos della natura.

Rumore Stocastico.
Gestire l'imprevedibiltà del dettaglio
Ci sono elementi in un'immagine, come la posizione esatta dei capelli o i pori della pelle, che non dovrebbero dipendere dallo stile globale della persona. Sono elementi casuali e stocastici
StyleGAN gestisce questa complessità attraverso il rumore.
Iniettando rumore casuale direttamente dopo ogni convoluzione, la rete può generare variazioni naturali senza alterare l'identità del soggetto.

Variazioni Locali
Realismo attraverso il caos
Il rumore viene aggiunto come una mappa di valori casuali pixel per pixel a ogni livello di risoluzione. Cambiando solo il seme di questo rumore, l'identità della persona rimane, ma i capelli si sposteranno leggermente o la grana della pelle cambia.
E' l'effeto delle immagini vecchie o troppo liscie che caratterizzava le vecchie GAN.
Invece di forzare la Mapping Network a imparare ogni singolo dettaglio, lasciamo che sia il rumore a decidere.
Insomma aggiungiamo il caos per imitare la realtà, Senza questo meccanismo, la rete sprecherebbe troppa energia per memorizzare dettagli che dovrebbero essere puramente casuali.
E il controllo di questo caoe è affidato alla rete stessa.
Questo tecnica previene l'effetto immagini troppo piatte o eccessivamente regolari tipico delle GAN precedenti.

Integrazoine del Dettaglio
Ogni blocco di convoluzione ha il suo ingresso di rumore dedicato e soprattutto un peso apprendibile, permettendo una variabilità specifica per ogni scala (input per layer).
La rete decide quanto impatto deve avere il rumore su quella specifica scala.
La forza del rumore è controllata da un parametro apprendibile che decide quanto impatto deve avere sui pixel finali (scaling dei pesi)
E' proprio questo maccanismo (tra la struttura rigida decisa da GAN e la variazione locale di rumore) a creare quella sensazione di fotorealismo, simulando le imperfezioni naturali della pelle e delle superfici.
E' una danza tra ordine e disordine che inganna l'occhio umano, facendogli credere che la pella ha una vera storia bioilogica.

Style GAN ha separato la struttura, decisa da W, dalla texture, decisa dal rumore. Questa distinzione è la chiave per ottenere modelli generativi che siano stabili e incredibilmente dettagliati e per la prima volta completamente esplorabili.

Oggi non usiamo StyleGAN solo per i volti, ma per farmaci, materiali, design industriale, grazie alla possibilità di controllo totale.
