Oltre l'addestrameto del modello
Nel software 'tradizionale' se il codice non cambia, il comportamento è lo stesso.
Il modello invece è come un organismo vivente che dipende dall'ambiente circostante.
E' necessario creare un flusso continuo dove i dati entrano, il modello si aggiorna, e l'output viene monitorato costantemente. 
La creazione del 'cervello' (il modello) rappresenta solo una piccola frazione dell'ecosistema totale.
Una pipeline end-to-end deve gestire il flusso continuo di dati, la trasformazione in codice eseguibile, il rilascio controllato e l'osservazione costante delle performance in tempo reale.

Quali sono le stazioni principali che compongono questo viaggio dei dati.
LA pipeline non è una linea retta, ma è un cerchio
- Tutto parte dalla data ingestion: recupero dei dati grezzi, versionamento e pulizia sistematica per garantire la qualità dell'input
- Poi passiamo al Model Developmente dove il cervello viene forgiato, fase di sperimentazione, tunning degli iperparametri e validazione su set di test rigorosi.
- Segue il Deploymente, dove mettiamo l'ai dentro un container, (esempio Docker) per servirla via API.
- Infine il monitoring, dobbiamo controllare che l'ai stia iniziando a dare risposte corrette o errate ed in questo caso intervenire di conseguenza. Perchè il mondo esterno cambia costantemetne.

Quest catena garantisce che il valore prodotto non sia solo un colpo di fortuna, ma per far girare l'ai senza intoppi, abbiamo bisogno di strumenti di automazione.

Infrastruttura e Automazione
Non possiamo permetterci di caricare i modelli a mano ogni volta, usiamo dei ciclo appositi per il ML,implementiamo testo automatizzati non solo per il codice, ma anche per la validatà statistica del modello prodotto.
Fondamentale è anche il feedback loop, se un utente corregge la risposta dell'ai, quel dato deve tornare indietro e deve diventare materiale per il prossimo allenamento.
Tutto questo serve a garantire una cosa sola: osservabilità del sistma

Osservabilità del sistema.
E' la capacità di sapere se il sistema sta bene, senza guardare dentro ogni singolo neurone.
Un'applicazione AI dfferisce dal softwere tradizionale perchè il suo comportamento può degradare se il codice non cambia, a causa del cambiamento dei dati nel mondo reale.
L'osservabilità deve coprire sia la salute dell'infrastruttura (CPU/RAM) che la salute del modello (accuratezza stimata)
L'osservabilità è la probabilità che l'accuratezza rimanga sopra ad una certa soglia durante il tempo.
Se classifichi un modello di moda, tra 6 mesi non saprà riconoscere le moderne installazioni.
Questo si chiama data drift
E' un applicazione ai professionale, deve avere dei sensori, dei monitor, che suonano un allarme quando il modello smette di essere sincronizzato con la realtà.
Ma attenzione aggiungere tutti questi pezzi ha un costo invisibile, parliamo del debito tecnico.

Il Debito Tecnico nei Sistemi ML
I costi nascosti dell'innovazione
Google ha evidenziato come i sistemi di Machine Learning abbiano una capacità unica di accumulare debito tecnico in modo rapido e silenzios.
Il debito tecnico è il lavoro di manutenzione e correzione che si rimanda quando si costruisce un sistema, rendendo più costoso modificarlo e mantenerlo in seguito.
Un modello può avere ottime prestazioni nei test e far parte di un sistema con molto debito tecnico. Accuratezza del modello e qualità del sistema son due cose diverse.
Il sistema oggi può funzionare, ma tra 6 mesi, quando i dati di partenza, di cui è alimentato il modello, cambiano, il modello può peggiorare nelle prestazioni.

Il debito è costituito dalle attività rimandate: versionamento, controlli sui dati, riproducibilità e monitoraggio. Gli “interessi” sono le ore aggiuntive necessarie per ogni diagnosi e modifica.

Nel software tradizionale il comportamento dipende soprattutto dal codice. Nel ML dipende anche dai dati, trasformazioni, addestramento e ambiente operativo.
Attenzione alla distinzione: il cambiamento dei dati, spesso chiamato data drift, non è di per sé debito tecnico. Non avere predisposto controlli e procedure per gestirlo può esserlo.

Nei sistemi ML il codice dell algoritmo puro è spesso solo una piccola scatola al centro di codice di supporto per la configurazione, quel mare è dove si nasconde il debito che potrebbe affondare la nave tra un anno.

Evitare la trappola del debito
- Entamglement: il fenomeno 'CACE' (changing anything changes everything) dovuto alle interdipendenze tra i dati di input. Immagina di cambiare solo il formato delle date in ingresso, improvvisamente i pesi della rete diventano innutili, tutto è collegato in modo non lineare
- Glue Code: scrivere kilometri di codice solo per far parlare due librerie diverse, finisce per rendere il sistema rigido.
- Pipeline Jungles: creazione di pipeline di preprocessing stratificate e non documentare che diventano impossibili da gestire.
Il debito tecnico aumenta il costo di manutenzione e riduce la velocità di innovazione futura del team.

Esistono anche altre forme di debito ancora più sottili
Erosione dei confine.
- Difficoltà nell'isolare i componenti a causa del fatto che i dati fluiscono tra i moduli, influenzano i pesi della rete in modi non lineari.
- Il debito derivante da migliaia di righe di configurazione (iperparametri, path, flag) che spesso non sono testaste o versionate correttamente.
- Dipendere da flussi di dati esterni che potrebbero cambiare formato o significato senza preavviso, rompendo le assunzioni del modello.

Come possiamo rendere tutto questo sostenibile a lungo termine?

Dobbiamo trattare i dati con lo stesso rigore con cui trattiamo il codice.
L'obbiettivo è creare astrazioni pulite che ci permettano di cambiare il modello senza dover ricostruire tutto.
Rendere il sistema modulare e sostenibile

Team e Responsabilità
Chi fa cosa nelle pipeline end-to-end
Le responsabilità si dividono.
Chi capisce i numeri, chi capisce il server, e chi capisce l'utente finale.
- Data Scientist: esplora i dati, testa le ipotesi, e decide l'architettura della rete neurale, la sua vittoria è una loss che scende
- ML Enginner: è specializzato nel trasformare i notebook in codice di produzione, ottimizzando l'inferenza e gestendo l'infrastruttura GPU. Obbiettivo tempi di inferenza di pochi millesecondi
- Full Stack Dev: responsabile dell'integrazione del modello nel prodotto, creando il frontend e gestendo le interazione utente. Integra l'AI nel prodotto reale, affinchè l'utente possa usarla
- La produttività del team AI è massimizzata quando i protocolli di comunicazione tra questi ruoli sono chiari e automatizzati.

Ma come avviene il passaggio di consegna tra questi mondi?
Il momento più critico è l'end over, il Data Scientist consegna la logica al ML Engineer per la 'produzionizzazione'.
Dobbiamo passare dati notebook sporchi del data scientist a script di produzione puliti, gestiti dal ML entineer, per farlo  usiamo il contratto API, definiamo esattamente come deve essere fatto il JSON input e quello di output. Se il contratto è chiaro il team di front-end può inizare a lavorare anche se il modello non è pronto.
Questa collaborazione sta facendo nascere una nuova figura professionale.
