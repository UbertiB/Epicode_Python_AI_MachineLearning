Perchp Stramlit è diventato lo standard del settore?
Rapid UI Prototyping con Streamlit
Tradizionalmente, esporre un modello AI richiedeva lo sviluppo di un'API backend e di un frontend complesso in Javascript. Streamlit elemina questa separazione permettendo di scrivere l'intera UI in python, è come passare dal dover costruire una casa mattone dopo mattone a disegnarla sul tablet e vederla apparire magicamente.
E' un framework per chi mastica dati.
Il framework trasforma ogni variabile o print del codice in un componente visuale reattivo, rendendo il ciclo di feedback tra sviluppatore e utente estremamente veloce.
Passiamo da un modello chiuso in un notebook ad un prodotto che chiunque può usare tramite browser. 

Vediamo qual'è la filosofia di sviluppo che rende questo possibile
La filosofia di Stramlit è lo scritp-to-app, non c'è separazione tra logica e grafica.
- Pure Python: non è richiesta conoscenza di Web Development tradizionale; ogni riga di codice Python corrisponde a un componente grafico
- Hot Reloading: ogni modifica al codice sorgente viene visualizzata istantaneamente nel browser, facilitando la sperimentaizone visiva
- Data Centricity: integrazione nativa con librerie come Pandas, Matplotlib, Plotly per la visualizzazione dei risultati dell'inferenza.
- La velocità di prototipazione 'V' è inversamente proporzionale alla complessità dello stack tecnologico utilizzato.

C'è un dettaglio tecnico sul funzionamento
Modello di Esecuzione
Il segrerto di Streamlit è il paradigma del Re-run
A differenza delle app web tradizionali che aspettano eventi specifici, Streamlit riesegue l'intero script dall'alto verso il basso ogni volta che un widget cambia stato.
Ogni volta che muovi uno slider, lo script viene rieseguito dall'alto verso il basso.
Questo approccio garantisce che la UI sia sempre lo specchio esatto dello stato attuale delle variabili nel codice.
Non dobbiamo preoccuparci di aggiornare manualmente un grafico, se la variabile cambia Stramlit disegna tutto il necessario per far si che la vista sia coerente con i dati.

Ma cosa succede se vogliamo che l'app ricordi qualcosa nel passato?

Gestione dello Stato: Session State
Persistenza in un mondo stateless
Poichè stramlit riesegue tutto ad ogni click, le variabili locali perdono il valore. I 'Session State' è la memoria dell'applicazione che sopravvive ai re-run. E' un dizionario speciale che sopravvive ai re-run. Senza di esso un chatbot dimenticherebbe la domanda precedente ogni volta che l'utente preme invio.
Il session state ci permette di costruire applicazioni complesse che mantengono uno stato interno coerente.

Come funziona tecnicamente questo meccanismo di memoria.
st.session_state
- State Persistence: memorizzazione di variabili tra diverse interazioni utente all'interno della stessa sessione browser
- Callbacks: funzioni associate ai widget che modificano il Session State prima che lo script venga rieseguito. sono funzioni che scattano appena un widget cambia, permettendo di aggionrare lo stato prima che l'app venga inizializzata
- Initialization: logica per definire di default solo il primo caricamente dell'app per evitare errori di chiave mancate.
- Lo stato dell'applicazione al tempo t+1 è una funzione dello stato precedente e dell'azione dell'utente.

ESempi pratici di cosa possiamo fare con questa memoria:
- Chat History: fondamentale per i chatbot: salvare la lista dei messaggi scambiati affinchè non scompaiono a ogni nuovo invio dell'utente
- Form Validation: tracciare se l'utente ha completato tutti i passaggi necessari prima di sbloccare il tasto finale di addestramento o inferenza.
- Cross-Widget Interaction: permettere a un widget di cambiare il valore suggerito in un altro widget basandosi su una scelta precedente (es. scegliere il dataset filtra i modelli disponibili). Se l'utente scegliere visione artificiale da un menù il session state può filtrare solo i modelli di computer vision nascondendo quelli di NLP

Gestione della Memoria
Dobbiamo distinguere tra memoria privata e risorse condivise
Mentre il session state è unico per utente (se hai 1000 utenti che salvano video pesanti il server esplode)
Il caching delle risorse è condiviso tra tutti gli utenti del server. Carichiamo il modello una volta sola e tutti gli utenti utilizzano la stessa istanza di memoria.
Questa distinzione è vitale per scalare l'app: i modelli pesanti sono in cache comune, i dati personali sono nello state privato.



```python
import streamlit as st
from openai import OpenAI

# --- CONFIGURAZIONE DEL CLIENT E DELL'INTERFACCIA ---

# LM Studio fornisce un endpoint locale che imita le API di OpenAI.
# Il client viene inizializzato per puntare all'indirizzo locale (localhost) sulla porta 1234.
# 'api_key' è necessaria per la struttura della libreria, ma LM Studio non la richiede realmente.
client = OpenAI(base_url="http://localhost:1234/v1", api_key="lm-studio")

# Impostazione della configurazione della pagina Streamlit
st.set_page_config(page_title="Educational Chat Streamer")

# --- BARRA LATERALE: CONTROLLI DEL MODELLO ---

# La sidebar viene utilizzata per separare i parametri di configurazione dalla chat principale.
with st.sidebar:
    st.title("Parametri Modello")
    st.info("Regola i parametri per modificare il comportamento del linguaggio naturale.")
    
    # Lo slider imposta la 'Temperature', che determina quanto la risposta debba essere creativa o deterministica.
    # Valori bassi (0.2) portano a risposte precise; valori alti (1.2) a risposte più varie.
    temp = st.slider("Temperature", 0.0, 1.5, 0.7, step=0.1)
    
    # Pulsante per resettare la conversazione.
    # In Streamlit, premere un pulsante attiva il blocco di codice sottostante.
    if st.button("Cancella Cronologia", use_container_width=True):
        # Svuotiamo la lista dei messaggi salvata nello stato della sessione.
        st.session_state.messages = []
        # Forza il ricaricamento dell'app per mostrare l'interfaccia pulita.
        st.rerun()

# Titolo principale dell'applicazione.
st.title("Chat Streaming con LM Studio")
st.markdown("Applicazione per interagire con un modello linguistico locale tramite streaming dei dati.")

# --- GESTIONE DELLO STATO (SESSION STATE) ---

# Streamlit riesegue l'intero script Python ogni volta che l'utente interagisce con un elemento.
# Per non perdere la cronologia della conversazione tra un'esecuzione e l'altra,
# utilizziamo 'st.session_state', un dizionario persistente durante la sessione dell'utente.
if "messages" not in st.session_state:
    # Se la chiave 'messages' non esiste, la inizializziamo come lista vuota.
    st.session_state.messages = []

# --- VISUALIZZAZIONE DELLA CONVERSAZIONE ---

# Iteriamo sulla lista dei messaggi salvati per renderizzarli nell'interfaccia.
# Ogni messaggio è un dizionario con chiavi 'role' (user/assistant) e 'content'.
for message in st.session_state.messages:
    # 'st.chat_message' crea un contenitore grafico adatto al ruolo specificato.
    with st.chat_message(message["role"]):
        st.markdown(message["content"])

# --- GESTIONE DELL'INPUT E GENERAZIONE RISPOSTA ---

# 'st.chat_input' cattura il testo scritto dall'utente. Se l'utente preme invio, il prompt viene popolato.
if prompt := st.chat_input("Scrivi un messaggio..."):
    
    # 1. Registrazione del messaggio utente.
    # Aggiungiamo il prompt alla cronologia nello stato della sessione.
    st.session_state.messages.append({"role": "user", "content": prompt})
    # Visualizziamo immediatamente il messaggio appena scritto.
    with st.chat_message("user"):
        st.markdown(prompt)

    # 2. Generazione della risposta dell'assistente.
    with st.chat_message("assistant"):
        # Creiamo un 'placeholder' (contenitore vuoto) che aggiorneremo dinamicamente.
        # Questo permette l'effetto di streaming, dove il testo appare parola per parola.
        response_placeholder = st.empty()
        full_response = ""
        
        try:
            # Chiamata al client per generare la risposta.
            # Impostando 'stream=True', il client restituisce un oggetto iterabile (generatore).
            stream = client.chat.completions.create(
                model="gpt-oss-20b", # Identificativo del modello caricato in LM Studio.
                messages=[
                    # Passiamo l'intera cronologia per dare contesto al modello.
                    {"role": m["role"], "content": m["content"]}
                    for m in st.session_state.messages
                ],
                temperature=temp,
                stream=True,
            )
            
            # Ciclo di lettura dei 'chunk' (frammenti di testo) inviati dal modello.
            for chunk in stream:
                # Verifichiamo la presenza di contenuto nel frammento attuale.
                if chunk.choices[0].delta.content:
                    # Accumuliamo il testo ricevuto nella stringa totale.
                    full_response += chunk.choices[0].delta.content
                    # Aggiorniamo il visualizzatore in tempo reale aggiungendo un cursore visivo.
                    response_placeholder.markdown(full_response + "|")
            
            # Una volta terminato lo streaming, mostriamo la risposta definitiva senza il cursore.
            response_placeholder.markdown(full_response)
            
            # 3. Salvataggio della risposta completa.
            # È fondamentale salvare la risposta intera in 'messages' per mantenerla nelle esecuzioni future.
            st.session_state.messages.append({"role": "assistant", "content": full_response})
            
        except Exception as e:
            # Gestione degli errori in caso di problemi di comunicazione con il server locale.
            st.error(f"Errore di connessione: {e}")
            st.warning("Assicurarsi che LM Studio sia attivo e che il Local Server sia in ascolto sulla porta 1234.")
```
