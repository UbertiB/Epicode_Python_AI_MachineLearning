CASO PRATICO: APPLICAZIONE COMPLETA AI PER ANALISI DOCUMENTI


Impariamo a comporre l'intero quadro.
Immagina di avere una biblioteca con migliaia di libri e vuoi trovare la risposta ad una singola domanda.

Passiamo dai bit grezzi di un pdf alla saggezza di un sistema che comprende il contenuto e risponde in modo accurato.

- Frontend reattivo: interfaccia reattiva per gestire l'upoload dei documenti
- Pipeline di ingestione: dall'astrazione testo (OCR) alla presistenza nel Vector Store
- Interaction Layer: implementazione del ciclo Chat-Q&A basato su logica RAG, ovvero dialogo vero a proprio basato sulla logica RAG.

L'interfaccia di caricamento e visualizzazione
In un'applicazione di analisi documenti,l'interfaccia deve gestire file binari pesanti e fornire un feedback immediato sull'elaborazione. Streamlit ci permette di creare questo ponte in pochi righe di codice Python. Stremlit è come un traduttore istantaneo tra python ed il browser. Immagina Stremlit come un potente prefabbricato.

Ma cosa succede tencicamente quando carichiamo un file di diversi megabyte.
Gestione dell'Input Binario
Flusso di byte dal browser al server
Come gestire un fiume in piena, se beviamo tutta l'acquia affoghiamo. 
- Buffer Management: lettura del PDF caricato come stream di byte per evitare il sovraccarico della memoria RAM sul server.
- UI Layout: utilizzo di colonne per affiancare l'anteprima del documento e l'area di chat migliorando la navigazione dell'utente.
- Status Indicators: implementazione di 'st.spinner' per segnalare le fasi critiche di OCR ed embedding, riducendo l'abbandono dell'utente.
- La velocità di caricamento dipende dalla larghezza di banda e dalla dimensione del file normalizzata.

Backend: Ingestione dei dati e Indicizzazione
Dal documento al database vettoriale
Il cuore tecnologico risiede nella pipeline ETL che trasforma pixel o byte in vettori matematici. Questo processo richiede precisione nella segmentazione del testo.
Utiliziamo PyMuPDF per l'estrazione e RecursiveCharacterTextSplitter per mantenere la coerenza semantica durante la frammentazione.
Ma non ci perfmiamo qui, se il pdf è una scansione di immagine attiviamo il nostro occhio artificiale. l'OCR per trasformare i pixel in parole.

Una volta estratto il testo come lo prepariamo per la ricerca semantica?
Processing e Archiviazione
Trasformazione semantica del dato
- Smart Splitting: divisione del testo in chunck con 'overlap' per evitare di troncare frasi a metà, preservando il contesto per l'LLM. Mantenere un'overlap significa mantenere una sovrapposizione per non perdere il contesto tra un frammento e l'altro.
- Embessing Generation: trasformiamo questi pezzi in vettori matematici densi, utilizzando modelli come 'Sentence-Transformers'
- Vector Store: caricamento degli embeddings in FAISS o ChromaDB per permettere la ricerca similarità in tempo reale.
La ricerca nel Vector Store si basa sulla minimizzazione della distanza angolare tra il vettore query e i chunck archiviati

E se il documento fosse solo un'immagine scansionata?
Entra in gioco la visione
OCR e Visione
- Tesseract Integration: uso di tecniche di Computer Vision per convertire pagine PDF scansionate (immagini) in testo leggibile prima della fase di chunking.
- Metadata Enrichment: associare a ogni chunk informazioni sul numero di pagine e sul nome del file per permettere al modello di citare le fonti.
- Data Cleaning: rimozione di boilerplate, numeri di pagine isolati e caratteri speciali che potrebbero sporcare la rappresentazione vettoriale.

Ma quante dimensioni servono per catturare il senso di una frase?
Dimensionalità dello Spazio Vettoriale
Trade-off tra precisione e velocità
La scelta del modello di embedding influenza la capacità di distinguere concetti simili. Modelli più grandi offrono una migliore risoluzione semantica, ma rendono la ricerca più lenta e costosa..
La complessità computazionale della ricerca aumenta con la dimensione 'd' dei vettori e il numero 'N' di frammenti.
In un'applicazione professionale dobbiamo trovare il punto di saturazione.

Ora che i dati sono archiviati ed indicizzati, vediamo come l'utente può finalmente interrogarli
Interacion: ChatQ&A e RAG
Interrogare il documento con linguaggio naturale.
L'interazione finale avviene tramite una pipeline RAG (Retrieval Augmented Generation) che fonde la conoscenza statica dell'LLM con i dati dinamici estratti dal PDF.
Il sistema va a rilevare i 3 e 4 frammenti che ritiene più rilevanti per la domanda specifica e di metterli davanti agli occhi del modello.
Questa tecnica riduce drasticamente le allucinazioni, perchè ancora la risposta alla realtà del documento caricato.
Dobbiamo orchestrare la query dell'utente affinchè recuperi i frammenti giusti e li presenti al modello come contesto privilegiato.

Come avviene questa funzione tra domanda e contesto
Il Ciclo RAG
Augmenting the prompot
Il ciclo RAG è una danza di probabilità. LA risposta non viene generata dal vuoto ma è condizionata sia dalla domanda che dai frammenti recuperati.
Usiamo il prompot engineering per dare oridni chiari. es. rispondo basandoti sul contesto fornito, se la risposta non è presente ammetti di non saperlo.
Mettiamo il contesto direttamente nel contesti di sistema.
-Semantic Retrieval: recupero dei 'top-k' frammenti più rilevanti rispetto alla domanda posta dall'utente.
- Context Injection: inserimento dei frammenti recuperati all'interno del prompot di sistema inviato all'LLM
- Prompt Engineering: istruire il modello a rispondere 'solo' sulla base del contesto fornito, citando la pagine di riferimento.
- La risposta finale è generata massimizzando la probabilità condizionata rispetto alla domanda e al contesto recuperato.

Ma come rendiamo questa interazione davvero naturale?
Dobbiamo gestire la memoria della chat
Ottimizzazione della Risposta
- Chat Memory: implementazione di un buffer di memoria per permettere all'utente di fare domande di follow-up (es. 'e riguardo a quello che hai detto prima?')
- Source Verification: visualizzazione dei frammenti originali sotto la risposta dell'AI per permettere all'utente di verificare l'attendibilità dell'informazione
- Modle Choice: valutazione del bilanciamento tra latenza e ragionamento: usare modelli locali leggeri o API potenti come GPT-4o
Per compiti difficili potrebbe scegliere tra modelli veloci o API cloud di classe superiore.

C'è però un'ultimao pericolo da monitorare
Dobbiamo essere i poliziotti delle allucinazioni
Valutazione delle Allucinazioni
Garantire la federtà al testo.
In contesti professionali, un'allucinazione è inaccettabile. Implementeremo filtri per assicurarci che il modello non inventi fatti non presenti nel PDF
La fedeltà al contesto viene monitorata analizzando la sovrapposizione informativa tra contesto 'c' e risposta 'y'
Se il modello inizia a invetare date o altro, il nostro sistema di validazione deve accorgersene.
Un'applicazione completa non produce solo risposte, ma garantisce che siano vere.
