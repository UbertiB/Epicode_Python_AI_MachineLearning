Passiamo ore ad addestrare modelli cambiando piccoli parametri per poi chiederci qual'era la configurazione che aveva data il 98% di accuratezza?

Come organizzare la ricerca dei modelli senza perdere traccia delle prove e senza scegliere il vincitore soltanto per un risultato fortunato.

- MLflow come diario di bordo digitale
- Optuna cerca per noi la combinazione perfetta di parametri
- Navighiamo nelle dashboard per vedere i risultati migliori e scegliere il nostro modello campione

MLFlow registra e organizza gli esperimenti. Optuna cerca buone configurazione degli iperparametri. Si possono  usare separatamente o insieme.
MLFlow: che cosa ho provato, con quali impostazioni e con quali risultati?
Optuna: quali impostazioni conviene provare per migliorare il risultato?

Ci sono due livelli di apprendimento:
1) Apprendimento dentro un addestramento: i pesi cambiano, questo apprendimento è gestito dall'algoritmo di training e optimizer
2) Apprendimento tra addestramenti diversi: tra addestramenti diversi learning rate, architettura, dropout, batch size possono cambiare, questi cambiamenti sono gestiti o personalmente annotando tutti i dati, o meglio e più professionale da Optuna.

* Optuna propone learning rate e dropout
* Il programma costruisce e addestra la rete
* L'optimizer aggiorna i pesi durante il trainig
* Calcoli i risultati in validazione
* MLFlow registra configurazione, metriche e file
* Optuna usa il risultato per orientare le prove successive
Optuna non aggiorna i pesi della rete. MLFlow non sceglie automaticamente come addestrarla

MLFlow
immagine di addestrare 10 modelli per prevedere i consumi ERP. Cambi algoritmo, feature e parametri. Dopo qualche giorno devi ricostruire quale modello funzionava meglio.
MLFlow Tracking permette di registare parametri, meteriche e file prodotti, inclusi i modelli, e confrontarli tramite un'interfaccia grafica.
Ogni esecuzione registrata si richiama run; più run possono essere raggruppate in un experiment.
MLFlow non registra tutto, registra quello ciò che esplicitamente è stato chiesto

Optuna
Automatizzare la ricerca degli iperparametri
Gli iperparametri sono impostazioni che scegli prima dell'addestramento: per esempio profondità massima degli albero, numero di albero o learning rate.
Con Optuna definisci:
1. Quali valori può provare
2. Come addestrare e valutare ogni configurazione
3. Quale metrica minimizzare o massimizzare
4. Quante prove eseguire.
Ogni prova si chiama trial; l'insieme delle prove è uno study. Optuna offre algoritmi di ricerca e maccanismi per interrompere prove poco promettenti, quando il processo fornisce valutazioni intermedie adeguate
Optuna cerca la configurazione migliore tra quelle valutate, secondo la meetrica e la validazione che hai definito. Non garantisce il miglior modello possibile.

MLFlow e Optuna possono lavorare insieme
Registri con MLFlow gli addestramenti del modello, cambiando gli iperparametri. Fai eseguire ad Optuna la ricerca del migliore.

Tracciabilità con MLFlow
Smettere di usare fogli Excel per i log
Nello sviluppo di modelli di DeepLearning, ogni tentativo (run) genera dati cruciali. Senza un sistema centralizzato, è impossibile ricordare quale combinazione di learnig rate e architettura ha prodotto il miglio risultato.
MLFlow agisce come un diario di bordo automatizzato, salvando ogni metrica e ogni peso del modello in modo strutturato e interrogabile.

Ma cosa registra MLFlow?
I Pilastri del Tracking
Cosa monitorare in ogni esperimento
Ogni esperimento in MLFlow di poggia su 3 pilastri
- Parameters: variabili di input statiche come il numero di layer, la dimensione del batch o il tipo di optimizer. Sono le impostazioni iniziali.
- Metrics: valori dinamici registrati durante l'addestramento come la loss o l'accuratezza per ogni epoca. Sono i risultati che cambiano nel tempo
- Artifacts: file di output pesanti inclusi i file '.pth' o '.h5' dei pesi, grafici di confusione e file di configurazione
- Ogni esperimento (run) viene registrato come un punto in uno spazio di configurazione multidimensionale.
Possiamo analizzare non solo se il modello è bravo, ma perchè lo è diventato.

Oltre al salvataggio manuale esistono modo più veloci per integrare questo strumento.

La storia delle metriche permette di calcolare la derivata della loss per decidere l'early stopping programmatico.
Ci permette di verificare se il modello ha smesso di imparare troppo presto e se ha iniziato a copiare i dati invece di capirli.

Ma come facciamo a trovare i parametri giusti, senza doverli inserire noi a mano uno per uno?
Andare oltre la Grid Search
Cercare manualmente la combinazione perfetta di iperparametri è inefficiente. La Grid Search è lenta e la Rendom Search è imprecisa.
Optuna, invece, utilizza algoritmi di ottimizzazione bayesiana per 'capire' quali regioni dello spazio degli iperparametri sono più promettenti, concentrando li le risorse computazionali. Guarda dove ha già cercato, impara dai fallimenti, e capisce dove andare.

Vediamo come Optuna organizza questa ricerca intelligente.
Uno 'Study' è l'obbiettivo finale, Un 'Trial' è una singola esecuzione del modello con parametri specifici.
TPE è l'algoritmo di default che modella la distribuzione dei parametri buoni rispetto a quelli cattivi.
Il Pruring è la capacità di interrompere automaticamente i Trial che mostrano risultati mediocri nelle prime fasi, risparmiando tempo GPU.
L'obbiettietivo è minimizzare o massimizzare una funzione obiettivo agendo su un set di variabili condizionate.
Il sistema cerca il punto x che minimizza la funzione obbiettivo f(x) adattando la ricerca in tempo reale.
PEr far funzionare questo dobbiamo avere uno spazio di manovra ben definito

Dobbiamo definire i confini del campo da gioco
Usiamo il metodo Suggest per dire ad Ottuna trova un learning rate tra 0.001 e 0.1, ma usa una scala logaritmica perchè i piccoli cambiamnti contano di più. 'trial.sugest_float' o 'trial_suggest_int' per definire range e distribuzioni (logaritmiche o lineari) dei parametri.
Possiamo anche fare ottimizzazione multi obbiettivo, magari un modello precisissimo ma anche molto leggero per funzionare su un cellulare. Optuna è flessibile.

Ma cosa succede se un obbiettivo sembra andar male fin dai primi secondi?
Efficienza del Pruning
Tagliare i rami secchi.
Uno dei super poteri di Optuna è il pruning.
Immagina di essere ad una gara di corsa, se dopo 100 mt un corridore è già stanchissimo ed ultimo perchè fargli fare l'intera maratona?
Il pruring confronta il trial corrente con i migliori del passato, se le speranze di successo sono sotto una soglia optuna taglia i rami secchi e ferma il trial.
Il pruning agisce come un guardiano: confronta il trial corrente con la cronologia dei trial passati.
Se la probabilità di superare il miglior risultato attuale è inferiore a una soglia, il trial viene terminato.

Una volta che è finita la missione, dopo aver lanciato centinaia di trial, ci troviamo con centinaia di run. Dobbiamo essere in grado di estrarre il 'vincitore' e capire quali parametri hanno influenzato maggiormente il risultato.

Analisi della Varianza
Non sempre il trial con il valore massimo assoluto è il migliore; cerchiamo modelli con una bassa varianza rispetto a piccoli cambiamenti dei semi casuali.
La robustezza del modello è inversamente proporzionale alla sensibilità rispetto a rumore negli iperparametri.
Un errore è scegliere un modello che ha avuto un picco di fortuna in un singolo trial, noi cerchiamo la stabilità,  un modello robusto deve dare ottimi risultati sempre. Analizzare la varianza significa capire se la nostra vittoria è meritata o frutto di rumore.

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
import mlflow
import mlflow.pytorch
import optuna
import logging
import warnings
from pathlib import Path

# -----------------------------------------------------------------------------
# SETUP LOGGING E WARNING
# -----------------------------------------------------------------------------
# Filtriamo i warning per mantenere l'output pulito durante l'esecuzione.
# Molti warning di MLflow sono informativi (es. versioni locali di PyTorch)
# e possono distrarre durante una prima lettura.
logging.getLogger("mlflow").setLevel(logging.ERROR)
warnings.filterwarnings("ignore")

# -----------------------------------------------------------------------------
# 1. PREPARAZIONE DEI DATI
#
# Questa funzione si occupa di generare un dataset sintetico e di prepararlo
# per l'utilizzo con PyTorch. Esegue tutti i passaggi necessari: generazione,
# split in train/val e conversione in tensori.
# -----------------------------------------------------------------------------
def get_data():
    # Generiamo un dataset "moons" (due semicerchi incastrati), un classico esempio
    # per testare problemi di classificazione non lineare.
    # n_samples=1000: numero totale di punti generati.
    # noise=0.2: aggiunge variabilità ai dati per renderli meno perfetti e più realistici.
    # random_state=42: fissa il seme casuale per garantire che i dati siano identici ad ogni esecuzione.
    X, y = make_moons(n_samples=1000, noise=0.2, random_state=42)

    # Dividiamo i dati in training set (per l'addestramento) e validation set (per la valutazione).
    # test_size=0.2: indica che il 20% dei dati verrà riservato per la validazione.
    X_train, X_val, y_train, y_val = train_test_split(X, y, test_size=0.2)
    
    # Conversione dei dati da array NumPy a Tensori PyTorch.
    # I modelli PyTorch richiedono dati sotto forma di tensori.
    # FloatTensor viene usato per le feature (numeri decimali), LongTensor per le etichette (interi delle classi).
    train_ds = TensorDataset(torch.FloatTensor(X_train), torch.LongTensor(y_train))
    val_ds = TensorDataset(torch.FloatTensor(X_val), torch.LongTensor(y_val))

    # Creazione dei DataLoader.
    # I DataLoader gestiscono l'iterazione sui dati, fornendoli in pacchetti (batch) per l'addestramento.
    # batch_size=32: ogni iterazione fornirà un blocco di 32 esempi.
    return DataLoader(train_ds, batch_size=32), DataLoader(val_ds, batch_size=32)

# -----------------------------------------------------------------------------
# 2. DEFINIZIONE DEL MODELLO (ARCHITETTURA DINAMICA)
#
# Definiamo una classe per la rete neurale. Questa classe è progettata per essere flessibile:
# il numero di neuroni nascosti e il numero di strati non sono fissi, ma vengono passati
# come parametri al costruttore.
# Questo design permette a Optuna di testare diverse architetture automaticamente.
# -----------------------------------------------------------------------------
class SimpleNet(nn.Module):
    def __init__(self, n_hidden, n_layers):
        super().__init__()
        layers = []
        in_features = 2  # I dati di input (make_moons) hanno 2 caratteristiche: coordinate x e y.
        
        # Creiamo dinamicamente gli strati nascosti in base al parametro n_layers.
        for _ in range(n_layers):
            # Aggiungiamo un livello lineare che trasforma l'input in 'n_hidden' neuroni.
            layers.append(nn.Linear(in_features, n_hidden))
            # Aggiungiamo una funzione di attivazione ReLU (introduce non-linearità).
            layers.append(nn.ReLU())
            # L'uscita di questo strato diventa l'ingresso (features) per il successivo.
            in_features = n_hidden
            
        # Aggiungiamo lo strato di output finale.
        # Collega l'ultimo strato nascosto ai 2 neuroni di uscita (corrispondenti alle 2 classi: 0 e 1).
        layers.append(nn.Linear(n_hidden, 2))
        
        # nn.Sequential avvolge la lista di strati in un unico modulo che li esegue in sequenza.
        # L'asterisco (*) serve a "spacchettare" la lista layers e passarne i contenuti come argomenti singoli.
        self.model = nn.Sequential(*layers)

    def forward(self, x):
        # Definisce il flusso dei dati attraverso il modello quando viene chiamato.
        return self.model(x)

# -----------------------------------------------------------------------------
# 3. FUNZIONE OBIETTIVO PER L'OTTIMIZZAZIONE
#
# Questa funzione è il cuore dello script. Viene chiamata ripetutamente da Optuna.
# Ogni chiamata rappresenta un "trial" (un esperimento) che testa una specifica
# combinazione di iperparametri.
# Inoltre, integriamo MLflow per tracciare e registrare i risultati di ogni esperimento.
# -----------------------------------------------------------------------------
def objective(trial):
    # --- DEFINIZIONE DEGLI SPAZI DI RICERCA (IPERPARAMETRI) ---
    # Usiamo l'oggetto 'trial' per suggerire valori per gli iperparametri da testare.
    
    # Range per il learning rate in scala logaritmica (esplora ordini di grandezza diversi).
    lr = trial.suggest_float("lr", 1e-4, 1e-2, log=True)
    
    # Range per il numero di neuroni nascosti (da 8 a 64).
    n_hidden = trial.suggest_int("n_hidden", 8, 64)
    
    # Range per il numero di strati nascosti (da 1 a 3).
    n_layers = trial.suggest_int("n_layers", 1, 3)

    # --- INIZIO TRACCIAMENTO CON MLFLOW ---
    # Avviamo una "run" di MLflow per registrare questo specifico esperimento.
    # nested=True indica che questa run è parte di un esperimento più grande (o gerarchia).
    with mlflow.start_run(nested=True):
        
        # Registriamo su MLflow i parametri che Optuna ha scelto per questo tentativo.
        mlflow.log_params(trial.params)
        
        # Configurazione del dispositivo di calcolo (GPU se disponibile, altrimenti CPU).
        device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
        
        # Istanziamo il modello SimpleNet utilizzando i parametri n_hidden e n_layers suggeriti.
        model = SimpleNet(n_hidden, n_layers).to(device)
        
        # Definiamo l'ottimizzatore (Adam) configurato con il learning rate suggerito.
        optimizer = optim.Adam(model.parameters(), lr=lr)
        
        # Definiamo la funzione di perdita per classificazione multiclasse (CrossEntropyLoss).
        criterion = nn.CrossEntropyLoss()
        
        # Otteniamo i dataloader per addestramento e validazione.
        train_loader, val_loader = get_data()

        # --- CICLO DI ADDESTRAMENTO ---
        # Eseguiamo l'addestramento per un numero fissato di epoche (es. 10).
        for epoch in range(10):
            model.train()  # Imposta il modello in modalità addestramento (abilita dropout, batchnorm, ecc.).
            
            # Iteriamo sui batch di dati di training.
            for X_batch, y_batch in train_loader:
                # Spostiamo i dati sul device (CPU o GPU).
                X_batch, y_batch = X_batch.to(device), y_batch.to(device)
                
                # Azzeriamo i gradienti accumulati nel passo precedente.
                optimizer.zero_grad()
                
                # Calcoliamo la loss: differenza tra le predizioni del modello e i valori reali.
                loss = criterion(model(X_batch), y_batch)
                
                # Calcoliamo i gradienti (backpropagation).
                loss.backward()
                
                # Aggiorniamo i pesi del modello basandoci sui gradienti calcolati.
                optimizer.step()

            # --- VALIDAZIONE ---
            model.eval()  # Imposta il modello in modalità valutazione (disabilita dropout, ecc.).
            correct = 0
            
            # Disabilitiamo il calcolo dei gradienti per risparmiare memoria e calcoli durante la validazione.
            with torch.no_grad():
                for X_val, y_val in val_loader:
                    X_val, y_val = X_val.to(device), y_val.to(device)
                    outputs = model(X_val)
                    
                    # Verifichiamo quante predizioni sono corrette.
                    # argmax(1) restituisce l'indice della classe con la probabilità più alta.
                    correct += (outputs.argmax(1) == y_val).sum().item()
            
            # Calcoliamo l'accuratezza su tutto il validation set per questa epoca.
            accuracy = correct / len(val_loader.dataset)
            
            # Registriamo l'accuratezza come metrica su MLflow, associandola all'epoca corrente.
            mlflow.log_metric("accuracy", accuracy, step=epoch)
            
            # --- PRUNING (POTATURA) ---
            # Comunichiamo a Optuna l'accuratezza intermedia ottenuta in questa epoca.
            trial.report(accuracy, epoch)
            
            # Optuna valuta se vale la pena continuare questo trial.
            # Se le prestazioni sono molto peggiori rispetto ad altri trial già eseguiti,
            # interrompiamo subito l'esecuzione (Pruning) per risparmiare tempo e risorse.
            if trial.should_prune():
                mlflow.set_tag("status", "pruned")  # Segnaliamo su MLflow che il trial è stato potato.
                raise optuna.exceptions.TrialPruned()

        # Se il trial arriva alla fine (non viene potato), salviamo il modello completo su MLflow.
        mlflow.pytorch.log_model(pytorch_model=model,name="model",serialization_format="pickle")
        
        # Ritorniamo l'accuratezza finale. Optuna userà questo valore per capire se ha trovato un buon set di parametri.
        return accuracy

# -----------------------------------------------------------------------------
# 4. ESECUZIONE DELLO STUDIO
# -----------------------------------------------------------------------------
if __name__ == "__main__":
    import os
    
    # Configurazione MLflow
    # Impostiamo il tracking URI in modo assoluto.
    # In questo modo, i log verranno salvati nella cartella "mlruns" all'interno
    # della directory di questo script, indipendentemente da dove viene lanciato il comando.

    from pathlib import Path
    import mlflow

    # Cartella di lavoro corrente del notebook
    current_dir = Path.cwd()

    # Database SQLite locale per esperimenti, parametri e metriche
    db_path = current_dir / "mlflow.db"

    # Su Windows produce, per esempio:
    # sqlite:///F:/EPICODE/progetto/mlflow.db
    tracking_uri = "sqlite:///" + db_path.as_posix()

    mlflow.set_tracking_uri(tracking_uri)

    # Crea l'esperimento se non esiste, altrimenti lo seleziona
    mlflow.set_experiment("Deep_Learning_Best_Practices")

    print("Directory di lavoro:", current_dir)
    print("Tracking URI:", mlflow.get_tracking_uri())
    print("Esperimento configurato correttamente")

    # Creazione Studio Optuna
    study = optuna.create_study(
        direction="maximize", 
        pruner=optuna.pruners.MedianPruner()
    )
    
    # Avviamo il processo di ottimizzazione.
    # Optuna eseguirà la funzione 'objective' per 20 volte (n_trials=20), cercando i migliori iperparametri.
    study.optimize(objective, n_trials=20)

    # Al termine, stampiamo i risultati migliori ottenuti.
    print(f"Miglior accuratezza: {study.best_value:.4f}")
    print(f"Migliori iperparametri: {study.best_params}")
    
    print("\n----------------------------------------------------------------")
    print("PER VEDERE LA DASHBOARD DI MLFLOW:")
    print("1. Apri un nuovo terminale")
    print(f'2. Esegui il comando: mlflow ui --backend-store-uri "{tracking_uri}"')
    print("3. Apri il browser all'indirizzo http://127.0.0.1:5000")
    print("----------------------------------------------------------------")
```
