Ha senso addestrate tutto da zero.
Il primo istinto di un appasionato di deep learning è quello di addestrare sempre un modello nuovo, ma in un'azienda il tempo è la risorsa più preziosa.
La prima domanda che un architetto AI deve porsi non è 'come lo addestro' ma 'devo davvero addestrarlo?'
Esistono ormai servizi maturi che coprono task standard in modo eccellente.
E' necessario valutare se usare API esterne come quelle di OpenAI o AWS, oppure costruire una pipeline proprietaria basata su librerie come PyTorch o Transformers di Hugging Face.

Se fai una semplice analisi di testo comprare l'accesso ad un API esistente è come noleggiare un get privato invece di costruirlo nel garage.
Dobbiamo imparare a distinguere task standard, dove l'intelligenze è una comodity, ed il task core, dove la personalizzazione è il vantaggio competitivo.

Analisi dei Costi del Controllo
Vantaggi e svantaggi delle due strade
Non è solo tecnica è puramente finanziaria.
- Buy (API): basso tempo di rilascio, nessun costo di manutenzione infrastrutturale, ma dipendenza totale dal fornitore e costi variabili scalabili con l'uso.
- Build (Custom): pieno controllo sui dati, privacy garantita, ottimizzazione specifica per il dominio, ma costi iniziali elevati in termini di GPU e ingegneria.
- Hybrid: utilizzo di API per la validazione rapida del prototipo (PoC) seguita da un fine-tuning locale per abbattere i costi in produzione
- Il valore di una scelta si riflette nel ritorno dell'investimento calcolato sulla vita utile del prodotto.

Scenari Decisionali Pratici
- Esempio: Analisi del Sentiment: utilizzare un modello 'studente' piccolo che imita un modello 'insegnante' gigante per mantenere alta la precisione riducendo la latenza. Esempio, analizzare i commenti sociale, conviene utilizzare API è un problema risolto, non perdere tempo
- Esempio: Manutenzione Predittiva: ridurre la precisione numerica dei pesi (es. da FP32 a INT8) per valocizzare i calcoli e dimezzare l'occupazione di memoria RAM. Se hai una turbina sperimentale nessuna AI di google o AWS può aiutarti, devi costruire un modello che impare quella specifica turnia.
- Esempio: Analisi Documentale Privata: scegliere l'architettura in base a dove girerà: un modello per server A100 è diverso da uno ottimizzato per un chip mobile ARM.
Se lavori per una banca o un ospedale i dati non possono uscire dal server, qui il build locale è l'unica strada percorribile.

Costruire non significa solo pagare lo sviluppatore oggi.

Total Cost of Ownership (TCO)
Addestare un modello è come comprare un cucciolo
Oltre al costo di acquisto, poi bisogna nutrirlo (con i dati), curarlo (mantenere il monitoraggio) ed assicurarci che non faccia danno.
Bisogna considerare i costi iniziali ed i costi continui. Un modello custom richiede un monitoraggio costante per rilevare quando l'ai inizia ad invecchiare perchè il mondo è cambiato.
Valuta sempre se il teams è pronto a gestire questa responsabilità nel tempo.

Se decidiamo di costruire, come facciamo a non scegliere una funzionalità obsoleta?

Identificare lo State Of The Art (SOTA)
Scegliere l'architettura vincente.
Il campo dell'AI evolve settimanalmente. 
Quello che era rivoluzionario l'anno scorso, oggi è superato.
Per non basare il proprio progetto su tecnologie obsolete, è fondamentale saper consultare le classifiche di performance globali sui dataset di riferimento.
'Paper with Code' e gli 'Hugging Face Leaderboards' sono diventati il punto di riferimento per selezionare il modello più adatto al proprio compito.
Dobbiamo imparare a leggere le classifiche mondiali ma con occhio critico, non sempre il modello primo in classifica è quello giusto per la nostra applicazione.

Come si misura questa superiorità tecnologica?

Benchmark e Metriche di Riferimento
Valutare la superiorità di un modello
Per confrontare i modelli, usiamo i Benchmark
- Dataset Standard: utilizzo di riferimenti come ImageNet per la visione o GLUE per il linguaggio per confrontare mele con mele
- Metriche SOTA: non guardare solo l'accuratezza, ma anche l'area sotto la curva (AUC), l'F1-score o metriche specifiche come la 'Perplexity' (per linguaggio naturale)
- Architetture Dominanti: riconoscere quando una tecnologia (es. Transformer) ha reso obsolete le precedenti (es. RNN) per determinati task
- Il miglioramento relativo di un modello rispetto alla baseline si misura in termini di riduzione dell'errore.

Per fortuna esiste un portale che raccoglie tutta questa conoscenza per noi.
Utilizzo di Papers with Code
Paper with Code è la nostra bussola, immaginalo come un dizionario dei modelli ai, dove ogni voce ha un link su github
Potete vedere il tradeoff tra la precisione e il numero di parametri. Spesso scopri che un modello che occupa la metà della memoria è quasi identico ed uno in cima alla classifica.

Attenzione a non innamorarsi di un modello troppo complesso per i dati a disposizione
Esiste un limite fisico che un modello può imparare.
Più un modello ha parametri (è più complesso) più è in grado di imparare funzioni difficili, ma c'è una troppola, richiede un volume di dati proporzionale per non cadere in un overfitting catastrofico.
Un modello con troppi parametri rispetto ai dati a disposizione rischia di imparare questi dati a memoria.
Scegliere l'architettura sota significa trovare il punto di equilibrio dove la capacità del modello è perfettamente allineata alla richezza del nostro dataset.

E qui entriamo nel vivo del mestiere, la gestione dei compromessi in produzione
Il Triangolo dei Compromessi
In università cerchiamo il 100% di precisione, in azineda il 100% è un miraggio e per di più un miraggio costoso. Immagina un chatbot che risponde perfettamente ma ci mette 15 secondi a farlo, l'utente ha già chiuso la finestra.
Dobbiamo pensare come ingenieri non come scienziati. In produzione dobbiamo fare i conti con la realtà. Si deve bilanciare la precisione con la velocità (latenza) e il costo computazionale necessario per ogni singola previsione.

Accuratezza vs Latenza vs Costo
La frontiera di Pareto del deployment
- Accuratezza: la capacità del modello di fornire la risposta corretta. Spesso l'ultimo 1% di miglioramento raddopia i costi computazionali. Vuoi più accuratezza?, preparati a pagara di più o ad aspettare di più
- Latenza: il tempo necessario per una singola inferenza. Fondamentale per applicazioni real-time come i chatbot o la guida autonoma. Vuoi un modello più veloce? preparati ad accettare un modello più piccolo, meno forzuto e meno intelligente.
- Costo Inferenza: quanto spendiamo in termini di GPU/CPU per ogni richiesta. Determinare la sostenibilità economica del prodotto su larga scala.
L'ottimizzazione ideale si trova sulla curva che massimizza le performance minimizzando l'uso delle risore.
Business sostenibile ed utenti felici.

Esistono delle tencniche magiche per barare e miglioare tutti e 3 i lati di questo triangolo (accuratezza, latenza, costo) 

Strategie di Ottimizzazione
- Modelli Distillati: utilizzare un modello 'studente' piccolo che imita un modello 'insegnante' gigante per mantenere alta la precisione riducendo la latenza.
- Quantizzazione: ridurre la precisione numerica dei pesi (es. da FP32 a INT8) per velocizzare i calcoli e dimezzare l'occupazione di memoria RAM
- Hardware Target: scegliere l'architettura in base a dove girerà: un modello per server A100 è diverso da uno ottimizzato per un chip mobile ARM

Analisi della Latenza
La latenza non è solo il calcolo dei neuroni, è anche il tempo che i dati impiegano per viaggiare tra RAM e GPU. Il rapporto tra gudadagno in velocità e perdita di accuratezza definisce l'efficienza della nostra tecnica di compressione.
