CONCETTO DI LATENT SPACE: NAVIGARE NELL'ESSENZA DEI DATI

Il concetto di latend space, o spazio latente, è uno dei concetti più importanti nei modelli generativi.
Lo spazio latente è una rappresentazione compatta e appresa dei dati, in cui il modello cerca di conservare le caratteristiche più utili e strutturali.
Non è una memoria 'nascosta' e non è una copia compressa perfetta. E' uno spazio numerico in cui i dati simili tendono spesso a essere rapprentati da punti relativamente vicini.
Partiamo da un immagine:
28x28 pixel = 784 numeri
L'immagine originale può essere rappresentata come:
[pixel1, pixel2, pixel3, ..., pixel784]
Un encoder potrebbe trasformarla in soli 2 numeri:
z=[1.4,-0.7]
Questi 2 numeri costituiscono la rappresentazione latente.
784 valori originale -> encoder -> 2 valori letenti
Il decoder fa il contrario
x originale -> encoder -> z -> decoder -> x ricostruito
z deve contenere abbastanza informazioni per permettere al decoder di ricostruire x

Perchè si chiama lentente?
Latente significa "non direttamente osservabile"

Quando osservi i pixel, dietro a quei pixel ci sono caratteristiche come: forma, orientamento, dimensione, colore, ecc
Il modello non riceve necessariamente queste etichette.
"Nessuno gli dice: questa variabile deve rappresentare l'ordinamento, questa il colore, ecc"
Le rappresentazioni emergono durante il training perchè risultano utili per ricostruire o generare i dati.

Però, attenzione, non è garantito che una dimensione latente rappresenti una caratteristica interpretabile in modo pulito. Potrebbe essere molto più mescolato (anzie è molto probabile così)

Se partiamo da un'immagine, e di usare uno spazio latente di solo 2 dimensioni, dopo il training le cifre simili occupano posizini vicine, solo per il fatto di condividere alcune caratteristiche grafiche.

La maggior parte dello spazio di un'immagine è vuoto o contiene rumore privo di senso
LA ManiFold Hypothesis suggerisce che i dati naturali giacciano su variatà a bassa dimensione immerse in questi spazi vasti. Imparare il 'latent space' significa trovar le coordinate di queste superfici.
Se prendiamo dei pixel a caso otteniamo rumore o spazio vuoto.
Estrarre lo spazio latente significa trovare la mappa della surficie con significato , non lo spazio vuoto
La 'Latent space' è la rappresentazione compressa che casttura solo i gradi di libertà essenziali del sistema, versione asciutta e densa dei dati, dove abbiamo rimosso tutto il superfluo.

Perchè si dice "navigare nello spazio latente"?
Immagina di avere due punti zA= scarpa e zB=stivale, possiamo prendere punti intermedi
z(t)=(1-t) zA+zB
esempio 
t = 0.0 -> scarpa
t = 0.25 -> scarpa leggeremente più alta
t = 0.50 -> forma intermedia
t = 0.75 -> quasi stivale
t = 1.0 -> stivale
Poi passiamo ciascun punto al decoder
z -> decoder -> immagine
Questa operazione si chiama "interpolazione nello spazio latente".
Ed è qui che nasce l'espressione "navigare nello spazio latente".

Non stai modificando direttamente migliaia di pixel
Stai modificando una rappresentazione compatta.

Se nello spazio dei pixel due immagini sono vicine solo se i loro colori si assomigliano matematicamete, nello spazio latente la vicinanza è semantica, riflette una somiglianza di contenuto.
Ogni asse dello spazio latente può arrivare a reppresentare una feature specifica, come l inclinazione di un volto o il colore di un oggetto. Un buon modello generativo 'impaccheetta' i dati in modo che lo spazio latente sia denso e privo di zone morte.
Ogni asse, ogni direzione in questo spazio può diventare un interruttore, spostandoci un poco potremmo cambiare solo i capelli del soggetto interpretato, mantendendo tutto il resto rimane invariato.

L'analogia del foglio accartocciato
Pensa a un foglio di carta (2D) appallottolalo in una stanza (3D). Se guardi solo le coordinate 3D, i punti sembrano complessi, ma la loro natura rimane 2D.
Estrarre lo spazio latente equivale (decoder) a 'spianare' quel foglio per recuperare le sue coordinate naturali e semplici.

Ecco quando se lo spazio è lineare, la matematica diventa intuitiva, in uno spazio latente ben addestrato, possiamo iniziare a fare aritmetica con i concetti, non stiamo sommando bit ma stiamo sommando idee.
La logica umana può essere catturata da semplici operazioni vettoriale
Re -Uomo +donna = regina

Perchè questo è potente?
supponiamo di modificare direttamente un'immagine pixel per pixel
E' facile ottenere: rumore, forme assurde, immagini incoerenti.
Nello spazio latente, invece, il modello può aver imparato una struttura del tipo:
zone plausibili -> volti, scarpe, automobili
Muovendoti in quelle zone, hai maggiori possibilità di ottenere risultati coerenti.

Questo è particolarmente importante nel VAE

Autoencoder normale vs VAE
Autoencoder classico:
x -> encoder -> z -> decoder -> rappresentazione di x
lo spazio latente può essere irregolare, cioò esserci dei buchi tra un vettore e l'altro. Ci possono essere zone dello spazio che non corrispondono a nessun dato sensato.
Se campioni casualmente, potresti capitare proprio in quel 'buco'
Risultato: decoder(z) -> immagine sensa senso.

Il VAE cerca invece di rendere lo spazio latente più regolare
I punti tendono a occupare uno spazio continuo e strutturato
Questo rende possibile: campionamento + interpolazione + generazione, in modo più affidabile

Ma come facciamo a isolare gli attributi?
Il traguardo ideale si chiama Disentanglement
Immagina una radio, hai una manopola per il volume ed una per la frequenza. non vuoi che cambiando il volume cambi la frequenza.
In uno spazio latente disentanglement ogni dimensione controlla un solo attributo indipendente. Spostare un pixel non ha senso, è come cercare di cambiare il finale di un libro cambiando una singola lettere.
Spostare una coordinata latente invece significa riscrivere il concetto.

Le reti neurali agiscono come linearizzatori di dati che in origine non sono lineari, sono profondi e complessi, e li proiettano in uno spazio dove le relazioni sono semplici rette.
Se lo spazio latente è ben ragolarizzato, i concetti simili si allineano lungo direzioni parallele, rindendo possibile l'uso dell'algebra lineare.
Si possono sommare concetti

Interpolazione e Morphing
L'interpolazione è il processo di creazioni di punti intermedi tra due vettori latenti noti. E' ciò che permette di generare transizione fluide, o 'morphing' tra due oggetti.
Ci si può muovere nello spazio latente per esplorare la varietà di campioni che il modello può generare.

Se durante l'interpolazione appaiono immagini prive di senso, significa che lo spazio latente è discontinuo,  ha dei buchi (tipico degli AE semplici). E' il testo definitivo del modello
Modelli come i VAE forzano lo spazio ad essere compatto, assicurando che ogni punto tra due campioni sia a sua volta un campione valido e pronto a generare realtà.
Non siamo limitati a linee rette: possiamo esplorare 'griglie' latenti per vedere come interagiscono due attributi diversi.

Traiettorie Latenti
Muoversi sulla varietà
- L'interpolazione lineare (LERP) è il percorso più semplice: una linea retta tra due punti nello spazio latente.
- Se lo spazio è sferico (tipico di molti modelli generativi), si preferisce l'interpolazione sferica (SLERP)
- Ogni punto lungo la traiettoria viene passato al decoder per visualizzare la trasformazione continua dei dati
- Transizione fluide indicano che il modello ha imparato una rappresentazione continua e non ha semplicemente memorizzato i campioni

Esempio con i volti
Supponiamo di addestrare un buon modello generativo su moltissimi volti.
Lo spazio latente potrebbe implicitamente rappresentare caratteristiche come:
orentameto testa, illuminazione, capelli, forma del viso, ecc
Hai:
z1 -> decoder -> volto A
Modifichi leggermente z1:
z2=z1 + piccola variazione
e potresti ottenere: stesso tipo di volto ma leggermente ruotato
Oppure: stesso volto ma espressione leggermente diversa.

Nei modelli ben strutturati, piccoli movimenti nello spazio latente producono piccoli cambiamente semantici nell'output.

Attenzione però tutto dipende dalla loss.
Se addestriamo il modello solo per ricostuire immagini, imparerà caratteristiche utili alla ricostruzione
Se addestriamo il modello per distinguere anomalie, potrebbe impararne altre
Il latent space non contiene una verità assoluta sul dato.
Contiene ciò che il modello ha imparato essere utile

Esempio industriale
Supponiamo di raccogliere: vibrazione, temperatura, pressione, corrente, ecc di una macchina
Input x = [2.1,7,4.2,12.5,1500]
L'encoder potrebbe trasformare queste 5 variabili in:
z = [0.4,-1.2]
Dopo migliaia di osservazioni potresti ottenere:
zona A - funzionamento normale
zona B - sovraccarico
zona C - comportamento anomalo
Non necessariamente perchè hai fornito quelle caratteristiche, ma perchè i pattern dei sensori sono diversi.
Questo rendo lo spazio latente utile per:
anomaly detection, clustering, visualizzazione, compressione

Esempio ERP
prendiamo un ordine rappresentato da: quantità, prezzo, sconto, frequenza ordini, lead time, valore ordine
Potresti avere
X = 6 variabilii
dopo l'encoder ottenere z = 2 variabili
Nello spazio latente potresti scoprire gruppi:
clienti regolari    ordini eccezionali     possibile anomalia

Questo può essere interessante perchè permette di ridurre molti variabili a una rappresentazione più compatta.

C'è però un punto da ricordare, se vuoi interpretabilità forte, PCA può essere spesso più semplice e trasparente di un autoencoder. Non userei un autoencoder solo perchè è più sofisticato.

Latent space vs PCA
PCA                 100 variabili -> 2 componenti principali
Autoencoder         100 variabili -> rete neurale -> 2 variabili latenti
La differenza principale:
PCA apprende trasformazioni lineari
Un autoencoder può apprendere trasformazioni non lineari
Quindi un autoencoder può rappresentare strutture molto più complesse, ma dall'altro lato ottengo: minore interpretabilità, meggiore complessità, più dati necessari, training, iperparametri
Per questo una PCA rimane assolutamente valida in molti problemi aziendali

```python
import os

# Configurazione del backend di Keras.
# Viene impostato PyTorch come motore di calcolo sottostante prima di importare Keras.
os.environ["KERAS_BACKEND"] = "torch"

import keras
from keras import layers, ops
import numpy as np
import matplotlib.pyplot as plt

# --- 1. PREPARAZIONE DEI DATI ---
# CARICHIAMIO il dataset MNIST composto da cifre scritte a mano.
# Ignoriamo le etichette, per farlo aggiungo _ (y_train, y_test) poiché in un autoencoder 
# l'obiettivo è ricostruire l'input stesso, rendendolo un compito di apprendimento non supervisionato.
(x_train, _), (x_test, _) = keras.datasets.mnist.load_data()

# NORMALIZZAZIONE dei dati nel range [0, 1] per facilitare la convergenza del modello.
# Le immagini originali hanno valori di pixel tra 0 e 255.
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# Appiattimento (RESHAPE)) delle immagini da matrici 28x28 a vettori di 784 elementi.
# Questo è necessario perché utilizzeremo strati densi (Fully Connected).
x_train = x_train.reshape((-1, 784))
x_test = x_test.reshape((-1, 784))

# --- 2. ARCHITETTURA DEL MODELLO (Autoencoder) ---
# Un AUTOENCODER è composto da due parti principali: un Encoder e un Decoder.
# L'obiettivo è comprimere l'informazione in uno spazio ridotto e poi ricostruirla.

# Dimensione dello spazio latente: rappresenta il numero di variabili che il modello
# userà per descrivere le caratteristiche essenziali di una cifra.
latent_dim = 32 

# Definizione dell'ENCODER: mappa l'input (784 pixel) in una rappresentazione compressa (Z).
# L'input riceve il vettore appiattito dell'immagine.
inputs = keras.Input(shape=(784,))
# Strato intermedio per estrarre caratteristiche di alto livello.
x = layers.Dense(128, activation="relu")(inputs)
# Lo strato "Latent_Space" riduce ulteriormente la dimensionalità a 32.
latent_repr = layers.Dense(latent_dim, activation="relu", name="Latent_Space")(x)
# Creazione del sottomodello Encoder.
encoder = keras.Model(inputs, latent_repr, name="Encoder")

# Definizione del DECODER: mappa la rappresentazione compressa (Z) di nuovo nello spazio originale.
# L'input del decoder è il vettore latente prodotto dall'encoder.
latent_inputs = keras.Input(shape=(latent_dim,))
# Strato intermedio che inizia a ricostruire la complessità dell'immagine.
x = layers.Dense(128, activation="relu")(latent_inputs)
# Lo strato finale riporta la dimensione a 784 pixel. 
# Si usa la "sigmoid" perché i pixel normalizzati sono compresi tra 0 e 1.
outputs = layers.Dense(784, activation="sigmoid")(x)
# Creazione del sottomodello Decoder.
decoder = keras.Model(latent_inputs, outputs, name="Decoder")

# AUTOENCODER COMPLETO: Unione di Encoder e Decoder.
# L'output dell'encoder viene passato come input al decoder.
autoencoder = keras.Model(inputs, decoder(encoder(inputs)))

# COMPILAZIONE del modello.
# Si usa 'adam' come ottimizzatore e l'errore quadratico medio (MSE) come funzione di perdita,
# che misura la differenza pixel per pixel tra l'immagine originale e quella ricostruita.
autoencoder.compile(optimizer="adam", loss="mse")
# mse confronta pixel originale con pixel ricostruito

# ADDESTRAMENTO del modello.
# Si noti che l'input (x_train) e il target (x_train) sono identici.
# Il modello impara a comprimere i dati cercando di perdere meno informazione possibile.
print("Inizio addestramento: estrazione delle caratteristiche salienti...")
autoencoder.fit(x_train, x_train, epochs=5, batch_size=256, verbose=0)

# --- 3. INTERPOLAZIONE E MORPHING ---
# Questa funzione permette di passare fluidamente da una cifra all'altra esplorando 
# lo spazio latente creato dall'autoencoder.
def interpolate_digits(digit_a_idx, digit_b_idx, steps=10):
    # Selezione di due immagini dal test set.
    img_a = x_test[digit_a_idx : digit_a_idx + 1]
    img_b = x_test[digit_b_idx : digit_b_idx + 1]
    
    # L'encoder trasforma le immagini reali in vettori nel "linguaggio" del modello (z).
    z_a = encoder.predict(img_a, verbose=0)
    z_b = encoder.predict(img_b, verbose=0)
    
    # Interpolazione Lineare (LERP): calcoliamo punti intermedi tra i due vettori latenti.
    # z(t) = (1-t) * z_a + t * z_b, dove t varia da 0 a 1.
    t_values = np.linspace(0, 1, steps)
    interpolated_points = [(1 - t) * z_a + t * z_b for t in t_values]
    interpolated_points = np.vstack(interpolated_points).astype("float32")
    
    # Il Decoder prende questi punti artificiali e li trasforma in nuove immagini visibili.
    reconstructions = decoder.predict(interpolated_points, verbose=0)
    # Conversione da tensore PyTorch a array NumPy per la visualizzazione.
    return ops.convert_to_numpy(reconstructions)

# Creazione della sequenza di trasformazione tra due cifre.
# Ad esempio, osserviamo come un '3' si trasforma gradualmente in un'altra cifra.
morph_sequence = interpolate_digits(18, 0, steps=12)

# --- 4. VISUALIZZAZIONE DEI RISULTATI ---
# Creiamo una striscia di immagini che mostra l'intera evoluzione del morphing.
plt.figure(figsize=(15, 3))
for i, img in enumerate(morph_sequence):
    ax = plt.subplot(1, 12, i + 1)
    # Ogni vettore di 784 pixel viene riportato alla forma 28x28 per la visualizzazione.
    plt.imshow(img.reshape(28, 28), cmap="gray")
    plt.axis("off")
    if i == 0: plt.title("Inizio")
    if i == 11: plt.title("Fine")

plt.suptitle("Morphing Semantico: Trasformazione fluida nello Spazio Latente")
plt.show()
```
