Il grande scostamento di paradigma
Per decenni l'ai è stata come un critico d'arte, doveva distinguere un falso dall'originale, o un gatto da un cane. Questo è l'approccio discriminativo
Ma i modelli generativi, introducono la capacità di sintesi. Immagina la differenza tra chi sa leggere uno spartito e chi sa comporre una sinfonia.
Significa che il modello non si limirta a trovare il confine tra i dati, ma ne comprende la genesi interna.
Non vogliamo che la macchina riconosca una sedia, ma che ne comprenda l'essenza per progettarne una nuova.
Ma cosa accade tecnicamente quando gli diciamo che una macchina sta immaginando.
Riconoscere un pattern è un lavoro di filtraggio. identifico feature note in input partendo da un contesto esistente.
Immaginare invece, significa campionare nello spazio latente, è come se la macchina avesse un sogno strutturato.
Mentre un classificatore riduce l'informazione ad una etichetta, un bit o pochi pixel, un generatore espande un seme casuale in un output ricco di dettagli.
La gereazione valida è la prova definiva che il modello ha appreso la realtà.
Se sai creare qualcosa di coerente, significa che ne possiedi le regole fisiche o logiche.

I modelli generativi, trasformano la geometria dei pixel in coordinate di conceetti, rendendo l'immagine un'operazione vettoriale.
La capacità di interpolare tra concetti distanti permette di esplorare soluzioni di design o artistiche impossibili per metodi tradizionali.
Imparare a disegnare costringe la rete a scartare il rumore e a concentrarsi sulla varianti fondamentali che definiscono un oggetto.
Immagina lo spazio semantico dove ogni coordinata è un idea. a nord il concetto di felicità a sud la tristezza. L'immaginazione diventa una operazione vettoriale, possiamo sommare o sotrarre concetti matematicametne. Questa creatività computazionale è la capacità di interpolare tra punti distanti, esplorando soluzione artistiche, che l'occhio umano, limitato dai propri bias, non vedrebbe mai.
Inoltre imparare a generare costringe la rete a scartare il rumore di fondo e concentrarsi solo sulle varianti fondamentali che definiscono un oggetto.

Il Salto Logico dell'AI
Perchè la sintesi è più difficile dell'analisi
Un modello che sa disegnare un volto deve necessariamente sapere dove sono gli occhi e come reagiscono alla luce, un classificatore potrebbe cavarsela guardando solo la texture della pelle o la forma del naso. La sintesi insomma richiede una fedeltà totale,  una conoscenza profonda. Ecco perchè matematicamente la perdita generativa è molto più complessa ed istruttura della perdita discriminativa.
Questa necessità di fedeltà totale rende i modelli generativi rappresentazioni molto più complete/fedeli del mondo reale rispetto ai semplici discriminatori.

Ma passiamo al secondo pilastro
Come alimentiamo queste menti artificiali?
Sel Supervised Learning (SSL)
Oggi la risorsa più abbaondante non sono i dati etichettati, ma i dati grezzi.
Nel self supervide learning la tecnica permette all'AI di addestraris su scala massima autonomamente.
Usiamo la generazione come un compito di addestramento, nascondiamo una parte di informazione alla macchina e chiediamo all'ai di generarla. E' il metodo con cui stiamo preaddestrando le ai più potenti del pianteta

MA come funzione l'addestramento senza insegnante
Il compito di ricostruzione funge da etichetta implicita (es. autoencoder e mascking).
Addestrando la rete a ricostruire immagini o testi, essa impara automaticamente concetti di gerarchia visiva o sintassi
I modelli pre-addestrati in questo modo richiedono pochissimi dati etichettati per essere adattati a compiti specifici.
Questa paradigma ha abbattuto i costi di annotazione manuale, permettendo di scalare su dataset di dimensioni planetarie.
Questo crea le base per il transfer learning, modelli che già conoscono il mondo e richiedono poi poichissimi dati per specializzarsi nei vari settori (es. medico, legale)

Un esempio calzante è il Masket Image Modeling. Nascondere porzione di un'immagine CIFRA-10 per forzare l'encoder a dedurre le parti mancanti della coerenza globale. Se vedo 3 zampe la quarta deve essere li. Le feature estratte da questi modelli sono spesso superiore rispetto all'addestramento supervisionate standard.
Le feature estratte sul bottleneck di un autoencoder in SSL sono spesso superiori a quelle ottenute con addestramento supervisionato (embedding universali).
Un modello che 'conosce' già la forma degli oggetti grazie alla generazione impara nuovi task in una frazione di tempo.

Ma come si è evoluta questa logica nel tempo.
La logica dell'autoencoder è l'antenata dei moderni Large Language Models: entrambi imparano prevedendo o ricostruendo.
Comprendere il collo di bottiglia informativo dell'autoencoder ci permette di capire come le macchine distillino la conoscenza pura.
