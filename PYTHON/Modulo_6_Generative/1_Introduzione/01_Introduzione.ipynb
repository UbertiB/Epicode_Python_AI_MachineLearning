INTRODUZIONE AI MODELLI GENERATIVI: OLTRE LA CLASSIFICAZIONE

Il Machine Learning visto finora serve a riconoscere, classificare o prevedere. I modelli generativi imparano la struttura dei dati e producono nuovi dati plausibili.

dati -> apprendimento della distribuzione -> generazione di un nuovo esempio

Significa che il modello genera qualcosa statisticamente compatibile con ciò che ha già vista, non che il modello diventa intelligente nel senso umano.

Un modello generativo è un modello di Machine Learning che apprende la distribuzione statistica dei dati di training e utilizza tale distribuzione per generare nuovi dati plausibili con caratteristiche simili a quelli osservati.
Un classificatore apprende a distinguere categorie, mentre un modello generativo apprende la struttura dei dati per produrre nuovi esempi.

Esemio
partendo dal prodtto chefbot
input: foto di una pizza
output: pizza (probabilità 97.4%)
quindi: immagine -> classe
Un modello generativo fa questo: testo -> nuova immagine
es. 
"generami un'immagine di una pizza margherita con basilico" -> nuova immagine
"elenco ingredienti" -> nuova riceva
foto della pizza -> ingredienti

Il modello generativo non recupera un dati esistente, genera un nuovo elemento sulla base delle regolarità apprese.

modelli discriminativi (modelli visti finora):
    "che cosa rappresenta questo dato?"
modelli generativi
    Come posso produrre un data nuovo ma compatibile con quello che ho imparato?

Che cosa possono generare? tantissime cose:
immagine, testo, audio, musica, video, codici, dati tabellari, struttura molecolari, serie temporali, modelli 3d, ecc

Nel Deep Learning tradizionale ci siamo concentrati sulla classificazione, ovvero nel mappare un input verso una specifica eticheteta (tracciare una linea di confine per separare i dati). Questo approccio viene definito DISCRIMINATIVO.
I modelli generativi cambiano prospettiva: non vogliono separare i dati, ma comprenderne la struttura profonda per poterli ricreare integralmente (non vuole dividere i dati, ma vuole capire i dati, la sua struttura, la sua essenza). Questo approccio viene definito GENERATIVO.

Vantaggio 1
Nei modelli discriminativi cerchiamo una linea o un iperpiano che separi le classi in modo ottimale (crea confini di decisoine).
Nel modelli generativi cerchiamo di capire dove i dti 'vivono' nello spazio multidimenionale (impara la distribuzione dei dati)
Conoscere la distribuzione permette di identificare campioni anomali che non appartengono alla distribuzione nota.

Vantaggio 2
Se conosciamo come i dati sono distribuiti (conosciamo la funzione di probabilità che ha generato i dati reali), possiamo non solo classificare un'immagine (es. quell'immagine è un gatto), ma possiamo anche chiderci quanto quell'immagine (o quanto quel dato) sia 'probabile' nel nostro mondo
Questo abilita due capacità:
    - capacità di gestione dei dati mancanti
    - capacità di generare nuovi scenari non presenti nel dataset originale
Cerchiamo di massimizzare la verosomiglianza.

Obbiettivo della Generazione
L'obbiettivo pratico della generazione è costruire una macchina capace di produrre nuovi dati, che sembrino provenire dalla stessa sorgente dei dati reali.
Vogliamo che i nostri campioni generati siano indistinguibili da quelli della distribuzione originale.
Non vogliamo che il computer sia un bibliocaterio che ci restituisce un libro già letto.
Vogliamo che il computer sia uno scrittore, che dopo aver letto migliaia di libri ne scriva uno nuovo.

Un modello generativo produce campioni che hanno le stesse caratteristiche stastistiche dai dati nel mondo reale. Il confine tra ciò che è catturato (reale) e ciò che è immaginato diventa sottile.
Possiamo addestrare altri modelli su dati sintetici, quando quelli reali sono troppo sensibili (es cartelle cliniche) o troppo rari (es. frode bancaria)

Esisteono diverse strade per arrivare alla generazione
    - modelli che cercano di calcolare esattamente la probabilità di ogni pixel. Modelli a verosomiglianza esplicita.
    - modelli che vogliono imparare a trasformare il rumore dei dati in qualcosa di sensato. Modelli impliciti.

Modelli espliciti: definiscono una funzione di densità. Sono come dei matematici pignoli, vogliono una formula che dica esattamente quanto è probabile ogni dato. Se devo valutare la probabilità di un dato. Se serve precisione statistica
Modelli impliciti: imparano un processo di campionamento tramite una competizione. Non sanno dire la probabilità esatta di un'immagine, ma sanno crearne una che inganna un critico. Se devo creare un nuovo dato. Se serve impatto visivo

I modelli generativi si divisono in 4 grandi famiglie:
    1) VAE Variational Autoencoder: es. immagine volto -> encoder -> decoder -> volto ricostruito. Il modello comprime le caratteristiche fondamentali dei dati in una rappresentazione numerica, poi possiamo campionare tale spazio per creare nuovi dati
    2) GAN: è un sistema competitivo dove Generation e Discrimination hanno obbiettivi opposti
        a) Generator: cerca di creare dati falsi. Esempio provo a creare un volto realistico
        b) Discriminator: cerca di capire vero oppure falso. Es. il volto creato da Generator è falso
    3) Autoregressivi (qui possiamo aggiungere anche i Transformer generativi). genera un elemento alla volta utilizzando quelle precedenti. Per un LLM token1 -> token 2 -> token 3 ... Quindi un modello come GPT, Llama, Qwen, ecc non genera normalmente l'intera frase contemporaneamente, produce sequenza di token
    4) Diffusion Models: si parte dal rumore per generare l'immagine pulita. Si prende l'immagine, aggiungo poco rumore, aggiunto altro rumore, ecc poi impare il percorso inverso cioè: rumore -> meno rumore -> ancora meno rumore -> ... -> immagine  (noise -> image)

GENERATIVA CONDIZIONATA
La maggior parte della Generative AI moderna è conditional generation
Non diciamo semplicemente: 
    "genera qualcosa"
ma
    "genera x dato y"
esempio:
    testo - genera immagine
    immagine - genera testo
    testo - genera testo
    ecc
