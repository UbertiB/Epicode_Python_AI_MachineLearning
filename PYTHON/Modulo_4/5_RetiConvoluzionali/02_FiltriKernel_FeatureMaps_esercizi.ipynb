Data un'immagine in input di forma 64 x 64 pixel, calcola manualmente la dimensione delle feature map in uscita prodotta da un layer convoluzoinale con le seguenti specifiche: Kernel size=5x5, Stride:2 Padding: 2 (equivalente a una cornice di 2 zeri per lato). Successivamente, verifica il risultato scrivendo un piccolo script keras che stampi il model.summary()

```python
import tensorflow as tf
from tensorflow.keras import layers, models
import numpy as np

# --- 1. SIMULAZIONE INPUT ---
# Creiamo un'immagine fittizia 64x64 con 3 canali (RGB)
# La forma richiesta è (Batch, Altezza, Larghezza, Canali)
input_image = np.random.rand(1, 64, 64, 3).astype(np.float32)

# --- 2. DEFINIZIONE LAYER CONVOLUZIONALE ---
model = models.Sequential([
    layers.Input(shape=(64, 64, 3)),
    #Cornice di 2 zeri per lato (Padding:2)
    layers.ZeroPadding2D(padding=2), # Aggiunge 2 pixel di padding su tutti i lati dell'immagine
    # LAYER 1: 
    # Formula 'same': 64 / stride(1) = 64
    # Output: (64, 64, 32)
    layers.Conv2D(filters=32, kernel_size=(5,5), strides=(2, 2), padding='valid', activation='relu'),
    #mettere filters=32 serve per avere 32 feature maps in output, uno individua le linee, l'altro i bordi, poi gli angoli, ecc
    #il numero di filters aumenta la capacità del modello di estrarre caratteristiche complesse, 
    #ma aumenta anche il numero di parametri e il rischio di overfitting
    #il numero di filters aumento layer per layer, perchè i primi layer estraggono caratteristiche semplici (linee, bordi), e bastando pochi filtri
    #mentre nei successivi layer la rete aumenta la complessità delle caratteristiche da estrarre (angoli, forme, texture), e servono più filtri per catturare queste informazioni complesse
    
    # LAYER 2:
    # Formula 'valid': floor((64 - kernel(5)) / stride(2)) + 1
    # floor(59 / 2) + 1 = 29 + 1 = 30
    # Output: (30, 30, 32)
    layers.Conv2D(filters=32, kernel_size=(5, 5), strides=(2, 2), padding='valid', activation='relu')
])

# --- 3. ANALISI DIMENSIONI (Punto chiave 3) ---
# Visualizziamo come cambiano le feature maps
model.summary()

# Output previsto per il primo layer (same padding): 64x64x32
# Output previsto per il secondo layer (valid padding, stride 2): 30x30x32
```
