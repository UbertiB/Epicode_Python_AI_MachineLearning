ARCHITETTURA RESNET: OLTRE I LIMITI DELLA PROFONDITA'


I limti della profondità.
Più si sale, più la vista è bella ed i tagli sono chiari. Ma ad un certo punto la struttura inizia a cedere sotto il suo stesso peso. Reti sempre più profonde per catturare concetti complessi, ma più intelligenza aggiungevamo (layer), portava a prestazioni peggiori (degradation problem).
La ResNet ha risolto questo paradosso con un'intuizione molto semplice.

Perchè le reti neurali diventano smemorate quando crescono troppo?

Fino al 2015 eravamo convinti che 'più profondo è meglio' (fino all'arrivo di ResNet), l'intuizione era semplice: più layer equivalgono a una maggiore capacità di astrazione. Tuttavia si notò che superata una certa soglia, non solo l'accuratezza smetteva di migliorare ma rischiava addirittura di precipitare.
Questo fenomeno non è dovuto all'overfitting, la rete smetteva di imparare perdendo le informazioni già accumulate. 
E' una difficoltà intrinseca nell'ottimizzazione: il segnale dell'errore, partendo alla fine della rete, non riusciva più a tornare all'inizio, attraverso troppi passaggi matematici, perdendo forza fino a scomparire.
Durante la backpropagation il gradiente attraversa tutti i layer, se i valori sono molti piccoli, il gradiente può diventare quasi zero (vanisching gardient) oppure enorme (exploding gradient). Quando il gradiente diventa trascurabile, i layer, soprattutto i primi, non ricevono istruzione sul come cambiare, di conseguenza i primi layer smettono di imparare, le rete converge lentamente, aggiungere layer peggiora il risultato invece di migliorarlo.

Come distinguiamo quesot problema (vanisching gradient) dal comune overfitting?
E' qui che i ricercatori hanno notato il paradosso del degrado.
In teoria, una rete profonda, dovrebbe essere in grado di fare tutto quello che fa una rete snella, semplicemente copiando l'identitità del layer precedente.
Invece l'errore di trading aumentava, questo ci dice che il problema non è la capacità della rete, ma la nostra incapacità di ottimizzarla.

Le Skip Connection
La soluzione proposta da He et al. nel 2015, se il segnale fatica a passare attraverso layer, dobbiamo creare delle  scorciatoie, questo connessioni, chiamate Identity Mappings o Skip Connections , permettono al gradiente di saltare alcuni blocchi di convoluzione.
La rete non è più obbligata a ricostruire l'informazione da zero in ogni passaggio.
Questo connessioni permettono al gradiente di saltare alcuni blocchi di convoluzione. In questo modo, la rete non è più obbligata a ricostruire l'informazione da zero, permettendo alla rete di imparare solo la 'differenza' (il residuo).
Questo cambia completamente il modo in cui la rete impara.
In una rete classica ogni layer cerca di imparare l'intera trasformazione, con ReseNet chiediamo alla rete di imparare solo la differenza tra l'input e l'output desiderato.
Se un layer non serve a nulla, l'ottimizzatore deve spingere i suoi pesi verso lo zero.
In questo caso, grazie alla skip connectoin, il segnale passa inalterato.
Oggi, grazie a quesi circuiti residuali, riusciamo ad addestrare reti con 50,100 e perfino 1.000 livelli, entrando nell'era del Deep Learning veramento profondo.

Anatomia del Blocco Residuale
Progettare l'unità fondamentale di ResNet
Un blocco residuale non è un singolo layer, ma un modulo composto da diverse operazioni disposte strategicamente per massimizzare il beneficio delle skip connections.
Attenzione a gestire le differenza di dimensione tra input eo output e come inserire le funzioni di attivazione in modo che in interrompano il flusso dell'identità.
I pezzi devono essere disposti in modo stretegico convoluzione + normalizzazione e attivazione

Struttura del Modulo
Componenti e connessioni
Il blocco residuali è l'unità fondamentale di ResNet, non è un singolo layer, ma una piccola squadra di operazioni che lavorando in armonia.

Il blocco standard prevede due convoluzioni 3x3 intervallate da Batch Normalization (per manetenere stabili i dati) e ReLu, con il salto che scavalca entrambi i layer e si ricongiunge alla fine.
Se le dimensioni spaziali o il numero di filtri cambiano, si applica una convoluzione 1x1 sul ramo della skip connection per pareggiare i tensori.
La ReLu finale non viene applicata dopo la somma tra input e residuo, permettendo alla rete di processare l'informazione combinata.
Il blocco residuale riduce il numero di parametri necessari rispetto a layer densi grazie all'uso efficiente della località
