ARCHITETTURA VGG: LA POTENZA DELLA SEMPLICITA' E DELLA PROFONDITA'

E' meglio aggiungere complessità o profondità?
La semplicità estrama e la profondità uniforme hanno trasformato le reti neurali in uno standard industriale.
La ripetizoine è la chiave della potenza di VGG

Dopo il successo di AlexNet (con struttura disorganizzata quasi artigianle), i ricercatori del Visual Geometry Group di Oxford si chiesero se fosse possibile ottenere prestazioni migliori non aggiungendo complessità, ma rendendo l'architettura più profonda e regolare.

VGG ha introdotto l'idea di blocchi ripetuti ed identici che si incastrano perfettamente, dove ogni sezione della rete utilizza la stessa configurazione, permettendo di scalare la profondità fino a 16 o 19 livelli ocn una logica cristallina.

VGG è una della architetture più importanti della storia del Deep Learning. E' stata sviluppata dal gruppo di ricerca della Visual Geometry Group dell'University of Oxford e presentato nel 2014 durante la competizione ILSVRC 2014.
Prima di VGG, molte reti neurali usavano architetture molto diverse tra loro:
- kernel 11x11, kernel 7x7, kernel 5x5, strutture complicate
I ricercatori si chiesero:
"Se invece di usare pochi layer complessi usassimo tanti layer semplici?"
La risposta fu VGG
L'idea
VGG utilizza:
- solo convoluzione 3x3
- stride 1
- padding same
- max pooling 2x2
Niente altro

VGG ha una struttura a blocchi ripetuti, la rete è composta da sequenza di layer convoluzoinali seguiti da un MaxPooling che dimezza la risoluzione spaziale.
Ogni volta che la dimensione dell'immagine si riduce, il numero di filtri raddoppia per compensare la perdita di risoluzione spaziale. Perdiamo risoluzione ma guadagnamo cancali di risoluzione semantica.
A differenza di architetture precedenti con kernel di dimensioni diverse, VGG utilizza un unico template per tutta la parte convoluzonale.
La profondità della rete diventa la variabile principale per miglioare l'accuratezza del riconoscimento (16, 19)

L'impilamento di più layer convoluzionali prima del pooling permette di apprendere feature più complesse e non lineari prima di ridurre i dati.
Mettere 2 o più convoluzioni una dietro l'altra, prima di ridurre i dati con il pooling permette alle rete di imparare caratteristiche molto più sofisticate.

VGG utilizza un padding tale da mantenere inalterata la dimensione spaziale durante le convoluzioni, delegando solo al pooling il compito del downsampling.
Questa architettura è diventata la base per molti modelli di Transfer Learning grazie alla sua semplicità nel caricare pesi pre-addestrati.

In VGG non ci sono salti bruschi, si osserva invece una transazione fluida: i primi blocchi si concentrano sulla sulla cattura di textuere e bordi, mentre quelli più profondi assemblano queste informazioni in strutture semantiche.
Questa progressione è facilitata dalla stabilità del design, che evita bruschi cambiamenti nella geometria dei dati attraverso i layer.

Il Segreto dei Filtri 3x3
Perchè piccolo è meglio
La decisone più radicale degli autori di VGG è stata l'eliminazione di filtri grandi come 7x7 o 11x11. tutta l'architettura si basa esclusivamente su kernl 3x3
Questa scelta non è solo estetica, ma si basa su un solido principio matematico riguardante il campo ricettivo e l'efficienza dei parametri.
Per guardare un panorama, userete un canocchiale potente, o una serie di lenti piccole e precise?
In passato si pensavo che per vedere oggetti grandi servissero filtri grandi, VGG usa esclusivamente filtri 3x3 rompendo il tabu dei filtri grandi.
Questa è una mossa matematica brillante per massimizzare l'efficienza.
impilare 3 layer 3x3 produce lo stesso campo ricettovo di unsingolo layer 7x7, ma abbiamo più non linearità (3 funzioni reu invece di 1) inoltre abbiamo un risparmio di parametri 7x7 maggiore di 3x3 x3volte.
VGG è lenta nel processo ma meticolosa nel risultato.

Ogni volta che aggiungiamo un layer 3x3 la rete aggiunge 2 pixel di contesto per lato rispetto al layer precedente. Questo permette di 'vedere' oggetti grandi scomponendoli in passaggi elementare.

VGG ci ha insegnato che per risolvere problemi complessi, bisogna dividerli in tanti piccoli passaggi semplici.
Se la parte convoluzionale è un capolavoro di efficienza, i layer finali sono un vero disastro, estramamente pesante, che costituisce la quasi totalità del peso del modello (ereditarietà da AlexNet) (oltre il 90% del peso risiede nei layer densi finali). Questo succedere perchè l'ultimo volume di convoluzione viene appiattivi passando da un valume 3D ad un vettore 1D.
Per questo dobbiamo utilizzare regolarizzazioni fortissime come il drop out.
Quindi non possiamo andare più in profondità se non cambiamo il modo dei layer finali.
Infatti, dopo VGG è cambiato radicalmnte (ResNet o Inception)


Un numero cos' alto di parametri spesso genera overfittin, inoltre il calcolo di matrici così grandi, rallenta  la fase di predizione rispetto ad architetture che utilizzano Global Average Pooling.

E' una gerarchia di feature, da semplice a complesso.

Lezione di VGG: PIU' LAYER SEMPLICI POSSONO ESSERE MIGLIORI DI POCHI LAYER COMPLESSI



```python


import tensorflow as tf
from tensorflow.keras import layers, models

# --- 1. CONFIGURAZIONE INPUT ---
input_shape = (32, 32, 3)

# --- 2. BLOCCO IN STILE VGG (Due 3x3) ---
# Ogni layer aggiunge una ReLU, aumentando la capacità di apprendimento
vgg_block = models.Sequential([
    layers.Input(shape=input_shape),
    layers.Conv2D(64, kernel_size=(3, 3), padding='same', activation='relu'),
    layers.Conv2D(64, kernel_size=(3, 3), padding='same', activation='relu'),
    layers.MaxPooling2D(pool_size=(2, 2))
], name="VGG_Style_Block")

# --- 3. BLOCCO CON KERNEL GRANDE (Singolo 5x5) ---
# Un solo layer, una sola attivazione ReLU
large_kernel_block = models.Sequential([
    layers.Input(shape=input_shape),
    layers.Conv2D(64, kernel_size=(5, 5), padding='same', activation='relu'),
    layers.Conv2D(64, kernel_size=(5, 5), padding='same', activation='relu'),
    layers.MaxPooling2D(pool_size=(2, 2))
], name="Large_Kernel_Block")

# --- 4. ANALISI COMPARATIVA ---
print(f"Parametri Blocco VGG (Due 3x3): {vgg_block.count_params()}")
print(f"Parametri Blocco Kernel Grande (5x5): {large_kernel_block.count_params()}")

# Ispezione della gerarchia
vgg_block.summary()
```

VGG ci regala la semplicita strutturale. Ci dimostra che kernel 3x3 sono superiori a quelli grandi grazie alla non linearità
