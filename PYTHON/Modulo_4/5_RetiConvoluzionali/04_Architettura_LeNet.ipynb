Yann Le Cun (padre di LeNet-5) disse: diamo i pixel direttamente alla rete e facciamo che sia lei ad imparare la geometria. Fu il 'Big Bang' delle reti convoluzionali.
Il successo di questa architettura risiedeva nella sua capacità di elaborare i pixel direttamente come tensori spaziali, superando la necessità di estrarre manualmente le caratteristiche geometriche dei tratti grafici.

Pioniere della Visione
Dall'automazione postale al Deep Learning
Il dataset MNIST è diventato lo standard di riferimento per i test grazie ai successi ottenuti originariamente da questa architettura su scala industriale.
LeNet ha dimostrato che un modello poteva imparare a riconoscere varianti dello stesso carattere senza essere programmato per ogni possibile inclinazione.
L'uso di piccoli kernel di convoluzione ha permesso di catturare le correlazioni locali dei pixel, un concetto allora rivoluzionario rispetto alle reti densamente connesse.
La struttura di base accetta input normalizzati di dimensione 32 x 32 pixel per gestire i margini delle cifre.
Cera un limite che allora (1998) era invalicabile, la potenza di calcolo.

Efficienza e Loclità
Questa rete girava su processori oggi lenti anche solo per un telecomando, l'efficienza era tutto. Grazie alla condivisione dei pesi, la LeNet riusciva a girare su processori della fine degli anni novanta, riducendo drasticamente le risorse rispetto a un layer denso equivalente.
L'architettua permetteva di rilevare la forma di un numero 7 indipendentemente dalla sua posizione esatta nel campo visivo della scansione.
Il modello è stato il primo a mostrare come la gerarchia di filtri potesse sistituire l'ingegneria manuale delle feature visive.

L'Hardware dell'Epoca.
Perchè LeNet è un capolavoro di ottimizzazione
Nel 1998 la potenza computazionale era miliardi di volte inferiore a quella attuale. LeCun dovette bilanciare il numero di filtri e la profondità dei layer per permettere un'inferenza in tempo reale.
Questo vincolo ha portato alla nascita della struttura Conv-Pool alternata (ancora usata), che oggi consideriamo lo standard ma che all'epoca era il frutto di una ricerca estrema sull'efficienza tensoriale.
LeNet è quindi il trionfo dell'ingenio sulla scarsità di risorse. Ogni layer è li perchè è assolutamente necessario.

La sintassi dell'alternanza architettonica.
Il cuore di LeNet-5 risiede nella sua sequenza di livelli. Non si limita a filtrare l'immagine, ma la condensa progressivamente attraverso l'alternanza tra layer di convoluzione e layer di subsampling (pooling).  
La convoluzione è la fase di osservazione attenta, trai i dettagli, cerca i bordi.
Il pooling è la fase di riflessione, riassume quello che abbiamno visto finora, scarta il superfluo e riduce la risoluzione.
E' come se un architetto prima guarda la trama dei mattoni, poi la finestra ed infine l'intero palazzo.
Questa alternanza permette alla rete di estrarre caratteristiche sempre più astratte riducendo contemporaneamente la risoluzione spaziale, costruendo quella che oggi chiamiamo gerarchia delle feature.
Senza questa alternanza la rete si perderebbe nei dettagli.

Estrarre significato dai pixel.
La convoluzione estrae pattern locali come bordi o angoli;
Il pooling riduce la sensibilità della rete alla posizione precisa di tali pattern.
In LeNet, ogni layer convoluzionale è seguito da un layer di average pooling (allora chiamato subsampling) che dimezza le dimensioni.
L'architettura classica segue lo schema: 
C1 (Conv) - S2 (Pool) - 3 (Conv) - S4 (Pool) - C5 (Conv) - F6 (Dense) - Output
L'alternanza produce una riduzione della risoluzione spaziale $W$ a favore di una maggiore profondità dei canali C
Scambiamo i pixel con le caratteristiche 
Perdiamo il dove esatto del pixel ma guadagnamo il cosa rappresenta.
Ma come facevano a riassumere i dati senza i moderni strumenti?

Subsampling e Connettività
A differenza delle rete moderne che usano il Max Pooling, LeNet utilizzava una media pesata delle finestre 2 x 2 per ridurre le mappe delle caratteristiche (average pooling).
Nonostrante l'average pooling oggi sia meno diffuso, in realtà funzionava benissimo per i caratteri in bianco e nero.
Il modello originale non usava molto padding, accettendo che l'immagine si rimpicciolisse layer dopo layer fino a diventare un vettore di caratteristiche.
L'alternanza permette al layer profondi di 'vedere' porzioni sempre più grandi dell'input originale, passando dai dettagli locali alla forma globale della cifra.
E per quanto riguarda l'attivazione? c'era un'altra differenza fondamentale.

Il Ruolo della Non-Linearità
Dalla Sigmoide alla ReLU
Originariamente, LeNet-5 utilizzava la funzione 'tanh' o la sigmoide come attivazione. Sebbene oggi usiamo esclusivamente la ReLu, l'efficacia di LeNet dimostrò al mondo che la non-linearità era il segreto per imparare concetti complessi.
Il design modulare ha permesso ai ricercatori successivi di sostituire semplicemente le funzioni di attivazione o i tipi di pooling mantenendo valida la struttura portante dell'architettura.
Cosa LeNet ha lasciato in eredità al mondo di oggi?

L'eredità di LeNet
Come è cambiato il design delle CNN negli anni.
Sebbene LeNet-5 sia piccola per gli standard odierni, ogni moderna architettura di Computer Vision è, in ultima analisi, un'espansione dei suoi principi fondamentali.
Esamineremo come la struttura finale di LeNet, che termina con layer densamente connessi, abbia stabilito il workflow classico: feature extraction spaziale seguita da classificazione globale.

Dalla Tradizione alla Modernità
Il ponte verso AlexNet e oltre
LeNet ha introdotto l'idea di aumentare il numero di filtri man mano che la risoluzione spaziale diminuisce per compensare la perdita di informazione.
Il passaggio dai layer convoluzonali a quelli densi (Flattening) è un concetto introdotto da LeCun che rimane invariato in moltissime rete contemporane.
Nonostante la semplicità, LeNet-5 contiene ancora oggi circa 60.000 parametri, una cifra enorme per il 1998 ma minuscola rispetto ai miliardi attuali.
Il numero di parametri totali è dominato dai layer densi finali, un limite che le architetture moderne cercano di mitigare.

Rifiniture Architetturali
LeNet utiizzava filtri 5 x 5; oggi preferiamo serie di filtri 3 x 3 perchè offrono la stessa capacità ricettiva con meno parametri e più non-linearità. Impoilando due filtri piccoli, otteniamo lo stesso campo visivo di uno grande, con meno parametri e più occasioni per inserire non linearità.
Il modello originale non usava molto padding, accettando che l'immagine si rimpicciolisse layer dopo layer fino a diventare un vettore di caratteristiche. 
Le varianti moderne di LeNet utilizzano la Softmax per l'output, mentre l'originale usava una funzione di costo basata su distnaze euclidee chiamata RBF

Perchè studiare LeNet oggi?
Il fondamento della compressione strutturale.
Studiare LeNet è come studiare l'anatomia di base prima di eseguire una chirurgia complessa. Senza capire come convivono convoluzione e pooling in LeNet, è impossibile apprezzare le innovazioni di ResNet o Inception.
Inoltre, LeNet rimane un modello eccellente per dispositivi embedded e microcontrolli dove la memoria è limitata e la velocità di esecuzione è critica. Se devi far riconoscere cifre a un micro controllore minuscolo, con pochissima memoria, non usi un trasformatore gigante, ma una variante della LeNEt.
E' l'essenziale che funziona sempre.



```python
import tensorflow as tf
from tensorflow.keras import layers, models

def build_lenet_modern(input_shape=(32, 32, 1)):
    """
    Costruisce l'architettura LeNet-5 con sintassi moderna.
    
    CALCOLO PARAMETRI (Esempio):
    Formula Conv2D: ((Kernel_W * Kernel_H * Input_Channels) + 1) * Filters
    Formula Dense:  (Input_Units + 1) * Output_Units
    """
    
    model = models.Sequential([
        # --- INPUT LAYER ---
        # Definiamo esplicitamente l'ingresso (Batch, 32, 32, 1)
        layers.Input(shape=input_shape),

        # --- PRIMO BLOCCO CONVOLUZIONALE ---
        # Calcolo: ((5 * 5 * 1) + 1) * 6 = 26 * 6 = 156 parametri
        # Output spaziale (Valid): (32 - 5)/1 + 1 = 28 -> (28, 28, 6)
        layers.Conv2D(6, kernel_size=(5, 5), activation='relu', name="Conv1"),
        layers.AveragePooling2D(pool_size=(2, 2), name="Pool1"), # Dim: (14, 14, 6)

        # --- SECONDO BLOCCO CONVOLUZIONALE ---
        # Calcolo: ((5 * 5 * 6) + 1) * 16 = (150 + 1) * 16 = 2416 parametri
        # Output spaziale (Valid): (14 - 5)/1 + 1 = 10 -> (10, 10, 16)
        layers.Conv2D(16, kernel_size=(5, 5), activation='relu', name="Conv2"),
        layers.AveragePooling2D(pool_size=(2, 2), name="Pool2"), # Dim: (5, 5, 16)

        # --- PASSAGGIO ALLA CLASSIFICAZIONE ---
        # Flatten: 5 * 5 * 16 = 400 neuroni
        layers.Flatten(name="Flatten"),

        # Layer Denso 1: (400 ingressi + 1 bias) * 120 = 48,120 parametri
        layers.Dense(120, activation='relu', name="FC1"),

        # Layer Denso 2: (120 ingressi + 1 bias) * 84 = 10,164 parametri
        layers.Dense(84, activation='relu', name="FC2"),

        # Output Layer: (84 ingressi + 1 bias) * 10 classi = 850 parametri
        layers.Dense(10, activation='softmax', name="Output")
        
    ], name="LeNet-5_Modern")

    return model

# 1. Istanziamo il modello
model = build_lenet_modern()

# 2. Ispezione dettagliata
# Somma totale: 156 + 2416 + 48120 + 10164 + 850 = 61,706 parametri
model.summary()
```
