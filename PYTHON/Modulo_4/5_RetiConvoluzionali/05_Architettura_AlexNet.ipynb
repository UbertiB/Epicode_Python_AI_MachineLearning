ARCHITETTURA ALEXNET: IL RINASCIMENTO DEL DEEP LEARNING

Se la LeNet è stata l'infanzia delle rete convoluzionale la AlexNet è stata il loro risveglio.
Immagina di aver vissuto per decenni cercando di descrivere alla macchina ogni cosa per farla vedere, per poi scoprire che la macchina può imparare tutto da sola molto meglio di noi.

La svolta storica arriva nel 2012 quando la rete AlexNet ha scioccato la comunictà scientifica introducendo la ImageNet Challenge.
Poi, con l'introduzione di ReLu e Dropout ha permesso di addestrare giganti senza che impazzisero.
Il matrimonio tra AI e GPU è stato il vero combustibile che ha reso possibile tutto questo.

Quando le CNN hanno scontitto le Computer Vision classica
Nel 2012, il panorama dell'intelligenza Artificiale è cambiato per sempre. Fino a quel momento, i sistemi di visione artificiale si basavano su caratteristiche estratte manualmente dai programmatori tramite algoritmi statistici.
AlexNex, progettata da Alex Krizhevsky, Ilya Sutsskevere e Geoffreu Hinton, partecipà alla ImageNet Large Scale Visual Recognition Challenge (ILSVRC), ottenendo un margine di vittoria così netto da convincere l'intera comunità scientifica del potenziale delle rete neurali profone, il futuro era appena arrivato.

Numeri che hanno cambiato la storia.
AlexNet non usa solo una matematica diversa ma anche una profondità mai vista prima, con ben 8 livelli pesati tra convoluzone e layer densi. Inoltre il modello fu addestrato su oltre 1.2 milioni di immagini ad alta risoluzione appartenenti a categorie diverse.
Avevamo capito che per fare AI servono due cose: 1) reti profonde 2) oceani di dati.

Il segreto è nel feature learning, invece di progettare noi i filtri AlexNet ha usato la backpropagation per imparà estrarre bordi, texture e forme (feature).
E' la differenza che passa tra disegnare una mappa a mano e usare un satellite GPS.
Per eviatare che la rete memorizzasse semplicemente le immagini, è stata utilzzata la data augmentation, rotazioniu e ritagli delle immagini per aumentare artificialmente la varietà del training set.
In questo modo la rete non imparava quella specifica foto ma il concetto di cane o gatto in ogni sua forma.

Architettura dei Layer
La struttura degli otto livelli.
AlexNet è composta da 5 layer convoluzonali seguiti da 3 layer densi (fully connected). I primi livelli estraggono caratteristiche generiche (vedono linee e texture), mentre gli ultimi si occupano della classificazione finale.
Questa separazione tra estrazione e decisione semantica è diventata il 'blueprint' per quasi tutte le architetture convoluzonali sviluppate negli anni successivi.
Questa struttura è diventata lo standard, lo sguardo spaziale prima e la divisione semantica dopo.
Ma far girare una rete così profonda non era affatto semplice, servivano nuovi trucchi.

ReLu e Dropout
Velocità e Robustezza
Per far funzionare una rete di tali dimensioni, i ricercatori dovettero risolvere due problemi critici: la lentezza dell'addestramento dovuta alla saturazione dei gradienti e l'overfitting massivo.
Le funzioni matematiche, come la sigmoide, erano troppo lente e facevano addormentare i gradienti durante l'addestramento.
Inoltre, con 60 milioni di parametri, la rete rischiava di diventare un pappagallo che impare a memoria il dataset senza capire nulla.
Per sconfiggere questi due mostri furono introdotti la ReLu ed il dropout
La soluzione ridiedeva nell'introduzione di una nuova funzione di attivazione non lineare: la ReLu, e di una tecnica di regolarizzazione brutale ma efficace: il Dropout.

Innovazoni Algoritmiche
Rompere i limiti della sigmoide.
Sostituire la tengente iperbolica con la ReLu permise un addestramento 6 volte più rapido.
La ReLu non satura per valori positivi, mitigando la scomparsa del gradiente che affliggeva le reti profonde.
Durante il training si 'spensero' casualmente il 50% dei neuroni nei layer densi per forzare la rete a non dipendere da singoli pesi.
La funzione ReLu trasforma gli input negativi in zero, mantenendo inalterati quelli positivi.

Efficienza e Generalizzazione
La semplicità matematica della ReLu (un semplice confronto) riduce drasticamente il tempo di calcolo rispetto alle funzioni esponenziali del passato.
Il Dropout impedisce che gruppi di neuroni imparino gli stessi errori, distribuendo la conoscenza in modo più uniforme nell'intera rete.
AlexNet introdusse il MaxPooling con stride inferiore all pool_size, facendo scorrere le finestre di riduzione in modo che si sovrapponessero, rendendo più difficile l'overfitting spaziale dei pattern rilevati.

La Regolarizzazione Moderna
Perchè Dropout è essenziale.
Con circa 60 milioni di parametri, AlexNet avrebbe facilmente memorizzato l'intero dataset. Senza il Dropout, il modello sarebbe stato inutile per predire l'immagini mai viste.
Sebbene oggi esistano tecniche come la Batch Normalization, la combinazione ReLu-Dropout introdotta in questa sede rimane uno dei pilastri fondamentali dell'ingegneria delle reti neurali profonde.
Abbiamo visto gli algoritmi, ma dove trovarono la potenza per farli girare?

L'Era della GPU
Sfruttare il parallelismo dell'hardware videoludico
Un'altra ragione del successo di AlexNet fu di natura puramente ingegneristica: l'uso massivo delle schede grafiche (GPU) per gestire i calcoli matriciali.
All'epoca, addestrare un modello simile su CPU avrebbe richiesto mesi; grazie a un'implementazione custom di CUDA, il team riuscì a completare l'addestramento in circa una settimana.
Ma come hanno fatto a stare tutto quel modello nelle schede video di allora?

Parallelismo Tensoriale
L'infrastruttura dietro il modello.
- GPU NVIDIA: il modello fu distribuito su due schede GTX 580 da 3 GB ciascuna, una quantià di memori allora considerata enorme.
- Slicing della rete: l'architettura della rete di AlexNet è stata divisa a metà, con una porzione della mappa delle caratteristiche su ogni GPU per superare i limiti di VRAM, le due metà comunicavano tra loro solo in determinati layer. Questo slacing è stato l'antenato di tutto cio che oggi permette di addestrare migliaia di dati, come GPT su diverse schede grafiche
- Libreria CUDA: fu necessario scrivere codice di basso livello per gestire la memoria e la comunicazione tra i due chip grafici.
- Il throughput computazionale di una GPU supera quello di una CPU grazie a migliaia di core dedicati ad operazioni aritmetiche semplici.
Il successo di AlexNet ha cambiato anche la storia di NVIDIA

Verso il Deep Learning Moderno
La divisione su due GPU non era una scelta estetica, ma una necessità fisica: i layer densi occupavano troppa memoria per una singola scheda dell'epoca.
Le due metà della rete comunicavano solo in determinati layer, un'idea che ha anticipato l'addestramento distribuito moderno (Cross-GPU Communication)
Il successo di AlexNet convinse NVIDIA a investire pesantemente nel settore AI, trasformando schede nate per i videogiochi in motori per l'intelligenza artificiale.
Le schede video non erano più solo per giocare con i videogiochi, erano diventate il motore della nuova rivoluzione industriale.
NVIDIA ha capito che il futuro non era nel rendering dei pixel ma nel calcolo dei tensori

Simulare AlexNet oggi
Dal codice C++/CUDA alle API Keras
Oggi possiamo implementare un'architettura ispirata a AlexNet in poche righe di codice. Vediamo come definire i layer convoluzionali e dove inserire i blocchi di Dropout.
Il primo layer in origina utilizzava filtri giganti (11 x 11), una pratica che con il tempo è stata sostituita da filtri più piccoli e ripetuti.



```python
import tensorflow as tf
from tensorflow.keras import layers, models

def build_alexnet(input_shape=(227, 227, 3), num_classes=1000):
    model = models.Sequential([
        # --- 1. INPUT LAYER (Approccio moderno) ---
        layers.Input(shape=input_shape),

        # --- LAYER 1: Conv + MaxPool   
        # Grande kernel (11x11) e stride elevato (4) per catturare caratteristiche a bassa frequenza spaziale
        # calcolo dei parametri: ((11*11*3)+1)*96 = 34,944 parametri
        layers.Conv2D(96, kernel_size=(11, 11), strides=(4, 4), activation='relu', name="Conv1"),
        layers.MaxPooling2D(pool_size=(3, 3), strides=(2, 2), name="MaxPool1"),

        # --- LAYER 2: Conv + MaxPool
        # Padding 'same' per mantenere le dimensioni spaziali, kernel più piccolo (5x5) per catturare dettagli più fini
        # Calcolo dei parametri: ((5*5*96)+1)*256 = 614656 parametri
        layers.Conv2D(256, kernel_size=(5, 5), padding='same', activation='relu', name="Conv2"),
        layers.MaxPooling2D(pool_size=(3, 3), strides=(2, 2), name="MaxPool2"),

        # --- LAYER 3, 4, 5: Conv
        # Kernel 3x3 per catturare dettagli ancora più fini, padding 'same' per mantenere le dimensioni spaziali
        # Calcolo dei parametri per Conv3: ((3*3*256)+1)*384 = 885120 parametri
        layers.Conv2D(384, kernel_size=(3, 3), padding='same', activation='relu', name="Conv3"),
        layers.Conv2D(384, kernel_size=(3, 3), padding='same', activation='relu', name="Conv4"),
        layers.Conv2D(256, kernel_size=(3, 3), padding='same', activation='relu', name="Conv5"),
        layers.MaxPooling2D(pool_size=(3, 3), strides=(2, 2), name="MaxPool5"),
        
        # --- CLASSIFICAZIONE ---
        layers.Flatten(name="Flatten"),

        # Dense 1: (Input Flatten (9216)+1) *4096 = 37,752,832 parametri
        layers.Dense(4096, activation='relu', name="FC1"),
        layers.Dropout(0.5, name="Dropout1"),
        
        # Dense 2: (4096+1)*4096 = 16,781,312 parametri
        layers.Dense(4096, activation='relu', name="FC2"),
        layers.Dropout(0.5, name="Dropout2"),

        # Output Layer: (4096+1)*num_classes = 4,097,000 parametri (per num_classes=1000)
        layers.Dense(num_classes, activation='softmax', name="Output")
    ], name="AlexNet")

    return model


def build_super_lenet(input_shape=(32, 32, 1)):
    model = models.Sequential([
        layers.Input(shape=input_shape),

        # --- PRIMO BLOCCO: Sostituito 5x5 con due 3x3 ---
        # Conv 1a: ((3*3*1)+1)*6 = 60 param. Output: (30,30,6)
        layers.Conv2D(6, kernel_size=(3, 3), activation='relu', name="Conv1a"),
        # Conv 1b: ((3*3*6)+1)*6 = 330 param. Output: (28,28,6)
        layers.Conv2D(6, kernel_size=(3, 3), activation='relu', name="Conv1b"),
        # Sostituito Average con Max Pooling
        layers.MaxPooling2D(pool_size=(2, 2), name="MaxPool1"), # Output: (14,14,6)

        # --- SECONDO BLOCCO: Sostituito 5x5 con due 3x3 ---
        # Conv 2a: ((3*3*6)+1)*16 = 880 param. Output: (12,12,16)
        layers.Conv2D(16, kernel_size=(3, 3), activation='relu', name="Conv2a"),
        # Conv 2b: ((3*3*16)+1)*16 = 2320 param. Output: (10,10,16)
        layers.Conv2D(16, kernel_size=(3, 3), activation='relu', name="Conv2b"),
        # Sostituito Average con Max Pooling
        layers.MaxPooling2D(pool_size=(2, 2), name="MaxPool2"), # Output: (5,5,16)

        # --- CLASSIFICAZIONE (Invariata) ---
        layers.Flatten(name="Flatten"),
        layers.Dense(120, activation='relu', name="FC1"),
        layers.Dense(84, activation='relu', name="FC2"),
        layers.Dense(10, activation='softmax', name="Output")
        
    ], name="Super-LeNet")

    return model

# Istanza e verifica
super_lenet = build_super_lenet()
super_lenet.summary()
```
