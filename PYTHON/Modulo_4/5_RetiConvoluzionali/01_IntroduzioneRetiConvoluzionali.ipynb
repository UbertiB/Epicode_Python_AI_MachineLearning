RETI CONVOLUZIONALI (CNN)
Le CNN (Convolutional Neural Networks) sono reti neurali progettate specificatamente per elaborare immagini.

Visione artificiale
Passiamo da un'intelligenza numerica ad un'intelligenza spaziale, dove le macchine hanno imparato a distinguere gli occhi dal naso dalla bocca.

Le architetture densamente connesse (ANN) falliscono nell'elaborazione delle immagini.
Nel layer Dense ogni neurone è collegato a ogni singolo input del livello precedente. Questo approccio, potente per dati tabulari, fallisce quando passiamo con immagini ad altra risoluzione.
Il problema risiede nella perdita della struttura spaziale: appiattire un'immagine in un unico vettore distrugge le relazioni di vicinanza tra i pixel, costringendo la rete a imparare da zero ogni singola variazione di posizone.
E' come avere un castello di mattoni lego, appiattire l'immagine significa prendere ogni singolo mattoncino e metterli tutti in fila. Il castello è sparito, abbiamo distrutto le informazioni spaziali.
Oltre a predere la forma perdiamo anche il controllo della memoria computazionale.

Un'immagine a colori da 244 per 244 pixel genera circa 150.000 ingressi. Un solo layer denso con 1.000 neuroni richiederebbe 150 milioni di pesi addestrabili.
Un numero così elevato di parametri rende il modello estremamente incline all'overfitting, richiedendo una quantità di dati di addestramento sproporzionata.
La memoria necessaria per memorizzare questi pesi e la potenza di calcolo per la backpropagation superano repidamente le capacità hardware standard.
La crescita dei parametri in una rete neurale classica segue una legge moltiplicativa tra input e output.

Nelle reti dense ogni neurone è un osservatore globale, cerca cioè di estrarre un'informazione dall'intera immagine contemporaneamente, ignorando sia i pattern visivi sia pattern più localizzati.
Se un pattern appreso si sposta di un solo pixel, una rete densa non è in grado di riconoscerlo perchè i pesi sono ancorati a specifiche coordinate del vettore di input. 
L'eccessiva complessità dei parametri rende il segnale dell'errore molo numeroso, rendendo difficile la convergenza verso il minimo globale della funzione di costo.

Le CNN si ispirano al funzionamento del sistema visivi animale. Gli studi di Hubel e Wiesel hanno dimostrato che i singoli neuroni della corteccia rispondono a stimoli presenti in una piccola regione limitata.

Questa scoperta ha portato alla creazione di modelli che non guardano tutto insieme ma scansionano l'ambiente per rilevare bordi, angoli e forme semplici prima di comporle in oggetti complessi.

Una CNN non guarda l'intera immagine contemporaneamente, guarda piccole zone (kernek o filtro)

Alcuni neuroni si attivano solo se esiste una retta verticale, altri reagiscono invece solo alle curve. Questo ha portato alla conoscenza delle reti convoluzionali.
Il segreto di tutto ciò sta nel layer convoluzionale. Invece di avere pesi diversi per ogni pixel, la rete utilizza un piccolo set di pesi, chiamato filtro o kernel, che viene fatto scorrere sull'intera immagine. E che avere una piccola lente di ingrandimento quadrata che facciamo scorrere su tutta la fotografica.
Questo introduce la 'condivisione dei parametri', non usiamo pesi diversi per ogni zona dell'immagine, ma usiamo la stessa. La conoscenza del nostro filtro se la porta dientro per scansionare ogni centimetro della foto. Da questo meccanisco di scorriemento deriva la proprietà cardine dellle CNN, INVARIANZA ALLA TRASLAZIONE

Riconoscere l'oggetto ovunque esso sia
- Grazie allo scorrere del filtro, se la rete impara a riconoscere un gatto in basso a destra, la rete saprà riconoscere il gatto anche se si sposta al centro dell'immagine, sarà quindi in grado di identificarlo in qualsiasi parte dell'immagine.
- L'invarianza è garantita dal fatto che la stessa trasformazione matematica viene applicata uniformemente a tutte le finestre locali dell'input.
- Questa proprietà rende le CNN incredibilmente robuste rispetto ai cambiamenti di inquadratura o allo spostamento dei soggetti.
- L'operazione di convoluzione applica un nucleo a ogni sotto-regione locale producendo una mappa delle attivazioni.
Macchine finalmente capaci di gestire il mondo reale, dove i soggetti non sono mai centrati o immobili, è la differenza tra una segnaletica rigista ed uno sguardo fluido sulla realtà.
Ma come fa un filtro così piccolo ad essere così efficiente?

Parameter Sharing in Dettaglio
Se uso un filtro 3 per 3, quindi crea un quadrato di 9 pixel che scorre su tutta l'immagine, stiamo usando solo 9 pesi per scansionare l'intera immagine, quei 9 numeri lavoano su milioni di pixer producendo una feature-map,ogni piano di questa mappa rappresenta la risposta dell'immagine ad un particolare rilevatore, uno cercerà le linee, uno le ombre, uno i colori. I primi layer vedono gli atomi, quelli successivi iniziano a vedere la struttura dell'immagine. 
C'è un altro segreto dietro a questa efficienza: La solitudine dei neuroni

Nelle CNN ogni neurone è orgogliosamente miope, dipende solo da un piccolo numero di pixel vicini. Questa 'connettività sparsa' elimina il rumore derivante da pixel distanti che non sono correlati al pattern locale. Un pixel nell'angolo sinistro, non ha alcuna correlazione con uno nell'angolo destro per definire le connesioni locali.
Riducendo le iterazioni inutili (le connesioni inutili), la rete diventa veloce da addestrare e molto più efficace nel generalizzare le informazioni visive su dati mai visti prima.
Ma se ogni neurone vede un solo pezzetto, come facciamo a vedere l'intero quadro?

Il campo Ricettivo
La finestra di attenzione del neurone
Il campo ricettivo definisce l'area dell'immagine di input che influenza l'attivazione di un determinato neurone. E' il concetto spaziale che governa la risoluzione della nostra rete.
Immagine un sistema di sorveglianza, ogni guardia guarda un solo monitor (campo ricettivo locale), ma il capo della sicurezza guarda tutti i monitr insieme (campo ricettivo globale).
Un neurone del primo livello vede solo un dettaglio minuscole, ma un neurone dei successivi livelli, riceve informazioni da molti neuroni dei livelli precedente espandendo gradualmente la sua vista.

Geometria del Campo Ricettivo
L'espansione della percezione attraverso i layer
- In un singolo layer, il campo ricettivo coincide con la dimensione del kernel/filtro, solitamente matrici piccole come 3 per 3 o 5 per 5.
- Man mano che scendiamo nei layer (man mano che andiamo in profondità) il campo ricettivo effettivo aumenta: un neurone nel decimo layer può avere un'influenza che copre l'intera immagine originale.
E' così che la rete percepisce la informazioni a lungo raggio, i primi layer vedono i peli di un cane, i successivi l'orecchio, i layer profondi vedono l'intero cane e capiscono che sta correndo in un prato.
Bilanciare questi filtri è molto delicato.

Progettazione dei Filtri
un campo ricettivo troppo piccolo, impedisce alla rete di vedere l'oggetto nel suo insieme (incapace di gestire il contesto), portanto a errori di classificazione contestuale.
Ma aumentare la prodondità della rete può essere pericoloso se abusato, il giusto equilibrio di espansione è il segresto per non far esplodere i parametri.
Aumentare la profondità della rete è il modo più efficiente per espandere il campo ricettivo senza aumentare esponenzialmente il numero di parametri.
Ogni pixel della feature-map finale porta con sè il riassunto informativo di tutto il suo campo ricettivo nel volume di input originale.

Confronto tra ANN e CNN
Confrontiamo il numero di pesi necessari per processare una piccola immagine cifrata 28 per 28 usando entrambi gli approcci, ANN e CNN, dimostrando matematicamente come la convoluzione permetta di ottenere una capacità rappresentativa superiore utilizzando una frazione minima delle risorse computazionali.
Una rete densa (ANN) usa milioni di parametri, una rete convoluzionale (CNN) grazie alla condivisione dei parametri, usa una frazione minima delle risorse

CNN Classica:
Input
↓
Conv2D: layers.Conv2d(filter=32,kernel_size=(3,3),activation="relu")  32 filtri, ognuno grande 3x3, ogni filtro cerca una caratteristica diversa
↓
ReLU: layer.ReLu() (introduce la non linearità)
↓
MaxPooling (riduce le dimensioni, serve a ridurre la memoria, i calcoli l'overfitting)
↓
Conv2D
↓
ReLU
↓
MaxPooling
↓
Flatten  (da dare ai Dense finali)
↓
Dense
↓
Output



```python
import tensorflow as tf
from tensorflow.keras import layers, models

# --- 1. CONFIGURAZIONE INPUT ---
# Immagine 28x28 a scala di grigio (1 canale)
input_shape = (28, 28, 1) 

# --- 2. APPROCCIO ANN (LAYER DENSO) ---
# Dobbiamo appiattire l'immagine in un vettore di 784 elementi
model_dense = models.Sequential([
    layers.Flatten(input_shape=input_shape),
    layers.Dense(32, activation='relu', name="Layer_Denso")
], name="Architettura_ANN")

# --- 3. APPROCCIO CNN (LAYER CONVOLUZIONALE) ---
# Manteniamo la struttura 2D. Usiamo 32 filtri da 3x3
model_conv = models.Sequential([
    layers.Input(shape=input_shape),
    layers.Conv2D(32, kernel_size=(3, 3), activation='relu', name="Layer_Convoluzionale")
], name="Architettura_CNN")

# --- 4. CONFRONTO ANALITICO ---
print("CONFRONTO PARAMETRI:")
print(f"{'Modello':<25} | {'Parametri Totali':<20}")
print("-" * 50)
print(f"{model_dense.name:<25} | {model_dense.count_params():<20}")
print(f"{model_conv.name:<25} | {model_conv.count_params():<20}")

# Ispezione dettagliata
print("\nDETTAGLIO ANN:")
model_dense.summary()

print("\nDETTAGLIO CNN:")
model_conv.summary()
```

CNN rispetto a ANN riesce a capire di più con meno neuroni.
