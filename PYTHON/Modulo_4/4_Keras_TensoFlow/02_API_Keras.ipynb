API FUNZIONALE DI KERAS: PROGETTARE ARCHITETTURE COMPLESSE

Il modello Sequential di Keras è semplice ma limitato, il flusso è rigidamente lineare
input -> layer1 -> layer2 -> layerx ->output

Keras collega automaticamente ogni layer al successivo. Non puoi avere: - due rami paralleli, concatenazione, input multipli, output multipli, riutilizzo layer, reti a grafo.
La Functinal API invece tratta la rete come un grafo computazionale.

Ogni leayer diventa una funzione che riceve tensori e restituisce tensori, quindi non impili layer ma colleghi tensori.

La rete non viene eseguita immediatamente, Keras costruisce una rappresentazione del flusso:
input tensor -> dense -> dense -> output tensor

- Limiti del modello Sequential
- Layer come funzioni matematiche richiamabili, non più come pezzi di una lista.
- Esercitazione (modello ha due rami, che riceve dati diversi)

LIMITI DELLA LINEARITA'
Quando la struttura a pila (Sequential) non è sufficente.
Il modello Sequential organizza i layer come una pila lineare, dove ogni strato ha esattamente un input ed un output. E' come un treno su un unico binario, parte dalla stazione A ed arriva alla stazione B passando per fermete intermedie fisse.
Questo approccio è perfetto per problemi semplici, ma mostra limiti invalicabili nella ricerca moderna, dove abbiamo bisogno di svincoli.

Esistono problemi che richiedono simultaneamente fonti di dati diverse, come problemi che richiedono contemporaneamente un'immagine ed una descrizione testuale o architetture che prevedono connessioni trasversali tra layer non consecutivi, per potare informazioni fresche direttamente alla fine della rete.
Il treno deve diventare una rete autostradale, gestendo ingressi ed uscite multipli.

Sistemi Multi-Input e Multi-Output
La flessibilità dei grafi computazionali.
Pensa ad un medico che deve fare una diagnosi, non guarda solo gli esami del sangue, guarda le radiografie, ascolta battiti del sangue, questo è un modello multi-input.

- Un modello multi-input permette di combinare feature eterogenee, ed esempio dati tabulari numerici e dati non strutturati come audio o video, in un unico decisore neurale.
- Le architetture a grafo aciclico diretto permetttono di creare ramificazioni dove l'informazione può essere divisa e poi riunita successivamente nel processo di addestramento.
- Le 'Skip Connections' sono fondamentali per le reti molto profonde: permettono al gradiente di saltare alcuni layer facilitando la convergenza ed evitando la degradazione del segnale. Risolvendo il problema delle reti molto profonde
Il passaggio dal modello sequenziale al funzionale trasforma la topologia di una lista a un grafo complesso.

Ma la flessibilità non finisce qui, ci sono casi d'uso ancora più complessi

Casi d'uso Avanzati
L'API funzionale è come un cortellino svizzero, ci permette di creare modelli multi-task con un solo addestramento impara a fare più cose come riconoscere un oggeetto e contemporaneamente capire in che posizione si trova.

- Modelli Multi-Task: è possibile addestrare una singola rete per risolvere più compiti contemporaneamente, fornendo output differenti basati sulla stessa elaborazione di base delle feature
- Condivisione dei Pesi: l'API funzionale consente di riutilizzare lo stesso layer in punti diversi del grafo, permettendo al modello di imparare pattern comuni tra diversi flussi di dati, come se avessi due occhi nella rete
- Modelli Feedback: sebbene Keras sia ottimizzato per grafi aciclici, l'API funzionale facilita la creazione di strutture  con connessioni residue tipiche delle architetture ResNet.

Dall'insieme alla Funzione
Logica di programmazione funzionale
Per padroneggiare tutto questo dobbiamo cambiare la logica di programmazione.
Passare all'API funzionale significa smettere di vedere il modello come un contenitore 'Sequential' e iniziare a vederlo come una vera e propria funziona matematica composta. Ogni layer è un operatore.

Matematicamente stiamo scrivendo una funzione composta gigante: y=f(g(h(m(..(x))))), questo ci da una libertà totale decidendo esattamente dove ogni tensore deve andare e come deve essere trasformato.

Ogni layer diventa un operatore che accetta un tensore in ingresso e ne restituisce uno in uscita, permettendoci di orchestrare il flusso dei dati con la stessa libertà con cui scriviamo codice Python puro

I Layer come Funzioni
Sintassi e richiamo dei tensori.
Nell'API funzionale, i layer sono oggetti richiamabili, come le funzioni di paython. Quando richiamiamo un layer su un input (tensore di input), non stiamo solo definendo la struttura ma creando un legame tra tensori, tutto parte dall'oggetto input.
Pensa alla porta di casa, devi definire la dimensione, e che ospiti (dati) può far entrare, senza questa definizione esplicita il grafo non può iniziare ad esistere.
Questo approccio esplicito richiede la definizione manuale dell'input tramite oggetto 'input' di Keras, che funge da segnaposto per la forma dei dati che la rete riceverà

Sintassi Funzione
Il collegamento dei nodi
La sintassi funzionale è elegante e potente.
Definiamo il layer e subito dopo (tra parentesi tonde) gli passiamo il tensore di input.
E' come il passsaggio di testimone in una staffetta, l'output del layer
 precedente diventa immediatamente input per il layer successivo.
 Alla fine il modello non è più una lista, ma un oggetto che sa dove inizia (input) e dove finisce (output), indipendentemente dai giri, tortuosi, che i dati hanno fatto nel mezzo.

- Ogni riga di codice definisce sia il layer che il suo collegamento immediato: l'output del layer precedente viene passato come argomento tra parentesi tonde al layer attuale..
- l'oggetto input definisce la 'shape' e il tipo di dato iniziale, agendo come radice del grafo computazionale che stiamo costruendo.
- L'istanza finale del modello viene creata specificando semplicemente quali tensori rappresentano gli ingressi e queli le uscite, indipendentemente dalla complessità interna.
La notazione funzionale permete di scrivere esplicitamente la dipendenza tra tensori di ogni livello.

Manipolazione dei Tensori
Una delle operazioni più potenti è la minipolazione dei flussi
Nell'API funzionale i tensori sono come fiumi, a volte vogliamo che i fiumi si uniscano per formare un corso d'acqua più grande, questa è la concatenazione. 
- Concatenazione: Possiamo fondere informazioni provenienti da rami diversi usando layer speciali come 'concatenate' o 'add', 
- Ispezione dei nodi: possiamo anche 'spiare' dentro la rete, accedendo agli output itermedi, come avere dei snesori lungo tutto il percorso del fiume per monitorare la qualità dell'acqua prima che arrivi alla foce. Questo facilita il debugging e permette la visualizzazione delle feature map in tempo reale.
Una volta disegnato il nostro grafo, dobbiamo renderlo pronto all'uso

Definizione del Modello Finale
L'oggetto Model
A differenza del Sequential che viene costruito aggiungendo pezzi, l'oggetto 'Model' funzionale viene istanziato solo alla fine, quando il grafo è completo. Dobbiamo solo dire a Kera, guarda questi sono i miei punti di ingresso e questi sono i miei traguardi.
Questa separazione (tra la definizione dei layer e la creazione del modello) è fondamentale, rende il modello un'entità autonoma e offre una flessibilità enorme nel modificare o estendere architetture esistenti senza riscrivere l'intero sistema. Massima espressione della modularità

Esercitazione: Modello a Due Rami
Sincronizzazione flussi indipendenti.
Metteremo in pratica quanto appreso costruendo una rete che riceve due input numerici differenti. Ogni input verrà elaborato da un ramo indipendente prima di essere fuso in un unico corpo centrale.
Questo esempio simula situazioni reali dove diverse feature richiedono elaborazioni preliminari distinte prima di poter contribuire alla classificazione finale.
Esempio vogliamo predire il prezzo di una casa utilzzando sia dati numerici, come il numero di stanze, metri quadri, sia dati non numerici come dati testuali o immagini, come la distinzione del quartiere

In questa simulazione utilizzeremo due input numerici, ma la logica è identica.
Ogni informazioni ha il suo canale preferenziale di elaborazione, prima di incontrarsi con l'altra informazione.

Fusioe dell'Informazione
L'unione dei rami.
- Il primo ramo processa il primo input attraversando una serie di layer densi, estraendo pattern specifici per quella porzione di dati
- Il secondo ramo lavora in parallelo sul secondo input, mantenendo pesi e bias indipendenti dal primo ramo durante questa fase.
(sono come due specialisti che studiano due aspetti diversi dello stesso problema)
- Il comando 'layers.concatenate' unisce i due vettori di feature risultanti in un unico tensore che contiene la conoscenza combinata di entrambi i flussi. I due specialisti mettono insieme i loro appunti, il risultato è un tensore unico che contiene la conoscenza combinata di entrambi i flussi e pronto per la decisione finale.
La fuzione tramite concatenazione somma le dimensioni delle feature dei due rami prendendone per il layer di output finale

Visualizzazione del Grafo
Un grafo può essere difficile da visualizzare solo con il codice, per questo Keras si offre strumenti per vedere, graficamente, come i rami si sviluppano.
- Topologia non lineare: usando strumenti di visualizzazione vedremo come i rami si sviluppano parallelamente per poi convergere in un unico imbuto decisionale.
- Gestione degli input: in fase di training, dovremmo passare una lista di input invece di un singolo array, rispettando l'ordine definito nella costruzione del modello.
- Stabilità del Merge: discuteremo come la dimensione dei layer nei singoli rami influenzi il peso relativo che ogni sorgente avrà nella decisione finale.

Dati Eterogenei in Prarica
oltre i numeri semplici
Sebben in questo esercizio useremo vettori numerici, la stessa logica si applica se un ramo fosse una rete convoluzionale per immagini e l'altro una rete ricorrente per testo.
L'API funzionale è il linguaggio universale di Keras che permette a queste diverse 'anime' della rete di comunicare e collaborare verso un obbiettivo comune.

E' qui che il Deep Learning smette di essere un semplice esercizio matematico e diventa un sistema di percezione multi-sensoriale.

```python
import tensorflow as tf
from tensorflow.keras import layers, Model
import numpy as np
from tensorflow.keras.utils import plot_model

# =============================================================================
# 1. PREPARAZIONE DEI DATI (Dataset Sintetico)
# =============================================================================

# Creiamo 500 esempi per ogni sorgente dati.
# input_tecnico_data: 10 colonne (es. parametri audio, frequenze).
input_tecnico_data = np.random.random((500, 10))

# input_social_data: 5 colonne (es. numero di condivisioni, commenti).
input_social_data = np.random.random((500, 5))

# targets: L'obiettivo è binario (0 o 1), ad esempio "Successo" o "Insuccesso".
# Usiamo shape (500, 1) per corrispondere all'output del neurone finale.
targets = np.random.randint(0, 2, size=(500, 1))

# =============================================================================
# 2. DEFINIZIONE DELL'ARCHITETTURA (API FUNZIONALE)
# =============================================================================

# Utilizziamo layers.Input per dichiarare la forma (shape) dei dati in ingresso.
# name aiuta a identificare i layer nel summary e nei grafici.
input_tecnico = layers.Input(shape=(10,), name="Ingresso_Tecnico")
input_social = layers.Input(shape=(5,), name="Ingresso_Social")

# --- RAMO TECNICO ---
# Elaboriamo i dati tecnici con layer densi. 
# Sintassi funzionale: Layer()(Input) crea un legame diretto tra i nodi.
x = layers.Dense(16, activation='relu')(input_tecnico)
x = layers.Dense(8, activation='relu')(x) # Riduce a 8 feature estratte

# --- RAMO SOCIAL ---
# Elaboriamo i dati social separatamente. Usiamo meno neuroni perché l'input è più piccolo.
y = layers.Dense(4, activation='relu')(input_social) # Riduce a 4 feature estratte

# --- FUSIONE DEI RAMI (CONCATENATE) ---
# Uniamo i due flussi. Il vettore risultante avrà dimensione 8 (da x) + 4 (da y) = 12.
# Questa fase permette alla rete di trovare correlazioni tra dati tecnici e social.
combined = layers.concatenate([x, y])

# --- TESTA DEL MODELLO (HEAD) ---
# Aggiungiamo layer comuni che lavorano sulle informazioni già fuse e filtrate.
z = layers.Dense(8, activation='relu')(combined)

# Output finale: 1 solo neurone con attivazione Sigmoid per dare una probabilità [0, 1].
output_finale = layers.Dense(1, activation='sigmoid', name="Output_Successo")(z)

# =============================================================================
# 3. CREAZIONE E COMPILAZIONE DEL MODELLO
# =============================================================================

# L'oggetto Model definisce i confini del grafo: dove inizia (lista input) e dove finisce.
model = Model(inputs=[input_tecnico, input_social], outputs=output_finale)

# Compiliamo specificando l'ottimizzatore e la funzione di perdita per classificazione binaria.
model.compile(optimizer='adam', 
              loss='binary_crossentropy', 
              metrics=['accuracy'])

# Visualizza la struttura: noterai come i due input iniziali convergono verso il layer 'concatenate'.
model.summary()

# =============================================================================
# 4. ADDESTRAMENTO
# =============================================================================

# Quando il modello ha più input, il parametro 'x' deve ricevere una LISTA di array.
# L'ordine nella lista deve corrispondere esattamente all'ordine definito in Model(inputs=[...]).
model.fit(
    x=[input_tecnico_data, input_social_data], 
    y=targets, 
    epochs=5, 
    batch_size=32
)

print("\nModello addestrato con successo gestendo input eterogenei.")
```

La flessibilità dell'API funzionale rompe la linearità di Sequential permettendo la creazione di grafi complessi, multi-input e multi-output
Ogni layer è una funzione che trasforma tensori, rendendo la progettazione dell'architettura esplicita e altamente personalizzabile.
L'unione di diversi rami di calcolo tramite concatenazione permette di modellare problemi dove i dati provengono da sorgenti multiple.
l'intretrazione di flussi diversi è la chiave per risolvere problemi del mondo reale
