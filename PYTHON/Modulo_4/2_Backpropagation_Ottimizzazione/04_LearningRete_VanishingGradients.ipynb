Per capire questi problemi bisogna analizzare lo strumento di controllo a disposizione il LEARNING RATE.

ANALISI DELLA DINAMICA DI OTTIMIZZAZIONE: cosa succede quando il passo di apprendimento è troppo lungo o troppo breve.

Il LEARNING RETE (ETA) è la manopola (iperparametro) più importante e critico del Deep Learning. Determina quanto profondamente modifichiamo i pesi a ogni iterazione basandoci sul gradiente calcolato.
Se il gradiente ci dice in che direzione muoverci, il learning rate ci dice quanto deve essere lungo il passo. Bilancia la velocità con la precisione.

Un valore errato può trasformare un modello promettente in un sistema che non converge mai o osclilla violentemente attorno alla soluzione ottimale senza mai raggiungerla.

LEARNING RATE ERRATO:
- OVERSHOOTING: se il passo è troppo lungo, l'algoritmo scavala il minimo dlela funzione di perdita, finendo in una zona con errore maggiore. E' come colpire la pallina troppo forte e finisce fuori dal campo
- DIVERGENZA: una tasso di apprendimento eccessivamente aggressivo può portare la funzione di perdita a crescere invece di diminuire, rendendo il modello instabile. La pallina non solo esce dal campo ma continua ad allontanarsi
- LENTEZZA ECCESSIVA: un valore troppo piccolo richiede un numero proibitivo di epoche per convergere, rischiando di rimanere intrappolati in minimi loscali piatti. E' come dover fare 1 km a passi di 1 cm
- OSCILLAZIONE: il modello rimbalda sulle pareti senza riuscire mai a fermarsi al suo punto più basso

Un ETA alto ci da variazione enormi nei parametri, utili all'inizio per sperimentare la mappa, ma letale quando cerchiamo una precisione chirurgica.
Un ETA basso ci da una discesa più pulita ma è estremamente inefficiente e soprattutto rischia di farci terminare in un minimo locale, che non è quello ottimale.

Immagina di dover scendere una montagna bendati, il gradiente è la pendenza della montagna mentra il learning rate è la lunghezza della falcata. Passi lunghi in cima e passi piccoli quando sentiamo che il terreno sta diventando pianeggiato, siamo sul fondo della valle, ecco perchè spesso utilizziamo il DECAY, ovvero una riduzione progressiva del passo man mano che ci avviciniamo alla soluzione, per permettere un fine-tuning finale dei parametri.

Ma cosa succede se il segnale della pendenza (gradiente) smette di arrivarci ai piedi?
Il silenzio dei layer profondi


FENOMENTO DEL VANISCHING GRADIENT: perchè le reti molto profonde smettono improvvisamente di imparare nei primi layer

Nelle reti neurali profonde, l'informzione sull'errore deve viaggiare all'indietro dall'output ai primi layer. Durante questo tragitto, il gradiente tende a rimpicciolirsi continuamente fino a diventare indistinguibile dallo zero.
Quando un gradiente diventa quasi zero, i pesi dei primi layer non vengono più aggiornati, rendendo di fatto inutile la profondità della rete.
Questo è il nemico numero 1 delle reti profonde.
La colpa di questo fenomeno è principalmente della Chain Rule: il gradiente finale è il prodotto delle derivate di tutti i layer successivi (risultato di una lunga catena di moltiplicazioni)
Se usiamo funzioni come la sigmoide, schiacciamo tutto tra 0 e 1, le loro derivate saranno sempre piccole (al massimo 0.25), moltiplicare 0.25 per se stesso (10,20 o 50volte) produce un numero sempre più piccolo (come il suono in fondo ad una galleria lungissima, il suona poco poco sparisce)

Questo porta il gradiente ad azzeraris ai primi layer (di una rete prodonda) e quindi questi layer non imparano nulla. Fortunamente sono state trovate delle soluzioni a questo problema
La prima soluzione è stata la ReLu che a differenza della Sigmoide, la derivata è 1 per tutti i valori positivi, la relu non schiaccia il segnale, se l'imput è positivo la sua derivata è 1

Oltre a questo, altre soluzioni comprendono la Batch Normalization e l'uso di architetture con 'skip connection' come ResNet, che permettono al gradiente di scavalcare alcuni layer.

Attenzione perchè non è un problame solo moltiplicare numeri piccoli ma anche moltiplicare numeri grandi.


PROBLEMA DELL'EXPLODING GRADIENT: instabilità numerica e divergenza dei pesi verso l'infinito

L'opposto del vanisching è l'exploding gradient. In questo caso, il gradiente accumulato durante la backpropagation cresce esponenzialmente ad ogni layer.
Un piccolo rumero entra, viene amplificato, rientra e diventa enorme, questo distrugge i nostri pesi.
Questo porta a aggiornamenti dei pesi enormi, che 'distruggono' la configurazione delle rete, causando errori tipo "Nan" nel software.
Le cause sono principalmente due
    1) inizializzazione dei pesi troppo agressiva, numero troppo grandi all'inizio
    2) uso di reti ricorrenti, dove lo stesso peso viene utilizzato centinaio di volte.

Se i pesi delle matrcici hanno autovalori maggiori di 1, il prodotto ripetuto durante la backpropagation, causa una crescita numerica di questo valore.
Questo fenomento si manifesta con oscillazioni estreme della Loss e con la comparsa improvvisa di valori non validi durante il training.
Il GRADIENT CLIPPING consiste nel riscalare il vettore gradiente se la sua norma supera una soglia prestabilita, mantenendo però la direzione originale.
