Derivata di y = sin(x^2)

Per calcolare la derivata di una funzione composta, dobbiamo "scomporla" in funzioni più semplici. Seguiamo i passaggi suggeriti:

1. Definizione delle variabili
    * Poniamo la parte interna della funzione come u:
    * u = x^2Quindi la funzione originale diventa: y = sin(u)

2. Calcolo delle derivate parziali
    * Ora calcoliamo la derivata di ogni pezzo rispetto alla propria variabile:
    * Derivata di y rispetto a u: dy/du = d/du(sin(u)) = cos(u)Derivata di u rispetto a x: du/dx = d/dx(x^2) = 2x

3. Applicazione della Regola della Catena
    * La regola della catena afferma che la derivata totale è il prodotto delle due derivate appena trovate: dy/dx = (dy/du) * (du/dx)
    * Sostituiamo i valori ottenuti: dy/dx = cos(u) * 2x

4. Risultato Finale
    * Sostituiamo nuovamente u = x^2 per esprimere tutto in funzione di x: dy/dx = 2x cos(x^2)

Analisi del caso particolare: x = 0

Vediamo cosa succede al coefficiente angolare della retta tangente quando la variabile x è uguale a zero.

Sostituiamo x = 0 nel risultato della derivata:

    1. dy/dx = 2(0) * cos(0^2)
    2. dy/dx = 0 * cos(0)
    3. Poiché cos(0) = 1, otteniamo: 0 * 1 = 0

Interpretazione: Quando x = 0, la derivata è nulla. Questo significa che nel punto (0, 0) la funzione ha un punto stazionario (in questo caso un minimo relativo) e la retta tangente al grafico è perfettamente orizzontale.
