LA CHAIN RULE
Come si scompone la complessita?
Le funzioni di influenzano a vicenda quando sono concatenate una dentro l'altra. Questo permette all'errore un viaggio a ritroso.
La Chain Rule (regola della catena), permette di calcolare la derivata di una funzione, moltiplicando le derivate delle singole funzioni che la costituiscono.

Nella rete naurale nulla accade in modo isolato, se cambiamo un peso sul primo layer quel cambiamento si ripercuote sino al layer di output, comprendere questa concatenazione permette di padroneggiare la back propagation.

Una rete neurale è una sequenza di trasformazioni.
Se y dipende da u e u dipende da x, allora ogni minima variazione di x si ripercuote su y tramite u.
Questo perchè stiamo lavorando a più step
Nelle reti neurali ogni neurone deve solo preoccuparsi del suo vicini immediato, il calcolo globale emerge poi da questi piccoli passi locali.
In una rete con milioni di parameetri sarebbe impossibile calcolare la derivato globale in un colpo solo, grazie alla chain rule, ogni neurone riceve un segnale di errore dall'alto, calcola la sua derivata locale, passa il risultato al layer precedente.
Non importa quanto sia profonda la rete, il gradiente scorre all'indietro come un fluido, scalando il suo valore attraverso gli strati, senza mai dover calcolare tutto da zero.

E se non è solo una linea ma un intreccio?
Un neurone non parla con un solo neurone, ma con decine di compagni, la catena si trasforma in un grafo, dobbiamo sommare i contributi che arrivano da ogni percorso.
Ogni freccia del nostro grafo rappresenta una derivata parziale che trasforma una quota dell'errore finale.

I pesi sono l'unica cosa che possiamo controllare: input (dati che arrivano) non possono essere cambiati, la Loss è il risultato, le manopole dei pesi sono sotto la nostra responsabilità e possiamo modificarli.

Il peso influenza l'output
l'output influenza l'attivazione
L'attivazione influenza la perdita

Quanto cambia l'errore se cambia l'attivazione
Quanto cambia l'attivazione se cambia l'output
Quanto cambia l'output se tocco il peso.

E' una catena
moltiplicando questi termini si ottiene la ricetta per correggere il parametro

Il segno della derivata è la nostra bussola.
Se è positivo, il peso e l'errore crescono insieme, dobbiamo quindi diminuire il peso.
Se è negativo, il peso e l'errore non crescono indieme, incrementare il peso diminuisce l'errore 
Se il gradiente è nullo, siamo in una zona piatta, questo è un pericolo, non riceve più segnali di miglioramento, non sta imparando nulla da quel parametro.

Un valore assoluto (magnitudo) grande indica che il peso ha una forte influenza sull'errore. Piccole correzioni portano a grandi miglioramenti.
