TENSORI: OPERAZIONI BASE IN PYTORCH

I Tensori sono la matrica prima di cui sono fatte le architetture delle reti neurali.
Sono mattoncini che sanno fare calcoli a velocità della luce.
Gestire delle immagini non è possibile con la programmazione standard, abbiamo bisogno di qualcosa che sia nato per la velocità, i tensori.

- Anatomia del Tensore
- Navigazione Multidimensionale:slicing e indexing
- Manipolazione delle Forma: reshape e transpose
