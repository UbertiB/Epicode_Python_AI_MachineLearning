TENSORFLOW VS PYTORCH: DIVENTA POLIGLOTTA DELL'AI

Passare da un modello all'altro, le parole cambiano, ma il risultato finale deve restare intatto
La stessa intuizione matematica può essere scritta con due dialetti diversi.

- Parità architetturale
- Verifica dei parametri
- Benchmarking delle Performance

Due linguaggi per lo stesso grafo
Sebbene TensorFlow e PyTorch utilizzino approcci diversi alla programmazione, l'obbiettivo finale è la costruzione di un grafo computazionale coerente. Tradurre un modello significa mappare ogni layer e funzione di attivazione tra i due framework.

Una struttura Sequantial di Keras si trasforma in una classe nn.Module in pyTorch, mantenendo inalterata la logica di trasformazione dei dati.
Se TensorFlow è come un set di istruzioni preconfezionate, PyTorch è come avere i passi e dover decidere, passo dopo passo, dove mettere le viti.

Sintassi a Confronto
Dalla dichiarazione alla definizione
- Keras Layers: utilizza un approccio ad alto livello dove 'layer.Dense' racchiude pesi, bias e spesso l'attivazione in un unico comando
- PyTorch Modules: separa nettamente la dichiarazione dagli organi (nel costruttore) dalla logica del movimento dei dati (nel metodo forward)
- Mapping dei nomi: imparare che 'Dense' equilave a 'Linear' e che 'CategoricalCrossentropy' corrisponde spesso a 'CrooEntropyLoss' è fondamentale per la migrazione
- Inizializzazione: i framework hanno default diversi (es. Glorot vs Kaiming). Per una comparazione equa, i pesi devono essere inizializzati con la stessa logica

Eager vs Lazy
Sia TensorFlow che PyTorch lavorano in modalità Eager, ovvero fai le cose appena le chiedo. PyTorch però esegue ogni passaggio mentro TensorFlow, con il metodo compile, astrae ogni processo.
Astrazione della Loss
In PyTorch dobbiamo spesso gestire manualmente il calcolo della loss, mentra in Keras il metodo 'compile' astrae l'intero processo di collegamento tra errore e parametri.
Pre-processing
I framework gestiscono i dati in modo differente (canali primo o dopo), un dettaglio che richiede attenzione per evitare errori di forma durante la conversione.


Ma perchè qualcuno deve preferire la complessità di PyTorch alla semplicità di Keras?
Mentre Keras punta alla velocità di sviluppo sacrificando a volte la granularità, PyTorch costirnge lo sviluppatore a comprendere ogni singolo passaggio del tensore, usanto la struttura ad oggetti si può intervenire su ogni elemento del calcolo.
Capire questa differenza permette di scegliere lo strumento giusto: Keras per la produzione rapida, PyTorch per la ricerca e l'innovazione strutturale.

Verifica dei parametri
Per essere certi che due modelli siano identici, non basta che abbiano lo stesso numero di layer. Devono avere esattamente lo stesso numero di parametri addestrabili.
Il conteggio dei parametri/pesi è una verifica matematica oggettiva che prescinde dal framework utilizzato e ci assicura la parità computazionale.

Il calcolo dei parametri
- Parametri dei pesi: calcolati come il prodotto tra il numero di neuroni in ingresso e quelli in uscita del layer corrente
- Parametri dei bias: ogni neurone in uscita possiede un bias additivo, quindi il numero di bias è uguale al numero di neuroni del layer
- Parametri totali: la somma algebrica di tutti i pesi e bias presenti in ogni strato addestrabile dalla rete
- Layer non Addestrabili: layer come Dropout o Flatten non possiedono parametri e non influenzano il conteggio totale dei pesi.


Ogni parametro è un n. deciamale a 32bit, conoscere il numero di parametri permette di stiamare quanto memoria VRAM sarà necessario per carica il modella nella GPU

Molti pensano che la velocità dipenda solo dalla GPU. Spesso i collo di bottiglia è il caricamente dei dati dalla CPU alla GPU
