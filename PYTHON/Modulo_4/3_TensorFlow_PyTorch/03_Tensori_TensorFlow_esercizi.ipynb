Crea due matrici casuali A e B con TensorFlow. La matrice A deve avere forma (3,4), la matrice B (4,2). Esegui la moltiplicazione matriciale e verifica che la forma del risultato sia (3,2). Successivamente, prova a sommare un vettore costante di bias di dimensione 2 sfruttando il broadcasting.
Suggerimento: usa tf.random.normal per generare i dati e l'operatore @ per la moltiplicazione

```python
import tensorflow as tf
import numpy as np
```

```python
#Crea due matrici casuali A (3,4) e B (4,2)
a=tf.Variable(tf.random.normal([3,4]),name="a")
b=tf.Variable(tf.random.normal([4,2]),name="b")
prodotto=a@b
print(f"a={a}\nb={b}\n\ta*b={prodotto}")
```

somma un vettore costante di bias di dimensione (2) sfruttando il broadcastring

```python
a=tf.constant([[1.0,2.0]])
somma=tf.add(prodotto,a)
print(somma)
```
