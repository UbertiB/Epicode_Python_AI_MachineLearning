```python
import torch
import numpy as np
```

```python
#1. IL PONTE TRA NUMPY E PYTORCH
#Crea un Tesore 3D di forma (3,4,5) contenente numeri casuali
tensor=torch.randn(3,4,5)
print(f"Tensore \n{tensor}")
```

```python
#Estrai la sesta riga della seconda matrice
slice_1=tensor[:,3,:] #non ho la sesta riga, estraggo la 4
print(slice_1)
```

```python
#trasforma il tensore originale in una forma (5,12) 
tensor_1=tensor.reshape(5,12)
print(tensor_1)
```

```python
#inverti l'ordine delle prime due dimensioni del tensore 3D iniziale
t_invertito=tensor.transpose(1,0)
print(tensor)
print(t_invertito)
```
