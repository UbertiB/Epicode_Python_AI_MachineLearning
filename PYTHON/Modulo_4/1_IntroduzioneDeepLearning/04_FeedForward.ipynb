STRUTTURA A STRATI (STRUTTUA A LAYER)
La vera potenza del Deep Learning i neuroni vengono organizzati in gruppi ordinati chiamati layer o strati (non si lasciano i neuroni sparsi alla rinfusa).
Questa organizzazione gerarchica è ciò che permette al Deep Learning di trasformare dati grezzi (come dei pixel) in decisioni, come per dire.. questo è un gatto.

Un'architettura Feed-Forward è composta da una sequanza di layer attraverso i quali l'informazione viene trasformata progressivamente, passando da una rappresentazione grezza a una decisione complessa.

Ogni rete ha un inizio, una parte centrale, ed una fine
Ogni strato ha un ruolo specifico in questa catena:
- Layer di Input: è come i nostri sensi, riceve i dati dal mondo esterno, non li elabora ma li distribuisce ai layer nascosti (che sono il vero cuore del nostro sistema). Non esegue calcoli ma serve come interfaccia che distribuisce i valori delle feature ai neuroni successivi.
- Layer Nascosti o Hiddens: si trovano tra l'input e l'output. Rappresenano la 'scatola nera' dove avviene l'estrazione delle caratteristiche e la creazione di astrazioni matematiche.
- Layer di Output è il giudice, quello che da il verdetto finale. 

La profondità di una rete indica quanti passaggi logici la macchina può compiere prima di arrivare a dare una risposta.
La struttura di una rete non dipende solo dal numero di strati, ma anche quanto questi strati sono larghi. 
Se la profondità è il numero di passaggi, la larghezza è la capacità di memoria a breve termine di quello strato.
Uno strato troppo stretto potrebbe ignorare dettagli importanti, mentre uno troppo largo potrebbe imparare a memoria il dataset invece di capirlo.
L'obiettivo dei layer nascosti è quello di craere una rappresentazione, una sorta di lingua interna della rete dove i dati diventano facili da classificare.
Tutto deve iniziare da una dimensione di input che rispetti federlmente i dati originali.

Ma è meglio avere pochi layer larghi o molti stretti, questo è il grande dilemma del deep learning.
Una rete poco profonda (rete 'shallow') ha pochi layer nascosti. Sebbene possa teoricamente approssimare molte funzioni, richiede un numero esponenziale di neuroni per farslo.
Le reti 'deep', invece, sfruttano la composizionalità (il primo strato impare i bordi, il secondo le forme, il terzo gli oggetti): ogni layer riutilizza le caratteristiche estratte dal precedente, rendendo il sistema drasticamente più efficiente in termini di parametri locali. Ed è questo il segreteo che permette alle reti moderne di funzionare con molte meno risorse rispetto ad un unico layer mastodontico, che comunque non è concettualmente sbagliato, anzi è stato dimostrato che un singolo layer enorme può approssimare qualsiasi funzione, il problema sta nelle risorse di calcolo.
Tanti piccoli layer sono molto più efficienti.

Come si collegano i vari strati.
Si passa al concetto di connessione
In una rete standard si utilizza il modello FULLY CONNECTED
Immagina di essere ad una festa dove ogni persona del tavolo A parla con tutte le persone al tavolo B, questo è connessione totale (fully connected).
Ogni filo che si vede in un diagramma è un peso ed ogni nodo è collegato ad un altro nodo.
Se abbiamo 10 neuroni che si collegano a 5 neuroni successivi abbiamo 50 pesi diversi (10 x 5)
Diventa una ragnatela fittissima dove ogni layer impara un pezzetto di verità
Inoltre ogni neurone riceve il suo baias personale che gli permette di decidere, autonamente, la propria soglia di attivazione. E' questa densità che da alla rete la sua flessibilità.
Gestire queste migliaia di fili sarebbe lungo per un computer,
Per questo ci rivolgiamo all'algerbra e raggruppiamo tutto in matrici di pesi, motivo per cui si utilizzano le schede GPU (le schede video sono abituate a lavorare con matrici)

Guardiamo la trasformazione nello spazio.
Unlayer fully connected, è un operatore di trasformazione, l'obbiettivi finale è che all'ultimo layer i dati sono così ordinati da essere separabili.
Il termine Feed-Forward descrive la dinammi principale del sistema: l'informazione si muove rigorosamente in una direzione, dall'input verso l'output, senza mai tornare indietro.
Garantisce che il calcolo sia stabile e prevedibile, un errore non può mai influenzare se stesso perchè passa l'informazione avanti.
Il layer 2 non può fare niente fintanto che il layer 1 non ha finito le sue attività ed ha passato i dati.
Grazie all'isolamente dei layer, anche se il layer sono dipendenti tra di loro, i neuroni dentro sono indipendenti, possono calcolare tutto quanto in parallelo (non si fa neurone 1, 2, 3, tutti i neuroni vengono calcolati in contemporaneamente, tramite le operazioni vettoriali). Mentre il segnale avanza la rete cerca di isolare il rumero e tiene solo le informazioni più utili.
In tutto questo c'è un elemento vitale per questo flusso, le funzioni di attivazioni, sono loro la colla tra i layer. La combinazione di strutture a leyer e ciò che permette alla rete di piegare la logica, piegare lo spazio, ogni layer aggiunge una piega e dopo molte pieghe anche il problema più complesso può sembrare semplice.
La rete neurale modifica lo spazio ad ogni layer per ottenere un risolutato migliore come obbiettivo.
