APPRENDIMENTO vs GENERALIZZAZIONE (overfitting)

Lo scopo di una rete neurale non è ripetere a memoria di dati che ha già visto, ma estrapolare regole generali da applicare a scenari futuri.
Questo passaggio è chiamato GENERALIZZAZIONE.
Con la generalizzazione la rete è in gradi di prevedere da dati mai visti prima.
La generalizzazione è la capacità di un modello di fornire risposte corrette su dati nuovi, non apparteneneti al set di addestramento.
L'OVERFITTING avviene quando il modello è troppo complesso e impara il rumore statistico dei dati, come se fosse una regola universale, impara a memoria i dati.
L'UNDERFITTING è il problema opposto: il modello è troppo semplice, troppo pigro e non riesce a catturare nemmeno le tenedenze principali dei dati.

La complessità del modello deve essere bilanciata rispetto alla quantità dei dati disponibili per evitare una memorizzazione inutile.

ERRORE DI TRAINING
Un errore di addestramento pari a zero non è sempre una buona notizia. Potrebbe indicare che la rete ha semplicemente fotografato il dataset. Deve essere un campanello di allarme, significa che il modello ha imparato a memoria.
ERRORE DI GENERALIZZAZIONE
Rappresenta la discrepanza tra la predizione del modello e la realtà su dati mai visti. E' il vero indicatore di qualità del sistema. 
Dobbiamo insegnare alla rete ad ignorare il rumore, ovvero le fluttuazioni casuali, che non si ripeteranno mai uguali e concentrarsi sul segnale strutturale. Un modello che spiega tutto, incluse le eccezioni più assure, di solito non sa prevedere nulla di nuovo.

RUMORE VS SEGNALE
I dati reali contengono imperfezioni casuali. Una rete che generalizza bene ignora queste fluttuazioni e si concentra solo sul segnale strutturale.

Tra più spiegazini possibili la più semplice è solitamente quella corretta.
Nelle rete neurali significa che se una rete piccola può risolvere il problema non dovremmo usarne una gigantesca. Se una spiegazione semplice è sufficiente a descrivere i dati, è probabile che sia quella corretta. Reti con troppi neuroni tendono a creare funzioni troppo contorte.
L'obbiettivo è trovare la configurazione più snella possibile che riesca comunque a risolvere il compito, massimizzando così la stabilità delle risposte su dati ignoti.

La Trinità dei dati: come dividere tra Train, Validation e Test
Per valutare onestamente una rete neurale, non possiamo usare gli stessi dati per l'addestramento e per la verifica. Dobbiamo simulare l'incontro con il mondo reale, dobbiao compartimentare i dati, creando delle zone sicure, dei dati vergini che la rete incontrerà solo per essere valutata.

Dividiamo il nostra dataset in 3 blocchidistinti e indipendenti, ognuno con un ruolo critico nel ciclo di vita dello sviluppo di un'intelligenza artificiale.
1) Training set: p il materiale di studio, i dati su cui il Gradient Descent lavora per aggiornare pesi e bias durante ogni epoca.
2) Validation Set agisce come un esame parziale, serve per monitorare l'andaento e scegliere i migliori iperparametri senza sporcare il test finale. Utilizzato dai programmatori
3) Test Set: è il verdetto finale, viene utilizzato una sola volta alla fine dello sviluppo per misurare le performance reale del modello. Il suo risultato è la verità definitiva sulla bontà del nostro lavoro. Non deve essere usato per coreggere il modello.

Serve il validation ci permette di provare diversi iperparametri e decidere quando smettere senza mai mostrare alla rete il verdetto finale del test test, è la nostra palestra.

Un errore comune è quello del datalake, se una sola riga del test finisce nel trainig il test è fallito, dobbiamo mescolare i datii (shuffle) per assicurarci di mantenere le stesse proporzioni tra le classi.

Una suddivisione classica dei dati garantisce che il modello non veda mai i dati di valutazione durante la fase di aggiornamento dei pesi.

Divergenza delle Loss
La Training Loss scende sempre man mano che la rete impara a minimizzare l'errore sui dati di studio.
Quando questa linea scende verso le zero, la rete pensa di aver imparato, ma potrebbe aver imparato a memoria, ed in questo ci viene in auto la linea della loss Validation, quando questa linea smette di scendere ed inizia a risalite, l'errore inizia ad aumentare, in quel preciso istante la rete ha smesso di imparare la rete ed inizia a memorizzare dettagli inutili. Il gap tra le due linee è la misura della nostra delusione, più è ampio meno possiamo fidarci del modello.
Il punto di minima della Validation Loss rappresenta il momento di massima generalizzazione possibile per quell'architettura.
L'Overfitting inizia quando la Validation Loss smette di scendere e inizia a risalire, nonostante la Training Loss continui a calare.
Il gap tra le due curve quantifica numericamente quanto il modello stia diventando specifico per il dataset di addestramento.
GAP=Loss(val)- Loss(train)

Early Stopping
E' l'atto di dire "basta fermiamo l'addestramento' non quando la loss del training è minima ma quando la Loss Validation inizia la sua risalita, ignorando le epoche successive.

Oscillazioni Elevate
Se la cura di validazione salta violentamente, potrebbe indicare un validation set troppo piccolo o un learning rate troppo aggressivo.

Plateau
Quando entrambe le curve si fermano a valori alti, siamo in presenza di un Underfitting: il modello non ha abbastanza capacità per il problema, il modello non ha capito come interpretare i dati. In questo caso dobbiamo aumentare la complessità del modello o dargli più dati.
