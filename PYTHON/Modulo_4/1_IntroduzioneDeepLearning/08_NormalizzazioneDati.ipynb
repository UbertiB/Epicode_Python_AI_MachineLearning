NORMALIZZAZIONE DEI DATI PER ANN

Riportare ogni dato un un campo di gioco equo affinchè la nostra rete neurale non venga ingannata dalla grandezza dei numeri.

Perchè grandezze diverse sono un ostacolo?
Immagina di voler prevedere il valore di una casa, hai l'età del fabbircato e il reddito del proprietario. La prima variabile oscilla tra sero e cento mentre il valore della seconda variabile oscilla tra ventimila e un milione.
Senza normalizzazione la rete darà molta più importanza alle variabili con valori numerici più grandi semplicemente perchè contribuiscono maggiormente alla somma pesata ignorando l'importanza reale del segnale.
Quindi il valore sarebbe dato quasi esclusivamente sul reddito senza contare, quasi, l'eta della casa.
Questo squilibrio non è solo un problema di attenzione ma distrugge l'efficienza del calolo.

Il Gradient Descent (la bussola dell'errore) fatica enormemente in presenza di feature non scalate, poichè la superficie di errore diventa molto allungata creando vallate strette dove l'ottimizzatore rimbalza senza convergere.
I pesi associati alle feature con range elevato richiederebbero aggiornamenti minimi mentre quelli con range piccolo ne richiederebbero di enormi rendendo impossibile l'uso di un unico 'learing rate' efficace.
L'obbiettvo è ottenere una funzione di costo più simmetrica che permetta al gradiente di puntare direttamente verso il minimo globale.

Valori troppo grandi possono causare problemi di overflow, il computer finisce lo spazio per memorizzare numeri grandi, rendendo instabili le funzioni di attivazione come la sigmoide (se riceva in input un valore enorme la sua curva diventa piatta).
Dati bene scalati mantengono il segnale nella zona attiva delle nostre funzioni, permettendo all'algoritmo di ottimizzazione di raggiungere l'errore minimo in un numero di epoche drasticamente inferiore risparmiando ore di calcolo.
Senza normalizzazione il modello è intrinsecamente prevenuto verso i dati con magnitudo superiore rendendo il processo di addestramento ingiusto e tecnicamente errato.

Come di standarizza?
Esistono diversi modi per riportare i dati su una scale comune. Le due tecniche più utilizzate nel Deep Learning sono il Mim-Max Scaling, che comprime i dati in un interallo fisso, e la Standarizzazione che lavora sulla distribuzione statistica.
Importante è capire quando preferire un e quando preferire l'altro in base alla natura del nostro dataset (da quano sono 'sporchi' i dati e da cosa ci si aspetta).

Mim-Max Scaling
mappa i dati in un intervallo specifico solitamente tra 0 e 1. E'molto utile quando la distribuzione dei dati non è necessariamente gaussiana.
x= (x-min(x))/(max(x)-min(x)))

Standarizzazione o Z-score
trasforma i dati in modo che abbiano media pari a zero e deviazione standard pari a 1. E' robusta è ampiamente usata. 
Sottraiamo la media e dividiamo per la deviazione standard. Non abbiamo limiti fissi ma i dati hanno una media di zero e una larghezza di 1. Tecnica robusta se abbiamo outlier che altrimenti schiaccerebbero tutti i dati normali.

E' fondamentale calcolare i parametri di scala solo sul 'training set' e applicarli poi al 'validation set' per evitare il fenomeno del data leakage (inquinamento dei dati).

Regola generale
Min-Max si sceglie spesso per i dati delle immagini dove i pixel sono già confinati in un range noto come quello tra zero e duecentocinquantacinque.
La standarizzazione è la scelta predefinita per la maggior parte delle reti neurali perchè mantiene la forma della distribuzione originale pur rendendo i dati centrati.
I presenza di molti outlier si usano varianti basate sulla media e sui quartili invece della media per non distorcere eccessivamente la scala.

La normalizzazione è parte del modello, si deve calcolare la standarizzazione solo sul training set e poi utilizzare gli stessi valori per pulire i dati di test. Se usi i dati di tutto il dataset dai alla rete informazioni sul futuro.

```python
import numpy as np
```

```python
def min_max_scaler(data):
    """
    Trasforma i dati nell'intervallo [0, 1].
    Formula: (x - min) / (max - min)
    """
    data_min = np.min(data)
    data_max = np.max(data)
    
    # Calcolo normalizzato
    scaled_data = (data - data_min) / (data_max - data_min)
    return scaled_data

def standardizer(data):
    """
    Trasforma i dati con media 0 e deviazione standard 1.
    Formula: (x - mean) / std
    """
    mean = np.mean(data)
    std = np.std(data)
    
    # Calcolo standardizzato (Z-score)
    standardized_data = (data - mean) / std
    return standardized_data

# --- TEST DELLE FUNZIONI ---

# Immaginiamo i prezzi di 5 prodotti in euro
prezzi_grezzi = np.array([50, 150, 200, 450, 900])

prezzi_normalizzati = min_max_scaler(prezzi_grezzi)
prezzi_standardizzati = standardizer(prezzi_grezzi)

print(f"Dati Originali: {prezzi_grezzi}")
print(f"Min-Max [0, 1]: {np.round(prezzi_normalizzati, 3)}")
print(f"Standardizzati: {np.round(prezzi_standardizzati, 3)}")

# Nota: Osservate come nella Standardizzazione i valori possano essere negativi
# (quando sono sotto la media) o maggiori di 1.def min_max_scaler(data):
    """
    Trasforma i dati nell'intervallo [0, 1].
    Formula: (x - min) / (max - min)
    """
    data_min = np.min(data)
    data_max = np.max(data)
    
    # Calcolo normalizzato
    scaled_data = (data - data_min) / (data_max - data_min)
    return scaled_data

def standardizer(data):
    """
    Trasforma i dati con media 0 e deviazione standard 1.
    Formula: (x - mean) / std
    """
    mean = np.mean(data)
    std = np.std(data)
    
    # Calcolo standardizzato (Z-score)
    standardized_data = (data - mean) / std
    return standardized_data

# --- TEST DELLE FUNZIONI ---

# Immaginiamo i prezzi di 5 prodotti in euro
prezzi_grezzi = np.array([50, 150, 200, 450, 900])

prezzi_normalizzati = min_max_scaler(prezzi_grezzi)
prezzi_standardizzati = standardizer(prezzi_grezzi)

print(f"Dati Originali: {prezzi_grezzi}")
print(f"Min-Max [0, 1]: {np.round(prezzi_normalizzati, 3)}")
print(f"Standardizzati: {np.round(prezzi_standardizzati, 3)}")

# Nota: Osservate come nella Standardizzazione i valori possano essere negativi
# (quando sono sotto la media) o maggiori di 1.
```
