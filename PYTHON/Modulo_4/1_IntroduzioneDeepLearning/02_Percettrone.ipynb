PERCETTRONE: UNITA' BASE DI UNA RETE NEURALE

PERCETTRONE
Il percettrone è un singolo mattoncino di una cattedrale.

Il percettrone rappresenta il mattoncino fondamentale di ogni architettura di Deep Learnig. Sebbene ispirato vagamente al neurone biologico, la sua struttura è un elegante modello matematico che trasforma segnali in ingresso in una decisione
Il permettrone Riceve dei segnali e li trasforma per prendere una decisione, è un modello elegante perchè ridue tutta la complessità del pensiero in una serie di operazioni aritmetiche.
I bulloni che tengono insieme questa struttura sono 3:
1) Input: gli inputi rappresentano i dati grezzi che entrano nel sistema. In un modello di classificazione potrebbero essere i valori dei pixel di un'immagine o i parametri fisici di un oggetto
2) Pesi: sono i parametri che il modello impara durante l'addestramento. Ogni peso determina quanto un specifico input influenza il risultato finale, agendo come un filtro di importanze.
Se un testimone è cruciale il suo peso è alto
3) Bais:é un termine aggiuntivo che permette al neurone di attivarsi anche quando gli input sono nulli. Sposta la soglia di decisione garantendo flessibilità al modello. E' come il giudizio iniziale. E' un abase di partenza che può facilitare o ostacalore la l'input
Il risultato di tutto ciò è la SOMMA PESATA combina tutti questi elementi (input, pesi, bias) in un unico valore scalare, rappresenta il verdetto del giudice che deve decidere (si o no). 

I pesi non sono solo numeri, sono filtri di importanza. Il peso negativo agisce come un freno, un segnale che va a dire, non attivarti. 
Il bias è fondamentale, immagina di dover attivare un interrutori, influenza la facilità con cui si può ottenere un risultato positivo o negativo.
Per ogni pezzetto di informazione in ingresso abbiamo un peso a lui dedicato.
Per calcolare la somma pesato in modo efficiente, utiliziamo il prodotto scalare tra il vettore degli input e quello dei pesi. Questa operazione è alla base di tutta l'algebra lineare del Deep Learning.
Questa forma compatta permette di sfruttare le librerie di calcolo parallelo, rendendo i calcoli istantanei anche con migliaia di connessioni (efficienza applicata all'algebra lineare).

Dallo scalare alla decisione bianaria
E' un interrutore on/off. Per dare risposte nette, si o no, non ci sono sfumature, o il segnale è abbastanza forte da superare la soglia e restituisce true altrimenti false.
Se il risultato della somma pesata (z) è maggiore di 0 l'output è 1 quindi il neurone è attivo, il neurone è 0 quindi spento nel caso opposto.
La fuazione ha pertanto un salti brusco a zero. Questo la rende difficile da gestire con algoritmi basati sulla derivata.
Il percettrone con funzione step, è un classificatore linerare. Può risolvere solo problemi linearmente separalbili, ovvero dove una retta può dividere perfettamente le due classi. Se i dati sono mescolati, un solo neurone non basterà.
Possiamo vedere la funzione step come un interrutore che scatta solo quando la pressione cambinata di input e pesi supera l'ostacolo posto dal bias.

L'iperpiano di decisione
La somma pesata uguale a zero in uno spazio definisce un confine decisionale, cambiare i pesi fa ruoatare questa linea nello spazio, mentre cambiare i bias significa spostare verso dx o verso sx, permettendo al neurone di posizionarsi correttamente per minimizzare l'errore.
Addestrare una rete significa trovare una posizione perfetta per questa linea in modo di separare i dati.


```python
import numpy as np

def heaviside_step(z):
    """
    Implementazione della funzione di attivazione Heaviside.
    Restituisce 1 se l'input è >= 0, altrimenti 0.
    """
    return 1 if z >= 0 else 0

def perceptron_output(inputs, weights, bias):
    """
    Calcola l'output di un singolo neurone (Percettrone).
    Formula: output = step(sum(w_i * x_i) + b)
    """
    # Trasformiamo le liste in array NumPy per usare il prodotto scalare
    x = np.array(inputs)
    w = np.array(weights)
    
    # Calcolo della somma pesata (z = W · x + b)
    # np.dot esegue la sommatoria dei prodotti w_i * x_i
    z = np.dot(w, x) + bias
    
    # Applicazione della funzione di attivazione
    y = heaviside_step(z)
    
    return z, y
```

```python
# --- ESEMPIO DI UTILIZZO ---

# Definiamo i parametri (es: un neurone che decide se uscire a correre)
# Input: [Meteo_Bello, Ho_Energia]
input_data = [1, 0] 
pesi = [0.8, 0.5]    # Il meteo conta più dell'energia
termine_bias = -0.6  # Soglia di "pigrizia" da superare

# 0.8*1+0.5*0-0.6=0.8-0.6=0.2
z_val, output = perceptron_output(input_data, pesi, termine_bias)

print(f"Somma pesata (z): {z_val:.2f}")
print(f"Decisione finale (Output): {output}")
```

Per padroneggia il Deep Learning è fondamentale saper calcolare l'output di un neurone 'su carta'. Questo aiuta a visualizzare il flusso dei dati e l'impatto dei singoli parametri.

- Si elencano gli input e i pesi associati.
- Sommiamo i risultati
- Aggiungiamo un baias alla somma dei risultati della moltiplicazione precedente
- Applichiamo la funzione Heaviside. Se il risultato è maggiore uguale a 0 restituire True altrimenti False
Il baias è potente, influenza la scelta.

Quello visto per un singolo neurone è alla base del Deep Learning.
L'output di un neurone è l'input per il neurone del livello successivo. E' come un passaggio di testimone, l'informazione viene passata strato dopo strato fino ad arrivare ad un risultato finale.
