LOOS FUNCTION_MSE, CROSS ENTROPY

Le loss funzion sono il feedback che dice alla macchina quantoha sbagliato, senza di esse l'inteligenza artificiale sarebbe un insieme di calcoli senza scopo.

Distinguiamo tra regressione e classificazione, utilizzano metri diversi.
La rete va punita per spingerla a migliorarsi.



Come scegliere la bussola giusta per il nostro problema.
Senza una metrica di errore una rete neurale sarebbe come un navigatore senza mappe. Nel Deep Learning, la loss funziont è come il navigatore. 
La Loss funzion definisce l'obbiettivo dell'addestramento quantificando numericamente quanto le previsioni del modello si discostano dalla realtà dei dati osservati. 

L'errore è una distanza numerica, se stiamo prevedendo il prezzo di una casa prezzo predetto e prezzo reale messo a confronto. 
Se stiamo distinguendo un gatto da un cane l'errore è una questione di certezza, sbagliare la funzione di costo significa dare alla rete l'obbiettivo sbagliato. Non posso sottrarre la predizione del cane rispetto alla predizione del gatto.

La scelta della funzione di costo non è arbitraria ma dipende strettamente dalla nutura dell'output che vogliamo ottenere. Un errore nel calcolo di un valore numerico continuo richiede una logica differente rispetto all'errore nella classificazione di un oggetto.

Abbiamo due standard per questi due mondi diversi.

MSE: è come un metro elastico, più ti allontani dal valore reale (per eccesso o per diffetto) più l'errore cresce, e grazie al quadrato, gli errori grandi pesano tanto.

La CROSS ENTROPY invece vive nel mondo delle probabilità. Non le interessa quanto siete lontani in termini di numero, ma quanto siete confusi. Se il bersaglio è gatto, e voi dite cane con molta sicurezza. la Cross Entropy punisce severamente, perchè la distribuzione di proabilità è l'esatto opposto di quella reale.

Le due funzione di comportano in modo diverso durante l'addestramento.
MSE è sensibile agli outlier, se un dato è fuori scala, il quadrato lo trasforma in un errore gigante che può destabilizzare tutta le rete. 
La Cross entropy è la regina delle velocità, predice gradinti forti anche quando il modello sbaglia di grosso, costringendo il modello ad imparare velocemente fin dall'inizio. Si sposa perfettamente con le funzioned softmax e sigmoide, creando un flusso matematico coerente.
Per visualizzare tutto questo dobbiamo immaginare un territoriio. Immaginiamo La Loss function come un paesaggio montuoso, le vette alte sono errori enormi e le valli profonde sono la perfezione. Addestrare la rete significa per scivolare la pallina verso il punto più basso a valle. Se scegliamo la loss sbagliata il paesaggio rischia di diventare una pianura nebbiosa e piatta e la pallina non sa dove rotolare Una buona funzione di costo crea un pendio deciso che guida la nostra pallina verso la soluzione ottimale (a valle). Ma perchè questo meccaniscmo funzioni dobbiamo capire la logica della punizione.

Punizione o costo
Usiamo terminimi come punizione o costo perchè l'apprendimento automatico è un processo di feedbakc negativo. Se la rete non sente il dolore matematico dell'errore non avrebbe motivo di cambiare i suoi pesi. E' come un bambino che impara di non toccare il fuoco, la scottatura è il segnale che modifica il comportamento futuro.
Vediamo come trasformiamo un errore concettale in un segnale numerico che squote la struttura della rete, come funziona questa punizione.

La Cross Entropy è asimmetrica. Se il modello è arrogante, ovvero è sicuro di una risposta sbagliata, la punizione tende all'infinito, vogliamo che la funzione sia convessa, forma a u, che garantisca la presenza di un fondo valle chiaro da raggiungere, inoltre aggiungiamo spesso la regolarizzazione, una piccola sanzione extra se la rete diventa troppo complessa. Come interpretiamo questi nunmeri in termini di probabilità? 
A livello statistico minimizzare la cross entproy significa fare la massima verosomigliamza, dicendo alla rete di cambiare i tuoi pesi finchè la realtà che osserviamo diventa la cosa più probabile sencodo la tua struttura. Usiamo il logaritmo non per complicarci la vita, ma perchè i computer odiano moltiplicare tante probabilità piccole che diventano quasi zero, preferiscono sommare i loro logaritmi.

Vedimao l'effetto reale della punizione sui pesi di una rete
Ogni volta che calcoliamo un loss inviamo un messaggio a ritroso nella rete. E' un processo di attribuzione della colpa, i pesi che hanno contribuito di pù all'errore vengono sanzionati maggiormente. Questo feed back continuo è ciò che permette di raggiungere la convergenza.

Senza una definizione rigorasa di costo la rete fluttuerebbe nel caos. provando combinazioni di pesi a caso, senza arrivare ad un conclusione intelligente.

Esempio di classificazione dove la rete deve scegliere tra 3 opzioni.
Un semplice vettore di probabilità si trasforma in un numero che rappresenta il nostro conto da pagare.
Questo è ciò che accade miliardi di volte al secondo dentro ad una GPU

Iniziano definindo i vettori

```python
import numpy as np
```

```python
def mean_squared_error(y_true, y_pred):
    """
    Calcola l'errore quadratico medio (MSE) per la regressione.
    """
    return np.mean(np.square(y_true - y_pred))

def cross_entropy_loss(y_true, y_pred):
    """
    Calcola la Cross-Entropy per la classificazione.
    y_true: vettore one-hot (es: [1, 0, 0])
    y_pred: probabilità della softmax (es: [0.9, 0.05, 0.05])
    """
    # Epsilon serve a evitare log(0) che restituirebbe -inf
    epsilon = 1e-15
    y_pred = np.clip(y_pred, epsilon, 1 - epsilon)
    
    # Formula: -somma(y_reale * log(y_predetto))
    loss = -np.sum(y_true * np.log(y_pred))
    return loss
```

# --- ESEMPIO DI UTILIZZO ---

```python

# Caso Regressione (Prezzo case in migliaia di euro)
prezzo_reale = np.array([200, 150, 300])
prezzo_predetto = np.array([210, 140, 250]) # Il modello sbaglia di molto sull'ultima casa

mse_val = mean_squared_error(prezzo_reale, prezzo_predetto)
print(f"Errore MSE: {mse_val:.2f}")

# Caso Classificazione (Cane, Gatto, Pesce)
target = np.array([1, 0, 0]) # È un Cane
predizione = np.array([0.9, 0.05, 0.05]) # Il modello è sicuro al 90%

ce_val = cross_entropy_loss(target, predizione)
print(f"Errore Cross-Entropy: {ce_val:.4f}")
```

Immaginiamo un problema di calssificazioni a 3 classi in cui la risposta corretta è la seconda classe
Utilizzando il one hot encodign il target reale diventa 0,1,0
La nostra rete però non è sciura dei calcoli dice che c'è il 70% di probabilità che è la classe 2.
Quando applichiamo la formula, gli 0 del target cancellano tutto ciò che non ci interessa, isolando l'errore sulla classe corretta. Il valore finale si calcola applicando il logaritmo con il segno meno.
In un esempio reale (addestramento reale), faremo la media di questo calcolo su centinaio di esempi, contemporaneamente, per ottener il costo del mini batch
Ma se il modello fosse convinto di un valore sbagliato, qui entra in gioco la potenza del logaritmo, se la probabilità scende dallo 0.7 allo 0.01, l'errore non aumenta di poco ma di 10 volte, la cross entrpoy odia l'eccessiva sicurezza nel torto, questo è il motivo per cui le reti neurali odierne sono così precise, vengono terrorizzate (a livello matematico) dall'idea di essere sicure di un errore.
