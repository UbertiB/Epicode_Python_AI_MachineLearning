Dato un peso attuale w=0.5, un valore del gradiente calcolato pari a 2.4 e un learning rate impostato a 0.1, calcola il nuovo valore del peso dopo un singolo step di aggiornamento.
Suggerimento: applica rigorosamente la formula w_new=w_old-(eta*grad). Il risultato dovrebbe essere minore del valore iniziale, riesci a spiegare perchè?

```python
import numpy as np
```

```python
def gradient_descent_update(w, grad, eta):
    """
    Esegue un singolo step di aggiornamento del peso.
    w: valore attuale del peso
    grad: valore del gradiente (derivata della loss rispetto a w)
    eta: learning rate
    """
    # La formula fondamentale: w = w - eta * grad
    w_new = w - eta * grad
    
    return w_new
```

```python
peso_iniziale = 0.5
gradiente_calcolato = 2.4  # Supponiamo che la pendenza sia verso l'alto
learning_rate = 0.1
#spostati del 10% della pendenza misurata, verso il basso
```

```python
# Primo aggiornamento
peso_aggiornato = gradient_descent_update(peso_iniziale, gradiente_calcolato, learning_rate)

print(f"Peso prima dell'aggiornamento: {peso_iniziale}")
print(f"Valore del gradiente: {gradiente_calcolato}")
print(f"Peso dopo l'aggiornamento: {peso_aggiornato:.2f}")

# Nota: Il peso è diminuito perché il gradiente era positivo. 
```

Perché il risultato è minore del valore iniziale?
Il gradiente è positivo (2.4 > 0). In analisi matematica, un gradiente positivo indica che la funzione di costo sta "salendo" in quel punto.
Per minimizzare l'errore (ovvero "scendere" verso il minimo della funzione), dobbiamo muoverci nella direzione opposta al gradiente.
Per questo motivo sottraiamo il valore, diminuendo il peso.
Nota bene: Se il gradiente fosse stato negativo, l'operazione di sottrazione (meno per meno) avrebbe trasformato il calcolo in un'addizione, aumentando il valore del peso.
