Immagina di avere due modelli, entrambi hanno indovinato la classe giusta, ma uno è molto più sicuro dell'altro.
Calcola la Cross Entropy per entrambi

```python
import numpy as np
```

```python
def mean_squared_error(y_true, y_pred):
    """
    Calcola l'errore quadratico medio (MSE) per la regressione.
    """
    return np.mean(np.square(y_true - y_pred))

def cross_entropy_loss(y_true, y_pred):
    """
    Calcola la Cross-Entropy per la classificazione.
    y_true: vettore one-hot (es: [1, 0, 0])
    y_pred: probabilità della softmax (es: [0.9, 0.05, 0.05])
    """
    # Epsilon serve a evitare log(0) che restituirebbe -inf
    epsilon = 1e-15
    y_pred = np.clip(y_pred, epsilon, 1 - epsilon)
    
    # Formula: -somma(y_reale * log(y_predetto))
    loss = -np.sum(y_true * np.log(y_pred))
    return loss
```

```python


# Caso Classificazione (Cane, Gatto, Pesce)
target = np.array([1, 0, 0]) # È un Cane
predizione_m1 = np.array([0.9, 0.05, 0.05]) # Il modello è sicuro al 90%
predizione_m2 = np.array([0.4, 0.03, 0.03]) # Il modello è sicuro al 40%

ce_m1 = cross_entropy_loss(target, predizione_m1)
ce_m2 = cross_entropy_loss(target, predizione_m2)

print(f"Errore Cross-Entropy (Modello 1): {ce_m1:.4f}")
print(f"Errore Cross-Entropy (Modello 2): {ce_m2:.4f}")
```

Il modello 2 viene punito molto di più rispetto al modello 1, nonostante entrmabi i modelli abbiano predetto la classificazione corertta.
Questo perchè il modello 2 è 'troppo incerto'. La Crosso Entropy punisce non solo la risposta sbagliata, ma anche la mancanza di confidenza con la risposta corretta.

Matematicamente man mano che la probabilità scende verso lo zero, il logaritmo scende verso l'infinito negativo: il segno meno trasforma questa discesa in una montagna di errore, che costringe l'ottimizzatore a cambiare i pesi con forza.
