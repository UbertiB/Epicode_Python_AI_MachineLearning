Il modello è in Overfitting.

Performance sul campo: L'accuratezza reale sarà scarsa. Il modello non sarà in grado di gestire nuove casi (non presenti nel database originale) perché cercherà di applicare loro dei pattern specifici e "rumorosi" che non hanno riscontro nella realtà generale.
