COME PENSAVANO LE MACCHINE UN TEMPO
Il programmatore era un legislatore, doveva prevedere ogni singolo scenario. Questo approccio basato su regole rigide si scontrava con la complessità del mondo reale dove le eccezioni superano spesso le norme definite a tavolino.
Solo fissando le regole sono da scrivere migliardi di righe, la realtà è troppo caotica per trovare spiegazione in un codice statico, ed è qui che il deep learning entra in gioco. Non siamo più noi a dare istruzioni ma noi forniamo i dati e poi l'algoritmo estrai dei pattern. Questo è possible anche ad un cambio di paradigma matematico.

Inizialmente avevamo in input più regole per ottenere un output. Avevi la ricetta ma non potevi cambiare niente, i casi non previsti non erano gestibili.
Poi arriva il ml classico, l'uomo aiutava la macchina creando delle features, ossia delle caratteristiche interessanti.
E' il deep learning a fare il passo finale, il dl va ad imparare la funzione e va a definire una gerarchia, è la macchina che decide cosa è importante osservare. Dal dato grezzo al risultato finale (end to end).

In un sistema a regole si richiede la perfetta conoscenza del problema. Se una condizone non è prevista dal codice, il sistema fallisce, rendendo impossibile la gesione dell'incertezza. Se il treno esce di un millimetro, deragla. Non gestiscono l'incertezza.
Il ML invece è più come un auto, ha bisogno di qualcuno che metta il carburante giusto, ma ha bisogno di qualcuno che metta il carburante giusto. Il DL è l'evoluzione finale, è un organismo vivente, scopre da solo correlazioni, anche quelle che noi umani non riusciamo a verbalizzare. Esempio l'ai è in grado di distinguere uomini e donne dalle iridi degli occhi, sembrava non fi fosse alucna differenza.

Come fa una rete a capire concetti complessi
DEEP si riferisce alla profondità o livelli di astrazione. Immagina un pittore che iniza con schizzi veloci di linee poi definisce le ombre e infine i dettagli del volto. Le reti neurali fanno più o meno la stessa cosa, nei primi strati vedono bordi e colori, poi negli strati centrali diventano forme geometriche e si finisce con il concetto di viso. E' questa gerarchia che rende il Deep Learning simile alla nostra corteccia visiva. 
Questa teoria esisteva già negli anni 80m, la viviamo solo ora per due supporti necessari:
1) cibo: dati
2) energia: processori veloci
Senza di essi il Deep Learning sarebbe rimasta teoria.

I big data sono il carburante della rete neurale. Più sono i dati , più diventa precisa, ma per gestire que4sti dati servono le GPU, nate per i video giochi. A questi si aggiungono i framework (TensoFlow e PyTorc) che permettono ai programmatori di costruire modelli molto complessi in poche righe di Python e di sfruttare a pieno il nostro albero.
La GPU è come un esercito di opeari che singolarmente sono meno intelligenti, ma lavorano insieme nello stesso istante, per le reti neurali non dobbiamo eseguire calcoli estremamente complessi, ma dobbiamo eseguirne tanti, ed è chi che vince la GPU. La cosa sorprendente è che questa potenza di calcolo non sembra avere un limite così vicino. Il ML tradizionale avevamo meno potenza, i modelli erano in grado di imparare da pochi dati, nel DL più dati forniamo più migliora il nostro modello. Dove lo troviamo tutto questo cibo?
