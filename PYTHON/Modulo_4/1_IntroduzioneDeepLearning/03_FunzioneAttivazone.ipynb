FUNZIONE DI ATTIVAZIONE - IL CUORE NON LINEARE DELL'AI

Come viene trasformata una calcolatrice in una intelligenza artificiale.
Parliamo delle funzioni di attivazioni. Senza di esse una rete neurale sarebbe come un orchesta che sa  suonare una sola nota senza essere in grado di comporre una melodia complessa.

Il percettrone, il neurone è come un contabile che fa una somma pesata degli input. 
Se ci fermassimo qui, avremmo un modello puramente lineare, la rete neurale rimarrebbe un semplice modello statistico perchè incapace di apprendere la complessità del mondo reale. Il mondo reale non è fatto di rette. Immagina di dover distinguere tra un email spam ed una importante, i confini non sono mai così netti.
La funzione di attivazione serve a distorcere il calcolo lineare per adattarlo alle curve ed alle pieghe della realtà.
Le funzioni di attivazione servono a:
- introdurre la non linearità
- permettere alla rete di modellare fenomeni reali (che NON sono lineari)
- decidere 'quanto' un neurone deve attivarsi
output=attivazione(w*x+b)
Ma c'è anche un motivo matematico ancora più profondo dietro a questa necessità: se mettiamo in fila mille funzioni lineari, il risultato finale sarà sempre comunque una funzione lineare.
Senza le attivazioni non lineari avere una rete con cento layer, come averne uno solo, sarebbe matematicamente la stessa cosa, sarebbe come cercare di scolpire una statua di marmo utilizzando solo il righello, come fai a fare una curva?.
La funzione di attivazione permette di piegare il segnale, decidendo non solo se il nuerone deve parlare ma anche con quale intensità deve parlare. Questo ci porta al teorema della approsimazione univarsale.

Teorema della approsimazione universale
Grazie alla non linearità una rete neurale può imparare qualsiasi cosa, ma c'è un trucco, la funzione deve essere derivabile.
La rete impara dai suoi errori tornando indietro, lungo la stessa curva, per aggiornare i pesi, se la curva ha dei tratti troppo piatti però la rete si addormenta, questo fenomeno di chiama saturazione, ed è il nemico numero 1 di ogni programmatori Deep Learning.

Cencetto: GRADIENTE
Il gradiente di una rete neurale è quanto cambia l'errore se cambio un peso. Se è grande: apprendo velocemente, se è vicino allo 0 non imparo nulla.
Quando il neurone entra in saturazione (vicino a 0 o a 1) la derivata tende a 0, il gradiente tende a 0, l'aggiornamento dei pesi tende a 0
In una rete con più layer la backpropagation fa una moltiplicazione di gradienti. Se goni gradiente è piccolo (tipo 0.1) dopo 5 layer ho 0.00001, dopo 10 layer praticamente zero. L'informazione muove andando dietro (venishing gradient problem)

Per capire meglio questo concetto, immagina di avere un foglio con tanti punti rossi e blu, se non posso piegare il foglio non riuscirò mai a separe i punti con una linea dritta, se invece posso piegare il foglio questa è la funzione di attivazione che mi permette di piegare il foglio, una volta piegato nel modo giusto, i punti rossi si troveranno da una parte e quelli blu dall'altra parte. Il Deep Learning è l'arte di piegare lo spazio (in questo caso matematico) per dare un senso ai dati. Ma quale piega dobbiamo scegliere è qui che entrano in gioco le funzioni di attivazione.
Non esiste  una funzione perfetta per ogni situazione/occasione. Non tutte le funzioni sono uguali. Nella storia del Deep Learning i ricercatori hanno sviluppato diverse funzioni, ognuna con vantaggi specifici e difetti strutturali che ne influenzano l'uso in base all'architettura.
Alcune son curve dolci altre spigolose, le più diffuse sono la sigmoide, la tangente iperbolica e la ReLu
Ognuna ha un carattere diverso e reagisce in modo diverso ai dati in input.

SIGMOIDE
La Sigmoide è la nonna delle attivazioni, prende qualsiasi numero e lo schiaccia tra o e 1. E' ideale per i layer di output dove serve una probabilità, ma soffre di gradienti che svaniscono.
Output tra 0 e 1, uso tipico, perfetta per probabilità:
- classificazione binaria (es. cliente abbandona si/no)

TANGENTE IPERBOLICA (Tanh)
E' simila alla sigmoide ma centrata sullo zero, con un range tra -1 e 1 (è più equilibrata). Risulta più equilibrata rispetto alla sigmoide

RELU
Se il numero è negativo lo cancella, se è positivo lo lascia invariato.
E' la regina del Deep Learning moderno per la sua semplicità e velocità, fa girare i modelli più potenti al mondo (oggi)
E' lo standard del deep learning
f(x)=max(0,x)

Alla somma pesata viene applicata la funzione di attivazione per l'output finale. La funzione di attivazione pertanto è quella che decide cosa esce dal neurone.

Ogni funzione hai i suoi problemi.
Sigmoide e Tanh soffrono del gradiente svanito, quando i numeri diventano troppo grandi la curva diventa piatta e la rete smette di imparare (l'aggiornamento dei pesi diventa quasi nullo durante l'addestramento), se un neurone riceve sempre segnali negativi, se spegne e si spegne per sempre. E' come cercare di dare istruzioni a qualcuno che è troppo lontano e sente solo un sussurro.
La ReLu può soffrore della morte del neurone. Se un neurone riceve sempre segnali negativi in input, il suo output e il suo gradiente restano zero: il neurone 'muore' e smette di imparare per sempre.
La Relu è veloce e non si appiattisce mai con i valori positivi.
Le funzioni centrate sullo zero (come la Tanh) aiutano l'ottimizzazione perchè mantengono la media della attivazoni vicina a zero, stabilizzando il calcolo dei pesi.

La Relu domina perchè è veloce come un fulmine e non si appiattisce mai per i valori positivi.
In una rete con milioni di campi questa velocità significa risparmiare giorni di elaborazione

```python
#immagina di aver calcolato la somma pesata e che valga 
```

```python
import numpy as np
```

```python
def sigmoid(z):
    """Schiaccia l'input nell'intervallo (0, 1)"""
    return 1 / (1 + np.exp(-z))
```

```python
def tanh(z):
    """Schiaccia l'input nell'intervallo (-1, 1)"""
    return np.tanh(z)
```

```python
def relu(z):
    """Soglia a zero: restituisce z se positivo, altrimenti 0"""
    return np.maximum(0, z)
```

# --- TEST DELLE FUNZIONI ---

```python
z_test = 2.0  # Valore di esempio (somma pesata)
```

```python
print(f"Risultati per z = {z_test}:")
print(f"1. Sigmoide: {sigmoid(z_test):.4f}")
print(f"2. Tanh:     {tanh(z_test):.4f}")
print(f"3. ReLU:     {relu(z_test):.4f}")
```

```python
# Esempio con valore negativo
z_neg = -1.5
print(f"\nRisultati per z = {z_neg}:")
print(f"1. Sigmoide: {sigmoid(z_neg):.4f}")
print(f"2. Tanh:     {tanh(z_neg):.4f}")
print(f"3. ReLU:     {relu(z_neg):.4f}")
```

Layer Nascosi
La regola generale è: usa sempre ReLU (o le sue varianti) nei layer interni per massimizzare la velocità e prevenire problemi di gradiente.

Output di Classificazione
Una la Sigmoide solo nell'ultimo layer e solo se devi prevedere una probabilità binaria (es. cane o gatto)

Output Regressione
Se devi prevedere un valore numerico (es. prezzo di una casa), spesso NON si usa alcuna funzione, lascia che il neurone sputi fuori il valore pureo.

L'output di un layer è il cibo del layer sucessivo. Se scegli ReLu, state propagando sparse attivazioni (molti zeri). Se scegli Tanh, stai propagando segnali densi e sbilanciati.

Scegliere una funzione di attivazione non è solo matematica ma gestione delle risorse, perchè decidiamo quanti segnali deve fluire attraversi i cavi dell'architettura neurale.

Senza non linearità il Deep Learning non esiste. Permette di modellare la complessità del mondo reale.
Ci sono 3 grandi funzioni di attivazione: Sigmoide (0,1); Tanh (-1,1) e ReLu (0,z). Ognuna ha un ruole specifico basato sulla posizione del layer.
ReLu è come standard per la sua efficienza e la risoluzione parziale del vanishing gradient, ReLu è la scelta predefinita per quasi ogni rete moderna.
