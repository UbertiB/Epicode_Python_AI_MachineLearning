OTTIMIZZAZIONE: GRADIENT DESCENT

Gradiente: Bussola per orientarci nel buio dell'errore.
Learning rate: quanto deve essere lunga la falcata mentre scendiamo verso la soluzione

Come facciamo a sapere in che direzione muoverci se non vediamo il fondo valle?
Immagina di trovarti in cima alla montagna, non vedi il rifugio a valle, ma senti con i piedi l'inclinazione del terreno, per scendere ti muovi nella direzione in cui il terreno scende cosi' arrivi prima alla base. 
Il GRADIENTE analizza la pendenza della montagna di errore e ci indica la via.
Senza questa indicazione la rete neurale cambierebbe i pesi a caso.

Il gradiente è un vettore che contiene le istruzione per ogni singolo peso della rete. Tecnicamente ci dice verso dove l'errore cresce di più, dato che noi vogliamo che l'errore diminuisca, dobbiamo muoversi esattamente nella direzione opposta.
La sua grandezza ci dice anche quanto è ripida la disceza, se il gradiente è enorme, siamo su un dirupo, se è vicino alla zero siamo quasi arrivati in pianura, cioè il nostro punto di minimo.

Questa idea si traduce in un contesto geometrico.
Visualizza questa superficie tridimensionale.
L'altezza è la nostra loss, il gradiente è una proiezione sul terreno che ci urla dove si trova la valle.
Si utilizano derivete parziali perchè lavoraiamo su ogni peso e ogni peso è come se fosse una coordinata diversa. 
Vogliamo isolare l'effetto di ogni singolo parametro: se cambio solo questo singolo peso, l'errore sale o scende? è un processo iterativo, viene fatto un passo dopo l'altro fino a quando non si raggiunge il fondo.

Nella formula c'è sempre un segno meno davanti al gradiente, se seguissimo il gadiente con il segno più andremmo a salire la nostra montagna e faremmo aumentare l'errore nel modello, il meno trasforma una formula matematica in un meccanismo di apprendimento, è la nostra volontà di fare meglio, di ridurre la discrepanza tra ciò che la rete crede e ciò che è vero.

Una volta trovata la direzione, quanto deve essere lungo il passo in quella direzione?
perchè il gradiente ci da la via, ma non ci dice di quanto dobbiamo avanzare, ed è qui che entra in gioco il LEARNING RATE, che è il parametro più influente di tutti. 
Immagina di dover centrare un bersaglio, il gradiente dice dove mirare, il learning rate dice con quanta forza scagliare la freccia. Sbagliare questa calibrazione significa condannare la rete all'instabilità o renderla immobile.
Se usiamo un passo troppo lungo, rischiamo overshooting, immagina di voler saltare un fosso e di voler fare salto così lungo da finire sulla parete opposta. se invce si fanno passi da formica non si supererà mail il fosso, il segreto è trovare una via di mezzo, una falcata corretta, un valore che permetta di scivolare velocemente verso il minimo ma che non sia abbastanza piccolo da non farci rimbalzare fuori dalla valle.

In un contesto reale questo valore di apprendimento, di solito si parte da valori standard come da 0,01 (ma non è una regola fissa).
Spesso usiamo il Decay, iniziamo a correre veloci quando siamo lontani dal minimo e poi man mano che il training avanza riduciamo il passo, permettendo alla rete di fare grandi salati all'inizio e piccoli aggiustamenti di precisione alla fine.

Diventa un lavoro di sintonizzazione fine che dobbiamo gestire noi programmatori, però esistono dei modi per rendere questo processo meno faticoso, come ottimizzatori intelligenti come ADAM che cambia la velocità del passo per ogni singolo peso. 

Prendiamo un peso, calcoliamo dove dobbiamo andare e lo modifichiamo.

```python
import numpy as np
```

```python
def gradient_descent_update(w, grad, eta):
    """
    Esegue un singolo step di aggiornamento del peso.
    w: valore attuale del peso
    grad: valore del gradiente (derivata della loss rispetto a w)
    eta: learning rate
    """
    # La formula fondamentale: w = w - eta * grad
    w_new = w - eta * grad
    
    return w_new
```

# --- SIMULAZIONE ---

```python
peso_iniziale = 0.5
gradiente_calcolato = 2.4  # Supponiamo che la pendenza sia verso l'alto
learning_rate = 0.1
#spostati del 10% della pendenza misurata, verso il basso
```

```python
# Primo aggiornamento
peso_aggiornato = gradient_descent_update(peso_iniziale, gradiente_calcolato, learning_rate)

print(f"Peso prima dell'aggiornamento: {peso_iniziale}")
print(f"Valore del gradiente: {gradiente_calcolato}")
print(f"Peso dopo l'aggiornamento: {peso_aggiornato:.2f}")

# Nota: Il peso è diminuito perché il gradiente era positivo. 
```

L'aggiornametno consiste nel sottrarre dal peso corrente il prodotto tra il learning rate e il gradiente calcolato nel punto attuale
Se il gradiente è positivo, significa che aumentare il peso aumenta l'errore. La formula quindi lo diminuisce correttamente.
Se il gradiente è negativo, aumentare il peso ridurrebbe l'errore. Sottraendo un valore negativo, la formula aumenta il valore del peso.
Il termine di aggiornamento totale rappresenta il piccolo spostamento che stiamo imponendo al nostro modello per renderlo più accurato.

Subito dopo il calcolo il peso cambia istantaneamente.
Il nuovo valore viene usato per la previsione successiva è un ciclo, previsione - errore -gradiente -aggiornamento. Ogni passo viene memorizzato nel peso creando una memoria storica. Attenzione ai segnali, un errore di distrazione del codice trasforma la discesa in una scalata infinita.
E se i pesi fossero milioni invece di uno solo?
non lo facciamo a mano ma entra in gioco la vettorizzazione di numpy, scala perfettamente anche con miliardi di parametri, utilizzando la stessa logica algebrica che applichiamo al singolo esercizio.

Il gradiente è la bussola 
Il learning rete decide quanto dobbiamo muoverci seguendo la bussola.
La regola di aggiornamento modifica realmente i parametri della rete.
