ESEMPIO ANN DA ZERO IN NUMPY (forward pass)

non usiamo libreria già pronte come keras, ma solo numpy per capire i passaggi (inizializzazione, vettorizzazione del calcolo, mini-batch)

In una rete neurale non calcoliamo un neurone alla volta, con il calcolo vettorializzato di numpy (operazioni tra matrici)

```python
import numpy as np
```

1) INIZIALIZZAZIONE
(si inizializa con valori casuali molto piccoli)

```python
#definiamo la rete
#3 feature in ingresso e 5 neuroni nello strato nascosto.
input_size = 3
hidden_size = 5
#creiamo i pesi per lo strato nascosto (pesi molto piccoli e diversi)
w = np.random.rand(hidden_size,input_size)*0.01
#creiamo i pesi per lo strato di uscita
b = np.zeros((hidden_size,1))
#definiamo una funzione di attivazione (ReLU)
w,b
```

```python
#eseguiamo il calcolo dello strato nascosto
#input di esempio (dato in ingresso)
x = np.array([[1],[2],[3]])
x=np.random.randn(input_size,1)
#calcolo dello strato nascosto
z = np.dot(w,x) + b #np.dot esegue il prodotto scalare tra matrice e vettore
z
```

```python
batch_size = 4
x_batch = np.random.randn(batch_size,input_size)
#calcolo vettorizzato, attenzione a trasformare l'array per moltiplicarlo con la matrice 
z_batch = np.dot(w,x_batch.T) + b
#applichiamo la funzione di attivazione ReLU
a = np.maximum(0,z_batch)
a  
```
