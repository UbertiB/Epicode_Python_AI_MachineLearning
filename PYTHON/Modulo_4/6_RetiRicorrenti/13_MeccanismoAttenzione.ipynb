MECCANISMO DI ATTENZIONE: FOCALIZZARE L'INTELLIGENZA NEUREALE

Abbiamo appena studiato Seq2Seq
input -> Encoder -> Context Vector -> Decoder -> Output
L'encoder legge tutta le frase, alla fine produce un sole vettore (Context Vector)
Quel vettore dovrebbe contenere tutto il significato della frase, è qui che nasce il problema.
Se la frase di input è breve, nessun problema, l'encoder riesce facilmente a riassumerla, ma ora immagina una frase molto più lunga, il context vector dovrebbe contenere tutte l'informazione compressa, è evidente che qualcosa di perde.

L'idea è rivoluzionaria
Perchè costringere il decoder a ricordarsi tutto? perchè non permettergli di guardare di nuovo l'input?
Se ascolto una storia molto lunga, ed ad un certo punto mi viene chiesto il nome del protagonista, il cervello non ripensa a tutta la storia ma va direttamente al punto deve è stato nominato il protagornista.
Nasce così il meccanismo dell'ATTENTION

Prima dell'Attention
il decoder vedeva solo il Context Vector
Dopo l'attention il decoder non riceve più soltanto l'ultimo strato, riceve tutti gli stati nascosti prodotti dell'encoder.

Ma non guarda tutta le frase allo stesso modo. Calcola quali parole siano importanti ottenendo, per ciascuna parola, una percentuale di probbabilità

Ecco perchè si chiama Attention
Il decoder dice continuamente:  
    "adesso quale è la parte importante?"
Ogni parola viene trasformata in un vettore, il decoder confronta il proprio stato interno con tutti i vettori prodotti dall'encoder, per ciascuno calcola un punteggio che vengono poi trasformati in probabilità tramite una funzione Softmax

Arriviamo poi al Context Vector Dinamico
Nel Seq2Seq classico avevamo un solo Context Vector, con l'Attentiopne il Context Vector cambia continuamente
quando il decoder produce 'I' il decoder produce 'eat'

ATTENTION IS ALL YOU NEED
e se eliminassimo completamente le RNN e usassimo solo l'Attenzione?
Nascono così i Transformer, modelli che non processano i dati in ordine sequenziale ma guardano l'intera sequenza simultaneamente tramite la Self-Attention.
Transformer è il vettore che oggi muovi giganti come GPT e BERT

Ma come può un modello capire l'ordine delle parole senza una sequenza temporale?
- Parallelizzazione totale: non dovendo aspettare lo step temporale precedente, i Transformer sono molto più veloci da addestrare
- Relazioni a lungo raggio: ogni parola può 'comunicare' con qualsiasi altra parola della frase istantaneamente.
- Query, Key e Value: il meccanismo con cui ogni token interroga il resto della sequenza per estrarre informazioni rilevanti. Ogni token lancia una ricerca (query), confronta i risultati (key) e scarica le informazioni (value). e per non peredere la posizione usiamo la positional encoding, una sorta di time-stamping
- Positional encoding: dato che non c'è ordine sequenziale naturale, aggiungiamo informazioni sulla posizione direttamente nei dati.

La rete, in questo modo, può prestasre attenzione a diversi aspetti della frase contemporaneamente (grammatica, semantica, ecc) usando più 'teste' di attenzione.
Questa architettura ha permesso la creazione dei grandi modelli di linguaggio moderni 'LLM' grazie alla sua capacità di scalare su hardware massicciamente parallelo.
La Query rappresenta 'cosa sto cercando', la Key 'cosa offro' e il Value 'l'informazione effettiva'. I match Query Key determina l'importanza del Value.

Il segreto dell'intelligenza non è ricordare tutto ma saper scegliere cosa ignorare.
L'attenzione è il ponte che permette ai dati di fluire senza ostacoli indipendentemente la loro distanza.
