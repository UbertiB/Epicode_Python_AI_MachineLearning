Hai mai provato a leggere un romanzo russo? magari a pagina 400 appare un personaggio citato solo al prima capitolo, come fa il cervello a capire chi è?
Le RNN classiche avrebbero dimenticato il personaggio dopo poche pagine, le LSTM permette alle macchine di fare esattamente quello che fa un umano per ricordare: dare un valore al passato per capire il presente.

Nelle RNN classica ogni nuovo hidden state viene calcolato partendo dal precedente, ogni volta il ricordo viene 'riscritto'. Dopo 1000 passaggi il ricordo iniziale è quasi sparito.
Il problema non è la memoria, il problema è la memoria che viene continuamente sovrascritta.

Perchè riscrivere tutta la memoria ogni volta?
Qui nasce LSTM
C'è una memoria separata chiamata Cell State
Immagina un treno, treno attraversa tutte le parole di una frase (come fossero stazioni), ad ogni stazione non viene ricostruito tutto, si aprono le porte, le persono possono scendere (forgot gate) o salire, ma il treno continua la sua corsa.
RNN= costruisco -> demolisco -> ricostruisco -> demolisco
LSTM: continua a viaggiare -> aggiunti -> togli -> continua
Nella LSTM il controllore del treno decide chi deve salire e chi scendere. Questa informazione mi serve ancora? se no diventa scende dal treno (forgot gate) se serve (input gate) la si inserisce nella Cell State
Quindi non si cancella tutto, ma si cancella solo alcune informazioni non ritenute necessarie.

Perchè usa la Sigmoide?
La sigmoide restituisce un valore tra 0 e 1. Qundi ogni informazione riceve un voto 0.99 tieni l'informazione
0.2 quasi inutile
Non è 0 o 1, ma un valore compreso tra 0 e 1, qindi mantieni al 99% (esempio precedente)

L'input Gate non salva direttamente, prima crea un candidato (con la Sigmoide), poi si usa la Tanh
La Tanh restituisce valori tra -1 e 1

La Cell State
L'autostrada dell'informazione nelle Long Shot-Term Memory
Immagina un nastro trasportatore che attraversa una fabbrica. nelle reti RNN tradizionali ogni operaio smonta e rimona il prodotto, alla fine del nastro l'oggetto originale è irriconoscibile. Nelle LSTM il nastro corre dritto, gli operai possono solo aggiungere piccoli componente, ma la struttura portante rimene integra.

Proprietà della Cell State
Perchè questa struttura risolve i limiti delle reti ricorrenti standard.
- Persistenza lineare: l'informazione scorre lungo la cella con interazioni lineari minime, riducendo il rischio di alterazioni drastiche.
- Superamento del gradiente svanente: grazie alla struttura a nastro, il gradiente può propagarsi all'indietro senza essere moltiplicato per pesi troppo piccoli.
Nelle RNN standard il gradiente svanisce perchè vengono moltiplicati ripeturamente per pesi non unitari. Nella Cell State l'informazione scorre attraverso operazioni lineari semplicissime, risolvendo il problema del vanisching gradient. Manteniamo l'informazione appresa all'inizio inalterata fino alla fine.
- Iterazione tramite gate: la Cell State non viene modificata direttametne dai pesi, ma solo attraverso operazioni puntuali regolate dai gate
- Memoria a lungo termine: permette di mantenere un'informazione utile appresa all'inizio di una frase fino alla sua conclusione.

Autonomia del Flusso Dei Dati
Operazioni puntuali: la cella interagisce con il mondo esterno solo tramite addisioni e moltiplicazioni elemente per elemento, evitando trasformazioni non lineari complesse che degradano il segnale
Se motiplico un numero per 0.9 per centinaia di volte, quel numero si azzera, se aggiungo a quel numero un piccolo segnale, solo quando server l'integrità del signale permane
L'analogia del nastro: immaginate un nastro trasportatore dove i gate possono aggiungere o rimuovere pacchetti di dati sena fermare il movimento principale del nastro stesso.
Stabilità del gradiente: questa architettura permette di calcolare la derivata dello stato attuale rispetto allo stato precedente in modo quasi unitario, prevedendo la scomparsa del gradiente

Formalizzazione della Cell State
Il valore della cella al tempo t dipende dallo stato precedente pesato dal gete di oblio e dal nuovo candidato pesato dal get di ingresso

Forget e Input Gate
I guardiano della memoria:
Se la Cell state è il nostro trasportatore, i gate sono le valvole che decidono quanto informazione può passare, essere aggiunta o essere eliminata.
I gete utilizzano funzioni di attivazione per mappare i valori tra zero e uno, agendo come filtri pecentuali sul flusso dei dati.

Meccanismi di filtraggio
Dall'oblio alla selezione dell'informazione
Sigmoide come selettore: produce un valore vicino allo zero per bloccare lìinformazione e vicino a uno per lasciarla fluire 
- Forget Gate: analizza l'input attuale e l'hidden state precedente per decidere quali parti del vecchio stato della cella non sono più utili.
- Input Gate: determina quali nuovi valori dell''imput corrente meritano di essere memorizzati nella cella.
- Candidato di memoria: un layer con attivazione tangente iperbolica  che crea un vettore di nuovi dati potenziali da aggiungere allo stato.

Logica delle Decisioni
Cosa dimenticare: se stiamo leggendo un testo e incroiano un nuovo soggetto possiamo, il forgot gate potrebbe decidere di azzerare informazione sul genere del sottetto precedente
- Vettore Conidato: la funzione 'tanh' assicura che i valori candidati siano normalizzati tra -1 e 1 permettendo sia aggiunte che sotrazioni alla memoria.
Dimenticare è importante tanto quando ricordasre

Matematica dei Gate
Le equazioni di controllo
Ogni gate ha il proprio set di pesi e bias che vengono addestrati tramite backpropagation per ottimizzare la selezione delle informazioni.
Tutto dipende dai pesi e dai bias
Il forgot gate ed l'input gate lavorano in parallelo per preparare la trasformazione dello stato della cella.
La decisione di dimenticare il passato, dipende sempre da ciò che stiamo vedendo nel presente. Una volta aggiornata la memoria interna, bisogna decidere cosa mostrare, cosa dire, al mondo esterno.

Output Gate e Identificazione
Dalla memoria alla predizione
L'ultimo passaggio consiste nel decidere cosa mostrare all'esterno, prendeno le informaizoni necessario dalla cell state. Non tutto ciò che è nella memoria a lungo termine è rilevante per l'output immediato.
L'output Gate filtra la Cell State per aggiornare l'Hidden State che verrà passato al passo successivo.

Come questo si trasfomra nel nuovo Hidden State
Come la cella comunica con il resto della rete
- Filtraggio dell'output: l'output Gate decide quale parte dello stato interno della cella deve influenzare l'output corrente
- Normalizzazione finale: lo stato della cella viene fatto passare attraverso una funzione 'tanh' prima di essere pesato dal gete
- Separazione dei ruoli: la Cell State conserva la memoria storica, mentra l'Hidden state trasporta solo ciò che è utile per il contesto immediato.
- Identificazione del gat di oblio: è l'unico gate che moltiplica direttamente il vecchio stato della cella, rendendolo responsabile delle eliminazioni.
