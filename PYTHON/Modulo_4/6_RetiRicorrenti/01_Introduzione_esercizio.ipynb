Nel mondo reale, una sequenz spesso non è composta da un solo valore (univariata), ma da più caratteristiche (multivariata). Immagina di avere due serie temporali: Prezzo di un'azione e Volume di scambi. Modifica la funzione crea_finestre o scrivi un nuovo script che riceva un array di forma (30,2) (30 giorni per 2 feature) e restituisca:
X: finestre di 5 giorni che includano entrambe le feature (forma desiderata: (N,5,2))
y: il solo prezzo del 6 giorno (forma desiderata: (N,1))

Suggerimento: usa lo slicing di NumPy su due dimensoni: array[i:i+5,:]

```python
import numpy as np

# =================================================================
# 1. GENERAZIONE DEI DATI (Multivariata: Prezzo e Volume)
# =================================================================
giorni = 30

# Feature 1: Prezzo (Trend + Rumore)
prezzi = 20 + np.linspace(0, 5, giorni) + np.random.normal(0, 1, giorni)
# Feature 2: Volume (Dati casuali simulati)
volumi = 100 + np.random.normal(0, 10, giorni)

# Uniamo le feature in un array di forma (30, 2)
# dati_multivariati[:, 0] è il Prezzo, dati_multivariati[:, 1] è il Volume
dati_multivariati = np.column_stack((prezzi, volumi))

print(f"Forma dataset originale: {dati_multivariati.shape} (Giorni, Feature)\n")

# =================================================================
# 2. DEFINIZIONE DELLA FUNZIONE DI FINESTRATURA
# =================================================================
def crea_finestre(serie, window_size):
    X = []
    y = []
    
    for i in range(len(serie) - window_size):
        # MODIFICA: Slicing su due dimensioni [righe, colonne]
        # Prendiamo 'window_size' righe e TUTTE le colonne (Prezzo e Volume)
        finestra = serie[i : i + window_size, :]
        
        # MODIFICA: Target solo sul Prezzo (colonna 0) del giorno successivo
        target = serie[i + window_size, 0]
        
        X.append(finestra)
        y.append(target)
        
    # Trasformiamo y in (N, 1) come richiesto usando reshape
    return np.array(X), np.array(y).reshape(-1, 1)

# =================================================================
# 3. APPLICAZIONE (Finestra di 5 giorni)
# =================================================================
WINDOW_SIZE = 5
X, y = crea_finestre(dati_multivariati, WINDOW_SIZE)

# =================================================================
# 4. ANALISI DEI RISULTATI
# =================================================================
# Ora X ha 3 dimensioni: (Campioni, Time Steps, Features) -> (25, 5, 2)
print(f"Forma di X (N, 5, 2): {X.shape}") 
# y ha forma: (Campioni, 1) -> (25, 1)
print(f"Forma di y (N, 1): {y.shape}")

print("\nEsempio primo campione:")
print(f"Input X[0] (5 giorni, 2 feature):\n{X[0]}")
print(f"Target y[0] (Prezzo del 6° giorno): {y[0]}")#
```
