```python
import torch
from sentence_transformers import SentenceTransformer
from scipy.spatial.distance import cosine


# 1. SCELTA AUTOMATICA DEL DISPOSITIVO

device = "cuda" if torch.cuda.is_available() else "cpu"

print(f"Dispositivo utilizzato: {device}")


# 2. CARICAMENTO DEL MODELLO

model = SentenceTransformer(
    "all-mpnet-base-v2",
    device=device
)


# 3. GENERAZIONE DEGLI EMBEDDING

words = [
    "king",
    "queen",
    "apple",
    "lemon",
    "computer",
    "mouse"
]

vectors = model.encode(words)

word_vectors = dict(zip(words, vectors))


# 4. RICERCA DELLA PAROLA PIÙ SIMILE

def find_most_similar(target_word, dictionary):
    if target_word not in dictionary:
        raise ValueError(
            f"La parola '{target_word}' non è presente nel dizionario."
        )

    best_word = None
    max_similarity = -1.0

    target_vector = dictionary[target_word]

    for word, comparison_vector in dictionary.items():
        if word == target_word:
            continue

        similarity = 1 - cosine(
            target_vector,
            comparison_vector
        )

        if similarity > max_similarity:
            max_similarity = similarity
            best_word = word

    return best_word, max_similarity


# 5. TEST

print("-" * 30)

for test_word in ["apple", "mouse"]:
    similar_word, score = find_most_similar(
        test_word,
        word_vectors
    )

    print(
        f"La parola più simile a '{test_word}' è "
        f"'{similar_word}' con score {score:.4f}"
    )
```
