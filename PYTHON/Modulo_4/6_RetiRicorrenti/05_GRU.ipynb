GRU: LA GATED RECURRENT UNIT E L'ARTE DELLA SEMPLIFICAZIONE

La Gru non introduce una nuova idea (rispetto alla LSTM), cerca di ottenere quasi gli stessi risultati della LSTM con meno complessità

Nella LSTM ho due memorie: 1) Cella State: memoria lunga 2) Hidden State: memoria breve. Son due memorie diverse
Nelle LSTM ho 3 gate: 1) Forgot Gete: (porta dell'oblio) 2) Input Gate (porta di ingresso) 3) Output Gate: (porta di uscite)

La GRU dice: Perchè avere due memorie?
Quindi sparisce la Cell State
La GRU dice: 3 gate sono troppi, tengo A) Update Gate: che fa il lavoro di Forgot e Input Gete B) Reset Gate: decide quanta memoria passata utilizzare

La GRU è l'arte di togliere ciò che non è strettamente indispensabile.
E' l'arte della semplificazione

La fusione dei fate
Dall'architettura a tre gate alla struttura binaria.
La GRU introdotta da Cho nel 2014 nasce come evoluzione della LSTM con l'obbiettivo di semplificare il flusso informativo senza perdere la capacità di gestire dipendenze a lungo termine.
La differenza fondamentale risiede nella rimozione del 'Cell State' separato, integrando tutta le memoria all'interno dell'Hidden State' tramite due soli gate principali: Update e Reset

I nuovi guardiani della memoria
Come Update e Reset Gate governano il flusso dei dati
- Update Gate (z): fonde le funzioni di 'forgot' e 'input' gate della LSTM, decidendo quanta parte dello stato precedente mantenere.
- Reset Gate (r): determina quanta parte della memoria passata deve essere ignorata nel calcolo del nuovo candidato di memoria
- Scomparsa del Cell State: l'informazione non viaggi apiù su due canali separati, ma viene aggiornata direttamente nello stato nascosto
- Memoria combinata: il nuovo stato è una combinazione lineare tra il vecchio stato e un candidato 'tilde h', regolata dal gete di aggiornamento.

Meccaniscmo di Aggiornameto lineare
Il ruolo del Reset Gate:
Se il reset gate è vicino a zero, la rete dimentica lo stato precedente e riparte dall'input attuale, permettendo di ignorare segmenti di sequenza irrilevanti.
L'eleganza dell'Update Gate
Questo gate agisce come un selettore bilanciato: se vale 1, manteniamo il passato; se vale 0, sovrascerive il passato con le nuove informazioni
Il candidato di stato:
Viene calcolato applicando una tangente iperbolica all'input attuale unito allo stato precedente 'resettato' del gate r.

Perchè dovremmo preferire una GRU?

Vantaggi Computazionali
Perche 'meno' spesso significa 'meglio'
L'efficienza è un fattore critico nel Deep Learning moderno. La GRU, avendo una struttura più snella rispetto alla LSTM, offre benefici immediati in termini di velocità e risorse.
La riduzione delle operazioni tensoriali influisce sui tempoi di training e sulla scalabilità dei modelli su larga scala.

Ma analiziamo dove risiede questa efficienza:

Efficienza e Performance
I punti di forza della Gated Recurrent Unit (GRU)
- Riduzione delle matrici di peso: una cella GRU richiede il calcolo di 3 set di pesi, contro i 4 richiesti dalla LSTM
- Minore overhead di memoria: l'assenza del 'cell state' riduce il consumo di memoria GPU durante il backpropagation through time
- Velocità di convergenza: spesso le GRU raggiungono una loss accettabile in meno epoche grazie alla struttura semplificata
- Efficacia sui piccoli dataset: la minore complessità del modello riduce il rischio di overfitting quando i dati sono limitati.

Scalabilità e produzione
- Inferenza più rapida: in scenari 'real-time' come il riconoscimento vocale, i millisecondi risparmiati per ogni passo temporale della GRU sono fondamentali. La GRU permette un'inferenza più rapide e l'utente non percipisce un ritardo
- Parallelismo limitato: sebbene rimanga una rete sequenziale, la riduzione del carico computazionale per cella permette di gestire batch di dimensioni maggiori
- Empirismo nelle performance: studi accademici dimostrano che LSTM e GRU ottengono performance simili, rendendo le GRU la scelta preferita per risparmiare risore.

In una LSTM  abbiamo 4 trasformazioni affini, mentre nella GRU ne abbiamo 3. Questo si traduce in una riducezione del 25% dei calcoli relativi

Confronto dei Parametri
Quando spazio occuperà il nostro modello nella nostra GPU o nello spazio dello smartphone. Questo è essenziale per il dimensionamento dell'hardware e per la prevenzione dell'overfitting

Il calcolo dei pesi
Analisi numerica della cella
- Struttura GPU: possibile tre matrici per l'input 'W' e tre matrici per lo stato ricorrente 'U', più i relativi bias
- Parametri totali GPU: il calcolo esegue la logica di 3 per 'h x h'+'h x d' + 'h' x bias
- Parametri totali LSTM: segue la stessa logica ma moltiplica per 4 gate invece di 3
- Risparmio strutturale: a parità di neuroni, la GPU è sempre strutturalmente più leggere della LSTM
