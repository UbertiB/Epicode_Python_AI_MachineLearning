Gestione della dimensione della sequenza
L'importanza del formato tridimensionale in PyTorch
A differenza dei modelli densi (tabelle dimensionali), le reti ricorrenti in PyTorch richiedono sensori a tra dimensioni.Questa struttura permette alla rete di processare non solo le caratteristiche, ma anche la profondità temporale. In una frase conta anche l'ordine con cui compaiono le parole, queta terza dimensione di Pytorch assolve a questo problema.
Comprendere l'ordine di queste dimensioni è il primo passo per evitare errori di architettura durante la fase di feed-forward
E' una struttura che permette alla rete di non vedere i dati come punti isolati, ma come una melodia  nel tempo.
Senza questa struttura, la ricorrenza non potrebbe esistere.

Come si compongono queste tre dimensioni?

Struttura dei Tensori 3D
Dimensioni fondamentali per le sequenze
- Sequence Length: rappresenta il numero di passi temporali che compongono ogni singolo campione della sequenza. Quanti batti ha il nostro cuore temporale
- Batch Size: definisce il numero di campioni indipendenti processati simultaneamente dalla rete durante un'iterazione. Numero di sequenza che analiziamo in parallelo, come se leggessimo più libri contemporaneamente
- Input Size: indica il numero di feature o variabili presente in ogni singolo istante temporale. Quante informazioni abbiamo per ogni singolo istante.

Attenzione al parametro Batch First: decide se dasre precedenza al batch a alla lunghezza della sequenza, identificando, uno dei due, come primo parametro.
Sbagliare questo ordine, vuol dire far fallire la rete

Configurazioe del Tensore di Input
Il default di PyTorch:
Di default, PyTorch si aspetta la lunghezza della sequenza come prima dimensione (per motivi di ottimizzazione hardware), un formato ottimizzato per le prestazioni ma controintuitivo per chi proviene da Keras.
Come se leggesse tutti i primi caratteri di ogni libro, poi tutti i secondi, e così via.
Tutta via, noi umani, ma anche Keras, preferiamo pensare per campioni, per questo c'è il parametro batch_first
Uso di batch_first:
Impostando questo parametro a True nell'istanza della RNN, possiamo utilizzare il formato Batch, Sequence, Feature, facilitando la gestione dei dati.
Reshaping dinamico:
Spesso i dati grezzi sono bidimensionali e richiedono l'uso dle metodo 'view' o 'unsqueeze' per aggiungere la dimensione temporale mancante.

Rappresentaizione matematica dei tensori
Dimensionamento dei dati di input
Ogni input x dal tempo t è un vettore di dimensione D. Una sequenza intera è composta da L di questi vettori, organizzati in un batch di dimensione B
La corretta allocazione spaziale nel tensore garantisce che i pesi della rete vengano applicati in modo coerente lungop l'asse temporale.

Una volta definita la forma del dato dobbiamo preoccuparci di chi quel dato deve ritornarlo Hidden State

Inizializzazione dell'Hidden State
Gestire la memoria iniziale della rete
L'Hidden State è la memoria a breve termine della nostra rete
In PyTorch (a differenza di Keras), lo stato nascosto non è gestito automaticamente come in altri framework di alto livello. Dobbiamo definire esplicitamente come la rete debba iniziare a ricordare.
Questo controllo manuale permette di implementare logiche avanzate, come stati iniziali apprendibili o il reset dello stato tra sequenze diverse.

Ma come si crea, tecnicamente, questa tabula rasa

Stato Nascosto e Memoria
Componenti del tensore h
- Inizializzazione a zero: la tencica più comune consiste nel creare un tensore di zeri per indicare che all'inizio la rete non ha memoria pregressa. Non sai nulla, inizia a imparare da qui (zeri). 
- Dimensione dello stato: lo stato nascosto richiede una forma specifica basata sul numero di layer e sulla dimensione dell'hidden state stesso. Questo tensore deve avere una forma che rispetti il numero di layer della rete e la dimensione della memoria interna
- Passaggio dello stato: la RNN riceve in input il dato corrente e lo stato precedente, restituendo il nuovo stasto aggiornato.
- Metodo detach: operazione necessaria per rompere il grafo computazionale durante il passaggio tra batch diversi ed evitare l'esplosione dei gradienti.

Esistono però diversi modi di inizializzare la rete (oltre lo zero)
Stati multistrato
Se la rete ha più di un layer ricorrente, lo stato iniziale deve contenere un vettore per ogni livello della gerarchia.
Inizializzazione Random: in alcuni casi particolare, inizializzare con rumore bianco può aiutare la rete a esplodere meglio lo spazio delle soluzioni. Inizializzare con valori casuali può aiutare la rete a non fossilizzarsi ed a esplorare meglio soluzioni possibili.
Uso di h_O nullo: se non forniamo uno stato iniziale esplicito alla RNN, PyTorch creerà internamente una stato di zeri, ma perderemmo il controllo sul monitoraggio della memoria. Significherebbe lasciare che Pytorch decida per noi, un bravo programmatore non si concede mai questo lusso
