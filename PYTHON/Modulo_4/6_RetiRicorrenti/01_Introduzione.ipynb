L'Importanza dell'Ordine nei Dati
Fino ad ora abbiamo trattato dati dove ogni esempio era indipendente dagli altri, come le immagini in un dataset di immagini, ogni campione era un sistema isolato, foto di un cane, foto di un aereo, ecc
Nelle sequenze questa indipendena scompare, ogni dato è un anello di una catena, se spezzi la catena o cambi l'ordine perdi l'informazione.

- Autocorrelazione: ciò che sentimao ora dipende da ciò che è accaduto poco fa
- Contesto: in una frase il significato di una parola è spesso chiaria dalle parole precedenti. Nel Deep Learning questo si traduce nella necessità di una memoria
- Lunghezza variabile. Le sequenza hanno spesso lunghezze diverse tra loro
- Rappresentazione matematica: una sequenza viene vista come un insieme ordinato di vettori estratti a intervalli discreti di tempo

Caratteristiche Strutturali delle Sequenze.
- Causalità: il principio per cui l'output al tempo 't' può dipendete solo dagli input avvenuti ai tempi precedenti o uguali a 't', mai nel futuro. Il domani non può influenzare ieri.
- Stazionarietà: una serie stazionaria mantiene media e varianza costanti, se cos' non è, il compito diventa molto più arduo
- Passo Temporale (Time Step): l'unità minima di avanzamento nella sequenza, fondamentale per definire la risoluzione dei dati (es. dati giornalieri vs ti orari)

Limiti dei modelli densi (MLP)
Le reti 'Feed-Forward' processano gli input in un unico blocco. Per gestire una sequenza, dovremmo 'appiattirla' in un unico vettore, perdendo la nozione di invarianza temporale e aumentando esponenzialmente i parametri. Sarebbe come leggere un intero libro schiaccianto tutte le pagine una sopra l'altra.
I modelli sequenziali devono invece condividere i pesi lungo ogni passo temporale, permettendo alla rete di imparare pattern indipendentemente da quando appaiono.
Se appiattiamo una sequenza, per darla in pasto ad un MLP, distruggiamo la struttura temporale.
Le reti ricorrenti invece sono come vettori che scorrono le pagine una ad una (passo temporale) usando la stessa funzione f per ogni passo.

Topologie delle Rete Ricorrenti
Mappare input e output in diverse dimensioni.
A volte abbiamo un singolo input che genera una sequenza, altre volte una sequenza che deve essere rissunta in un unico valore.
Questi schermi definiscono la struttura della rete e determinano come l'informazione fluisce attraverso i layer nel corso del tempo.
- Uno a Molti: un singolo input (es. un'immagine) produce una sequenza di output (es.la descrizione testuale dell'immagine)
- Molti a Uno: Una sequenza di input (es. una recensione) produce un singolo output categorico (es. sintiment positivo o negativo)
- Molti a Molti (Sincrono) una sequenza di input produce una sequenza di output di pari lunghezza (es. classificazione di ogni frame in un video o i sottotitolli che devono apparire esattamente mentre si parla)
- Molti a Molti (Asincrono): una sequenza produce un'altra sequenza di lunghezza indifferente tipica dei modelli 'Encoder-Decoder' per la traduzione (traduttore che ascolta l'intera frase prima di iniziare a parlare in un'altra lingua)

Analisi delle Topologie
- Bottleneck informativo:
Nel molti a uno, l'ultimo stato della rete deve riassumere tutta la storia della sequenza; se la sequenza è troppo lunga l'informazione inziale rischia di andare persa. E' come se gli chiedessi di leggere 'guerra e pace' ed alla fine di riassumere tutto in una unica parola, perdiamo dettagli prezioni dell'inizio del libro.
- Allineamento temporale:
nei modelli molti a molti sincroni, l'output al tempo 't' spesso è vincolato a essere calcolato non appena l'input 't' è disponibile, in altri modelli invece possiamo permetterci il lusso di aspettare.
- Flessibilità dell'output:
I modelli uno a molti sono spesso usati nella generazione generativa, dove l'output di un passo viene usato come input per il passo successivo.

Scegliere la topologia corretta significa bilanciare memoria, velocità e precisione.

Come fa la rete a rimenere efficiente pur processando sequenza lunghissime.

Condivisione dei Parametri
In segreto dell'efficienza ricorrente.
Indipendentemente dalla topologia scelta, la caratteristica chiave è l'uso degli stessi pesi 'W' per ogni passo temporale. Questo permete alla rete di generalizzare su sequenze di lunghezza arbitraria.
Questa struttura è ciò che permette di estrarre feature temporali coerenti, riducendo drasticamente il numero di parametri rispetto a un approccio denso.
Invece di imparare pesi diversi per ogni sencod della giornata, la rete usa lo stesso insieme di pesi per ogni passo temporale.

Ma i dati grezzi raramente sono pronti per la rete, dobbiamo prima preparali

Preprocessing per sequenze.
- Finestratura (Windowing): tecnica per trasformare una singola serie temporale lunga in molti campioni più brevi sovrapposti, pronti per l'addestramento.
- Padding: serve per l'eleganza informatica. aggiunta di valori neutri (solitamente zeri) per rendere tutte le sequenze di un batch della stessa lunghezza.
- Tokenizzazione: processo di scomposizione di una sequenza di testo in unità atomiche (parola o caratteri) convertibili in indici numerici. Trasformiamo le parole in numeri perchè le reti neurali riescono a gestire solo la matemtica non le lettere 

La sfida della Predizione
la predizione del futuro.
Modellare una sequenza significa stimare probabilità.Ogni volta che il telefono suggerisce la parola successiva di un sms, sta calcolando una probabilità condizionata. Nei modelli complessi, cerchiamo di approssimare questa distribuzione tramite uno stato nascosto che funge da memoria.
Il successo della predizione dipende dalla capacità del modello di catturare dipendenze a lungo termine senza fasi distrarre dal rumore locale.


```python
import numpy as np

# =================================================================
# 1. GENERAZIONE DEI DATI (Simulazione serie temporale)
# =================================================================
# Iniziamo creando un dataset sintetico. In un contesto reale, 
# questi potrebbero essere prezzi di borsa, temperature o segnali audio.

giorni = 30

# Trend Lineare: Rappresenta la componente deterministica. 
# Usiamo np.linspace(start, stop, num) per creare una crescita costante.
# Matematicamente: T_base(t) = 20 + m*t
base_temp = 20 + np.linspace(0, 5, giorni) 

# Rumore Gaussiano: La realtà non è mai una linea retta. 
# Aggiungiamo 'sporcizia' statistica (media=0, deviazione standard=1).
# Questo costringe il modello a imparare a distinguere il segnale dal rumore.
rumore = np.random.normal(0, 1, giorni)

# La serie finale è la somma di Trend + Rumore.
temperature = base_temp + rumore

print(f"Serie originale (primi 5 giorni): {temperature[:5]}\n")

# =================================================================
# 2. DEFINIZIONE DELLA FUNZIONE DI FINESTRATURA (Sliding Window)
# =================================================================
def crea_finestre(serie, window_size):
    """
    Questa funzione è il 'cuore' della preparazione dati per RNN/LSTM.
    Trasforma un problema di sequenza in un problema di apprendimento supervisionato.
    
    CONCETTO CHIAVE:
    Dato un vettore $X = (x_1, x_2, \dots, x_n)$, vogliamo creare coppie $(input, target)$ 
    dove l'input è una 'finestra' di osservazioni passate e il target è il valore futuro.
    """
    X = []
    y = []
    
    # Il loop si ferma a (lunghezza - window_size) perché l'ultimo target 
    # deve avere una finestra completa che lo precede.
    for i in range(len(serie) - window_size):
        
        # SLICING: Estraiamo un blocco di dati da i fino a i + window_size.
        # Se window_size=7 e i=0, prendiamo gli indici [0, 1, 2, 3, 4, 5, 6].
        finestra = serie[i : i + window_size]
        
        # TARGET: È il valore all'indice successivo (i + window_size).
        # Sigifica: "Dati i 7 giorni precedenti, predici l'ottavo".
        target = serie[i + window_size]
        
        X.append(finestra)
        y.append(target)
        
    # Convertiamo in array NumPy per l'efficienza computazionale (richiesto dalle librerie come TensorFlow/PyTorch)
    return np.array(X), np.array(y)

# =================================================================
# 3. APPLICAZIONE (Preprocessing per il Modello)
# =================================================================
# Scegliamo una finestra di 7 giorni (una settimana). 
# Questo è un iperparametro critico: troppo corta perde il contesto, 
# troppo lunga rende il modello lento e difficile da addestrare.
WINDOW_SIZE = 7
X, y = crea_finestre(temperature, WINDOW_SIZE)

# =================================================================
# 4. ANALISI DELLE DIMENSIONI (Shapes)
# =================================================================
# X.shape restituirà (N, T). 
# N = Numero di campioni (quante finestre siamo riusciti a scorrere)
# T = Time Steps (lunghezza della finestra)
print(f"Forma di X (Campioni, Time Steps): {X.shape}") 
print(f"Forma di y (Target): {y.shape}")

print("\nEsempio primo campione:")
print(f"Input (X[0] - primi 7 giorni): {X[0]}")
print(f"Target (y[0] - valore atteso all'8° giorno): {y[0]}")
```
