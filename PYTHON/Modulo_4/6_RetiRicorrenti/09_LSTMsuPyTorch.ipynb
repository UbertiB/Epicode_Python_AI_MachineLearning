LSTM SU PYTORCH: CONTROLLO TOTALE DEL TRAINING E INFERENZA.

Su Pytorch, lo stesso problema IMDb cambia soprattutto per un motivo: Keras nasconde gran parte del ciclo di addestramento, PyTorch ti obbliga a gestirlo direttamente.

Questo significa più codice ma anche più controllo.

Il problema resta identico:
'This movie was excellent'
Outoput:
'positivo'
L'architettura concettuale non cambia rispetto a Keras, cambia il modo in cui gestisci:
- batch
- passaggio forward
- calcolo dell loss
- gradienti
- aggiornamento dei peis
- modalità training e inferenza
- uso CPU o GPU
- salvataggio modello
Con PyTorch se tu a scrivere esplicitamente questi passaggi.

In Keras scrivi:
model.fit(...)
e il framework gestisce: forward pass, loss, backward pass, optimizer, batch, metriche
In PyTorch scriti tu:

optimizer.zero_grad()
output = model(x)
loss = criterion(output, y)
loss.backward()
optimizer.step()

Significa che puoi intervenire in qualunque punto.
Per esempio puoi:
- usare loss diverse per categorie diverse
- bloccare alucni layer
- modificare i gradienti
- eseguire gradient clipping
- gestire più output contemporanemante
- ispezionare i tensori intermedi
- costruire training loop non standard
