CONFRONTO TENSORFLOW E PYTORCH IN CASI REALI

TensorFlow e PyTorch servono entrambi a costruire, addestrare ed eseguire reti neurali. Non sono modelli: una CNN, una GRU o un Transformer posssono essere implementati con entrambi.
La scelta del framework, da sola, non rende il modello più preciso. A parità di architettura, dati, inizializzazione e addestramento, i risultati possono essere molto simili. La differenza reale riguarda soprattutto:
- come scrivi il codice
- quanto facilmente lo controlli e lo correggi
- come distribuisci il modello
- quali librerie devi utilizzare
- quali competenze possiede il gruppo di lavoro.

La vecchia frase 'TensorFlow per la produzione, PyTorch per la ricerca' è ormai superata. 

La differenza fondamentale nel modo di lavorare.

*** TensorFlow con Keras ***

model = keras.Sequential([keras.layers.Dense(64, activation="relu"),keras.layers.Dense(1)])
model.compile(optimizer="adam",loss="mse")
model.fit(X_train,y_train,epochs=20,validation_split=0.2)
Con TensorFlow normalente lavori ad un livello molto alto.
Poche istruzioni per chiedere al framework di gestire:
- passaggio in avanti - calcolo della loss - backpropagation - aggiornamento pesi - metriche - validatione - batch - callback - chekpoing
Il principale vantaggio: PRODUTTIVITA'
Il rischio che si utilizzi model.fit() senza capire cosa ci stia diestro, ma questo è solo un cattivo uso dello strumento

*** PyTorch nativo ***
model.train()
for X_batch, y_batch in train_loader:
    optimizer.zero_grad()
    predictions = model(X_batch)
    loss = loss_function(predictions, y_batch)
    loss.backward()
    optimizer.step()
Con Pytorch il ciclo di addestramento è normalmente più esplicito.
Qui vedi direttamente: - azzeramento gradienti - forward pass - calcolo della loss - backward pass - aggiornamento pesi
Questo lo rende molto utile per capire il deep learning e per realizzare addestramenti non standard.
Non significa che Pytorch sia più complicato. Per un modello normale il codice è solo più esplicito. Per un modello molto particolare, questa esplicitazione può diventare un vantaggio enorme.

Ma perchè esistono queste due tecnologie?

Dinamismo vs Staticità
Eager execution e grafi computazionali
- Eager Execution: PyTorch opera nativamente in modelità imperativa, usa la Eager Execution, è come parlare una lingua, se fai un errore grammaticale te ne accorgi subito. Questo facilita il debugging tramite i classici strumenti Python come 'pdb'
- Grafi Statici: TensorFlow ottimiza l'esecuzione tramite grafi che possono essere serializzati e ottimizzati indipendentemente dall'interprete Python
- TFX e TF Sarving: TensorFlow offre una suite completa per la gestione dei metadati, il versionamento e il serving ad alte prestazioni
- TorchScript e Compile: le recenti evoluzioni di PyTorch permettono di compilare i modelli per ottenere Performance vicine a quelle dei grafi statici.

E quando passiamo dal laboratorio al cloud cosa cambia?
in cloud la parola d'ordine è portabilità, TensorFlow eccede qui
- Portabilità dei modelli: TensordFlow facilita l'esportazione verso formati come SavedModel, pronti per essere caricati in container Docker ottimizzati per Google Cloud o AWS. Questo formati, una volta chiusi, possono essere esportati ed importati in qualsiasi ambiente
- Flessibilità nella ricerca: nella ricerca vince PyTorch (oltre 80% a NeurIPS) perchè permette di modificare l'architettura della rete durante l'esecuzione stessa.
- Rappresentazione del Grafo: la rappresentazione astratta di un modello può essere vista come un grafo diretto aciclico dove i nodi sono operazioni e gli archi sono tensori.

Community e Librerie Satelliti

La scelta dello stack spesso non dipende dal framework core, ma dalla disponibilità di un modello pre-addestrato o di una libreria di utility specifica su una determinata piattaforma.

E se volessimo cambiare casa?
esiste un traslocatore universale?
esiste e si chiama ONNX (Open Neural Network Exchange), funge da ponte permettendo di addestrare un modello in PyTorch ed eseguire il deploy in ambienti ottimizzati per TensorFlow. E' il traduttore ufficiale, permette di scrivere codice nel confort di PyTorch e di trasferirlo ad una fabbrica che usa solo TensorFlow.
Librerie come 'Fastai' per PyTorch hanno rivoluzionato l'insegnamento, mentre TensorBoard' rimane lo standard de facto perla visualizzazione delle metriche.
L'aggiornamento dei pesi durante il training segue regole stocastiche che devono essere consistenti tra le diverse implementazioni delle librerie satelliti.

Metriche Decisionali
Indicatori chiave di prestazione (KPI)
- Time-to-Market: PyTorch vince speso nella fase di prototizazione grazie alla facilità di scrittura e debug del codice
- Runtime Efficienty: TensorFlow grazie a XLA (Accelerated Linear Algebra), può offrire performance superiori in inferenza su hardware specifico. Può compilare codice su hardware specifici, come le TPU di Google
- Support Hardware: TF ha un supporto nativio eccellente per le TPU (Tensor Processin Unit) di Google
- Curva di apprendimento: Keras (TF) è generalmente considerato più semplice per i principipanti, mentre PyTorch richiede una compressione più profonda dei tensori

Ma i costi non sono solo hardware, sono anche umani

Oggi è possibile scrivere il codice una sola volta, e con una sola riga di codice decidere se farlo girare in TensorFlow, Pytorch o JAX

```python
import os

# Configurazione del Backend: Nel 2026, Keras 3 permette di scegliere il motore di calcolo.
# Impostiamo 'torch' prima di importare keras.
os.environ["KERAS_BACKEND"] = "torch"

import keras
import torch
import torch.nn as nn
import numpy as np

# 1. GENERAZIONE DATI SINTETICI (Standard per entrambi)
# Creiamo un problema di regressione semplice: y = 3x + 2 + rumore
X = np.random.rand(1000, 1).astype("float32")
y = 3 * X + 2 + np.random.randn(1000, 1).astype("float32") * 0.1

# --- APPROCCIO A: KERAS 3 (Focus: Velocità di Deploy e Semplicità) ---
print("\n--- Training con Keras 3 (Backend: PyTorch) ---")

# Definizione del modello: Astrazione di alto livello.
# Teoria: L'approccio dichiarativo nasconde la complessità del grafo computazionale.
keras_model = keras.Sequential([
    keras.layers.Input(shape=(1,)),
    keras.layers.Dense(1)
])

# Compilazione: Nel 2026 'adamw' è spesso il default per una migliore regolarizzazione.
keras_model.compile(optimizer="adamw", loss="mse")

# Training Loop: 'fit' gestisce automaticamente batching, callback e log.
# Best practice 2026: usare verbose=2 per log puliti in ambienti cloud/CI-CD.
keras_model.fit(X, y, epochs=5, batch_size=32, verbose=0)

# Salvataggio: Il formato '.keras' è lo standard universale (V3).
# È un archivio compresso che include architettura, pesi e configurazione dell'ottimizzatore.
keras_model.save("my_model_2026.keras")
print("Modello Keras salvato in formato .keras")


# --- APPROCCIO B: PYTORCH NATIVO (Focus: Ricerca e Controllo Granulare) ---
print("\n--- Training con PyTorch Nativo ---")

# Conversione dati in Tensori PyTorch
X_pt = torch.from_numpy(X)
y_pt = torch.from_numpy(y)

# Definizione del modello: Approccio imperativo (Object-Oriented).
# Teoria: Ereditare da nn.Module permette il controllo totale sul 'forward pass'.
class LinearResearchModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(1, 1)
        
    def forward(self, x):
        return self.linear(x)

pt_model = LinearResearchModel()
optimizer = torch.optim.AdamW(pt_model.parameters(), lr=0.01)
criterion = nn.MSELoss()

# Training Loop Manuale: Tipico della ricerca per inserire logica custom (es. gradient clipping).
# Teoria: Qui vediamo esplicitamente il ciclo di 'Zero-Grad', 'Backward' e 'Step'.
for epoch in range(5):
    # Forward pass
    pred = pt_model(X_pt)
    loss = criterion(pred, y_pt)
    
    # Backpropagation: Calcolo esplicito dei gradienti tramite Autograd.
    optimizer.zero_grad() # Reset dei gradienti accumulati
    loss.backward()      # Calcolo derivate parziali dLoss/dWeight
    optimizer.step()     # Aggiornamento dei pesi tramite l'ottimizzatore
    
print(f"Ultima Loss PyTorch: {loss.item():.4f}")

# Salvataggio PyTorch: Best practice 2026.
# Si preferisce salvare lo 'state_dict' (solo i pesi) per massima portabilità.
torch.save(pt_model.state_dict(), "pt_weights_2026.pt")
print("Pesi PyTorch salvati in formato .pt")

# 3. ANALISI DEI TRADE-OFF (Esempio di Inferenza)
# Keras è pronto all'uso, PyTorch richiede la ricostruzione della classe.
print("\n[INFO] Keras: model.predict(X)")
print("[INFO] PyTorch: pt_model(X_pt) con torch.no_grad()")
```
