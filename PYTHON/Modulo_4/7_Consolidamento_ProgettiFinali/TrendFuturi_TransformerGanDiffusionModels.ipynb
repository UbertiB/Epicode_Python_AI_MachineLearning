Il Meccanismo dell'Attenzione
Focalizzare le risorse computazionali
- Self-Attention: permette a ogni parola in una sequenza di relazionarsi con tutte le altre, indipendentemente dalla distanza fisica nel testo. Ogni parola può parlare con ogni altra parola del testo simultaneamente. Non c'è più quindi un ordine sequenziale che ci rallenta. Possiamo far analizzare il tutto sulle GPU
- Parallelizzazione: a differenza delle LSTM, i Transformers consentono l'addestramento massimo su GPU eliminando il collo di bottiglia temporale.
- Scaling Laws: l'osservazione che l'aumento dei parametri e dei dati porta a miglioramenti prevedibili e talvolta a priorità emergenti
- Positional Encoding: tencica per iniettare informazioni sull'ordine delle parole in un sistema che processa tutto in parallelo.
