ANALISI DI SENTIMENT CON HUGGING FACE

L'analisi dei sentiment con Hugging Face serve a classificare un testo in base al tono espresso, per esempio:
positivo, negativo, neutro.
Oppuere, a seconda del modello: 1 stella, 2 stelle, 3 stelle, 4 stelle, 5 stelle.
Il punto importante è che Hugging Face non è il modello: è l'ecosistema che ti permette di scaricare e usare modelli trasformer addestrati.

La pipeline è:
testo -> tokenizer -> token id -> modello transformer -> classificazione -> sentiment

Esempio:
"il prodotto è fantastico e la consegna è stata velocissima" potrebbe diventare POSITIVE score=0.98
mentre
"prodotto difettoso e servizio pessimo" potrebbe diventare NEGATIVE score=0.99
il risultato è pertanto una label (classe prevista) e uno score (confidenaza del modello)
Una score di 0.97 non significa che il modello sia 'corretto al 97%' in senso generale, significa che, per quella predizione, e secondo il modello, qualla classa ha un punteggio/confidenza molto alta del 97%

La cosa comoda di Hugging Face è la funzine pipeline()
Hugging Face si occupa, dietro le quinte, di parecchie cose
testo -> tokenizzazione -> token IDs -> padding/trunction -> modello -> logits -> probabilità -> label

Questoè quello che abbiamo già visto con spaCy, molta complessità incapsulata nella libreria.

Devi però scegliere un modello, in questo caso adatto al sentiment analysis.
Non basta prendere un qualunque BERT
Per esempio un modello BERt potrebbe essere stato pre-addestrato principalmente per comprendere il linguaggio, ma non necessariamente addestrato per distinguere positivo/negativo.
Per fare sentiment analysis normalmente prendi un modelo che sia stato successivamente fine-tuned per text classification/sentiment analysis
quindi:
BERT base -> conosce rappresentazioni linguistiche
+ 
fine-tuning su recensioni/testi con etichette positive/nevative
= 
BERT per sentiment analysis

Quindi hai due fasi:
1) PRETRAINING: il modello impara il linguaggio
2) FINE-TUNING: impare un task specifico come il sentiment
Quando scarichi un modello già tine-tuned di Huggin Face, entrambe queste fasi sono già state fatte da altri.
Tu fai solo inferenza.
cioè:
nuovo testo -> modello già addestrato -> predizione

Mentre con TF-IDF e Logistic Regression se tu il protagonista, sei tu che fai model.fit(...) con un transformer preaddestrato e fine-tuned lavori su un qualcosa che ha già fatto training/fine tuning. Qualcunaltro ha pubblicato il modello su Huggin Face e tu lo scarichi semplicemente per fare classifier.
In questo caso potresti non aver bisogno di addestrare nulla.

Un Transformer analizza le parole nel loro contesto, quindi puoi distinguere meglio frasi del tipo: "il prodotto è buono" da altre del tipo "il prodotto non è buono" oppure "pensavo fosse pessimo, invece è fantastico"
Perchè Attention e rappreentazioni contestuali consentono al modello di mettere in relazione parole anche distanti.

HUGGIN FACE
Hugging Face è come una super car già pronta per la pista.
Hugging Face ha democratizato l'accesso ai modelli Transformer attraverso la libreria 'transformers'. Il componente principale per i principianti e per la prototipazione rapida è la funzione 'pipeline', che incapsula tutte le fasi di preprocessing, inferenza e post-processing in un unico oggetto di facile utilizzo.
E' come un nastro trasportatore, voi inserite un testo grezzo da un lato, e quasi istantaneamente ricevi un'analisi strutturato dall'altro lato. Non devi preoccuparti di spezzare le parole o di quanti strati neuroni ci siano, la pipeline incapsula modelli importanti permettendoti di concentrarti sul valore del vostro business.
Invece di gestire manualmente la tokenizzazione o l'architettura della rete neurale, la pipeline permette di invocare un compito specifico come 'sentiment-analysis' e ottenere risultati immediati.
Questo approccio 'balck-box' è costruito sopra modelli complessi come BERT o RoBERTa, garantendo prestaszioni allo stato dell'arte.

Quando invochi una pipeline di sentiment analysis, accadono 3 cose in pochi millisecondi:
- Task-Specific: ogni pipeline è ottimizzata per un compito preciso, eliminando la necessità di configurare manualmente i layer finali della rete. Cioè il testo viene trasformato, automaticamente, in numeri comprensibili per la rete.
- Auto-Model Loading: la libreria scarica automaticamente i pesi del modello predefinito o di uno specificato dall'utente. Questi numeri attraversano diversi tensori neurali
- Tokenizzazoine Integrata: il testo viene convertito nel formato tensoriale corretto senza passaggi intermedi visibili al progremmatore. Il modello emette il suo verdetto.

Il risultato è una lista di dizionari contenenti l'etichetta predetta e il punteggio di confidenza calcolato dal modello.
E' il modo in cui il modello dice quanto è sicuro della sua predizione

Nell'industria moderno la velocità è tutto, le pipeline di Huggin Face sono ottimizzate per l'uso sia su CPU che su GPU, permettendo di scalare l'analisi di migliaia di documenti con poche righe di codice.
La flessibilità è totale, se il modello predefinito non vi soddisfa, potete puntare ad uno dei tantissimo modelli presenti  sull'Hub di Hugging Face (es. 'distilbert-base-uncase d') per adattare l'analisi a domini specifici (personalizzazione del Modello). Come cambiare la lente di una macchina fotografica per adattarla alla luce.
La pipeline gestisce automaticamente l'applicazione della funzione Softmax per trasformare i logit grezzi in probabilità leggibili.

Il cervello elettronico non ragiona il probabilità ma in logits. Dietro le quinte, il Trasformer emette dei valori chiamati logit. Questi non sono probabilità, ma punteggi gressi che indicano la forza della predizione per ogni classe, questi punteggi possono essere di qualsiasi numero che a noi umano non direbbe molto. La pipeline applica matematicamente la funzione Softmax per normalizzare questi valori. Immagina la softmax come un correttore che schiaccia tutto tra 0 e 1
La probabilità di una classe i-esima viene calcolato rapportando l'esponenziale del suo punteggio alla somma degli esponenziali di tutti i punteggi possibili.

Ma cosa succede se volessimo classificare una cosa che il modello non ha mai visto?

Classificazoine Zero-Shot
Analisi semantica senza dati di addestramento
Una delle innovazioni più sbalorditive dei modelli moderni è la capacità Zero-shot. Questa tecnica permette di classificare un testo in categorie arbitrarie definite dall'utente al momento dell'inferenza, senza che il modello sia mai stato addestrato specificatamente su quelle etichette.
Non siamo più limitati a positivo o negativo
Mentre il sentiment analysis classica distingue solo tra positivo e negativo, lo zero-shot classification può determinare se un testo esprime 'frustazione', 'entusiasmo' o 'ironia', semplicemente fornendo queste parole come etichette candidate durante l'esecuzione. Il modello ci risponde senza aver mai avuto bisogno di un addestramento specifico su quelle categorie.

Ma qual'è la logica che permette questo ragionamento?

Logica dello Zero-Shot
Ragionamento per implicazioni testuale.
Il segreto è la Natural Language Inference (NLI Basis). 
La flessibiiltà totale del modello consente di combiare la tassonomia di classificazione senza ri-addestrare o modificare il codice neurale.
L'utente fornisce una lista dinamica di categorie che il modello deve valutare. 
Il modello fa classificazione su categorie per cui non è mai stato addestrato specificatamente con esempi etichettati dal tuo programma.
Esempio
prendiamo una mail:
"vorrei ricevere un preventivo per 500 pezzi"
tu vuoi classificare in una di queste categorie:
ordine, richiesta offerta, reclamo.
Con un approccio classico supervisionato dovresti avere esempi etichettati:
"confermiamo ordine..." -> ordine
"chiediamo quotazione ..." -> richiesta offerta
"la merca è danneggiata..." -> reclamo
Poi addestare TF-IDF + Logistic Regression
oppure
TF-IDF + SVM
Nello Zero-Shot, invece, non devi necessariamente addestare un classificatore sulle tue tre categorie.
Prendi un modello già addestrato per NLI e trasformi le categorie in ipotesi linguistiche
Esempio
"vorrei ricevere un preventivo per 500 pezzi"
e poi prove:
ipotesi 1
"questo testo riguarda un ordine"
ipotesi 2
"questo testo riguarda una richiesta offerta"
opetesi 3
"questo testo riguarda un reclamo"
Il modello valuta ciascuna coppia
potrebbe ottenere
ordine: entailment=0.12
richiestsa offerta: entailment=0.91
reclamo: entailment=0.03
e quindi scegliere "richiesta offerta"
Quind il testo da classificare lo considero la premessa, le categorie le trasformo in ipotesi, poi applico il modello NLI che valuta, per ogni ipotesi, quanto la premessa supporta quella ipotesi. Alla fine scelgo la categoria con entailment maggiore.
La cosa interessante è che il modello non ha mai visto una classe chiamata "richiesta offerta" durante il suo training, ma conosce il linguaggio abbastanza bene da confrontare: "vorrei un preventivo" con "questo testo riguarda una richiesta offerta?" (confronta la premessa con l'ipotesi 1) e riconosce se sono semanticamente compatibili.

Riassumento, con Zero-Shot parti da un modello già preaddestrato, definisci le etichette, e lui fa la predizone, non devi fare model.fit(...) sul tuo dataset

Hugging Face lavora linguisticamente, tipicamente costruisce un hypothesis template. concettualmente costruisce qualcosa tipo: "questo esempio è {...}" e di volta in volta sostituisce il contenuto delle parentesi graffe con le categorie scelte.
quindi non sempre "ordine" è altrettanto efficace di "ordine cliente" oppure "una richiesta di inserimento ordine"
nel  nostro esempio laberl="ordine" quindi la frase che il modello valuta diventa: "questo esempio è {ordine}", piuttosto ""questo esempio è {ordine cliente}".
Puoi anche personalizzare questo templete di frase con: classifier(testo,candidate_labels=etichette,hypothesis_template="Questo testo riguarda {}.").
In questo modo ottieni "questo testo riguarda un ordine"
Spesso una buona formulazione di etichette migliora il risultato
Questo fa emergere un limite importante di Zero-Shot: il risultato può dipendere da come definisce le categorie.
Per esempio: "commerciale" è molto generico, mentre "richiesta commerciale di preventivo" è più descrittivo, quindi le laberl devono essere: chiare, distinte, semanticamente ben definite

Un altro vantaggio di Zero-Shot è che puoi aggiungere una categoria senza riaddestrare il modello.
Esempio, prima hai queste categorie: ordine, offerta, reclamo
Successivamente vuoi aggiungere: sollecito di pagamento
Non devi rifare training, puoi semplicemente aggiornare le etichette che passi al modello.

Zero-Shot non è sempre la scelta migliore, se hai migliaia di esempi reali ben etichettati un classificatore supervisionato addestrato sul tuo dominio può diventare più preciso.
Zero-Hot: ottimo quando hai pochi o zero esempi, molto flessibile, rapido da provare
Supervides Classification: richiede dataset etichettato, meno flessibile, può diventare molto specializzato.

La flessibilità di Zero-Shot apre scenari incredibili per le varie applicazioni

Applicazioni Strategiche
- Dynamic Sentiment: possiamo estendere la sentiment analysis a emozioni specifiche come 'rabbia' 'gioia' o 'paura' senza dover cercare dataset etichettati. 
- Filtro Argomenti: oltre al tono emotivo, lo zero-shot può identificare l'argomento (es. 'assistenza tecnica' vs 'fatturazione') nello stesso passaggio logico. Possiamo filtrare il sentiment dei clienti filtrando, automaticamente, se si lamentano dei costi e se eloggiano il personale, tutto con un unico passaggio logico.
- Hypothesis Template: è possibile personalizzare il modo in cui il modello formula l'ipotesi interna (es. 'il tono di questa recensione è {}") per migliorare la precisione.

Il modello valuta 3 stati, implicazoine, contraddizione e neutralità
quando facciamo zero-shot ci interessa solo implicazione, più il modello è convinto che la classificazione porta all'ipotesi, più il punteggio sarà alto.
Quindi lo zero-shot trasforma la classificazione in un compito di verifica delle ipotesi
Il punteggio finale per ogni classe è la probabilità normalizzata che la relazione logica tra premessa e ipotesi sia di implicazione positiva.
Non si base su statitiche di parole, ma su una vera e proprio comprensione delle relazioni logiche tra concetti.

E se tutto questo dovesse funzionare in un contesto globale, con 10 lingue diverse per esempio?

Sentiment Multilinge
Superare le barriere linguistiche con i Transformer
Nel mercato globale attuale, l'analisi del sentiment deve essere in grado di processare feedback in italiano, inglese, spagnolo o cinese senza perdere precisione. I modelli Transformer multilingue sono addestrati su corpora che comprendono oltre 100 lingue contemporaneamente.
Hugging Face permette di caricare modelli multilingue che sfruttano lo spazio latente comune tra le linghe. Questo significa che la conoscenza appresa su testi positivi in inglese, può essere trasferita con successo all'analisi su testi in italiano, fenomeno noto come 'Cross-Lingual Transfer'
Il concetto di gioia è lo stesso sia che venga espresso in italiano, inglese o altra lingua

Ma quali sono i giganti che ci permettono questo multilingua?

Modelli Multilingua
XLM-RoBERTa o mBERT sono i padroni di questa categoria di modelli.
La loro particolarità è che mappano lingue diverse in uno spazio matematico comune.
Se prendiamo la parola 'buono' in italianom, inglese ed altre lingue, i loro vettori numerici finiranno vicinissimi. Questo permette di costruire un unico sistema di analisi che server tutto il mondo e tutte le 100 lingue previste dal modello, risparmiando risorse e semplificando l'architettura di questi software.
- XLM-RoBERTa: uno dei modelli più potenti per l'inferenza multilingue, capace di gestire testi complessi in decine di idiomi
- Language Agnostic: i vettori di parole con significato simile in lingue diverse vengono mappati in posizioni vicine nello spazio multidimensionale.
- Convenienza: un unico modello in produzione può servire utenti da tutto il modno senza necessità di traduzione automatica preventiva.
- Il modello opera su un vocabolario condiviso basato su sub-word (SentencePiece) che include token comuni a più lingue.

Ma attenzione, non è sempre tutto così semplice quando cambiamo lingua.
Sfide del Multilingua
- Code-Switching: i modelli multilingue gestiscono efficacemente frasi dove l'utente mescola parole in lingue diverse es. "l'esperienza è stata davvero cool". Il modello deve essere abbastanza intelligente da passare da una lingua all'altra senza confondersi.
- Sfumature locali: sebbene potenti, questi modelli devono essere monitorati per catturare espressioni idiomatiche o sarcasmo specifici di una determinata cultura.
- Dimensione del modello: i modelli multilingue tendono ad essere più pesanti in termini di memoria rispetto ai modelli monolingua, richiedendo un bilanciamente delle risorso hardware.

Vediamo come questi modelli riescono a far parlare tra loro lingue diverse

Cross-Lingual Alignment
Allineamento dello spazio latente.
L'efficacia del sentiment multilingue deriva dall'allineamento dei vettori nello spazio latente. Se la parola 'buono' in italiano e 'good' in inglese hanno lo stesso ruolo semantico, il classificatore finale imparerà a riconoscerli come segnali positivi indipendentemente dalla lingua.
La distanza semantica tra concetti viene minimizzata durante l'addestramento multilingue per garantire la coerenza dell'analisi. Il significato diventa indipendente dall'idioma specifico.

```python
"""
====================================================================================================
CORSO DEEP LEARNING: GUIDA DEFINITIVA AI TRANSFORMERS (HUGGING FACE)
====================================================================================================
INTRODUZIONE:
Utilizziamo la libreria 'transformers' di Hugging Face per dimostrare come i modelli 
allo stato dell'arte (BERT, BART, RoBERTa) possano essere utilizzati senza scrivere 
manualmente migliaia di righe di codice neurale.

INTERAZIONI PRINCIPALI:
1. La classe 'pipeline' agisce come ORCHESTRATORE: coordina un Tokenizer (che trasforma il testo 
   in numeri) e un Model (la rete neurale vera e propria) in un unico flusso continuo.
2. PyTorch funge da 'MOTORE' di calcolo, gestendo i tensori e le operazioni matematiche (come la Softmax).
3. L'utente interagisce solo con l'interfaccia di alto livello, inserendo testo e ricevendo logica.
====================================================================================================
"""

import torch                      # Importiamo PyTorch: il "muscolo" matematico che muove i tensori
import numpy as np                # Importiamo NumPy: per la gestione dei dati numerici standard
from transformers import pipeline # Importiamo Pipeline: la "bacchetta magica" di Hugging Face

def analizza_sentiment_semplice():
    """
    Dalla Slide 3: Dimostrazione del 'Nastro Trasportatore' (Pipeline).
    Questa funzione mostra come automatizzare l'analisi del sentiment in 2 passaggi.
    """
    
    print("\n--- [Esempio 1: La Semplicità delle Pipeline] ---")

    # RIGA 1: Creiamo l'istanza della pipeline. 
    # 'sentiment-analysis' specifica il task, 'model' specifica quale supercar pre-addestrata usare.
    # Aggiungiamo framework="pt" per forzare l'uso di PyTorch ed evitare conflitti con Keras.
    analizzatore = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english", framework="pt")

    # RIGA 2: Definiamo il testo grezzo ricevuto dall'utente o da un database.
    testo = "This Hugging Face library is absolutely life-changing for developers!"
    
    # RIGA 3: Passiamo il testo alla pipeline. 
    # In questo istante il testo viene spezzettato (tokenizzato), passa nei neuroni e viene classificato.
    # Il risultato è una lista di dizionari, prendiamo il primo [0].
    risultato = analizzatore(testo)[0]

    # STAMPA: Mostriamo l'etichetta (POSITIVE/NEGATIVE) e la sicurezza del modello (SCORE).
    print(f"Testo: {testo}")
    print(f"Etichetta: {risultato['label']} | Confidenza: {risultato['score']:.4f}")



def classificazione_zero_shot():
    """
    Dalla Slide 7: Intuizione Artificiale (Zero-Shot).
    Qui il modello usa la logica pura per classificare testi su argomenti mai studiati.
    """
    
    print("\n--- [Esempio 3: Il Miracolo dello Zero-Shot] ---")

    # RIGA 1: Carichiamo un modello specializzato in NLI (Natural Language Inference).
    # Questa pipeline non cerca parole chiave, ma "ragiona" sulle implicazioni semantiche.
    # Forziamo framework="pt" (PyTorch) per stabilità e prestazioni.
    classificatore = pipeline("zero-shot-classification", model="facebook/bart-large-mnli", framework="pt")

    # RIGA 2: Un testo tecnico complesso.
    sequenza = "Il nuovo telescopio spaziale ha catturato immagini di una galassia a 13 miliardi di anni luce."
    
    # RIGA 3: Definiamo le categorie desiderate (possiamo cambiarle in tempo reale senza ri-addestrare!).
    categorie = ["scienza", "politica", "cucina", "sport"]
    
    # RIGA 4: Esecuzione del "duello logico" (Slide 10).
    # Il modello confronta la premessa (sequenza) con l'ipotesi "Questo testo parla di [categoria]".
    risultato = classificatore(sequenza, candidate_labels=categorie)

    # LOOP: Stampiamo ogni categoria con il relativo punteggio di attinenza logica.
    print(f"Testo: {sequenza}")
    for label, score in zip(risultato['labels'], risultato['scores']):
        print(f"  -> Categoria: {label:10} | Probabilità di attinenza: {score:.4f}")


def modelli_multilingue_allineati():
    """
    Dalla Slide 11-14: Abbattimento delle Frontiere Linguistiche.
    Il modello capisce il significato profondo, indipendentemente dalla lingua usata.
    """
    
    print("\n--- [Esempio 4: Sentiment Multilingue e Code-Switching] ---")

    # RIGA 1: Carichiamo BERT multilingue.
    # Questo gigante ha studiato 100+ lingue simultaneamente, allineando i concetti in uno spazio comune.
    modello_multi = pipeline("sentiment-analysis", model="nlptown/bert-base-multilingual-uncased-sentiment", framework="pt")

    # LISTA: Testiamo diverse sfumature, incluso il 'Code-Switching' (misto lingua).
    esempi = [
        "Il pranzo era delizioso, tornerò sicuramente!", # Italiano puro
        "The software is buggy and constantly crashes.",  # Inglese puro
        "L'esperienza è stata davvero cool!"              # Mistura Italiano/Inglese (Slide 13)
    ]

    # ESECUZIONE: Cicliamo sugli esempi.
    for es in esempi:
        # La pipeline riconosce la lingua e applica i pesi corretti nello spazio latente.
        res = modello_multi(es)[0]
        # Il task restituisce un rating da 1 a 5 stelle (label).
        print(f"Testo: {es:45} | Valutazione (Rating): {res['label']}")


# PUNTO DI INGRESSO (Main): Dove tutto prende vita.
if __name__ == "__main__":
    analizza_sentiment_semplice()       # Dimostra l'astrazione
    classificazione_zero_shot()         # Dimostra l'intuizione logica
    modelli_multilingue_allineati()      # Dimostra la globalità dei modelli

# ====================================================================================================
# CONCLUSIONI E INTERAZIONI TRA COMPONENTI
# ====================================================================================================
# 1. IL FLUSSO DATI: Testo utente -> Pipeline -> Tokenizer (Numeri) -> Transformer (Tensori) 
#    -> Logit (Punteggi) -> Softmax (Probabilità) -> Risultato Finale.
#
# 2. PERCHÉ HUGGING FACE? Perché ci permette di usare modelli con MILIARDI di parametri 
#    come se fossero semplici funzioni Python, nascondendo la complessità dei gradienti e 
#    delle matrici (Slide 1).
#
# 3. IL RUOLO DI PYTORCH: In questo script, PyTorch è l'infrastruttura silenziosa che 
#    permette ai Transformers di "pensare". Gestisce la memoria della GPU/CPU e calcola 
#    velocemente le funzioni di attivazione come la Softmax.
# ====================================================================================================
```
