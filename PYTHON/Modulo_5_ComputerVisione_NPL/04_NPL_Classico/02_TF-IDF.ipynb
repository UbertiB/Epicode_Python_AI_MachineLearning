TF-IDF - LOGICA E APPLICAZIONE

TF= Term Frequency (per misurare la rilevana locale di un termine)
IDF= Inverse Document Frequency (per pesare l'importanza globale e la rarità)

Sappiamo come trasformare testo in numero contando le parole, ma questo processo porta con se tutta una serie di problemi, in primis, non tutte le parole hanno lo stsso valore.

Vediamo come far pesare di piu' le parole rare, quelle importanti, e come mettere a tacere il rumore delle parole troppo comuni.

Come capiamo di cosa parla un testo?

Term Frequency: L'Importanza Locale
Misurare quanto una parola descrive un singolo documento.
Abbiamo visto come il modello Bag of Words conti semplicemente le occorrenze delle parole. La Term Frequency (TF) è l'evoluzione naturale del Bag of Words.
Nel Bag of Words contavamo e basta
Qui iniziamo a dare una proporzione.
L'idea è semplice: piu' una parola appare in un documento, più è probabile che quel documento tratti quell'argomento. 
Tuttavia, il conteggio assoluto può essere fuorviante se i documenti hanno lunghezze diverse, rendnedo necessaria una normalizzazione.

Logica della Frequenza Locale
Normalizzare il segnale linguistico
La frequenza grezza è il numero pure di volte che un termine appare, utile come base di partenza.
Dividere il conteggio per documento per il numero totale di parole del documento (normalizzare per lunghezza) serve a evitare bias verso testi lunghi. Il valore ottenuto non dipende più dalla lunghezza del singolo testo, in questo modo il segnale linguistico diventa puliti e soprattutto confrontabile tra documenti di dimensione diversa.
La TF indica quanto un termine è 'pesante' per quel contesto specifico, isolandolo dal resto del corpus.
La formula standard del Term Frequency (TF) rapporta le occorrenze del termine 't' alla somma di tutte le occorrenze nel documento 'd'

Esistono però diversi modo per calcolare questa intensità locale.

Variante della TF
- TF Binaria: in alcuni casi, ci interessa solo sapere se una parola è presente (1) o assente (0), ignorando il numero di ripetizioni.
- Log-scaling: dato che una parola che appare 100 volte non è necessariamente 100 volte più importante di una che appare solo una volta, si usa spesso una scala logaritmica.
- TF Aumentata: utilizzata per prevenire il bias in documenti molto lunghi, rapportando la frequenza al termine più frequente nel documento stesso.

Ma perche logaritmo è così amato da Data Scienticts?

Approfondimento: Scaling Logaritmico
Smorzare la frequenza
Perchè usiamo il logaritmo? perchè l'importanza di un termine non cresce linearmente con la sue frequenza. Se il termine 'neurale' appare 20 volte, il documento è certamente rilevante, ma la 21esima apparizione aggiunge meno informazione della prima.
La trasformazione logaritmica permette di gestire questa saturazione dell'informazione locale.

Ma ora che abbiamo il peso locale, dobbiamo alzere l'orizzonte verso l'intero dataset

Inverse Documento Frequency (IDF): L'Importanza Globale
Identificare i termini che distinguono i documenti.
Se la TF guarda all'interno del singolo documento, la Inverse Documento Frequency (IDF) guarda all'esterno, verso l'intero dataset o corpus. Il suo scopo è penalizzare le parole troppo comuni e premiare quelle rare e distintive.
Il suo scopo è identificare termini che distinguono un documento da un altro.
Parole come 'il', 'essere', o 'che' hanno una TF altissima quasi ovunque, ma non aiutano a distinguere un testo i medicina da uno di sport. L'IDF agisce come un filtro che 'zittisce' il rumore comune ed amplifica quelle uniche e preziose.

Come si traduce questa rarità in un peso matemartico?

Logica della Rarità Globale.
Pesare la specificità dle termine.
- Documento Frequency (DF) il numero di documneti nel corpus che contengono almeno una volta il termine specifico.
- Inversione del peso: più un termine è diffuso nel corpus (DF alta), più il suo peso IDF sarà basso.
- Specificità Informativa: L'IDF assegna un valore alto a termini tecnici o spefifici che compaiono in pochi documenti
- L'IDF è calcolato come il logaritmo del rapport tra il numero totale di documenti 'N' e il numero di documenti che contengono il termine 't'

Più una parola è diffusa, meno è specifica e se appare in tutti i documenti (o quasi) il suo valore discriminante è zero, l'IDF premia così la rarità.

L'IDF è di fatto un creatore automatico di stop words, non serve più che io elenchi a mano tutte le preposizionio gli articoli, l'algoritmo si accorge che sono ovunque e le spegna automaticamente. Termini che appaiono in ogni documento avranno un IDF pari a log(1) ovvero 0 

Cosa succede se carchiamo una parola che non esiste nel nostro archivio?

Approfondimento: Smoothing dell'IDF (aggiungere un piccolo 1 simbolico)
Evitare divisioni per zero.
Nelle implementazioni pratiche, si aggiunge spesso '1' al denominatore. 
Questo serve a gestire termini che potrebbero non apparire mai nel corpus di addestramento ma he compaiono durante l'inferenza.
Si aggiunge anche '1' al risultato finale per assicurarsi che i termini con DF uguale a N non vengano completamente annullati.

Finalmente siamo pronti per l'unione dei due mondi

Integrazione: Perchè TF-IDF Supera il BoW
Sinergia tra rilevanza locale e rarità globale.
Il punteggio TF-IDF finale è il prodotto dei due componenti. Questa combinaione bilancia perfettamente l'esigenza di sapere quanto una parola è importante in un testo e quanto è unica nel contesto generale del dataset.
A differnza del semplice Bag of Word (Bow) che tratta ogni occorrenza allo stesso modo, il TF-IDF trasforma la matrice dei conteggi un una matrice di pesi statistici, rendendo i vettori molto più rappresentativi della semantica profonda.

Applicazioni pratiche
Il TF-IDF è il motore storico nei sistemi di ricerca, se cercate in un motore di ricerca i risultati vengono mostrati in base alla pertinenza della query rispetto ai documenti indicizzati.
Selezionando i termini con puntegio TF-IDF più alto in un documento, possiamo riassumere il contenuto in poche parole chiave (keyword-Extraction)
Inoltre fornire vettori pesati con TF-IDF a modelli come SVM o Logistic Repression migliora drasticamebte l'accuratezza rispetto al semplic BoW

Controntare documenti
Nello spazio TF-IDF due docuenti sono simili se condividono parole con pesi elevati. Poichè hanno lunghezze diverse, usiamo la similarità coseno per misurare l'angolo tra di essi 
Questa misura è robusta rispetto alla lunghezza del testo, concentrandosi sulla direzione semantia dei vettori pesati.


```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
import joblib # Utilizzato per la persistenza del modello (salvataggio su disco)

"""
PIPELINE DI ELABORAZIONE TESTUALE CON TF-IDF
-------------------------------------------
Questo script implementa una pipeline completa per trasformare testo non strutturato
in rappresentazioni numeriche pesate tramite l'algoritmo TF-IDF (Term Frequency-Inverse Document Frequency).

Interazioni principali:
1. TfidfVectorizer: Classe core che analizza il corpus, crea il dizionario e calcola i pesi.
2. DataFrame (Pandas): Utilizzato per strutturare l'output numerico (matrice sparsa -> densa) 
   rendendolo leggibile e analizzabile come una tabella.
3. Joblib: Gestisce il ciclo di vita del modello, permettendo di riutilizzarlo in produzione.
"""

# 1. PREPARAZIONE DEL CORPUS (Dataset di Input)
# Un "corpus" è semplicemente una lista di stringhe (documenti).
# L'algoritmo ha bisogno di vedere l'intero insieme per determinare quanto una parola sia "rara".
corpus = [
    "Il Deep Learning rivoluziona l'intelligenza artificiale.",
    "L'intelligenza artificiale e le reti neurali sono il futuro.",
    "Le reti neurali sono ispirate al cervello umano.",
    "Il Deep Learning richiede grandi quantità di dati."
]

# 2. CONFIGURAZIONE DEL VETTORIZZATORE TF-IDF
# Inizializziamo l'oggetto TfidfVectorizer che incapsula tutta la logica di pre-processing.
vectorizer = TfidfVectorizer(
    lowercase=True,      # Converte tutto in minuscolo per evitare differenze (es: "Il" vs "il")
    stop_words=None,     # In questo esempio includiamo tutto per studiare l'effetto dell'IDF sui connettivi
    use_idf=True,        # Abilita il calcolo del peso inverso della frequenza nei documenti
    smooth_idf=True,     # Aggiunge 1 ai conteggi (laplace smoothing) per prevenire divisioni per zero
    norm='l2'            # Normalizzazione Euclidea: rende i vettori di lunghezza 1, 
                         # fondamentale affinché documenti lunghi non abbiano pesi sproporzionati.
)

# 3. FIT E TRASFORMAZIONE (Core Logic)
# Questa riga esegue due operazioni distinte ma correlate:
# - .fit(): Scansiona il 'corpus', estrae le parole univoche (Vocabulary) e calcola l'IDF globale.
# - .transform(): Prende ogni singolo documento e lo converte in un vettore di numeri basato sul vocabolario.
# Risultato: Una 'sparse matrix' dove ogni riga è un documento e ogni colonna è una parola.
tfidf_matrix = vectorizer.fit_transform(corpus)

# 4. ANALISI E VISUALIZZAZIONE DEI RISULTATI
# Estraiamo i nomi delle colonne (le parole del dizionario ordinato alfabeticamente)
feature_names = vectorizer.get_feature_names_out()

# Poiché la matrice TF-IDF di scikit-learn è ottimizzata per la memoria (CSR format),
# la convertiamo in 'dense' (matrice standard) per caricarla in un DataFrame Pandas.
dense_matrix = tfidf_matrix.todense()
df_tfidf = pd.DataFrame(dense_matrix, columns=feature_names)

print("--- Matrice TF-IDF (Rappresentazione Tabellare) ---")
# Visualizziamo i pesi: valori alti indicano parole "caratterizzanti" per quel documento.
print(df_tfidf.head())

# 5. ISPEZIONE DELL'IDF (Importanza Semantica Globale)
# L'IDF misura la "rarità" di una parola: IDF = log(N/n) dove N è il tot documenti.
# Creiamo un dizionario mappando parola -> suo peso IDF calcolato durante il .fit()
idf_values = dict(zip(feature_names, vectorizer.idf_))

print("\n--- Analisi della Rarità (Top 5 Termini con IDF più alto) ---")
# Più alto è il valore IDF, più la parola è unica e informativa nel contesto del corpus.
for word, idf in sorted(idf_values.items(), key=lambda x: x[1], reverse=True)[:5]:
    print(f"Termine Tecnico/Raro: '{word}' -> Peso IDF: {idf:.4f}")

# 6. PERSISTENZA DEL MODELLO (Best Practice per Produzione)
# Non salviamo i dati trasformati, ma il 'vectorizer' stesso.
# Perché? Se domani arriva un nuovo documento, dobbiamo trasformarlo usando gli STESSI 
# parametri e lo STESSO vocabolario imparato oggi per mantenere la coerenza dimensionale.
model_filename = "tfidf_processor_v2026.pkl"
joblib.dump(vectorizer, model_filename)

print(f"\n[INFO] Pipeline salvata in: {model_filename}")
print("[INFO] Il file può ora essere caricato in una web-app per processare nuovi input utente.")
```
