CLASSIFICAZIONE CON TF-IDF+ LOGISTIC REGRESSION

La regressione logistica nell'NLP è un classico intramontabile anche se sono nati nuovi algoritmi.
La regressione logistica è uno standard industriale per la classificazione bianaria dei testi.

non sempre serve una rete neurale profonda con migliardi di parametri, a volte basta pesare correttamente le parole chiave.

TF-IDF
-> Trasforma il testo in numeri

Logistic Regressione
-> usa quei numeri per classificar eil testo.

E' molto simile a Bow + SVM solo che cambiano sia la rappresentazione sia il classificatore.

La pipeline è:
testo -> TF-IDF -> Vettore numerico -> Logistic Regression -> Classe

Immagina di voler classificare delle mail in: ordine, reclamo, offerta.
Abbiamo per esempio:
testi = [
    "vorrei ordinare 500 pezzi",
    "confermiamo ordine per 300 pezzi",
    "la merce ricevuta è danneggiata",
    "il prodotto presenta un difetto",
    "chiediamo quotazione per 100 pezzi",
    "potete inviarci la vostra offerta"
]

categorie = [
    "ordine",
    "ordine",
    "reclamo",
    "reclamo",
    "offerta",
    "offerta"
]
TF-IDF prende queste frasi, costruisce il vocabolario e trasforma ogni documeto in un vettore.
Supponiamo che alcune feature siano:
              ordine  pezzi  danneggiata  difetto  quotazione  offerta

email 1        0.72    0.35      0          0        0          0
email 2        0.68    0.31      0          0        0          0
email 3        0       0         0.81       0        0          0
email 4        0       0         0          0.76     0          0
email 5        0       0.22      0          0        0.79       0
email 6        0       0         0          0        0          0.83
Questi numeri sono l'input della Logistic Regression
TF-IDF  dice al modello quali caratteristiche ha il documento; Logistic Regressoin impara queli caratteristiche sono associate alle varie classi.
Per esempio, durante il training potrebbe imparare che:
ordine       → forte indizio per ORDINE
confermiamo  → indizio per ORDINE

danneggiata  → forte indizio per RECLAMO
difetto      → forte indizio per RECLAMO

quotazione   → forte indizio per OFFERTA
offerta      → forte indizio per OFFERTA

Naturalmente non sono regole che scrivi, sono appresi dal modello.
Ed è proprio questo che fa Logistic Regression
Anche se si chiama Regressone, la Logistic Regression viene usata per la classificazione
Con una classificazion ebinaria (esempio: spam/non spam) la logistic Regression produce un valore che può essre interpretato come probabilità della classe positiva. Poi viene applicata una soglia, qualcosa come:
>=0.5 -> spam
<0.5 -> non spam
La funzione logistica, detta anche sigmoide, serve proprio a portare il risultatato in un intervallo tra 0 e 1

Riassumetno:
TF-IDF produce le FEATURE
Logistic Regressione assegna dei PESI alle feature
Combina i pesi
Calcola la probabilità/classe

I coefficienti della Logistic Regression sono appresi dai dati.

Non impara a memoria i dati ma impare a generalizzare.

Perchè questo modello è così amato?
Perchè scegliere la regressione logistica quando abbiamo, per esempio, i transformer?

Un Classico della Classificazone
Nonostrante la semplicità architettonica, la Regressione Logistica rimano uno degli algoritmi più potenti per compiti come la sentiment analysis o la spam detectoin. La sua forza risiede nella capacità di mappare uno spazio vettoriale ad alta dimensionalità, come  quello generato dal TF-IDF, in una probabilità discreta.
A differenza dei modelli generativi, questo approccio discriminativo si concentra direttamente sul confine di decsione. Quando accoppiato con la pesatura statistica del TF-IDF, il modello è in grado di dare meno importanza alle parole frequenti e focalizzarsi sui termini che spostano effettivamente l'ago della bilancia verso una classe o l'altra.
In molto scenari azienda la velocità e la semplicità battono la complessita.
La regressione legistica è un modello discriminativo, non cerca di capire il linguaggio naturale,.
Quando poi usiamo il TF-IDF abbiamo migliaria di coordinate per trovare la retta, o meglio l'iperpiano che taglia fuori il rumore e tiene solo il segnale.

Dalla Regressione alla Probabilità
il segresto sta tutto nella sigmoide.
Invece di avere un interrutore di acceso o spento, la sigmoide prende un numero qualunque e lo schiaccia tra 0 e 1, non vogliamo che il computer ci dica solo spam e non spam, vogliamo che ci dica al 98% è spam.
E' questa confidenza che ci permette di gestire meglio gli errori.
La vera magia però avviene quando questa logica incontra TF-IDF
una combinazione perfettta
TF-IDF pulisce il campo per noi, toglie importanza alle parole inutili (come il o di) e accende i riflettori su parole più caratteristiche, riduce il peso dei termini comuni, che solitamente creano rumore nei modelli lineari, pulendo il segnale per la Regressione Logistica.
Un documento lungo non avrà un vantaggio sleale dato dalla verbosità dei testi, grazie alla normalizzazione.
In ambito industriale questa accopiata (TF-IDF + Logistic Regression) è la prima tecnica da testare prima di passare a reti neurali profonde, poichè offre prestazioni competitive con costi minori.

Come il computer impara a non sbagliare durante l'addestramento.
La Funzione Costo
Log-Loss e Ottimizzazione
Per addestrare il modello, utilizziamo la "Cross Entropy" o "Log Loss". Questa funzione penalizza pesantemente le predizioni sicura ma errate, spingendo i pesi del modello verso una configurazione che massimizzi la verosomiglianza dei dati osservati. Questo spinge i pesi del modello verso la verità
Il calolo dell'errore per 'm' campioni guida l'aggiornameto dei parametri tramite la discesa del gradientie.

Interpretabilità
Analisi delle Feature Predittive
Identificare le parole che guidano il modello
Uno dei vantaggi principali della Regressione Logistica è la sua 'interpretabilità'. 
Ogni parola del vocabolario ha i suii pesi che possono essere analizzati.
Analizzando i pesi assegnati a ciascuna parola nel vocabolario, possiamo letteralmente vedere quali termini il modello considera più importanti per distinguere le classi.
Parole con coefficienti positivi elevati sono forti predittori per la classe target (es. 'eccellente' per recensioni positive), mentre coefficienti negativi spingono il documento verso la classe opposta. Questo ci permette di validare se il modello sta imparando concetti sensati o semplicemente rumore.

Ma come leggiamo tecnicamente questi pesi per capire l'impatto reale.
Coefficienti e Significato
Estrattore di conoscenza dai pesi.
Più il valore assoluto di un coefficiente è alto,  maggiore è l'impatto di quella parola sulla decisione finale (sia positiva sia negativa).
La direzioe del peso, cioè il segno del coefficiente, determina a quale classe la parola conferisce con maggiore proabibiltà (se verso il positivo o il negativo)
L'analisi top-k vuol dire ordinare i termini per valore dei pesi permettendo di creare classifiche delle parole più caratterizzanti.
L'odds ratio esprime come cambia la probabilità di appartenenza a una classe all'aumentare unitario del valore di una feature. Se aumentiamo di poco il valore di TF-IDF la proabibilità che la parola appartenga alla classe target cresce in modo esponenziale.
E' come se il modello ci dicesse, ecco le parole che classificano questo concetto.

Analizzare i pesi è il nostro test delle verità
Visualizzazione e Validazione
A differenza dei Transformer, possiamo spiegare esattamente perchè un'email è stata marcata come spam, indicando le parole specifiche che hanno alzato il punteggio.
Una regolarizzazione forte aiuta a ignorare le parole rare che potrebbero apparire solo casualmente in una classe durante il training
L'analisi dei pesi ci aiuta a identificare nuove 'stopwords' specifiche per il dominio che non portano informazione utile.
Questo ci permette di capire cosa disturba il modello, fare data cleaning e riaddestare, è un sistema molto più equo, meno influenzati da solo statistica. 
Geometricamente la regressione logistica traccia una linea dritta nello spazio dell iperpiano. Questa struttura lineare è ciò che permette di rendere il calcolo così veloce, è la bellezza della matematica semplica applicata a problemi complessi.

Ma cosa succede se il vocabolario è troppo esteso?
Tunning della Regolarizzazione
Evitare che il modello impari a memoria.
Con migliaia di parole nel vocabolario, è facile che il modello diventi troppo complesso e inizi a memorizzare refusi o coincidenze statistiche del training set. Questo fenomeno è chiamato 'overfitting' e porta a scarse prestazioni sui nuovi dati. Nel NLP è facile incapparci perchè abbiamo migliaia di feature. 
La regolarizzazione è la tecnica che utilizziamo per 'punire' i coefficienti troppo grandi, costringendo il modello a mantenere una struttura più semplice e generale. Il parametro 'C' in scikit-learn controlla questo delicato equilibrio tra precisoine e capacità di generalizzazione.
Abbiamo due tecniche regine per aiutare il modello a mantenere una struttura semplice, L1 e L2 e il parametro C
- Regolarizzazione L2 (Ridge): tende a distribuire i pesi tra tutte le parole, dicuendo i valori estremi senza annullarli del tutto. Impedisce ai pesi di diventare troppo grandi, distirbuenzo importanza su tutte le parole.
- Regolarizzazione L1 (Lasso): tende ad azzerare i pesi delle parole meno importanti, eseguendo una 'feature selection' automatica. E' come un esperto di minimalismo, utilissimo se volete un modello leggero.
- Parametro C: in scikit-learn, è l'inverso della forza di regolarizzazione. Un valore di C basso significa regolarizzazione forte, Controlla quanto forte vogliamo che sia questo effetto.
- Il termine di regolarizzazione L2 aggiunge una penalità proporzionale al quadrato della norma del vettore dei pesi alla funzione di costo.

Ma quando conviene usare L2 o L1
L2 se volete la massima accuratezza e non preoccupa la dimensione del modell
L1 se avete bisogno di interpretabiltà estrama o se devi far girare il modello su dispositivi con poca memoria.
co L1 le parole inutile scompaiono dal modello lasciando solo le parole con vero significato.



```python
"""
Classificazione con TF-IDF + Logistic Regression
Questo script illustra l'implementazione della Regressione Logistica utilizzando 
scikit-learn, focalizzandosi sull'interpretabilità dei coefficienti e sulla regolarizzazione.
"""

import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

def get_educational_dataset():
    """
    Crea un dataset di esempio per illustrare il sentiment analysis.
    
    Returns:
        tuple: (testi, etichette)
    """
    texts = [
        "Il film è stato assolutamente fantastico ed eccellente",
        "Pessima esperienza, tutto molto brutto e noioso",
        "Lo raccomando a tutti, una visione incredibile",
        "Non mi è piaciuto per niente, uno spreco di tempo",
        "La trama era geniale e gli attori bravissimi",
        "Terribile, non guardatelo, storia pessima",
        "Un capolavoro del cinema moderno",
        "Disastro totale, sceneggiatura imbarazzante"
    ]
    # 1 = Positivo, 0 = Negativo
    labels = [1, 0, 1, 0, 1, 0, 1, 0]
    return texts, labels

def analyze_model_weights(vectorizer, model):
    """
    Estrae e visualizza i coefficienti del modello per spiegare l'interpretabilità.
    
    Args:
        vectorizer: Il TfidfVectorizer addestrato.
        model: Il modello di LogisticRegression addestrato.
    """
    # Otteniamo i nomi delle feature (le parole nel vocabolario)
    feature_names = vectorizer.get_feature_names_out()
    # Otteniamo i coefficienti (i pesi assegnati a ogni parola)
    coefficients = model.coef_[0]
    
    # Creiamo una lista di coppie (parola, peso)
    feature_importance = list(zip(feature_names, coefficients))
    
    # Ordiniamo per peso crescente
    feature_importance.sort(key=lambda x: x[1], reverse=True)
    
    print("\n--- ANALISI DEI COEFFICIENTI (Interpretabilità) ---")
    print("Top 3 parole che spingono verso il POSITIVO (Coefficienti +):")
    for word, weight in feature_importance[:3]:
        print(f" -> {word}: {weight:.4f}")
        
    print("\nTop 3 parole che spingono verso il NEGATIVO (Coefficienti -):")
    for word, weight in feature_importance[-3:]:
        # Le parole negative avranno i pesi più bassi (più negativi)
        print(f" -> {word}: {weight:.4f}")

def main():
    # 1. Caricamento Dati
    texts, labels = get_educational_dataset()

    # 2. Vettorizzazione TF-IDF (Slide 5)
    # Trasforma il testo in una matrice numerica pesata
    # min_df=1 permette di includere anche parole che appaiono una sola volta (ideale per piccoli dataset)
    vectorizer = TfidfVectorizer(min_df=1)
    X = vectorizer.fit_transform(texts)
    y = np.array(labels)

    # 3. Creazione del Modello con Regolarizzazione (Slide 11-12)
    # 'C' è l'inverso della forza di regolarizzazione. 
    # Un valore basso (es. 0.1) = Regolarizzazione forte (modello semplice)
    # Un valore alto (es. 10.0) = Regolarizzazione debole (modello complesso)
    model = LogisticRegression(
        penalty='l2',    # Tipo di penalità (Ridge)
        C=1.0,           # Bilanciamento Bias-Varianza
        solver='liblinear'
    )

    # 4. Addestramento (Slide 6)
    model.fit(X, y)
    print("Modello addestrato con successo.")

    # 5. Analisi delle Feature (Slide 7-10)
    analyze_model_weights(vectorizer, model)

    # 6. Esempio di Predizione (Inferenza)
    test_reviews = [
        "Un film incredibile e fantastico",
        "Esperienza pessima e noiosa"
    ]
    
    # Trasformiamo i nuovi testi usando il vettorizzatore già addestrato
    X_test = vectorizer.transform(test_reviews)
    
    # Prediciamo le probabilità (Slide 4 - La funzione Sigmoide)
    probabilities = model.predict_proba(X_test)
    predictions = model.predict(X_test)

    print("\n--- TEST DI INFERENZA ---")
    for i, review in enumerate(test_reviews):
        class_label = "Positivo" if predictions[i] == 1 else "Negativo"
        prob_pos = probabilities[i][1]
        print(f"Testo: '{review}'")
        print(f"Probabilità Positivo: {prob_pos:.2%} -> Risultato: {class_label}\n")

if __name__ == "__main__":
    main()
```
