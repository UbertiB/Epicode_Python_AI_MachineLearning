WORD2VEC - LOGICA E GENSIM

Word2Vec è uno dei metodi classici per costruire i word emnedding, come TF-IDF

Immagina di voler spiegare ad un bambino il significato di una parola senza usare il dizionario ma semplicemente mostrandogli dove quella parola 'abita' abitualmente

Word2Vec non è un singolo algoritmo, ma una famiglia di archichetture proposte da Mikolov nel 2013. L'idea rivoluzionaria non è stata aggiungere complessità, bensi togliere. Invece di usare reti neurali profonde e lentissime Mikolov ha proposto una rete superficiale con un unico layer nascosto, per imparare la rappresentazione delle parole basandosi sul loro contesto locale.
Il concetto è: conoscerai una parola dalla compagnia che tiene (contesto locale).
Le due varianti principali differiscono nel modo in cui pongono il problema di predizione. Mentra una cerca di indovinare la parola centrale partendo dal contesto, l'altra rovescia la prospettiva cercando di prevedere le parole circostanti partendo da un unico termine fulcro.

- CBOW (Continuous Bag of Words): utilizza le parole di contesto all'interno di una finestra fissa per prevedere la parola target centrale. La rete vede le parole circostanti e la rete deve indovinare quella mancante al centro.
- Skip-Gram: riceve in input una parola target e tenta di prevedere le parole che la circondano nel testo originale. Riceve una sola parola e deve lanciare una previsione su tutto ciò che la circonda.

La finestra di contesto è il parametro che definisce quanti termini a destra e a sinistra vengono considerati durante la scansione del corpus
L'output dello strato finale viene normalizzato tramite una funzione di attivazione che produce una distribuzione di probabilità sull'intero vocabolario.


Il salto concetuale rispetto a TF-IDF:
TF-IDF
parola/documento -> vettore basato su frequenze
Word2Vec
parola -> vettore APPRESO dal contesto

Word2Vec non decide il significato di una parola leggendo un dizionaraio. Impara osservando quali parole tendono a comparire vicino tra loro in un grande corpus. E' una famiglia di architetture per apprendere word embedding, in particolare CBOW e Skipgram

Immagina di avere le seguenti frasi:
il cliente ordina il prodotto
il cliente acquista il prodotto
il cliente paga il fattura
il compratore ordina il prodotto
il compratore acquista la merce
il compratore paga la fattura

Word2Vec nota che cliente e prodotto compaiono spesso in contesi simili: ordina, acquista, paga, prodotto, fattura.
Durante il training modifica progressivamente i numeri dei loro vettori affinchè parole che hanno contesti simili finiscano in una zona vicino dello spazio vettoriale.

Alla fine potresti avere:
cliente -> [0.23,-0.51,0.72,0.06]
compratore -> [0.25,-0.48,0.70,0.19]
montagna -> [-0.81,0.34,-0.15,0.67]
cliente e compratore risultano matematicamente più vicini rispetto a cliente e montagna

Quindi Word2Vec sta imparando
parola -> contesti in cui compare -> vettore denso
Non gli dici mai esplicitamente che il cliente è simile al compratore, è il modello che ricava questa relazione dai dati.

Ci sono due logiche fondamentali: CBOW e Ski-gram

CBOW

Partiamo dalla frase "Il  ____ ordina il prodotto"
gli dai il contesto: il, ordina, il, prodotto e il modello cerca di PREVEDERE la parola centrale: "cliente"
CWOW -> contesto -> parola
CBOW è la previsione della parola centrale a partire dalle parole circostanti

SKIP-GRAM
Skip-gram fa l'opposto, gli dai la parola "cliente" e durante il training cerca di prevedere quali parole possono comparire nelle sue vicinanze: ordine, prodotto, paga, fattura
Quindi 

skip-gram parola -> contesto
tecnicamdnte vengono costruite coppie del tipo:
(cliente, ordina)
(cliente, prodotto)
(cliente, il )
e il metodo impara a distinguere contesti plausibili da compinazioni non pertinenti. 

Quale scegliere tra CBOW e SkipGram?
dipende dal contesto

Il CBOW è generalmente più veloce da addestrare e mostra una precisione leggermente superiore per le parole che appaiono frequentemente nel dataset
Lo Skip-gram richiede più tempo ma è noto per funzionare meglio con dataest piccoli e per rappresentare meglio le parole rare.
La rete è così veloce perche il layer nascosto è lineare e nessuna funzione di attivazione non lineare viene usata nel layer nascosto (nè per CBOW nè per Skip-Gram), questo permette alla rete di essere estremamente efficiente su miliardi di parole. Non ci sono attivazioni complicate è pura proiezione matematica.

LaFunzione Obbiettivo
Matematicamente è ottimizzare la verosomiglianza
L'obiettivo dell'addestramento è trovare pesi che massimizzano la probabilità di osservare le coppie di parole realmente presenti nel testo. Nel caso dello Skip-Gram, cerchiamo di massimizzare la log-probabilità media.
Sia T la sequenza di parole, l'algoritmo minimizza la perdita negativa logaritmica calcolata sulla finestra di ampiezza c.

da tutto questo emerge qualcosa di sbalorditivo: la semantica

Relazioni Semantiche
L'algebra dello spazio vettoriale.
La caratteristica più affascinante di Word2Vec è la regolarità linguistica che emerge spontaneamente. Poichè le parole vengono mappate in base al contesto, termini che condividono relazioni logiche simili finiscono per avere una disposizione geometrica coerente.
Questo significa che possiamo letteralmente sommare e sottrarre concetti usando i vettori delle parole. Se lo spazio è ben addestrato, la distanza tra i concetti rimane costante attraverso diverse coppie di termini

Questa regolarità si trasforma in vera e propria algebra.

Linearità e Analogice
Ragionare con i vettori
L'esempio più famoso di tutti: re - uomo + donna = regina
- Relazioni di genere: lo spostamente vettoriale da uomo a donna è simile a quello che va da re a regina o da attore ad attrice.
- Relazioni Geopolitiche: la differenza tra Italia e Roma è geometricamente vicina alla differenza tra Francia e Parigi.
- Tempi Verbali: lo spazio cattura anche relazioni morfologiche come il passaggio dal presente al passato.
- Un'analogia del tipo 'A sta a B come C sta a D' può essere risolta cercando il vettore che massimizza la similarità 

** GENSIM **

Gensim è una libreria Python che offre, tra le altre cose, un'implementazione pronta di Word2Vec e può essere addestrato direttamente passando una sequenza di frasi già tokenizzate.
Gensim accetta frasi già tokenizzate
Progettato per la velocità e per gestire dataset così grandi da non entrare nella RAM, offre interfacce semplici per addestrare modelli Word2Vec da zero.
Inoltre permette di scaricare e caricare modelli famosi già addestrati su dataset gigantesci (come Google News o Wikipeida) consentendoci di usare vettori di alta qualità senza dover spendere settimane di calcoli.

from gensim.models import Word2Vec
frasi = [
    ["cliente", "ordina", "prodotto"],
    ["cliente", "paga", "fattura"],
    ["compratore", "ordina", "prodotto"],
    ["fornitore", "consegna", "merce"]
]
model = Word2Vec(
    sentences=frasi,
    vector_size=100,
    window=2,
    min_count=1,
    sg=1
)
accetta un iterabile di frasi, dove ogni frase è a sua volta un iterabile di token
Parametri:

windows=2: stabilisce quanto lontano guardare intorno alla parola corrente. cioè la distanza minima gra la parola corrente e quella da predire. Per esempio nella frase "il cliente ordina subito il prodotto" se la parole è "oridna" con una finestra locale considererei vicine: cliente, subito, il 
E' il concetto visto per gli n-gram, ma qui il contesto è utilizzato per imparare gli embedding
min_count=1: significa di considerare anche parole che compaiono una sola volta, se mettessi min_count=5 Gensim ignorerebbe le paroleo che compaiono meno di 5 volte (simile al concetto di min_df visto per TF-IDF)

Ma dove troviamo questi modelli pronti all'uso?

Modelli Pre-addestrati
Il modulo downloader di Gensim fornisce un catalogo di modelli pronti all'uso, identificati da stringhe come "word2vec-google-news-300'
E' importante selezionare il modello corretto, un word2vec addestrato su twitter capità il gergo giovanile ma fallirà su un testo medico.

Con Word2Vec non è sempre necessario addestrare i modelli partendo da zero. Esistono molti embedding già pre-addestrati e pronti da usare.
In questi casi non addestrin ulla, carichi embedding già appresi su un corpus enorme (es. wikipeida)
Puoi fare model("computer") ed ottiene il vettore della parola "computer"
Gensim permette poi di fare direttamente confronti di similarità, ricercando le parole più vicine e analogie data una specifica parola (es. "computer")

Il modello generale potrebbe conoscere molto bene parole come: cliente, fornitore, ordine, fattura, ma non necessariamente capire bene termini estremamente specifici della tua azienda o codici tecnici.
Qui c'è la grende differenza tra General Domain e Domain Specific, un embedding pre-addestrato conosce bene la lingua italiana. Un embedding addestrato sui tuoi dati conosce meglio il tuo dominio.
In alcuni progetti si parte da vettori pre-addestrati e poi si lavora sul dominio specifico, quando la tecnica e il modello lo consentono.

Word Embedding pronto all'uso: 
    - Word3Vec Google News (inglese, 300 dimensioni)
    - Glove (diversi corpora/dimensioni)
    - FastText (molte lingue, italiano disponibie, subword)

Word2Vec / Glove / FastText sono però embedding STATICI, sono ancora molto usati per capire gli embeddig e per determinati progetti leggeri.
Oggi però, per molti problemi NLP moderni, si utilizzano anche: BERT, Sentence Transformers, LLM, 

C'è però una sfida tecnica, la gestione della memoria fisica

Caricamente Efficiente
Memory Mapping
Per modelli molto grandi (milioni di parole), Gensim supporta il caricamente in 'mmap' che carica solo le parti del file necessarie alla query corrente, preservando la memoria di sistema.
Leggi il dato dal disco, solo quando mi serve quella specifica parola.
La cardinalità del vocabolario V e la dimensionalità d dell'embedding determinano l'ccupazione di memoria totale dei pesi.
Con Keras3 possiamo persino spostare questi pesi sulla GPU per accellerare ancora di più i calcoli.

```python
# python -m pip install gensim
import os
os.environ["KERAS_BACKEND"] = "torch"

import keras
import gensim.downloader as api
from gensim.models import KeyedVectors
import numpy as np

# ======================================================================================
# 2. CARICAMENTO DELLO SPAZIO VETTORIALE (KNOWLEDGE RETRIEVAL)
# ======================================================================================
# Utilizziamo 'gensim' per scaricare un modello di Word Embedding pre-addestrato.
# Modello scelto: 'glove-twitter-25' (25 dimensioni, addestrato su Twitter).
# Logica: Le parole vengono trasformate in vettori numerici dove la vicinanza spaziale 
# indica una somiglianza semantica nel contesto del linguaggio naturale.
print("Download e caricamento dello spazio vettoriale in corso...")
word_vectors = api.load("glove-twitter-25") 

# ======================================================================================
# 3. INTERROGAZIONE SEMANTICA (ANALISI DELLE RELAZIONI)
# ======================================================================================
# Calcolo della similarità tramite Coseno (Cosine Similarity).
# Matematicamente: Misura l'angolo tra due vettori nel multi-spazio a 25 dimensioni.
# Più il valore è vicino a 1, più le parole sono state usate in contesti simili.
word_a, word_b = "pizza", "pasta"
sim = word_vectors.similarity(word_a, word_b)
print(f"\nSimilarità semantica tra '{word_a}' e '{word_b}': {sim:.4f}")

# ======================================================================================
# 4. ALGEBRA VETTORIALE (ANALOGIE SEMANTICHE)
# ======================================================================================
# Word2Vec permette operazioni matematiche sulle parole. 
# Equazione: Re - Uomo + Donna = Regina
# Spiegazione: Sottrarre il vettore 'man' da 'king' isola il concetto di 'regalità'. 
# Aggiungendo 'woman', ci spostiamo nel punto dello spazio che rappresenta la 'regalità femminile'.
print("\nRisoluzione analogia: Re - Uomo + Donna = ...")
result = word_vectors.most_similar(positive=['king', 'woman'], negative=['man'], topn=1)
print(f"Risultato calcolato: {result[0][0]} (Confidenza: {result[0][1]:.4f})")

# ======================================================================================
# 5. PONTE FRA GENSIM E KERAS (FEATURE TRANSFER)
# ======================================================================================
# Questa funzione è il "cuore" dell'integrazione: trasforma un dizionario di parole 
# (Gensim) in un layer neuronale (Keras) pronto per una rete profonda.

def create_keras_embedding(gensim_model):
    # Passaggio A: Recupero metadati dallo spazio Gensim
    # vocab_size: quante parole uniche conosce il modello (es. 1.2 milioni).
    # vector_dim: quanti numeri compongono ogni parola (qui 25).
    vocab_size = len(gensim_model.index_to_key)
    vector_dim = gensim_model.vector_size
    
    # Passaggio B: Estrazione della matrice dei pesi
    # 'weights' è un array NumPy di forma [vocab_size, vector_dim].
    # Ogni riga 'i' contiene il vettore della parola all'indice 'i' del vocabolario.
    weights = gensim_model.vectors
    
    # Passaggio C: Definizione del Layer Embedding
    # input_dim: Raggio d'azione del vocabolario (ID delle parole).
    # output_dim: Dimensione del vettore di uscita (25).
    # trainable=False: "Congeliamo" i pesi. Non vogliamo che la rete modifichi 
    # la conoscenza linguistica già appresa durante l'addestramento.
    embedding_layer = keras.layers.Embedding(
        input_dim=vocab_size,
        output_dim=vector_dim,
        trainable=False,
        name="word2vec_pretrained"
    )
    
    # Passaggio D: Inizializzazione fisica del Layer
    # keras.layers.build() alloca la memoria necessaria per i pesi.
    # set_weights() inietta i numeri presi da Gensim dentro il layer Keras.
    embedding_layer.build((None,))
    embedding_layer.set_weights([weights])
    
    return embedding_layer

# ======================================================================================
# 6. ARCHITETTURA DEL MODELLO (PIPELINE DI CLASSIFICAZIONE)
# ======================================================================================
# Costruiamo una rete che accetta testo (sotto forma di ID numerici) e produce un output.
model = keras.Sequential([
    # Input Layer: Accetta sequenze di lunghezza variabile (None) di interi.
    keras.layers.Input(shape=(None,), dtype="int32"),
    
    # Pretrained Embedding: Trasforma gli ID in vettori semantici da 25 dimensioni.
    create_keras_embedding(word_vectors),
    
    # GlobalAveragePooling1D: Se abbiamo una frase, calcola il vettore "medio" della frase.
    # Trasforma una matrice [numero_parole, 25] in un unico vettore [25].
    keras.layers.GlobalAveragePooling1D(), 
    
    # Dense Layer: Un singolo neurone con attivazione Sigmoide per classificazione binaria.
    keras.layers.Dense(1, activation="sigmoid")
])

# ======================================================================================
# 7. PERSISTENZA E SALVATAGGIO (FORMATO .KERAS)
# ======================================================================================
# Il nuovo formato .keras è un pacchetto compresso che contiene sia l'architettura 
# (come sono connessi i layer) sia i pesi pre-caricati da Gensim.
model_path = "word2vec_pipeline_2026.keras"
model.save(model_path)

print(f"\n[INFO] Architettura e Pesi integrati correttamente.")
print(f"[INFO] Modello salvato in: {model_path}")
print(f"[INFO] Dimensione vocabolario importato: {len(word_vectors.index_to_key)} parole.")
```
