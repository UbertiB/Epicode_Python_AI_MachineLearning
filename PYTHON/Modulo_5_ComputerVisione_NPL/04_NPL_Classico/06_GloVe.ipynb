GLOVE EMBEDDINGS

GloVe (Global Vectors for Word Representation) è un altro modello classico per crare word embedding, come Word2Vec.

La differenza principale è il modo in cui impara i vettori. 

Word2Vec (predizione locale) ragiona soprattutto sul contesto locale, osserva quali parole compaiono vicino a una parola target e impara da quelle relazioni. E' vero che esistono modelli enormi già preaddestrati, ma, durante il training, impare usando una finestra locale di conteto attorno a ogni parola.

GloVe (conteggio globale) invece usa statistiche globali di co-occorrenza del corpus.
In pratica costruisce una grande matrice che dice, più o meno: quante volte la parola A compare vicino a parola B?
GloVe usa queste statistiche per imparare vettori densi.
L'idea è che parole che compaiono in contetsi simili tendano ad avere vettori simili
Quindi il principio è identico a Word2Vec ma cambia il modo per arrivarci

Meglio guardare da vicino è guardare tutto?

Finestra Locale vs Statistiche Globali
Due paradigmi per mappare il significato
Mentre Word2Vec utilizza una rete neurale per prevedere le parole in una finestra locale (usando una finestra scorrevole), GloVe (Global Vectors) adotta un approccio basato sul conteggio esplicito. La sfida è combinare i vantaggi dei metodi di fattorizzazione di matrici con l'efficienza dell'apprendimento basato sul contesto. 
GloVe non si limita a scorrere il testo, ma aggrega le statistiche dell'intero corpus in una struttura globale prima di iniziare l'addestramento. 
Glove invece di scorrere tutto il testo per finestre locali, aggrega tutto quello che sa in un'unica grande fotografica (struttura globale) prima di iniziare l'addestramento.
Questo permette di catturare non solo chi è vicino a chi, ma quanto frequentemente determinati concetti co-occorrono nell'intero dataset.

Predizione (Word2Vec) contro Conteggio (GloVe)
- Approccio predittivo: Word2Vec impare i pesi cercando di minimizzare l'errore di predizione di un termine target data la sua finestra locale.
- Approccio Statitico: GloVe sfrutta le statistiche globali del corpus analizzando i rapporti di probabilità di co-occorrenza tra le parole, guardando l'intero dataset.
GloVe risulta spesso più veloce  da addestrare su grandi volumi di dati, poichè i calcoli sono basati sulla matrice pre-compilata. Una volta costruita la matrice dei conteggi, l'addestramento è rapidissimo perchè i calocli pesanti sono già stati fatti a monte.
L'obbiettivo di GloVe è modellare il prodotto scalare dei vettori affinchè corrisponda al logaritmo della probabilità di co-occorrenza.

Anche un gigante come GloVe ha i sui punti deboli
Limiti e Opportunità
Sebbene GloVe utilizzi statistiche globali, deve gestire una matrice di co-occorrenza potenzialmente enorme e molto sparsa se il vocabolario è vasto. Se avete un vocabolario di 10.000 parole la matrice ha 10 miliardi di celle.
Inoltre, sia Glove sia Word2Vec faticabo a distinguere i sensi diversi di una stessa parola, fornendo un unico vettore medio (confuso) tra i due signigicati per ogni lemma.

GloVe eccelle nel catturare relazioni semantiche globali che potrebbero sfuggire a finestro di contesto troppo strette

Rapporti di Probabilità (la vera intuizione geniale che sta dietro a GloVe)

La vera intuizione di GloVe risiede nell'osservare che il signifidato emerge non dalla probabilità singola, ma dal rapporto tra probabilità. 
Il significato non sta nella probabilità che una parola appaia, ma nel suo rapporto con le altre
Ad esempio, la parola 'ghiaccio' è più correlata con 'solido' che a 'gas', e questo rapporto definisce il vettore. 
Se indichiamo con P la probabilità che la parola k appaia nel contesti di i, Glove modella il rapporto tra queste probabilità

Sia GloVe sia Word2Vec  producono embedding statici. Quindi 'banca' ha lo stesso vettore in queste due frasi: "la banca condede un mutuo" e "la banca del fiume"
Il contesto locale non cambia il vettore
Questo limite accomuna GloVe e Word2Vec, mentre i Transformer producono embedding contestuali

GloVe viene spesso usato già pre-addestrato ocn Gensim puoi caricare modelli GloVe pronti

La differenza rispetto a TF-IDF è molto netta
Con Tf-IDF la parola "computer" e la parola "laptop" sono due feature distinte. Il sistema non sa automaticamente che sono correlate.
Con GloVe invece "computer ha il proprio vettore, "laptop" ha il proprio vettore" e i due vettori possono essere vicini perchè il modello ha imparato che le parole compaiono in contesti simili

Per fare tutto questo GloVe ha bisogno di una base solida, la matrice di co-occorenza.

La Matrice di Co-occorrenza.
Il fondamento statistico del modello.
Per implementare GloVe, il primo passo è la costruzione della matrice di co-occorrenza X (che è il nostro archivio universale). 
Ogni volta che due parole si incontrano entro una certa distanza, noi mettiamo un segno nella cella corrispondnete, alla fine avremmo una mappa simmetrica che ci dice chi sono gli amici stretti di ogni termine. 
Questa matrice rappresenta la memoria collettiva del dataset, invece di iterare per gionri sulle frasi, glove legge direttamente questa mappa che risiede già in memoria.

Ogni cella X_ij indica quante volte la parola j è apparsa nel contesto della parola i all'interno del corpus.
Questa matrice è simmetrica e rappresenta la memoria collettiva del dataset. Invece di iterare milioni di vole sulle fasi, l'algoritmo di apprendimento lavorerà direttamente su questi conteggi aggregati, ottimizzano la rappresentazione vettoriale delle parole

Analisi dei Conteggi
Dalla frequenza alla geometrica

- La Co-occorenza è il legame statistico tra due termini che compaiono frequentemente entro una cerca distanza  l'uno dall'altro, termini vicini
- Glove è intelligente, non da lo stesso valore a tutte le vicinanzae, una parola che sta subito accanto alla nostra riceve un peso maggiore rispetto ad una che sta a 5 o 6 posizioni di distanza. Questo permette di capire la centralità di una parola all'interno del discorso globale.

Ma come evitiamo che parole comuni come 'il' o 'e' rovinino tutto il lavoro?

Funzione di Pesdatura
Per evitare che parole estremamente frequenti dominino il modello GloVe introduce una funzione di costo pesata che limita l'impatto dei conteggi troppo elevati.
E' come un limittore di volume, su una parola urla troppo forte è prechè è ovunque e noi limitiamo la sua voce.
Una volta superata una certa frequenza critica, l'incremento di importanza di una coppia di parole smette di crescere stabilizzando l'addestramento.

Efficienza del Training
A differenza dei metodi neurali iterativi che devono passare attraverso ogni frase, GloVe esegue una sola scansione del testo per creare la matrice. Successivamente, l'ottimizzazione è slegata dalla lungheza totale del testo.
Il tempo di addestramento del modello GloVe scala con il numero di elementi non nulli nella matrice di co-occorrenza, non con il numero di parole.

Perchè GloVe è così amato nell'industria?

Vantaggi e Scalabilità
Perchè GloVe domina nei grandi dataset
GloVe è progettato per brillare quando i dati abbondano. Grazie alla sua natura statistica globale, produce vettori più stabili e robusti rispetto a Word2Vec quando il corpus raggiunge dimensioni nell'ordine dei miliardi di token
Oltre alla stabilità, la linearità intrinseca dello spazio vettoriale di GloVe rende le relazioni di analogia estremamente precise, permettendo di mappare con successo proprietà linguistiche e semantiche complesse.

Stabilità Semantica
Robustezza delle rappresentazioni
- Invarianza del Corpus: GloVe tende a mantenere relazioni simili anche se addestrato su porzioni diverse dello stesso grande dataset
- Precisione nelle Analogie: le regolarità lineari sono spesso più marcate, facilitando compiti come il ragionamento semantico
- Uso di Pesi Pre-addestrati: molti modelli industriali iniziano con vettori GloVe pre-addestrati su Common Crawl per risparmiare tempo e risorse 
- La similarità tra due vettori si misura tramite il coseno dell'angolo, dove valori vicini a 1 indicano massima affinità semantica globale.

Ecco perchè molti modelli moderni, prima di iniziare imparare, caricano i pesi pre-addestrati di Glove, come dare al modello una laurea in lingue prima ancora di iniziare il primo gionro di lavoro

```python
"""
================================================================================
Implementazione  di GloVe (Global Vectors)
================================================================================
Questo script è una simulazione completa del flusso di lavoro di GloVe, dall'analisi 
del testo "sporco" alla creazione di uno spazio vettoriale dove le parole simili 
sono vicine tra loro.

ARCHITETTURA DEL CODICE:
1. [ELABORAZIONE DATI]: Scarichiamo testi reali e costruiamo la "Mappa dei Vicini" 
   (Matrice di Co-occorrenza). È qui che applichiamo la statistica globale.
2. [IL MODELLO]: Creiamo una rete neurale "minima" che ha un solo compito: 
   regolare i vettori affinché il loro prodotto scalare rispecchi i conteggi fatti.
3. [ANALISI]: Interroghiamo la mappa creata per vedere se le parole "spazio" 
   e "orbita" sono effettivamente finite vicine.

REQUISITI: scikit-learn, keras, torch, numpy
"""

import os

# --- STEP 0: CONFIGURAZIONE INFRASTRUTTURA ---
# Impostiamo PyTorch come motore di calcolo prima di caricare Keras.
# Questo garantisce che tutte le operazioni 'ops' usino i tensori di PyTorch.
os.environ["KERAS_BACKEND"] = "torch"

import numpy as np
import keras
from keras import layers, ops
from sklearn.datasets import fetch_20newsgroups # Per scaricare testi reali
import re # Per la pulizia del testo (Regex)
from collections import Counter, defaultdict # Per contare le parole in modo efficiente


# ==============================================================================
# FASE 1: COSTRUZIONE DEL CERVELLO STATISTICO (LA MATRICE DI CO-OCCORRENZA)
# ==============================================================================

def prepare_real_cooccurrence(vocab_size=1000, window_size=5):
    """
    Legge il testo e crea la struttura di dati richiesta da GloVe.
    
    Interazione: Questa funzione 'alimenta' il modello Keras trasformando 
    frasi umane in triplette numeriche: (Indice_Parola_1, Indice_Parola_2, Forza_Legame).
    """
    
    # 1.1 ACQUISIZIONE DATI
    print("1. Scaricamento testi reali (Articoli scientifici su spazio e medicina)...")
    data = fetch_20newsgroups(subset='train', categories=['sci.space', 'sci.med'])
    texts = data.data[:500] # Limite a 500 per non rallentare la lezione

    # 1.2 PULIZIA E TOKENIZZAZIONE (Trasformiamo "Ciao!" in ["ciao"])
    all_tokens = []
    word_pool = []
    for text in texts:
        # re.findall(r'\b[a-z]{3,}\b') -> Estrae solo parole di almeno 3 lettere, ignorando numeri e punteggiatura
        tokens = re.findall(r'\b[a-z]{3,}\b', text.lower())
        all_tokens.append(tokens)
        word_pool.extend(tokens) # Mettiamo tutte le parole in un unico 'secchio' per contarle

    # 1.3 CREAZIONE VOCABOLARIO (Mappa: Parola -> Numero)
    # Prendiamo le 1000 parole più comuni per evitare di gestire milioni di termini rari
    most_common = Counter(word_pool).most_common(vocab_size)
    word_to_id = {word: i for i, (word, count) in enumerate(most_common)}
    id_to_word = {i: word for word, i in word_to_id.items()} # Inversa: serve per leggere i risultati

    # 1.4 IL CUORE DI GLOVE: SCANSIONE DEI VICINI (Slide 7-8)
    print("2. Scansione testo: Analisi della 'Sliding Window' (Chi sta vicino a chi?)...")
    cooc_matrix = defaultdict(float)

    for tokens in all_tokens:
        for i, word in enumerate(tokens):
            if word not in word_to_id: continue # Saltiamo parole fuori dal top-1000
            
            # Definiamo i confini della finestra (context window)
            start = max(0, i - window_size)
            end = min(len(tokens), i + window_size + 1)
            
            for j in range(start, end):
                if i == j: continue # Una parola non è vicina a se stessa
                neighbor = tokens[j]
                if neighbor not in word_to_id: continue
                
                # Slide 8: Pesatura dei vicini. Le parole adiacenti (distanza 1) 
                # contano 1.0, quelle più lontane (distanza 2) contano 0.5.
                dist = abs(i - j)
                cooc_matrix[(word_to_id[word], word_to_id[neighbor])] += 1.0 / dist

    # 1.5 FORMATTAZIONE PER IL MODELLO
    # Trasformiamo il dizionario in vettori NumPy pronti per l'addestramento
    idx_a, idx_b, counts = [], [], []
    for (a, b), weight in cooc_matrix.items():
        idx_a.append(a)
        idx_b.append(b)
        counts.append(weight)

    idx_a = np.array(idx_a, dtype="int32")
    idx_b = np.array(idx_b, dtype="int32")
    counts = np.array(counts, dtype="float32")

    # TRASFORMAZIONE LOGARITMICA (Slide 4)
    # GloVe non cerca di indovinare il conteggio esatto, ma il suo LOGARITMO.
    # Questo perché i conteggi crescono in modo esponenziale (legge di Zipf).
    log_targets = np.log(counts + 1e-10) # + 1e-10 evita il log(0) che è errore

    # FUNZIONE DI PESO (Slide 9-10)
    # Diciamo al modello: "Impegnati molto se vedi conteggi alti, ma ignora il rumore di fondo"
    sample_weights = np.minimum(1.0, (counts / 100.0)**0.75)

    return (idx_a, idx_b), log_targets, sample_weights, id_to_word


# ==============================================================================
# FASE 2: COSTRUZIONE DELLA RETE NEURALE (IL MOTORE GEOMETRICO)
# ==============================================================================

def build_glove_model(vocab_size, dimension=50):
    """
    Crea il modello Keras seguendo la logica Log-Bilinear.
    
    Interazione: Il modello prende due numeri (ID parole) e cerca di 
    far sì che i loro vettori interni, se moltiplicati tra loro, restituiscano 
    il logaritmo del conteggio che abbiamo calcolato prima.
    """
    
    # Input: Rappresentano un indice di parola (es. "space" è l'indice 45)
    input_target = layers.Input(shape=(1,), name="Parola_A")
    input_context = layers.Input(shape=(1,), name="Parola_B")
    
    # LAYER EMBEDDING: È il cuore del Deep Learning per NLP. 
    # È una tabella gigante dove ogni parola ha una riga di 'dimension' numeri.
    # Questi numeri all'inizio sono casuali, alla fine saranno 'significato'.
    embedding_layer = layers.Embedding(vocab_size, dimension, name="Mappa_Vettoriale")
    
    # Estraiamo i vettori per le due parole in input
    vec_a = embedding_layer(input_target)
    vec_b = embedding_layer(input_context)
    
    # PRODOTTO SCALARE (Dot Product): Misura la sovrapposizione tra i due vettori.
    # Slide 4: L'obiettivo è dot(vec_a, vec_b) = log(conteggio_reale)
    dot_product = layers.Dot(axes=2)([vec_a, vec_b])
    
    # Puliamo la forma dei dati (da [Batch, 1, 1] a [Batch, 1])
    output = layers.Reshape((1,))(dot_product)
    
    # Configurazione finale: Usiamo Adam (ottimizzatore intelligente) e MSE (errore quadratico)
    model = keras.Model(inputs=[input_target, input_context], outputs=output)
    model.compile(optimizer=keras.optimizers.Adam(learning_rate=0.005), loss="mse")
    
    return model


# ==============================================================================
# FASE 3: TRAINING E INTERROGAZIONE (VERIFICA DEI RISULTATI)
# ==============================================================================

# 3.1 Prepariamo tutto
(inputs, targets, weights, id_to_word) = prepare_real_cooccurrence()
model = build_glove_model(vocab_size=len(id_to_word))

# 3.2 ADDESTRAMENTO
print("\n3. Inizio fase di apprendimento... (Il modello sposta i vettori nello spazio)")
# Passiamo sia la coppia di parole che i pesi (weights) per dare importanza ai dati giusti
model.fit(
    x=[inputs[0], inputs[1]], 
    y=targets, 
    sample_weight=weights, 
    epochs=12, 
    batch_size=512, 
    verbose=1
)

# 3.3 ESTRAZIONE CONOSCENZA
# Ora che il modello ha imparato, prendiamo i vettori definitivi
final_vectors = model.get_layer("Mappa_Vettoriale").get_weights()[0]

def find_similar(word_to_test, n=5):
    """Calcola la similarità tra una parola e tutto il resto del vocabolario."""
    # Cerchiamo l'indice della parola (se esiste)
    word_id = None
    for i, w in id_to_word.items():
        if w == word_to_test:
            word_id = i
            break
    
    if word_id is None: return f"'{word_to_test}' non trovata."

    # Prendiamo il suo vettore
    target_v = final_vectors[word_id]
    
    # SIMILARITÀ DEL COSENO (Slide 12)
    # È la formula standard per vedere se due frecce (vettori) puntano nella stessa direzione:
    # cos(theta) = (v1 dot v2) / (||v1|| * ||v2||)
    norms = np.linalg.norm(final_vectors, axis=1) # Lunghezza di ogni vettore nel vocabolario
    target_norm = np.linalg.norm(target_v)        # Lunghezza del nostro vettore test
    
    # Calcoliamo la vicinanza con TUTTI gli altri vettori contemporaneamente (efficienza NumPy)
    similarities = np.dot(final_vectors, target_v) / (norms * target_norm + 1e-10)
    
    # Ordiniamo e prendiamo i migliori (saltando il primo perché è la parola stessa)
    best_indices = np.argsort(similarities)[::-1][1:n+1]
    
    return [(id_to_word[idx], similarities[idx]) for idx in best_indices]

# TEST FINALE: Vediamo se il modello ha capito la semantica del dataset
print("\n--- ANALISI SEMANTICA GLOBALE ---")
for word in ["space", "health", "orbit", "system"]:
    results = find_similar(word)
    print(f"Parole simili a '{word}': {results}")


# ==============================================================================
# CONCLUSIONI DIDATTICHE
# ==============================================================================
"""
COSA ABBIAMO DIMOSTRATO:
1. GloVe non "legge" il testo durante l'allenamento: legge solo la matrice pre-calcolata. 
   Questo lo rende incredibilmente veloce su grandi mole di dati.
2. La funzione di pesatura garantisce che parole comuni (come 'the', ignora qui 
   ma presente in testi grossi) non schiaccino i termini tecnici come 'orbit'.
3. Quello che vedi come 'Similarità' è la prova che la statistica si è trasformata 
   in geometria: parole che co-occorrono spesso nell'intero corpus finiscono per 
   avere vettori simili nello spazio multidimensionale.
"""
```
