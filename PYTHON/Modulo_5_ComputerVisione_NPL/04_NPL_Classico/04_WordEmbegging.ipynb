RAPPRESENTAZIONI DENSE: WORD EMBEDDINGS

Con Bag of Words e con TF-IDF ottengo una rappresentazione del documento basata sul vocabolario.
Questi vettori risultanti, sono spesso molto lunghi e pieni di zeri. 
Per questo si chiamano sparse.

Limiti del One-Hot Encoding
Perchè contare le parole non basta per capire il senso.
Finotra abbiamo usato tecniche basate sulla frequenza e sull'indicizzazione univoce, come il One-Hot Encoding. Qui ogni parola è un'isola a se stante, rappresentata da un vetotre cheha un unico valore '1' in una foresta di '0'
Questo approccio soffre di due problemi: 1) 'maledizione della dimensionalità, poiche la lunghezza dei vettori cresce con il vocabolario, 2) e l'assenza totale di relazioni semantiche. 
Per un sistema 'One-Hot', le parole 'gatto' e 'cane' sono distanti quanto 'gatto' e 'aspirapolvere'

Vediamo quali sono le conseguenze matematiche di questo 'isolamento'
I vincoli matematici delle rappresentazioni discrete.
- Vettori Sparsi: la memoria viene sprecata per memorizzare migliaia di zeri che non portano alcuna informazione utile al modello
- Indipendenza Lineare: tutti i vettori 'one-hot' sono ortogonali tra loro, rendendo impossibile calcolare una somiglianza semantica. Matematicamente è come se la rete dicesse al modello che non esiste alcuna relazione tra i termini.
- Esplosione dei Parametri: nelle reti neurali, l'input layer diventa enorme, richiedendo una potenza computazionale eccessiva per pesi spesso nulli.
- In una rappresentazione one-hot il prodotto scalare tra due terminini distinti 'i' e 'j' è sempre nullo, indicando assenza di correlazion

Nelle rappresentazioni discrete (come one-hot) non catturano sinonimi o relazioni gerarchiche. 'Medico' e 'Dottore' sono trattati come concetti totalmente differenti l'uno all'altro (mancanza di gerarchia), è un limite enorme per la generalizzazione.
Se il mio modello impara qualcosa con la parola 'cane' vorrei che quella conoscenza fosse trasferita anche con la parola 'cucciolo'. Con il hone-hot encoding questo però è impossibile, il modello deve re-imparare tutto per ogni singola parola.
Gestire vettori di dimensione 50.000 per ogni singola parola di un testo è inefficiente e limita la capacità di addestramento su grandi corpus.
Un modello basato su 'one-hot' non può intuire che un verbo incontrato al passato abbia lo stesso valore di quello al presente se non li ha visti entrambi esplicitamente.
Se proviamo a calcolare la distanza euclidea tra due vettori 'one-hot' otterremo lo stesso valore costante. Questo significache lo spazio non è curvato dal significato, ma è un ipercubo rigido.

Dobbiamo cambiare completamente il modo in cui rappresentiamo i dati. Ecco che entrano in scena i vettori Densi

I Word Embeddings, invece, sono rappresentazioni dense
Per superare i limiti visti con il one-hot encoding introduciamo i word embeddins. Invece di vettori sparsi ed enormi, usiamo vettori densi di piccole dimensioni, tipicamente tra 50 e 300, composti da numeri reali continui.
Questi vettori sono chiamati 'rappresentazioni distribuite' perchè il significata di una parola non è più concentrato in un singolo indice, ma è distribuito tra tutte le dimensioni del vettore. Ogni dimensione può potenzialmente catturare una sfumatura diversa: genere, tempo, astrazione, o funzione grammaticale.

Ma come fa il computer a gestire questa complessità in modo efficiente?

L'Essenza della Densità
Come i numeri reali modellano il linguaggio
- Compressione informativa: riduzione drastica dello spazio di input (da 50.000 a 128 dimensioni) mantenendo intatto il segnale.
- Valore Continui: l'uso di 'float' permette di mappare le parole in uno spazio fluido dove le transazioni sono possibilil.
- Embedding Layer: struttura neurale che funge da tabella di lookup per convertire istantaneamente ID in vettori densi.
- Un 'word embedding' è una mappatura 'f' che proietta il vocabolario 'v' in uno spazio vettoriale continuo a bassa dimensionalità 't'

I vettori densi, portano con sè l'ereditarietà semantica, attraverso l'addestramento, parole che appaiono in contesti simili tendonon ad avere coordinate simili nel vettore denso.
L'uso di spazi pre-addestrati permette a modelli con pochi dati di 'ereditare' la compressione linguistica appresa da modelli più grandi (efficienza neurale).
Gli embeddings possono essere appresi da zero per un compito specifico o pre-addestrati su enormi corpora come Wikipedia.
I gradineti fluiscono meglio attraverso valori continui che tra 0 e 1

La proprietà della Vicinanza
Geometria del significato e relazioni vettoriali.
La vera magia dei 'Word Embeddings' emerge quando osserviamo la loro disposizione nello spazio. Parole con significati simili finiscono per raggrupparsi in cluster vicini. Questa vicinanza non è casuale, ma riflette la somiglianza semantica del mondo reale.
I colori staranno vicino ai colori, le città vicino alle città, i sentimenti positivi vicini ai sentimenti positivi. Questo è la prova che il modello ha capite le relazioni del mondo reale.

Come misuriamo questa 'vicinanza' in modo oggettivo?

Metriche di Similarità
Misurare l'affinità semantica.
- Similarità Coseno: misura l'angolo tra due vettori, più l'angolo è piccolo, più le parole sono correlate.
- Relazioni di analogia: proprietà per cui la differnza tra 'uomo' e 're' è simile alla differenza tra 'donna' e 'regina'
- Cluster Semantici: formazione natuale di gruppi nello spazio (es colori, aniamli, nomi di città)
- La similarità Coseno è calcolato come il prodotto scalare normalizzato delle lunghezze dei due vettori.

L'equazione più celebre dell'NLP moderno è:
Re - Uomo + Donna = Regina
le direzioni nello spazio vettoriale corrispondono a concetti logici. 

Supponiamo essere presente, nel vocabolario, la parola 'cliente' con id 5
La parola cliente può diventare un vettore esempio: [0.21,-0.48,0.73,0.15,...]
magari con 100 o 300 o 768 dimensioni, a seconda del modello.

La cosa interessante non è solo che il vettore sia più corto.
E' che gli embedding cercano di rappresentare anche relazioni semantiche tra le parole.
Supponiamo di avere nel vocabolario, tra le altre, le parole 'cliente' e 'acquirente'
Le dua parolo potrebbero avere vettori realmente vicini, mentre 'cliente' e 'montagna' potrebbe essere molto lontani.

Word Embedding: rappresenta una parola come un vettore numerico appreso, in uno spazio in cui parole più usate in contesti simili, tendono ad avere rappresentazioni simili (vettori simili)

Questo è il grande salto rispetto a TF-IDFm dove le singole parole sono semplicemente colonne diverse del vocabolario, non sapendo se sono o no concetti vicini.

Con embedding invece, la parola 'cliente' e la parola 'acquirente' possono risultare matematicamente vicini.
PAROLE USATE IN CONTESTI SIMILI TENDONO AD AVERE SIGNIFICATI SIMILI

Per esempio in un corpus potresti avere le seugneti frasi:
- il cliente ordina il prodotot - l'acquirente ordina il prodotto - il cliente paga la fattura - l'acquirente paga la fattura.
Il modello osserva che 'cliente' e 'acquirente' compaiono spesso vicino a parole simili (ordina, prodotto, paga, fattura) e quindi impara vettori simili.

```python
import os

# --- CONFIGURAZIONE AMBIENTE ---
# Configurazione del Backend per Keras 3 (Standard 2026)
# INTERAZIONE: Keras 3 è agnostico rispetto al backend. Impostando "torch", istruiamo 
# Keras a utilizzare PyTorch per gestire i calcoli tensoriali e i grafi di computazione.
os.environ["KERAS_BACKEND"] = "torch"

import keras
import numpy as np
from keras import ops # Keras Ops: operazioni matematiche che funzionano su tutti i backend (Torch, JAX, TF)

# --- 1. DEFINIZIONE DEL VOCABOLARIO E TOKENIZZAZIONE ---
# INTERAZIONE: Iniziamo trasformando il testo in numeri. Ogni parola è mappata a un ID intero.
# Questo è il primo passo per trasformare concetti astratti in dati leggibili dalla macchina.
vocab = {"gatto": 0, "cane": 1, "astronave": 2}
dim_voc = len(vocab)
embedding_dim = 2  # Dimensione dello spazio semantico (2D per facilità di comprensione)

# --- 2. IL LIMITE DELLA RAPPRESENTAZIONE ONE-HOT ---
# TEORIA: I vettori One-Hot sono "ortogonali" (perpendicolari). 
# Questo significa che il prodotto scalare tra due parole diverse sarà sempre 0.
# LIMITE: Il sistema non può capire che "cane" è più vicino a "gatto" rispetto a "astronave".
def get_one_hot(word_id, size):
    """Crea un vettore di zeri con un unico '1' nella posizione dell'ID parola."""
    vec = np.zeros(size)
    vec[word_id] = 1
    return vec

# Estraiamo i vettori One-Hot per il confronto
v_gatto_oh = get_one_hot(vocab["gatto"], dim_voc) # [1, 0, 0]
v_cane_oh = get_one_hot(vocab["cane"], dim_voc)   # [0, 1, 0]

# Calcolo Prodotto Scalare One-Hot: Risultato 0 -> Zero similarità matematica nonostante la vicinanza semantica.
print(f"Similarità scalare One-Hot (Gatto-Cane): {np.dot(v_gatto_oh, v_cane_oh)}")

# --- 3. WORD EMBEDDINGS (Rappresentazione Densa e Continua) ---
# INTERAZIONE: Il layer 'Embedding' agisce come una tabella di lookup gigante.
# Invece di vettori sparsi (0 e 1), usa coordinate reali (es. 0.9, 0.1).
embedding_layer = keras.layers.Embedding(input_dim=dim_voc, output_dim=embedding_dim)

# Inizializziamo i pesi manualmente per simulare un addestramento avvenuto.
# NOTA: Gatto e Cane hanno coordinate molto vicine nello spazio 2D, Astronave è su un altro quadrante.
custom_weights = np.array([
    [0.9, 0.1],  # Coordinata Gatto
    [0.8, 0.2],  # Coordinata Cane (Vicina a Gatto)
    [-0.5, 0.8]   # Coordinata Astronave (Lontana dagli animali)
])

# Costruiamo il layer e iniettiamo i nostri pesi personalizzati
embedding_layer.build()
embedding_layer.set_weights([custom_weights])

# --- 4. CALCOLO DELLA SIMILARITÀ SEMANTICA (Cosine Similarity) ---
# INTERAZIONE: La similarità coseno misura l'angolo tra due vettori. 
# Più l'angolo è piccolo, più il valore si avvicina a 1 (massima similarità).
def cosine_similarity(v1, v2):
    # Usiamo 'keras.ops' per garantire che la funzione funzioni indipendentemente dal backend scelto (Torch/TF)
    dot_product = ops.sum(v1 * v2)
    norm_v1 = ops.sqrt(ops.sum(ops.square(v1)))
    norm_v2 = ops.sqrt(ops.sum(ops.square(v2)))
    # Formula: (A dot B) / (||A|| * ||B||)
    return dot_product / (norm_v1 * norm_v2)

# PASSAGGIO CHIAVE: Trasformiamo gli ID in vettori densi passando attraverso il layer di Embedding
# 'embedding_layer' riceve un input numerico e restituisce il vettore di 'embedding_dim' componenti.
vec_gatto = embedding_layer(np.array([vocab["gatto"]]))
vec_cane = embedding_layer(np.array([vocab["cane"]]))
vec_astronave = embedding_layer(np.array([vocab["astronave"]]))

# Calcoliamo le distanze semantiche
sim_animali = cosine_similarity(vec_gatto, vec_cane)
sim_misto = cosine_similarity(vec_gatto, vec_astronave)

print(f"\nSimilarità Coseno (Gatto - Cane): {float(sim_animali):.4f} (Alta: sono entrambi animali)")
print(f"Similarità Coseno (Gatto - Astronave): {float(sim_misto):.4f} (Bassa: concetti distanti)")

# --- 5. PERSISTENZA DEI DATI ---
# Creiamo un modello 'Sequential' che incapsula il nostro layer.
# Questo permette di salvare non solo i pesi, ma l'intera architettura del sistema di embedding.
model = keras.Sequential([embedding_layer])
model.save("word_embeddings_2026.keras")
print("\n[INFO] Modello salvato: la mappa semantica è ora pronta per essere riutilizzata.")
```
