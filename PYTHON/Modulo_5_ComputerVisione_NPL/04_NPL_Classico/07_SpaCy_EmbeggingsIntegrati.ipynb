SPACY EMBEDDINGS INTEGRATI

SpaCy permette di usare vettori già inclusi in alcuni suoi modelli linguistici senza dover costruire a mano nè Word2Vec nè GloVe (o altro).

Word2Vec / GloVe -> tecniche per CREARE word embeddings
SpaCy -> può FORNIRTI e USARE word embeddings già inclusi in una pipeline

SpaCy non ha un suo algoritmo equivalente a Word2Vec o GloVe per addestrare da zero una tabella di word vectors statistici. Può invece incorporare vettori preaddestrati provenienti da tecniche come GloVe, FastText o Gensim e usarli nella propria pipeline

Quindi, senza fare: - matrice do co-occorrenza - training - embedding layer - ottimizzazione, hai già a disposizione i vettori, questo è il senso degli embedding integrati

Tutta le complessità gi GloVe (o di Word2Vec o di Transformer, ecc) non è scomparsa, è semplicemente stata fatta prima da chi ha costruito la pipeline di SpaCy

spaCy può integrare word embeddings preaddestrati direttamente nelle proprie pipeline. Nei modelli che includono una tabella di vettori, questi sono accessibili tramite .vector e possono essere utilizzati per calcolare la similarità semantica tra Token, Span e Doc.

Cosa c'è dentro i modelli che scarichiamo?
Vettori Statistici Integrati
Sfruttare la potenza dei modelli pre-addestrati.
Nelle lezioni precedenti abbiamo visto come funzionano Word2Vec e GloVe a livello teorico. Con SpaCy, questi vettori non sono più entità esterne, ma sono integrati direttamente nell'oggetto nlp. 
Questo approccio permette di accedere a rappresentazioni dense di alta qualità senza dover addestrare modelli da zero.
I modelli SpaCy sono distribuiti in diverse dimensioni: 'sm' (small), 'md' (medium), 'lg' (large). E' fondamentale comprendere che solo le versioni 'md' e 'lg' includono veri 'static word vectors' derivati da GloVe, necessari per operazioni di similartià semantica accurata.

Ma come facciamo, tecnicamente, ad estrarre queste coordinate?

Accesso ai Vettori
Interrogare l'attributo .vector
sua che tu stia guardando una singola parola o un intero paragrafo, spacy restituisce un array nympy denso, una stringa di 300 numeri (vettore di 300 nunmeri).
* Modelli 'md' e'lg': il primo bilancia memoria e precisione, il secondo offre un vocabolario vettoriale più vasto per una copertura lessicale superiore.
* Attributo .vector: ogni oggetto 'Token', 'Span' o 'Doc' possiede la proprietà .vector che restituisce un array Numpy denso.
* Proprietà .has_vector: un flag booleano essenziale per verificare se una specifica parola è presente nel dizionario pre-addestrato del modello.
* La mappatura trasforma un termine in uno spazio vettoriale a 300 dimensioni, dove la posizione definisce il contesto semantico.

Caratteristiche dei Vettori restituiti da spaCy
I vettori che restitiusce SpaCy sono solitamente normalizzati e hanno una dimensione fissa tipicamente di 300, che rappresenta uno standard nel settore per catturare sfumature semantiche complesse.
Se una parola non è presente nel modello, spaCy assegna un vettore di soli zeri (OOV, ut-of-Vocabulary). E' best practice verificare sempre la presenza del vettore prima di eseguire calcoli matematici.
Mentre il modello 'sm' utilizza tensori di contesto, solo 'md' e 'lg' permettono di eseguire operazioni di algebra lineare coerenti basate sulla distribuzione globale delle parole.

Pensa a questi vettori non come a numeri ma alla geometria di una parola

Vettori come coordinate
Ogni dimensione del vettore può essere vista come una coordinata in uno spazio semantico. Parole vicine in questo spazio condividono contesti d'uso simili nel mondo reale, permettendo al computer di 'intuire' la parentela tra termini.
La rappresentazione densa elimina la sparsità del 'one-hot encodig', comprimento l'informazione in un formato nunmeri continuo.

La Similarità Semantica (di due parole)
Misurare la vicinanza tra concetti.
Una volta ottenuti i vettori, la funzione più potente a nostra disposizione è .similary(). Questo metodo implementa il calcolo della similarità coseno, restituendo un valore compreso tra 0 e 1 che indica quanto due entità linguistiche siano correlate semanticamente.
A differenza dei metodi basati sui caratteri, la similarità semantica di spaCy riconosce che 'cane' e 'cucciolo' sono simili anche se non condividono nessuna lettera in comune.

Questa funzione però, puo' essere applicata a diversi livelli di astrazione.

Metodologie di Calcolo
Confrontare Token e Documenti.
* Token Similarity: confronto diretto tra vettori di due singole parole per trovare sinonimi o correlati
* Doc Similarity: calcolata mediando i vettori di tutti i token contenuti nel documento, fornendo un 'centroide' semantico del testo.Per confrontare due interi documenti. Il centrodide è la media delle media di ogni singolo vettore.
* Indipendenza dalla lunghezza: la similarità coseno si focalizza sulla direzione tra vettori piuttosto che sulla loro magnitudo assoluta
* La formula esprime il coseno dell'angolo tra due vettori, normalizzando il prodotto scalare per le loro norme.

Interpretazione dei Punteggi
Soglie di confidenza: Un punteggio di 1.0 indica identità, mentre valori sopra lo 0.7 indicano solitamente forte correlazione semantica nel linguaggio naturale.
Similarità tra Doc e Token: in sistemi di ricerca, è prassi comune pre-calcolare i vettori 'md' e memorizzarli in un database vettoriale per interrogazioni ultra-rapide
Limitazioni del contesto: poichè questi sono vettori statici, la parola 'banca' avrà lo stesso vettore sia in un contesto finanziario che fluviale, mediando i significati, questo limite di supera con i Transformer.

Vediamo più da vicino come spaCy calcola la media di un testo

Medie Vettoriali e Centroidi
Rappresentare insiemi di parole.
Quando spaCy confronta due documenti, esegue internamente una media dei vettori dei token. Questo vettore 'medio' rappresenta il baricentro semantico del documento nello spazio multidimensionale.
Dati n vettori, il vettore risultante per il documento è la media aritmetica dei singoli componenti.
Se un documento parla per il suo 90% di tecnologia, il suo centroide sarà trascinato verso quella galassia, anche se aggiungiamo una parola fuori contesto.

Calcolare media e coseni sempra faticoso, in realtà è abbastanza immediato

Efficienza e Performance
Perchè spaCy è la scelta industriale
In ambito professionale, la velcoità di calcolo è critica quanto la precisione. spaCy è progettato con un backend in Cython (linguaggio che traduce il codice python in C), il che gli permette di eseguire operazioni su vettori quasi alla velocità del linguaggio C pur mantenendo l'interfaccia Python
Questa ottimizzazione consente di processare migliaia di documenti al secondo, rendendo spaCy lo strumento ideale per pipeline di produzione in tempo reale.

E se avessimo milioni di documeti da analizzare?

Uso in Produzione
in produzione usiamo il metodo nlp.pipe, permete di parallelizzare l'estrazione dei vettori su più core della CPU massimizzando il throughtput del sistema. Invece di processare un testo alla volta, mandiamo i dati in gruppi, spaCy sfrutta il multitrading della CPU per lavorare su più documenti in contemporanea
In sistemi di ricerca moderno, è prassi comune pre-addestrare i vettori 'md' e memorizzarli in un database vettoriale per interrogazioni ultra-rapice (pre-computazione). Così la ricerca diventa un semplice confronta tra vettori.
Scegliere il modello 'md' rispetto 'lg' può ridurre l'occupazione di memoria della metà con una perdita di precisione spesso trascurabile per compiti standard.

Scalabilità del Calcolo
Tutto si traduce, come sempre, al prodotto scalare.
SpaCy utilizza librerie di basso livello, sono algoritmi scritti decenni fa ed ottimizzati per ogni tipo di processore esistente. E' questa forza bruta, unita all'eleganza linguistica, che lo rende uno strumento imabttibile.

Per l'italiano puoi scegliere tra:
it_core_news_sm → modello piccolo (non ha tabella completa di static word vectors)
it_core_news_md → modello medio + vectors
it_core_news_lg → modello grande + vectors
Per installarli usa per esempio: python -m spacy download it_core_news_md

```python
# python -m spacy download it_core_news_lg
"""
=============================================================================
         ANALISI DOCUMENTALE AVANZATA: IL BARICENTRO DEL SIGNIFICATO
=============================================================================

Questo script è un framework per la gestione e il confronto di interi DOCUMENTI (oggetti `spacy.tokens.Doc`)
invece di singole parole. Nel machine learning moderno, questo approccio è alla base 
dei sistemi RAG (Retrieval-Augmented Generation) e della Document Intelligence.

ARCHITETTURA DEL CODICE:
1. LOGICA DI ESTRAZIONE: La classe `Doc` interagisce con i suoi `Token` per estrarre vettori densi 300D.
2. FILTRAGGIO SEMANTICO: Pulizia del rumore (stop-words/punteggiatura) per isolare il significato puro.
3. MOTORE DI RICERCA: Interazione tra `nlp.pipe` (efficienza) e `scipy.spatial` (geometria).
4. ANALISI DEI LIMITI: Dimostrazione di come i vettori statici (Word2Vec/GloVe) falliscono nel contesto.

"""

import spacy                # Core NLP: gestisce la pipeline di analisi e i modelli linguistici
import numpy as np          # Matematica vettoriale: usato per aggregare i vettori (medie e baricentri)
from scipy import spatial   # Geometria spaziale: fornisce l'algoritmo per calcolare la distanza coseno

def get_cleaned_vector(doc):
    """
    Calcola il vettore rappresentativo di un documento (Doc) pulendolo dal 'rumore'.
    
    INTERAZIONE FRA CLASSI:
    - Riceve un oggetto `Doc` (un'intera frase o paragrafo).
    - Itera su ogni oggetto `Token` all'interno del `Doc`.
    - Accede alle proprietà booleane del Token (`is_stop`, `is_punct`) per decidere cosa scartare.
    - Estrae l'attributo `token.vector` (un array NumPy 300D) dai termini validi.
    """
    
    # List Comprehension: Filtriamo i token in base a criteri di rilevanza semantica
    # 1. token.has_vector: Verifica se il modello ha un embedding associato (evita OOV - Out Of Vocabulary)
    # 2. not token.is_stop: Esclude congiunzioni, articoli, preposizioni (rumore statistico)
    # 3. not token.is_punct: Esclude virgole, punti, ecc. che non hanno peso semantico
    vectors = [token.vector for token in doc if token.has_vector and not token.is_stop and not token.is_punct]
    
    if not vectors:
        # Edge Case: Se il testo è vuoto o fatto solo di stop-words (es. "che il lo"), 
        # restituiamo un vettore nullo della dimensione corretta (300 per il modello _lg)
        return np.zeros((300,))
    
    # OPERAZIONE MATEMATICA: Calcolo del Centroide
    # Sommiamo tutti i vettori e dividiamo per il numero di parole rilevanti.
    # axis=0 assicura che la media venga fatta componente per componente (300 dimensioni)
    return np.mean(vectors, axis=0)

def motore_di_ricerca_semantica(nlp, query, documenti):
    """
    Esegue una ricerca basata sul concetto (vettore), non sulla coincidenza di caratteri.
    
    PARAMETRI:
    - nlp: L'oggetto Language caricato (es. it_core_news_lg).
    - query: La stringa cercata dall'utente.
    - documenti: Una lista di stringhe che compongono il nostro database di conoscenza.
    """
    print(f"\n--- [LOG]: AVVIO RICERCA SEMANTICA ---")
    print(f"Query Digitata: '{query}'")
    
    # 1. TRASFORMAZIONE QUERY: Convertiamo la query in un vettore pulito
    # nlp(query) crea un oggetto Doc, che passiamo alla nostra funzione di pulizia
    query_vector = get_cleaned_vector(nlp(query))
    
    risultati = []
    
    # 2. OTTIMIZZAZIONE SCALABILE: nlp.pipe
    # Invece di chiamare nlp(doc) in un loop (lento), usiamo nlp.pipe:
    # - Elabora i testi in batch (gruppi).
    # - Sfrutta il multithreading e le ottimizzazioni C di spaCy.
    # - Restituisce un iteratore di oggetti Doc.
    for doc in nlp.pipe(documenti):
        # Per ogni documento nel database, calcoliamo il suo baricentro semantico
        doc_vector = get_cleaned_vector(doc)
        
        # 3. CALCOLO SIMILARITÀ (1 - Distanza Coseno)
        # La distanza coseno misura l'angolo tra i due vettori nello spazio 300D.
        # Se l'angolo è 0, i vettori sono sovrapposti (uguaglianza semantica).
        # Convertiamo la distanza (0 a 2) in uno score (0 a 1) dove 1 è il massimo.
        score = 1 - spatial.distance.cosine(query_vector, doc_vector)
        risultati.append((doc.text, score))
    
    # 4. RANKING: Ordiniamo i risultati dal più simile al meno simile
    risultati.sort(key=lambda x: x[1], reverse=True)
    
    # Output dei risultati formattati
    for testo, score in risultati:
        status = "MATCH" if score > 0.6 else "RELEVANT" if score > 0.4 else "LOW"
        print(f"[{score:.4f} - {status}] {testo[:70]}...")

def analisi_contrasto_documentale(nlp, doc_a, doc_b):
    """
    Confronta due documenti evidenziando la differenza tra approccio standard e raffinato.
    """
    # .similarity() è il metodo integrato di spaCy: fa la media Semplice di TUTTE le parole
    similarity = doc_a.similarity(doc_b)
    
    print(f"\n--- [DEBUG]: ANALYSIS COMPARISON ---")
    print(f"A: '{doc_a.text[:40]}...'")
    print(f"B: '{doc_b.text[:40]}...'")
    
    # Mostriamo il vantaggio del filtraggio stop-words
    vec_a = get_cleaned_vector(doc_a)
    vec_b = get_cleaned_vector(doc_b)
    clean_sim = 1 - spatial.distance.cosine(vec_a, vec_b)
    
    print(f"Similarità Standard (Include rumore): {similarity:.4f}")
    print(f"Similarità Raffinata (Focus concetti): {clean_sim:.4f}")

def main():
    # SETUP: Carichiamo il modello Large. Contiene 500.000+ vettori unici.
    # È indispensabile per analisi di similarità; il modello 'sm' non ha vettori reali.
    try:
        nlp = spacy.load("it_core_news_lg")
    except OSError:
        print("Modello non trovato. Eseguire: python -m spacy download it_core_news_lg")
        return

    # DATABASE: Il nostro piccolo 'Knowledge Graph' testuale
    conoscenze = [
        "Il motore a scoppio utilizza combustibili fossili per generare energia meccanica.",
        "Le energie rinnovabili come il solare e l'eolico sono il futuro del pianeta.",
        "La cucina mediterranea si basa su olio d'oliva, verdure fresche e cereali.",
        "Il machine learning permette ai computer di apprendere dai dati senza programmazione esplicita.",
        "Investire in borsa richiede una profonda conoscenza dei mercati finanziari."
    ]

    # TEST 1: Ricerca di un concetto non presente testualmente
    # Cerchiamo "veicoli a motore". Notate come troverà il documento sui motori a scoppio
    # anche se sono parole diverse: questo è il potere del Word Embedding.
    motore_di_ricerca_semantica(nlp, "veicoli a motore", conoscenze)
    
    # TEST 2: Concetti astratti
    motore_di_ricerca_semantica(nlp, "intelligenza artificiale", conoscenze)

    # TEST 3: Limiti dei vettori statici (Il problema "NON")
    # I Word Embeddings classici sono una 'zuppa di parole': l'ordine non conta.
    doc_neg = nlp("Il film non era affatto bello, mi sono annoiato.")
    doc_pos = nlp("Il film era molto bello, non mi sono annoiato.")
    
    print("\n--- [WARNING]: IL LIMITE DEI VETTORI STATICI ---")
    print(f"Sim. Semantica: {doc_neg.similarity(doc_pos):.4f}")
    print("ANALISI: Poiché contengono quasi le stesse parole, il baricentro è quasi identico.")
    print("SOLUZIONE: Per distinguere 'Bello' da 'Non Bello' servono i Transformer (BERT/GPT).")

    doc1 = nlp("Il sistema di propulsione richiede carburante.")
    doc2 = nlp("Un motore ha bisogno di benzina per funzionare.")
    
    # Ora la chiamiamo davvero!
    analisi_contrasto_documentale(nlp, doc1, doc2)

if __name__ == "__main__":
    main()

# =============================================================================
#  NOTE TECNICHE PER LO STUDENTE
# =============================================================================
# 1. IL DOC (DOCUMENTO) COME VETTORE:
#    In spaCy, doc.vector è la media aritmetica dei vettori dei token componenti.
#    È come se prendessimo le coordinate GPS di 10 persone in una piazza e 
#    calcolassimo il punto medio: quel punto rappresenta la "posizione del gruppo".
#
# 2. PERCHÉ PULIRE LE STOP-WORDS?
#    Le parole come "il", "di", "per" sono ovunque. Matematicamente, hanno vettori
#    che 'tirano' ogni frase verso il centro dello spazio vettoriale. 
#    Togliendole, permettiamo alle parole chiave (es. "motore", "cucina") di 
#    esprimere la loro vera direzione senza interferenze.
#
# 3. PERFORMANCE CON nlp.pipe:
#    In produzione, non iterare mai manualmente con un for se hai più di 100 testi.
#    nlp.pipe è thread-safe e scritto in Cython. Gestisce la memoria in modo 
#    estremamente efficiente, evitando l'overhead del garbage collector di Python.
# =============================================================================
```
