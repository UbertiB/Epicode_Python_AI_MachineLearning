POS TAGGING E PARSING SINTATTICO

Il POS tagging significa Part-of-Speech tagging: assegnare a ogni token la sua categoria grammaticale. In spaCy usa le categorie universali di Universal Dependencies come NOUN, VERB, ADJ, ADV, PROPN, DET, PRON, ADP, AUD, ecc
Prediamo la frase: "il cliente ordina il prodotto."
il -> DET
cliente -> NOUn
ordina -> VERB
il -> NOUNT
prodotto -> PUNTCT
Quindi POS risponde alla domanda:
"che tipo di parola è questo token?"
non ti dice ancora quale funzione abbiam quella proale nella frase.
Per esempio:
cliente -> NOUN
prodotto -> NOUN
Sono entrambi sostantivi. Ma nella frase hanno ruoli sintattici differenti: uno è il soggetto e l'altro è l'oggetto.
Ed è qui che il parsing sintattico entra in gioco.

Il dependency parsing cerca di ricostruire le relazioni grammaticali fra le paroel. spaCy rappresenta queste relazioni tramite un albero di dipendenze: ogni token, salvo la radice, dipende da un altro token chiamato head, e possiede una relazione di dipendenza accessibile con token.dep
Prendiamo ancora la frase: "il cliente ordina il prodotto"
una rappresentazioe potrebbe essere:
        ordina
            |
    cliente prodotto
        |       |
        il      il
Qui "oridna" è il centro sintattico principale della frase.
Possiamo leggerlo così:
ordina -> ROOT
cliente -> soggetto di 'ordina'
prodotto -> oggetto di 'ordina'
il -> determinante di 'cliente'
il -> determinante di 'prodotto'
Quindi la differenza fondamentale è:
POST TAGGING
"che COSA è questa parola?"
cliente -> NOUN
ordina -> VERB
mentre:
DEPENDENCY PARSING
"che RUOLO ha questa parola e da quala parola dipende?"
cliente -> soggetto di ordina
prodotto -> oggetto di ordina
Universal Dependencies definisce proprio un insieme standarizzato di POS tag, caratteristiche morfologiche e relazioni sintattiche per rappresentare la grammatica tra lingue diverse.

L'etichettatura grammaticale
Part-of-Speech-Taggin
immagina le parole come mattoncini lego, alcuni sono pezzi importanti altri no. 
Il Part-of-Speech(POS) taggin è il processo di assegnazione di una categoria grammaticale a ogni parola in un testo. Non si tratta solo di distinguere un nome da un verbo, ma di comprendere il ruolo morfologico del termine basandosi sia sulla sua forma che sul contesto circostante.
I tagger moderni utilizzano schemi standarizzati come le Universal POS tags (UPOS), che permettono di avere un set di etichette coerente tra lingue diverse, facilitando lo sviluppo di pipeline multilingue robuste.

Per standarizzare questo processo a livello universale, usiamo set di etichette universali.
Dalle etichette universali a quelle specifiche
- Universal POS (UPOS): set semplificato di 17 tag (es. NOUN, VERB, ADJ) per una classificazione generale
- XPOS: tagset specifici della lingua che includono dettagli morfologici come genere, numero e tempo verbale.
- Amniguità: una stessa parola può cambiare catagoria (es. 'canto' come nome o verbo) a seconda delle posizione
- Il modello calcola la probabilità di una sequenza di tag T data una sequenza di parole W massimizzando la verosomiglianza condizionata.

Come come fa spaCy a raggiungere una precisione così alta in una frazione di secondo?

Modellazione Morfologica
L'analisi dei token adiacente è cruciale per risolvere le omonimie e determinare correttamente se un termine funge da aggettivo o sostantivo (contesto locale)
I modelli moderni analizzano prefissi e suffissi per dedurre la categoria di parole mai viste prima (nel training set, Out-of-Vocabulary).
SpaCy implementa tagger basati su reti neurali convoluzoinali estremamente veloci che garantiscono un'accuratezza superiore al 95% su testi standard.

Ma sapere che è un nome e che è un verbo non basta, dobbiamo capire come interagiscono
Parsing Sintattico delle Dipendenze.
Mappare le relazioni tra testa e dipendente.
Il parsing delle dipendenze analizza come le parole si legano tra loro per formare il significato. A differenza del POS tagging che guarda la singola parola, il parsing identifica chi compie l'azione (soggetto), su cosa ricade (oggetto) e quali sono i modificatori.
Ogni relazione è asimmetrica: esiste una parola 'testa' (head) e una parola 'figlio' (child). Questa struttura permette di visualizzare la frase come un albero gerarchico dove il verbo principale funge solitamente da radice.
Ogni parola è legata ad una testa, è una gerarchia. questo sturttura gerarchica visualizzato con un albero

Albero delle Dipendenze
- Root: si trova al centro di tutto, è il nodo centrale dell'albero. Solitamente il verbo reggente della proposizione principale. Il motore che genera l'azione, da li partono le freccie.
- nsubj/dobj: etichette comuni che identificano rispettivamente il soggetto nominale e l'oggetto diretto.
- Archi direzionati: le frecce indicano la direzione della dipendenza dalla testa verso il dipendnete.
- Un albero di dipendenza valido deve essere un grafi diretto aciclico dove ogni nodo ha esattamente un genitore tranne la radice.
Se una parola avesse due genitori, la logica della frease crolla.
Questa struttura permette alle macchine di rispondere a chi ha fatto cosa?

Il parsing ci permette quindi di risolvere problemi che la semplice vicinanza di parole non può gestire.

Una delle capacità del parsing è la capacità di gestire le relazioni a lunga distanza.
Un modello basato solo sulla vicinanza fallirebbe.
Il persing permette di collegare parole distanti nel testo, ma legate logicamente, superando i limiti delle semplici finestre di contesto.
Capire a quale sostantivo si riferisce un aggettivo in una frase complessa è possibile solo analizzando l'albero delle dipendenze.
Alcune lingue permettono incroci tra gli arhi delle dipendenze, una sfida computazionale che i parser moderni gestiscono tramite algoritmi su transizioni.

Ma come fa un computer a costruire questo albero mentre legge la frase?
Il segreto è Transition-based Parsing
Immagina un magazziniere con uno scaffale (lo stack) ed un nastro trasportatore (buffer), il magaziniere prende una parola dal nastro (stack) e la mette sullo scaffale e decide: queste due parole si incastrano? se si crea un arco altrimenti continua a spostare i pezzi.
 
Estrarre valore dalla struttura sintattia
Il POS tagging e il parsing non sono fini a se stessi, ma sono strumenti per distillare informazioni. Invece di trattare tutte le parole come uguali, possiamo decidere di mantenere solo quelle che portano il maggior carico semantico.
E' possibile filtrare i nomi e gli aggettivi per creare riassutni automatici o come usare le dipendenze per estrarre triplette 'soggetto-verbo-oggetto' utili per popolare databse di conoscenza strutturata.

In un mondo sommerso dai dati, non abbiamo il tempo per leggere tutot, il pos taggin ed il parsing ci permettono di non analizzare ogni singola posizione ed articolo, possiamo istruire il modello per guardare solo i nomi ed i verbi, è come guardare solo i contorni e le figure principali di un quadro ignorando lo sfondo.

Tecniche specifiche per estrarre il cuore del messaggio
Tecniche di Estrazione
Isolare il cuore del messaggio
- Filtraggio per POS: selezione esclusiva di NOUN o PROPN per identificare i concetti principali trattati.
- Analisi dei Modificatori: utilizzare gli aggettivi (ADJ) legati a specifici nomi per l'analisi di sentiment granulare.
- Subtree Navigation: estrarre intere porzioni dell'albero sintattico per catturare sintagmi complessi invece di singole parole.
- La densità informativa può essere migliorata riducendo il testo originale alla sua struttura portante definita dai tag POS rilevanti.
Possiamo estrarre non solo singole parole ma interi concetti complessi come intelligenza artificiale moderna
Questo aumenta la densità informativa, riduciamo la lunghezza del testo ma aumentiamo il valore di ogni singolo token che passiamo alla rete neurale.

Applicazioni
Sistemi di domanda/risposta: identificare il soggetto e l'oggetto permette ai chatbot di capire esattamente 'chi ha fatto cosa' e rispondere con precisione. Questo permette di capire che "il cane morde l'uomo" è l'opposto di "l'uomo morde il cane" questo anche se le parole sono le stesse.
Analisi dell'opinione: Legando i verbi ai loro avverbi, possiamo determinare l'intensità di un'azione o di una critica in una recensione. leghiamo un aggettivo negativo esattamente al prodotto che viene giudicato.
Data Augmentation: il parsing permette di sostituire sinonimi mantenendo la correttezza grammaticale, generando nuovi campioni per addestrare modelli di Deep Learning.

Il traguardo finale è l'estrazione di triplette.
Una tripletta è l'unità base della conoscenza.
Soggetto - Predicato - Oggetto
Navigando l'albero sintattico, possiamo estrarre automaticamente queste relazioni, identificando il nodo radice (verbo) e i suoi figli etichettati come nubj e dobj. Costruire delle nodlege graph ovvero delle mappe del sapere dove i concetti sono collegati tra loro.
Questo approccio trasforma il testo libero in una forma relazionale standarizzata, base per la costruzione di grafi semantici complessi.
Questo trasforma la letteratura in un database esplorabile è così' che lavorano i motori di ricerca moderni per dare risposte dirette invece di semplici bit.


```python
"""
DESCRIZIONE: Guida pratica al Part-of-Speech (POS) Tagging e al Dependency Parsing.
             Questo script utilizza spaCy, lo standard industriale per l'NLP rapido e accurato,
             per mostrare come trasformare un testo "piatto" in una struttura gerarchica logica.
OBIETTIVO: Spiegare come le macchine identificano le categorie grammaticali e le relazioni
           tra le parole (Soggetto, Verbo, Oggetto) per "capire" il senso di una frase.
"""

import spacy       # Libreria principale per l'NLP (la nostra "fabbrica" di analisi)
import os          # Usato solo per scaricare il modello linguistico se manca

# --- FASE 1: PREPARAZIONE DEL MOTORE (Slide 5: Velocità Industriale) ---

def carica_modello_linguistico():
    """
    Carica il cervello linguistico di spaCy per l'italiano.
    
    Interazione: Il modello contiene pesi statistici e regole grammaticali 
    pre-addestrate su milioni di frasi per riconoscere i pattern della lingua.
    """
    try:
        # 'it_core_news_sm' è un modello piccolo (sm), veloce e ottimo per imparare.
        # È il nostro "detective" che analizzerà il contesto delle parole.
        return spacy.load("it_core_news_sm")
    except OSError:
        # Se il modello non è installato sul computer, lo scarichiamo al volo.
        print("Modello non trovato. Download in corso...")
        os.system("python -m spacy download it_core_news_sm")
        return spacy.load("it_core_news_sm")

# Inizializziamo il nostro "motore" nlp globally per usarlo nelle funzioni
nlp = carica_modello_linguistico()

# --- FASE 2: ANALISI MORFOLOGICA (Slide 4: Identità delle Parole) ---

def mostra_identita_parole(testo):
    """
    Esegue il POS Tagging: assegna un'etichetta grammaticale a ogni parola.
    
    Argomenti:
        testo (str): La frase che vogliamo analizzare.
    """
    # 1. Il motore 'nlp' elabora il testo e crea un oggetto 'doc' (documento).
    # Il 'doc' non è solo testo, è un contenitore di oggetti 'token' (parole) ricchi di dati.
    doc = nlp(testo)
    
    print(f"\n[ANALISI POS] Identificando i mattoncini della frase: '{testo}'")
    
    # Cicliamo su ogni parola (token) del documento
    for token in doc:
        # token.text: La parola originale
        # token.pos_ : L'etichetta UPOS (Universal Part-of-Speech) come VERB, NOUN, ADJ.
        # token.lemma_: La forma base (es: "mangiato" -> "mangiare"). Utile per normalizzare i dati.
        print(f"-> Parola: '{token.text:12}' | Tag POS: {token.pos_:8} | Base (Lemma): {token.lemma_}")

# --- FASE 3: ANALISI SINTATTICA (Slide 7-8: Gerarchia e Relazioni) ---

def mostra_relazioni_sintattiche(testo):
    """
    Esegue il Dependency Parsing: trova i legami "genitore-figlio" tra le parole.
    
    Argomenti:
        testo (str): La frase da mappare.
    """
    doc = nlp(testo)
    
    print(f"\n[ANALISI PARSING] Costruendo l'albero delle dipendenze:")
    
    for token in doc:
        # token.dep_ : Il tipo di relazione (es: nsubj = soggetto, obj = oggetto).
        # token.head : Il "genitore" sintattico della parola (la 'Testa' a cui è legata).
        # Nelle relazioni sintattiche, ogni parola punta a chi la governa.
        print(f"-> '{token.text:12}' è un {token.dep_:10} di --> '{token.head.text}'")

# --- FASE 4: ESTRAZIONE CONOSCENZA (Slide 14: Il Setaccio d'Oro) ---

def estrai_triplette_chiave(testo):
    """
    Naviga l'albero sintattico per isolare il cuore del messaggio: CHI FA COSA.
    Questo filtra il rumore (articoli, preposizioni) e tiene solo i concetti puri.
    """
    doc = nlp(testo)
    print(f"\n[SETACCIO D'ORO] Estraendo triplette di conoscenza (Soggetto, Azione, Oggetto):")
    
    for token in doc:
        # Identifichiamo il Verbo principale (che solitamente è la ROOT, la radice dell'albero)
        if token.pos_ == "VERB" or token.dep_ == "ROOT":
            
            # Cerchiamo tra i "figli" (children) del verbo chi ha il ruolo di soggetto e oggetto
            soggetto = ""
            oggetto = ""
            
            for child in token.children:
                if "subj" in child.dep_:    # Cerchiamo 'nsubj' (nominal subject)
                    soggetto = child.text
                if "obj" in child.dep_:     # Cerchiamo 'obj' (direct object)
                    oggetto = child.text
            
            # Se abbiamo trovato sia chi compie l'azione che chi la riceve, stampiamo la tripletta
            if soggetto and oggetto:
                print(f"   >>> Conoscenza trovata: [{soggetto}] --({token.lemma_})--> [{oggetto}]")

# --- ESECUZIONE DEL PROGRAMMA ---

if __name__ == "__main__":
    # Esempio 1: Una frase semplice che mostra la differenza di ruoli
    frase_1 = "Il programmatore scrive il codice."
    mostra_identita_parole(frase_1)
    mostra_relazioni_sintattiche(frase_1)
    estrai_triplette_chiave(frase_1)
    
    # Esempio 2: Una frase più complessa per vedere il parsing in azione (Slide 9: lunghe distanze)
    frase_2 = "Gli studenti di Deep Learning analizzano con attenzione i modelli linguistici."
    mostra_identita_parole(frase_2)
    mostra_relazioni_sintattiche(frase_2)
    estrai_triplette_chiave(frase_2)

    print("\n--- CONCLUSIONE DIDATTICA ---")
    print("1. Il POS Tagging ha identificato i ruoli (Nomi, Verbi).")
    print("2. Il Parsing ha collegato le parole tra loro creando una gerarchia.")
    print("3. Il Setaccio ha isolato i concetti chiave ignorando il 'rumore' grammaticale.")
```
