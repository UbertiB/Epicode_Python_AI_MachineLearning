La Luce
Dalla luce ai pixel
Immagina la luce come una pioggia di particelle, i nostri fotoni
Quando scattiamo una foto stiamo usanto un dispositivo chiamato CMOS o CCD, che converte la radiazione elettromagnetica in segnali elettrici; conversione di segnale analogico in un segnale digitale.
Questo processo trasforma una fenomeno continuo, la luce, in un dato discreto che il computer può processare.

Come avviene esattamente questa cattura riga per riga?

Elementi della cattura digitale
Dal fotone al bit

Il FOTODIODO è l'elemento sensibile che accumula carica elettrica proporzionalmente all'intensità luminosa ricevuta durante il tempo di esposizione.
Una fotocamera non vede oggetti, misura quanta luce arriva in ogni punto del sensore.
La luce è composta da fotoni, cioè quantità discrete di energia elettromagnetica.
Un oggetto viene visto perchè:una sorgente luminosa emette luce, la luce colpisce l'oggetto, una parte viene assorbita, una parte viene riflessa, i fotoni riflessi entrano nella fotocamera.
Il colore quindi, non è una proprietà digitale del pixel. E' il risultato della distribuzione delle lunghezze d'onda della luce ricevuta.

Prima di raggiungere il sensore, la luce attraversa l'obbiettivo.
L'OBBIETTIVO  raccoglie la luce, focalizza la luce, proietta la scena sul sensore, determina campo visivo e ingrandimento.

L'APERTURA determina quanta luce entra.
Un'apertura maggiore fa entrare più luce, riduce il tempo di esposizione necessario, diminuisce la profondità di campo
Un'apertura minore fa entrare meno luce, aumenta la profondità di campo, può richiedere tempi più lunghi.

In Computer Vision Industriale questo conta molto, Se il pezzo non è interamente a fuoco, il modello riceve informazioni peggiori.

Il TEMPO DI ESPOSIZIONE è il tempo durante il quale il sensore raccoglie luce
Tempo lungo: più fotoni, immagine più luminosa, maggiore rischio di movimento sfocato
Tempo breve: immagine più scura, minore sfocatura

Su un nastro trasportatore si preferisce un tempo breve, compensato da un'illuminazione forte.

il SENSORE normalmente di tipo CMOS (o meno frequente CCD) è composta da una griglia di elementi sensibili alla luce, chiamati fotositi
Ogni FOTOSITO raggoglie i fotoni che arrivano su una piccola area
Ogni fotosito contribuirà alla creazione di un pixel dell'immagine finale, anche se il processo reale può essere più complesso

Dal fotone all'elettrone
Quando un fotone colpisce il materiale fotosensibile del sensore, può generare un elettrone
Più fotoni arrivano, maggiore sarà normalmente la carica accumulata.

La RISOLUZIONE è la capacità di distinguere due punti vicini, è determinata dalla densità dei fotorecettori sul sensore. 
Nel Deep Learining vogliamo che il sensore catturi più sfrumature possibili. Se usiamo 8 bit avremmo solo di 256 livelli grigio, sembrano molti, ma in campo medico potrebbero non bastare.
Una profondità maggiore permette di rappresentare differenze più piccole di luminosità

Non significa automaticamente un'immagine migliore, è migliore se il sensore ha una qualità sufficiente, il rumre è contenuto, il sw mantiene la precisione.

Purtroppo però la fisica non è perfetta ed ogni processo di cattura introduce rumore. Nel Deep Learning comprendere il rapporto segnale-rumore è vitale per la robustezza degli algoritmi
Perchè ci interessa?, perchè se addestriamo un ai solo su immagini perfette, al primo accenno del mondo reale il sistema fallirà.
Immagine = Segnale puro + Rumore

DAL FOTOSITO MONOCROMATICO AL COLORE 
Un fotosit misura principalmente l'intensità luminosa. Non riconosce direttamente il colore.
Per ottenere un'immagine RGB si sua spesso un filtro di Bayer
Ogni fotosito misura quiondi una sola componente RGB (rosso, verde, blu)
Il FILTRO DI BAYER: filtro colore che permette ai sensori monocromatici di catturare informazioni cromatiche alternando pixel rossi, verdi, blu.
