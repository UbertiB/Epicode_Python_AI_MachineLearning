Come possiamo cambiare tutti i colori di una foto?

Immagina di voler tradurre un libro parola per parola. puoi usare un glossario, una LUT è esattamente questo, un glossario per ogni pixel. Invece di calcolare una funzione matematica per ognuno di milioni di pixel nella nostra immagine, calcoliamo il risultato solo per i 256 valori possibili e li mettiamo in una tabella (Look-Up Table o LUT)
La LUT è un array di indicizzazione che permette di trasformare i valori dei pixel senza ricalcolare complesse funzioni matematiche per ogni punto.
Invece di applicare una formula a milioni di pixe, leggiamo semplicemente il valore pre-calcolato nella tabella, riducendo drasticamente il carico computazionale.
Quando dobbiamo processare un foto il computer non pensa, si limita a leggere la risposta corretta.
E' l'efficienza del estrema del video editing moderno.

Come questo glossario ci permette di scolpire le curve di luce.

Manipolazione delle Curve
Mapping lineare e non lineare
* L'indice della LUT corrisponde al valore di intensità originale (0-255), mentre il valore contenuto nell'array è il nuovo colore.
* Le curve di contrasto (come la S-curve) vengono create alterando i valori della LUT per schiarire le luci e scurire le ombre
* Possiamo applicare LUT diverse per ogni canale (RGB) per creare viraggi cromatici specifici (es. effetto Sepia o Teal and Orange)
* La funzione cv2.LUT di OpenCV è altamente ottimizzata per eseguire questa mappatura visiva su arrary NumPy
