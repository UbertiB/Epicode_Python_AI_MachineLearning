```python
import spacy

# Caricamento del modello per l'italiano
try:
    nlp = spacy.load("it_core_news_sm")
except OSError:
    import os
    os.system("python -m spacy download it_core_news_sm")
    nlp = spacy.load("it_core_news_sm")

def analizza_frase_medica(text: str):
    doc = nlp(text)
    
    print(f"Analisi della frase: '{text}'\n")
    print(f"{'Token':<12} | {'Lemma (spaCy)':<15} | {'POS Tag'}")
    print("-" * 45)
    
    for token in doc:
        if not token.is_punct:
            print(f"{token.text:<12} | {token.lemma_:<15} | {token.pos_}")

# Esecuzione
frase_medica = "I medici curano i pazienti con cura"
analizza_frase_medica(frase_medica)
```
