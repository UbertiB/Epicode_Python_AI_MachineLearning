COS'E' IL NLP E DIFFERENZA CON NLU/NLG

Isegnamo al nostro computer a comunicare

NLP è il contenitore generale
NLU e NLG sono due grandi aree al suo interno

* NLP: Natural Languade Processing (elaborazione del linguaggio naturale)
E' l'insieme delle tecniche che permettono a un computer di lavorare con il linguaggio umano: testo, email, documenti, chat, recensioni, frasi.
Quindi NLP è il termine più ampio, dentro NLP trovi problemi molto diversi, per esempio: classificazione di testo, sentiment analysis, traduzione, estrazione informazioni, riassunto, riconoscimenti di entità, questin answering, generazione di testo.
Se prendiamo una mail, un sistema NLP può lavorare su quel testo, ma può fare due cose profondamente diverse: 1) capire cosa c'è scritto 2) produrre testo nuovo

Il NLP non è un blocco unico, ma un ecosistema che vive di due processi speculari. Il primo mira a decodificare l'intento umano partendo da un segnale testuale grezzo, mentre il secondo trasforma concetti astratti in frasi grammaticalmente corrette e naturali
Da qui nascono NLU e NLG
Questi due processi convivono in quasi tutti i sistemi moderni, come gli assistenti vocali, dove la macchina deve prima capire la richiesta e poi formulare una risposta coerente.

* NLU: Natural Language Understranding (cerca di comprendere il significato del linguaggio)
Capacità di capire
Input: vorremmo ordinare 500 pezzi dell'articolo A
Output: ordine cliente articolo A quantità 500. Deve capire che 500 è la quantità mentra A è l'articolo. Questo è NLU
Alcuni compiti tipici sono: intent classification, entity recognition, sentiment analysis, semantic analysis, information extraction
Esempio: "vorrei annullare l'ordine 123"
intent: annullamento_ordine
numero_ordine 123

Oppure: "la merce ricevuta è danneggiata"
intent: reclamo
problema: merce_danneggiata

* NLG: Natural Language Generation
Capacità di creare
fa praticamente l'opposto del NLU 
parti da informazioni: cliente: Rossi, Ordine: 123, stato: pronto, data_spedizione: 12/08
e vuoi produrre:
Buongiorno, l'ordine 123 è pronto e verrà spedito il 12/08

Quindi
NLU: linguaggio -> informazione strutturata
NLG: informazione strutturata -> linguaggio

Ma come interagiscono questi due processi all'interno di un sistema AI?

NLU, NLG e il loop NLP
Input, elaborazione e output
- Natural Language Understanding (NLU): si occupa della disambiguazione dell'estrazione di significato, trasformando il testo in una rappresentazione strutturata. Prende un testo disordinato e cerca di estrarne una struttura logica, un mapping semantico
- Natural Language Generation (NLG): è il processo di sintesi che parte da dati o conoscenze per produrre enunciati in lingua naturale. E' come un romanziere, riceve i dati grezzi e deve pianificare come raccontarli scegliendo le parole giusto e l'ordine corretto.
- Mapping Semantico: il compito principale della NLU è mappare stringhe di caratteri in concetti logici comprensibili per un software
- Pianificazione del Testo: la NLG deve decidere l'ordine delle informazioni e la scelta lessicale puù appropriata per il contesto.

Senza NLU la macchina sarebbe come un sordo che urla
Senza NLG un saggio muto.

Il vero segresto di un sistema NLP moderno è l'equilibrio tra questi due mondi.

La Natura Probabilistica del Linguaggio
Prevedere la sequenza ottimale
Il linguaggio, per un computer, è una sequenza di scommesse. 
Sia nella comprensione che nelle generazione, il computer ragiona in termini di probabilità. La generazione di una frase può essere vista come la massimizzazione della probabilità di una sequenza di parole dato un contesto iniziale.
La modellazione linguistica cerca di calcolare quanto sia verosimile che una determinata parola segua una sequenza data, basandosi su enormi dataset testuali.
L'ai moderna non sa cosa sia un gatto nel senso filosofica, ma sa calcolare la probabilità che due parole stiano insieme.

I Livelli del Linguaggio
Dall'unità minima al significato complesso
Per elaborare il linguaggio, dobbiamo scomporlo in strati gerarchici. Ogni strato aggiunge un pezzo di informazione: dai suoni e i simboli alla struttura delle parole, fino ad arrivare alle relazioni sintattiche e al significato finale dell'enuciato
Analizzare il linguaggio significa navigare attravrso questi livelli per risolvere ambiguità che per noi umani sono istintive ma per una macchina rappresentano sfide enormi.

Analizziamo questa gerarchia partendo dalla base della piramide

Morfologia, Sintassi e Semantica
La piramide linguistica
- Livello Morfologico: studia la struttura interna delle parole e come queste cambiano per esprimere genere, numero o tempo. Guarda dentro la parola, cantavamo descrive chi e cosa
- Livello Sintattico: analizza come le parole si combinano per formare frasi coerenti seguendo regole grammaticali. Decide l'ordine delle parole e controlla che soggetto e verbo vadano d'accordo
- Livello Semantico: si occupa del significato letterale delle parole e delle loro combinazione all'interno di un contesto. E' il cuore, cosa significa davvero quella frase?
- Livello Pragmatico: analizza come il contesto extra-linguistico influenza l'interpretazione del messaggio (es. ironia). E' il livello più difficile per l'ai

Il problema principale che riscontriamo in questi livelli è l'Ambiguità

Risoluzione delle Ambiguità
- Ambiguità Lessicale: una parola può avere più significati (polisemia). Il sistema deve decidere quale lemma è corretto basandosi sulle parole vicine.
- Ambiguità Strutturale: frasi con la stessa sequenza di parole possono avere alberi sintattici diversi portando a interpretazioni logiche totalmente differenti
- Entità Nominate: riconoscere che 'Apple' in  una frase è un'azienda e non un frutto è un compito che integra morfologia e semantica contestuale.

Ma come misuriamo quanto incertezza c'è in una frase?

Incertezza e Informazione
Misurare la complessità del contesto
L'entropia (è una misura dell'incertezza) di una sorgente linguistica ci dice quanta incertezza è presente nella scelta della parola successiva. 
Un livello sementico efficace riduce questa incertezza restringendo li campo delle interpretazioni possibili.
L'analisi della struttura sintattica permette di ridurre lo spazio di ricerca delle interpretazioni corrette applicando vincoli grammaticali formali.
Se dico "vado al" c'è molta entropia perchè le opzioni sono infinite
Se aggiungo "cinema" l'entropia crolla perchè il contesto ha ristretto le 
Un buon sistema lingiustico sa ridurre l'entropia usando la sintassi come guida.
Più informazioni abbiamo dai livelli inferiori, più è facile prevedere il significato ai livelli superiori.

Vediamo il passato

Evoluzione dei Modelli
Dalle regole scritte a mano ai dati
La storia del NLP è segnata da un cambiamento di rotta radicale. All'inizio, i ricercatori cercavano di codificare ogni regola grammaticale manualmente. Oggi, lasciamo che siano i modelli a estrarre queste regole osservando miliardi di frasi reali.
Questa transizione ha permesso di passare da sistemi fragili e limitati a modelli capaci di gestire la variabilità infinita e creativa del linguaggio umano naturale.
Il linguaggio è troppo vivo e troppo creativo per essere ingabbiato in regole rigide.
La svolta è arrivata quando abbiamo smesso di dare ordine ed abbiamo iniziato a dare esempi.

Vediamo questo scontro tra i due paradigmi, regole contro dati

Sistemi Simbolici vs Statistici
Il tramonto degli approcci deterministici.
- Rule-based NLP: sistemi basati su istruzioni if-then e grammatiche formali rigide, accurati ma non scalabili
- Corpus Linguistics: lo studio del linguaggio basato su grandi collezioni di testi reali (corpora) per trovare pattern ricorrenti. Invece di una regola, usiamo la frequenza, se in un miliardi di frasi apple è seguita da iphone il computer impara questa associazione, non perchè lo abbiamo detto noi, ma perchè l'ha visto accadere
- Machine Learning Statistico: modelli che imparano a classificare o generare testo basandosi sulla frequenza di occorrenza nei dati
- Deep Learning: l'uso di reti neurali per creare rappresentazioni continue del linguaggio, superando i limiti del conteggio discreto.

Ma quali sono stati i primi tentativi famosi

La Rivoluzione dei Dati
- Il Fallimento di ELIZA: i primi chatbot (anni 60) simulavano la comprensione tramite semplici sostituzioni di pattern, ma non possedevano alcuna rappresentazione del significato reale. Riprendeva le tue parole e te le girava sotto forma di domanda. Non capiva nulla
- Modelli Markoviani: hanno introdotto l'idea che lo stato di una parola dipenda solo da quella precedente, gettando le basi per i suggeritori di testo moderni. E' il nonno del correttore automatico sullo smartphone. E' semplice ma ha aperto le porte alla generalizzazione.
- Generalizzazione: a differenza delle regole, i modelli statistici possono gestire parole mai viste o errori ortografici trovando la corrispondenza più vicina nello spazio dei dati

Inferenza Statistica nel Testo
Classificazione Bayesiana
Uno dei primi successi dei modelli linguistici è stato il filtro antispam. Utilizzando il teorema Bayes, il sistema non legge più la mail per capirla, ma per calcolare le probabilità, per decidere se una mail è spam calcolando la probabilità delle parole contenute data la classe. Se la parola gratis o vinci compare troppo spesso, la proabiltà che sia spam aumenta.
Questo approccio segna l'inizio dell'era in cui il contesto globale del documento prevale sulla singola regola grammaticale locale.


```python
# --- 1. MOTORE NLU POTENZIATO (Natural Language Understanding) ---
def simple_nlu_engine(text: str):
    """
    Analizza il testo e calcola un punteggio di confidenza per ogni intento.
    Teoria: Passiamo da un matching booleano a un sistema pesato per gestire l'ambiguità.
    """
    # Pre-processing: normalizzazione del testo
    normalized_text = text.lower().strip()
    
    # Mappatura Intenti/Parole Chiave
    intents_map = {
        "meteo": ["tempo", "piove", "sole", "gradi", "previsioni", "caldo", "freddo"],
        "saluto": ["ciao", "buongiorno", "salve", "ehi", "sveglia"],
        "domotica": ["accendi", "spegni", "luce", "lampada", "clima", "casa", "dispositivo"]
    }
    
    # Inizializziamo il dizionario dei punteggi (Confidence Scores)
    scores = {intent: 0 for intent in intents_map.keys()}
    
    # Calcolo delle occorrenze
    words_in_text = normalized_text.split()
    for intent, keywords in intents_map.items():
        for word in words_in_text:
            if word in keywords:
                scores[intent] += 1
                
    # Determiniamo l'intento con il punteggio massimo
    best_intent = "sconosciuto"
    max_score = 0
    
    for intent, score in scores.items():
        if score > max_score:
            max_score = score
            best_intent = intent
            
    # Log diagnostico della confidenza (Best Practice)
    if max_score > 0:
        print(f"[NLU LOG] Score distribuzione: {scores}")
        
    return best_intent

# --- 2. MODULO NLG (Natural Language Generation) ---
responses = {
    "saluto": "Bentornato! Come posso aiutarti oggi?",
    "meteo": "Al momento ci sono 22°C e il cielo è sereno.",
    "domotica": "Comando ricevuto: sto agendo sui dispositivi della casa.",
    "sconosciuto": "Scusami, non ho abbastanza informazioni per agire. Puoi essere più specifico?"
}

def generate_response(intent: str):
    """Mappa l'intento alla risposta pragmatica più coerente."""
    return responses.get(intent, responses["sconosciuto"])


# --- TEST DELLA NUOVA LOGICA (Scenario Ambiguità) ---
# Caso 1: Comando senza verbo (ora funzionante grazie al punteggio)
test_input_1 = "La luce della cucina"
# Caso 2: Comando misto (prevale la domotica per numero di keyword)
test_input_2 = "Ciao, per favore accendi la luce e il clima"

for user_input in [test_input_1, test_input_2]:
    intent_detected = simple_nlu_engine(user_input)
    final_response = generate_response(intent_detected)

    print(f"Input Utente: '{user_input}'")
    print(f"Intento Rilevato: {intent_detected.upper()}")
    print(f"Risposta AI: {final_response}")
    print("-" * 50)
```
