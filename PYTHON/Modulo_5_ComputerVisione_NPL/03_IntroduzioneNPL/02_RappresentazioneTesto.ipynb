RAPPRESENTAZIONE DEL TESTO: STRINGHE VS NUMERI

Passaggio dalle stringhe ai numeri

Un modello non capisce direttamente stringhe. Prima il testo deve essere trasformato in numeri
Pensa ad una frase:
"il cliente ordina 100 pezzi!
Per noi è una strniga, per un modello, così com'è, non è ancora utilizzabile.
Il passaggio è:
testo -> token -> numeri -> modello
Un modello di Machine Learning lavora con valori numerici o meglio con vettori di numeri

La domanda è: come trasformo le parole in numeri?
La frase : "il cliente ordina 100 pezzi"   STRINGA
può diventare:
["il","cliente","ordina","100","pezzi"] TOKEN
poi
["15,87,146,532,64] qui i numeri sono solo ID (15=il, 87=cliente,...)
Il numero in questa fase descrive l'elemento ma non descrive il significato
La seconda trasformazione invece ogni singolo ID viene trasformato in un vettore con significato
quindi la frase potrebbe diventare
[[0.21,-0.40,0.78,...],[0.8,-0.3,0.41,...]...[...]] EMBEDDING
Questi vettori possono risultare relativamente vicini perchè i concetti hanno qualche relazione semantica.
stringa -> token -> ID -> embedding

Perchè una macchina non può semplicemente leggere come noi?

Le reti neurali sono motori matematici non letterali
Immagina la rete come un'immensa calcolatrice che sa solo sommare e moltiplicare. 
Dall'astrazione del simnbolo alla concretezza del segnale.
Le reti neurali sono motori matematici basati sull'algebra lineare e sul calcolo infinitesimale. Le operazioni fondamentali come la moltiplicazione di matrici richiedono che i dati di input siano tensori numerici. Una lettera o una parola, per un computer, è un simbolo discreto privo di valore intrinseco finchè non viene mappato in un dominio numerico.
Mentre l'occhio umano percepisce la forma di una lettera, il neurone artificiale percepisce solo la magnitudo di un segnale. Senza conversione numerica, non potremmo applicare l'algoritmo di retropropazione per aggiornare i pesi del modello.

Ma qual'è la bariera tecnica che separa l'alfabeto dai numeri?
La bariera è la differenziabilità
Per addestrare una rete usiamo l'errore che corregge i pesi tramite il calcolo infinitesimale.
Immagine di dover calcolare quanto sia sbagliata la parola cane rispetto a gatto, senza numeri non abbiamo una metrica, non possiamo dire che due parole sono vicine o lontate, abbiamo di trasformare il sengnale in  un valore continuo
Questo processo avviene all'interno del neurone artificiale. (riceve dei input, li moltiplica per dei pesi e sommma tutto)

Simboli discreti vs Segnali continui
La barriera tra linguaggio e computazione
- Incompatibilità algebrica: non è possibile calcolare il gradiente di una stringa di testo rispetto a un errore
- Assenza di metrica: nella loro forma raw, non esiste una distanza matematica definita tra la stringa "cane" e la stringa "gatto"
- Differentiabilità: l'addestramento richiede funzioni differenziabili applicabili solo a variabili numeriche continue
- Input Layer: il primo strato di ogni rete NLP deve fungere da convertitore che traduce token in vettori.

Una volta trasformati in numeri, i dati testuali devono essere scalati per evitare che valori troppo grandi destabilizzino la rete durante il training.

Come aviene, matematicamente, questa traduzione magica?

La trasformazione fondamentale
Mappatura nello spazio reale
La conversione da testo a numero avviene definendo una funzione di mapping che associa ogni elemento di un vocabolario discreto V a un elemento in uno spazio vettoriale di dimensione N
In termini formale, stiamo proiettando un insieme finito di simboli in un sottospazio delle spazio euclideo.
Ogni parola del dizionario riceve un indirizzo numerico unico

Ma una volta che abbiamo questi indirizzi, come facciamo a dare a loro un senso geometrico?

Il Vector Space Model
Geometria del significato
Il Modello a Spazio Vettoriale trasforma le parole in punti all'interno di uno spazio geometrico. In questo modello, la vicinanza fisica tra due punti (vettori) corrisponde alla loro similarità semantica nel mondo reale. E' una topologia del pensiero umano resa calcolabile.
Questo paradigma permette di utilizzare la geometria analitica per risolvre problemi linguistici complessi, come la ricerca di sinonimi o la classificazione automatica di documenti.

Ma di cosa sono fatti questi indirizzi?
parliamo di vettori e coordinate

Linguistica come topologia
per ongi parola è associato un vettore di numeri
immagina ogni numero come fosse una caratteristica nascosta.
La dimensionalità (spesso 300 o 768) determina quanto è ricco il vocabolario del nostro modello, più dimensioni abbiamo, più sfumature di siginificato possiamo catturare.
In questo spazio possiamo fare vera e propria algebra.
- Embedding: il processo di rappresentazione di una parola come vettore di numeri reali, solitamente densi e a bassa dimensionalità
- Direzione e Verso: in uno spazio semantico, la direzione del vettore può rappresentare relazioni come il genere o il tempo verbale.
- Similarità di Coseno: la metrica principale per misurare quanto due vettori puntano nella stessa direzione, ignorando la loro lunghezza
- Dimensionalità: il numero di componenti di un vettore (es. 300 o 768) determina la ricchezza delle sfumature che il modello può catturare.

E' una sottrazione e addizione di coordinate

Possiamo usare la similarità di coseno per capire se due concetti puntano nella stessa direzione, ignorando quanto sono comuni nel testo.
E' il modo più elegante che abbiamo per misurare l'affinità tra le idee

Vediamo la formula che ci permette di calcolare questa affinità

Calcolo della Similarità
La similarità del coseno guarda l'angolo tra due vettori  
definita come il prodotto di due vettori diviso il prodotto dei loro valori assoluti.
Il risultato è un valore che indica quanto i concetti siano correlati indipendentemente dalla frequenza.
Se due vettori puntano esattamente nella stessa direzione, l'angolo è zero ed il cose è 1 = somiglianza totale.
Usiamo il coseno perchè in NLP una parola, più frequente può apparire migliaia di volte, mentre un'altra (più rara) solo due. Le due parolo però potrebbero avere significato molto simile o uguale, quindi puntano nella stessa direzione.

Encoding: ASCII e Unicode
La codifica hardware del linguaggio
Prima ancora di parlare dei vettoriali semantici, il computer deve essere in grado di mappare i singoli caratteri in byte. 
Questo processo è l'encoding. L'encoding è il dizionario che dice all'hardware, quando vedi questo pattern di byte disegna la lettere A
Storicamente, siamo passati da sistemi molti limitati a standard globali capaci di rappresentare ogni alfabeto esistente.
Capire come il testo viene memorizzato fisicamente è essenziale per evitare errori di decodifica e per gestire correttamente dataset multilingue in Python

Come siamo passati a pochi caratteri nel dizionario a una vasta gamma di parole come oggi

Standard di rappresentazione
Dall'alfabeto latino agli Emoji
Siamo partita da Ascii, sapeva a male pena scrivere in ingelse, poi arrivato Unicode, che assegna un numero ad ogni carattere. Poi arrivata UTF-8 intelligente ed occupa poco spazio per caratteri comuni.
- ASCII: lo standard originale a 7 bit. Può rappresentare solo 128 caratteri, limitandosi all'inglese e ai comandi di controllo base
- Unicode: un framework universale che assegna un numero univoco (Code Point) a ogni carattere di ogni lingua dle mondo
- UTF-8: la codifica a lunghezza variabile più diffusa. Utilizza da 1 a 4 byte per carattere, garantendo compatibilità con ASCII
- Byte Order Mark (BOM): un segnale opzionale all'inizio del file per indicare l'ordine dei byte nella codifica Unicode

Gestione dell'Encoding
Pronelma dei Mojibake: si verifica quando interpretiamo il testo con un encoding sbagliato. Esempio testo scritto con Unicode ed interpretato con UTF-8. Questo causa visualizzazione di caratteri errati.

Vediamo come queste stringhe trasformano i flussi binari

Mappatura Binaria
La struttura del byte
In UTF-8 la struttura dei byte è dinamica. I caratteri 'base' (caratteri ascii) sono più leggeri 1 byte, mentre quelli complessi sono più pesanti (+ byte). Caratteri più complessi utilizzano bit di controllo nei byte successivi per segnalare la continuazione della sequenza. 
Questo meccanismo di controllo permette al computer di capire se un byte è l'inizio di una nuova lettera o la continuazione della precedente. Garantendo al contempo l'estensibilità a milioni di simboli.


```python
import numpy as np

# --- 1. GESTIONE ENCODING (ASCII vs UTF-8) ---
def demonstrate_encoding():
    # Teoria: ASCII usa 7 bit (max 128 caratteri). UTF-8 è a lunghezza variabile (1-4 byte).
    testo = "AI nel 2026 😛" # Carattere speciale (Emoji) non presente in ASCII
    
    # Codifica in byte (UTF-8 è lo standard universale)
    utf8_encoded = testo.encode('utf-8')
    print(f"Testo Originale: {testo}")
    print(f"Byte UTF-8: {list(utf8_encoded)} (Lunghezza: {len(utf8_encoded)} byte)")
    
    try:
        # Questo fallirà perché l'emoji non esiste nello standard ASCII
        ascii_encoded = testo.encode('ascii')
    except UnicodeEncodeError:
        print("Errore: ASCII non può rappresentare i simboli moderni.")

# --- 2. VECTOR SPACE MODEL: SIMILARITÀ DI COSENO ---
def calculate_similarity(vec_a, vec_b):
    """
    Calcola la similarità di coseno tra due vettori.
    Teoria: Misura l'angolo tra due vettori. Se l'angolo è 0°, il coseno è 1 (massima somiglianza).
    Formula: (A · B) / (||A|| * ||B||)
    """
    # Prodotto scalare (Somma dei prodotti delle componenti corrispondenti)
    dot_product = np.dot(vec_a, vec_b)
    
    # Norma Euclidea (La 'lunghezza' del vettore nello spazio)
    norm_a = np.linalg.norm(vec_a)
    norm_b = np.linalg.norm(vec_b)
    
    # Calcolo della similarità
    return dot_product / (norm_a * norm_b)

# Esecuzione dimostrativa
demonstrate_encoding()

# Simulazione di due parole nello spazio 3D: 'Cane' e 'Gatto'
# In un vero modello, queste coordinate sarebbero apprese dai dati
v_cane = np.array([0.9, 0.1, 0.05])
v_gatto = np.array([0.85, 0.12, 0.07])
v_pietra = np.array([0.1, 0.8, 0.9])

sim_animali = calculate_similarity(v_cane, v_gatto)
sim_pietra = calculate_similarity(v_cane, v_pietra)

print(f"\nSimilarità Cane-Gatto: {sim_animali:.4f} (Vicini)")
print(f"Similarità Cane-Pietra: {sim_pietra:.4f} (Lontani)")
```
