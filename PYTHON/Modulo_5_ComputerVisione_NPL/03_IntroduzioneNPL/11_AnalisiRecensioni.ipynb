Sappiamo smontare il linguaggio nei suoi componenti minimi: token, lemmi e stopword, oggi impariamo a rimontare tutto per estrarre valore reale dai dati.

Come si costruisce la catena di montaggio del dato?

La Pipeline di Preprocessing Integrata
Concatenare le fasi per ottenere dati normalizzati
Una pipeline efficiente trasforma un testo grezzo e rumoroso in  un insieme di token semantici pronti per l'analisi.
Il flusso inizia con la rimozione del rumore, passa per la scomposizione in token e si conclude con la lemmatizzazione garantendo che ogni parola sia ricondotta alla sua forma base per eliminare la ridondanza morfologica.
Il pre-processing deve essere visto come una catena di montaggio, se un passaggio lavora male, l'intero processo ne risentità. 

Quali sono gli stadi di questra trasformazione?

I 3 stadi delle trasformazione del dati:
- Pulizia Modulare: rimozione selettiva di tag HTML, URL e caratteri non alfanumerici per sanificare l'input. Puliamo il terreno
- Tokenizzazione Atomica: suddivisione delle stringa in unità minime significative, preservando la struttura logica della frase. Dividiamo il testo in token
- Riduzione Canonica: applicazone della lemmatizzazione per unificare sinonimi e flessioni verbale sotto un unico lemma.

Possiamo modellare la pipeline come una composizione di funzioni dove l'output di una fase diventa l'input della successiva.
Per far si che questi ingranaggi girano bene dobbiamo seguire alcune regole.

Integrazione e Best Practice

Durante la pulizia è vitale non rimuovere punteggiatura che potrebbe servire al lemmatizzatore per il Part-of-Speech tagging, posticipando la rimozione definitiva dei simboli a valle.
L'uso di oggetti 'nlp' di spaCy permette di eseguire tokenizzazione e lemmatizzazione in un unico passaggio computazionale ottimizzato (efficienza del codice)
Il filtraggio delle parole 'vuote' deve avvenire dopo la lemmatizzazione per garantire che le forme base siano correttamente confrontate con la lista delle stopwords.

Una volta definita la pipelina dobbiamo assicurarci che sia stabile.

In informatica usiamo il termine idempotenza, significa che se passasse un testo già pulito attraverso la pipeline il risultato non deve cambiare, l'obbiettivo finale è quello di ridurre la varianza.
Meno varianza innutile abbiamo nei nostri dati piu' facile è per la rete neurale trovare i pattern semantica per la rete.

Passiamo al secondo capitolo

Analisi delle Frequenze
Estrarre il valore statistico dalle recensioni
dopo la pulizia, il testo non è più una giungla, contare quante spesso un utente dice ottimo o deluso, è la forma più pura di business intelligence.
Ma non basta contare tutto, dobbiamo distinguere tra parole che appaiono spesso, e che magari sono necessarie per la grammatica, da parole che appaiono spesso perchè rappresentano il cuore per il messaggio dell'utente.

Come pesiamo l'importanza di un termine?

Distribuzione Lessicale
Metriche per pesare l'importanza dei termini
usiamo metriche precise:
- Term Frequency: conteggio assoluto delle occorrenze di un lemma nel dataset per identificare i trend macroscopici.
- Top-N Analysis: analizzando i termini più ricorrenti possiamo definire il profilo semantico delle recensioni positive o negative.

Normalizzare la frequenza diventa fondamentale per non dare troppo peso alla recensioni più lunghe rispetto a quelle brevi.

Ma c'è una trappola statistica in cui è facile cadere, la coda lunga

Interpretazione dei Risultati
Il fenomeno della Long Tail: se guardi la distribuzione delle parole vedi che poche parole dominano la scena (frequenza totale), mentra esiste una coda lunghissima di parole che appaiono una sola volta. Questo ci conferma che il filtraggio delle stopword è una necessità statistica.
Analisi dei contrasti: confrontare le frequenze tra recensioni a 5 stelle e recensioni a 1 stella rivela istantaneamente quali lemmi fungono da predittori di sentiment.
- Data Shifting: l'analisi delle frequenze agisce come un setaccio che separa l'oro informativo dalla sabbia linguistica della sintassi comune.

Questo fenomeno ha un nome preciso nella linguistica computazionale
Parliamo della legge di Zipf
Ci dice che la seconda parola più frequente apparirà circa la metà delle volte rispetto la prima la teza 1/3 e così via.
La maggior parte della struttura di una frase è puro servizio grammaticale, rimuoverlo ci permette di concentrarci sulla sostanza

Ma come possiamo comunicare tutto questo a qualcuno che non è un esperto di statistica?

Visualizzazione con WordCloud
Trasformare i numeri in immagini intuitive.
I dati numerici, sebbene precisi, non sempre comunicano il messagio in modo immediato. Le WordCloud affrono una sintesi visiva dove la dimensione di ogni parola è direttamente proporzionale alla sua importan
Impariamo a generare queste nuvole per comunicare efficacemente i risultati della nostra analisi delle recensioni rendendo evidenti i concetti chiave a colpo d'occhio

Meccanica della WordCloud
Dietro un immagine semplice c'è un algoritmo per visualizzare l'importanza numerica delle varie parole
- Scaling Dimensionale: algoritmo che mappa la frequenza del lemma alla dimensione del font nel canvas (point size)
- Gestione layout: disposizione non sovrapposta dei termini, spesso basata su un algoritmo che parte dal centro e si muova a spirgale
- Mappatura Cromatica: uso del colore per aggiungere una dimensione informativa, come la polarità del sentiment o la categoria del prodotto. Magari il verde per i termini positivi ed il rosso per i negativi, in un istante comunichiamo non solo di cosa si parla ma anche come se ne parla
- L'area occupato da una parola nuvola riflette la sua rilevanza relativa all'interno dell'insieme dei dati processati

Ma attenzione: una WordCloud fatta male può comunicare male

Ottimizzazione Visiva
Visualizzare troppe parole genera caos visivo; è fondamentale impostare un parametri 'max_words' per mantenere la leggibilità (selezione del limite).
Le nuvole possono essere modellate tramite maschere (mask) per adattarsi a loghi o forme pertinenti al dominio di analisi (personalizzazione della forma). La WordCloud è solo un punto di partenza non la prova finale
Una WordCloud è utile per l'esplorazione, ma deve sempre essere accompagnata da analisi quantitative per evitare interpretazioni soggettive errate.

E se volessimo visualizzare non solo parola ma concetti piu' ampi?
Qui entra in gioro il concetto di n-gramma

Oltre la Singola Parola
N-grammi nelle WordCloud
E' possibile generare nuvole basate su bigrammi o trigrammi per catturare concetti composti, evitando che termini come 'servizio' e 'clienti' vengano separati visivamente (per vedere servizio clienti in unica nuvola, separate non hanno significato)
L'inclusione del contesto locale arrichisce drasticamente il valore informativo della visualizzazione finale.


```python
# python -m pip install wordcloud
import os
import re
import spacy
import matplotlib.pyplot as plt
from wordcloud import WordCloud
from collections import Counter


# Caricamento del modello linguistico spaCy per l'italiano
# Teoria: 'it_core_news_sm' è un modello statistico multi-task che esegue
# simultaneamente tokenizzazione, POS tagging e lemmatizzazione.
try:
    nlp = spacy.load("it_core_news_sm")
except OSError:
    os.system("python -m spacy download it_core_news_sm")
    nlp = spacy.load("it_core_news_sm")

def full_nlp_pipeline(text_data):
    """
    Pipeline Integrata: Pulizia -> Tokenizzazione -> Lemmatizzazione.
    
    Teoria: La pipeline trasforma il segnale testuale da 'unstructured' a 'structured'
    riducendo l'entropia e la varianza morfologica del dataset.
    """
    processed_docs = []
    
    for text in text_data:
        # 1. Pulizia con Regex (Noise Removal)
        # Rimuoviamo URL e tag HTML prima che entrino nel modello linguistico.
        text = re.sub(r'<.*?>|https?://\S+', '', text)
        # Rimuoviamo caratteri speciali mantenendo solo lettere e spazi.
        text = re.sub(r'[^a-zA-Zàèìòù\s]', '', text).lower()
        
        # 2. Processing con spaCy (Tokenization + Lemmatization)
        # Il metodo nlp() trasforma la stringa in un oggetto Doc ricco di annotazioni.
        doc = nlp(text)
        
        # 3. Estrazione Lemmi e Rimozione Stopwords
        # Teoria: Filtriamo le parole ad alta frequenza ma basso valore semantico.
        # Estraiamo il .lemma_ per ricondurre ogni parola alla sua forma canonica.
        lemmas = [token.lemma_ for token in doc if not token.is_stop and not token.is_space]
        processed_docs.append(" ".join(lemmas))
        
    return processed_docs

# --- ESEMPIO DI DATASET: RECENSIONI HOTEL ---
raw_reviews = [
    "Il soggiorno è stato fantastico! <br> Personale gentilissimo e camere pulite.",
    "Esperienza pessima. La camera era sporca e il personale maleducato. Sconsiglio.",
    "Ottima posizione, ma il check-in è stato lentissimo. Camere nella media.",
    "Colazione incredibile! Tornerò sicuramente in questo hotel fantastico.",
    "Non mi è piaciuto nulla. Sporco ovunque e troppo rumore di notte."
]

# Esecuzione della Pipeline
print("Esecuzione Pipeline in corso...")
cleaned_reviews = full_nlp_pipeline(raw_reviews)

# --- ANALISI DELLE FREQUENZE ---
# Teoria: La Legge di Zipf suggerisce che pochi termini domineranno il corpus.
# L'analisi delle frequenze ci permette di validare la qualità della pulizia.
all_words = " ".join(cleaned_reviews).split()
word_freq = Counter(all_words)
common_words = word_freq.most_common(10)

print(f"\nTop 10 Lemmi più frequenti: {common_words}")

# --- VISUALIZZAZIONE: WORDCLOUD ---
# Teoria: La WordCloud è una mappatura spaziale dove l'area occupata dal testo
# è proporzionale alla sua frequenza relativa nel dataset.
wordcloud = WordCloud(
    width=800, 
    height=400, 
    background_color='white',
    colormap='viridis'
).generate(" ".join(cleaned_reviews))

plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis("off")
plt.title("Visualizzazione Semantica delle Recensioni")
plt.show()
```
