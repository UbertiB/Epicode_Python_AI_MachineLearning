** NLTK **

Laboratorio accademico
Nata in ambito universitario, questa libreria non cerca di essere la più veloce ma la più dettagliata, è il posto dove andare se vuoi studiare o se hai bisgno di un controllo granulare,  è lo strumenti didattico per eccellenza
La sua forza è la modularità, ogni mattoncino è un algoritmo diverso
A differenza di altri sistemi che obbligano a seguire un percorso prestabilito, qui è tutto libero.
Inoltre regala accesso a centinaio di testi pronti per essere analizzati.
Tuttavia questa ricchezza richiede una gestione oculata delle risorse

NLTK (Natural Language Toolkit) è una delle librerie storiche di NLP in Python ed è utile per imparare i concetti: tokenizzazione, stemming, lemmatizzazione, POS tagging, analisi semantica, corpora, WordNet ed altre elaborazioni linguistiche. 


```python
import nltk
# Necessario solo la prima volta
nltk.download("punkt_tab")
testo = "Il cliente ha ordinato 100 pezzi."
tokens = nltk.word_tokenize(testo,language="italian")
print(tokens)
```

Quindi NLTK è utile quando vuoi vedere i singoli passaggi.
Se però l'obbiettivo non è lo studio ma produzione di massa, dobbiamo cambire NLTK con altro


*** spaCY ***

spaCY ha una filosofia completamente diversa.
E' brutalmente efficiente.
Ma mentre NLTK offre 10 modi per fare una singola cosa, spaCy offre solo il migliore, quello che funziona meglio. E' la scelta obbligata se devi processare milioni di commenti social al secodo senza far esplodere il servere.

Il cuore di SpaCy è l'oggetto DOC
- Oggetto Doc: immagina un contenitore super intelligente ,che appena riceve un testo, esegue, automaticamente, una serie di operazioni (memorizza token, lemmi, e annotazioni) in modo estremamente ottimizzato per la memoria.
- Pipeline Pre-Configurata: tutto avviene in un unico passaggio, caricando un modello linguistico SpaCy esegue automaticamente taggin, parsing e riconoscimento delle entità, in una unica pipeline pre-configurata automatizzata ed ottimizzata.
- Velocità di esecuzione: grazie al core in Cython, è in grado di processare milioni di parole al secondo su hardware standard.
- Support Word Vectiors: integrazione nativa per vettori densi che permettono di calcolare similarità semantiche in modo immediato.

La cosa bella è che spaCy ci permete di vedere quello che la macchina capisce.

Integrazione e Visualizzazione
Uno dei tesori di spaCy è diplaCy, un tool di visualizzazione integrato, include struemnti per generare graficamente alberi sintattici e identificare visualmente le entità nominate (NER) all'interno del browser.
Invece di guardare freddi numeri è possibile visualizzare questi grafici per vedere come queste parole sono collegate tra loro.

Strumento fondamentale per creare motori di ricerca intelligenti o sistemi di raccomandazioni.

*** Huggin Face ***

Huggin Face non è una semplice libreria NLP ma un ecostistema, una delle sue librerie principali è Transformers, ma esistono anche: datasets, tokenizers, accelerate, huggingface_hub.

Se oggi l'ai è ovunque è merito di Huggin Face e i Transformer.

La nuova frontiera del Deep Learning
Huggin Face è diventata rapidamente l'ecosistema di riferimento per l'intelligenza artificiale linguistica moderna. Attraverso la libreria 'transformers', permette l'accesso a modelli giganti come BERT e GPT che hanno richiesto infinite risorse per essere addestrati.

Il suo valore principale risiede nella democratizzazione della tecnologia: permette a chiunque di scaricare e utilizzare modelli che hanno richiesto enormi risorse computazionali per l'addestramento iniziale.
E' quello che potremo definire il git-hub dell'ai

Vediamo come questo ecosistema ha cambiato le regole del gioco

Modelli Pre-addestrati
Dall'Hub alla produzione

Haggin Face non è solo codice è un vero e proprio Hub, un portale dove migliaia di ricercatori condividono i propri modelli.
La libreria Transforers permette di utilizzare moltissimi modelli pre-addestrati per testo, immagini, audio, video e multimodalità, oggi quindi Huggin Face va ben oltre il solo NLP
- Model Hub: un portale centralizzato che ospita migliaia di modelli ottimizzati per compiti che vanno dalla traduzione alla classificazione
- AutoModel Class: classi intelligenti che deducono automaticamente la configurazione corretta del modello partendo semplicemente da una stringa identificativa
- Fast Tokenizers: implementazioni scritte in Rust che preparano i dati per i modelli Transformer con latenza minima.
- Integrazione Deep Learning: pieno supporto per le principali librerie di calcolo tensoriale, garantendo flessibilità nello sviluppo

Ma come si usa tutta questa potenza senza perderci nei dettagli tecnici?

Ecosistema e Workflow
la rispota è nell API pipeline

Vuoi sapere se una recensione è positiva?, vuoi catagorizzare le mail ricevute? vuoi tradurre dall'italiano al cinese? con Huggin Face basta una riga di codice, la pipeline di occupa di tutto.
Scarica il modello, prepara il testo, esegue l'inferenza e restituisce il risultato.
E' come avere un maggiordomo super-intelligente, a cui dai il comando e lui esegue il lavoro sporco.

- Pipeline API: rappresenta l'astrazione di più alto livello per eseguire task complessi come il riassunto di testi o la risposta a domande in una sola riga di codice
- Dataset Library: fornisce un accesso standarizzato a migliaia di dataset pubblici, facilitando la fase di caricamento e pre-processing dei dati per l'addestramento
- Huggin Face Hub: funziona come un GitHub per l'AI, dove gli sviluppatori possono condividere pesi dei modelli, configurazioni e metriche di valutazione

Ma come fa la rete a scegliere la risposta corretta tra mille possibilità?

La rispota è nella formula della softmax
La Funzione Softmax
Interpretazione delle decisioni neurali
Negli strati finali dei modelli Transformer, è necessario convertire i valori numerici grezzi (logits), difficili da interpretare per noi,  in una distribuzione di probabilità comprensibile.
La funzione softmax assicura che la somma di tutte le proabilità d'uscita sia parta a 1, evidenziando la scelta più verosimile del modello.

pipe=pipeline("task",model="autore/modello",framwork="pt"
task= coda deve fare
model= chi deve farlo
framework= con quale motore viene eseguito


E' il momento in cui il calcolo neurale diventa una decisione comprensibile.


Vediamo questi 3 approcci
NLTK, SpaCy, Huggin Face

```python
import os
import nltk
import spacy
from transformers import pipeline


# 1. APPROCCIO NLTK (Accademico e Granulare)
# Teoria: NLTK è una libreria 'symbolic-first'. Ogni operazione è un modulo isolato.
def run_nltk_demo(text):
    # Scarichiamo i tokenizer necessari. 
    # 'punkt_tab' è richiesto dalle versioni più recenti di NLTK per la tokenizzazione.
    nltk.download('punkt', quiet=True)
    nltk.download('punkt_tab', quiet=True) # FIX: Aggiunto per risolvere l'errore Resource Not Found
    
    # Tokenizzazione: divide il testo in unità minime (token) usando regole grammaticali
    tokens = nltk.word_tokenize(text)
    print(f"[NLTK Tokens]: {tokens}")

# 2. APPROCCIO SPACY (Industriale 'Production-Ready')
# Teoria: spaCy crea un oggetto 'Doc' che contiene tutte le annotazioni linguistiche
# calcolate in un unico passaggio tramite una pipeline ottimizzata in Cython.
def run_spacy_demo(text):
    # Carichiamo il modello per l'italiano (small)
    # BEST PRACTICE: Usare modelli pre-installati per evitare latenza al primo avvio
    try:
        nlp = spacy.load("it_core_news_sm")
    except OSError:
        # Fallback nel caso il modello non sia presente nell'ambiente
        print("Modello spaCy non trovato. Download in corso...")
        os.system("python -m spacy download it_core_news_sm")
        nlp = spacy.load("it_core_news_sm")
    
    doc = nlp(text)
    
    # NER (Named Entity Recognition): Identifica entità come Persone, Luoghi o Organizzazioni
    # Teoria: spaCy usa modelli statistici per inferire il ruolo dei nomi nel contesto
    entities = [(ent.text, ent.label_) for ent in doc.ents]
    print(f"[spaCy NER]: {entities}")

# 3. APPROCCIO HUGGING FACE (SOTA Deep Learning)
# Teoria: Hugging Face astrae i modelli Transformer (BERT, GPT, ecc.)
# L'oggetto 'pipeline' gestisce automaticamente tokenizzazione e inferenza neurale.
def run_transformers_demo(text):
    # Sentiment Analysis: determina l'emozione prevalente nel testo
    # Teoria: I modelli Transformer usano l'Attention Mechanism per pesare l'importanza delle parole
    print("Inizializzazione Transformer Pipeline...")
    
    # FIX: Aggiunto framework="pt" per forzare PyTorch ed evitare il conflitto con Keras 3
    classifier = pipeline("sentiment-analysis", model="dbmdz/bert-base-italian-xxl-cased", framework="pt")
    
    result = classifier(text)
    print(f"[Hugging Face Sentiment]: {result}")


# ESECUZIONE
if __name__ == "__main__":
    test_text = "Leonardo da Vinci ha dipinto la Gioconda a Firenze, è un'opera meravigliosa."
    print("--- INIZIO ELABORAZIONE NLP ---")
    run_nltk_demo(test_text)
    print("-" * 30)
    run_spacy_demo(test_text)
    print("-" * 30)
    run_transformers_demo(test_text)
```
