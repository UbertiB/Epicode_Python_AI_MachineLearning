LEMMATIZZAZIONE

La lemmatizzazione serve a riportare una parola in una forma base corretta dal punti di vista linguistico, chiamata lemma
Esempi
ordinato ->ordinare
ordini -> ordine
pezzi -> pezzo
migliori -> buoni (a seconda del constesto/modello)
sono -> esere
Rispetto allo stemming dove lo stem può essere una parola che non esiste nel vocabolario, il lemma dovrebbe essere una parola valida nel vocabolario
La lemmatizzazione è quindi èiù 'linguistica' dello stemming

Per fare la lemmatizzazione non guardo solo la parola, serve anche capire il suo ruolo grammaticale cioè il POS (Part Of Speech)
Per esempio "ordine" può non essere un sostantivo ma in altri contesti una forma simile può avere un ruolo diverso
Questo è il motivo per cui la lemmatizzazione è più sofisticata del semplice "taglia un suffisso"

Con la lemmatizzazione riduce diverse forme grammaticali a una rappresntazioe comune.
Esempio: ordino, ordini, ordina, ordinavamo, ordineremo, possono esswre ricondotte a "ordinare"
In questo modo il modello può capire, più facilmente, che i concetti di fondo sono identici

Questo riduce la variabilità del vocabolario

A differenza dello stemming la lemmatizzazione produce una parola esistente nel vocabolario mantentendo la leggiiblità e la congruenza liguistica.

Unificazione semantica
Riconducendo "migliore", "meglio" e "ottimo" allo stesso concetto, la lemmatizzazione riduce la sparsità del vocabolario senza distruggere la leggibilità
Precisione linguistica
A differenza degli stemmer che producono 'stems' illegibili, i lemmi sono interpretabili dagli umani e utili per analisi di topic modeling
Gestione delle irregolarità
Le forme verbali irregolari (es. "sono" che diventa "essere") vengono risolte correttamente solo tramite dizionari, dove lo stemming fallirebbe sistematicamente.

Come questo si riflette sul nostro spazio lessicale?

Morfologia e Vocabolario
Mappatura dello Spazio Lessicale
Ogni lemma funge da "centroide" per una famiglia di parole flesse. 
Nello spazio vettoriale del linguaggio, tutte le forme flesse (andiamo, andare, andassi, ecc), sono punti che ruotano attorno ad ununico sole (il lemma andare).
In questo modo stiamo concentrando il segnale e riducendo il rumore statistico senza perdere la bussola del significato.
Questo permette di collassare diverse dimensioni dello spazio vettoriale in un unico asse semantico coerente.
Se indichiamo con 'S' l'insieme di tutte le forme flesse di un lemma 'l' l'operazione di lemmatizzazione assicura la convergenza del segnale.

Ma come facciamo se una parola può avere più lemmi?

Il Ruolo del POS Taggin
Perchè il contesto cambia tutto
Una parola può avere lemmi diversi a seconda della sua funzione grammaticale. Ad esempio 'lavori' può essere il plurale di un sostantivo o la seconda persona di un verbo. Senza conoscere il Part-of-Speech (POS), il lemmatizzatore non può scegliere la forma corretta.
L'analisi del contesto non riguarda solo la parola singola, ma le relazioni con i termini adiacenti. Assegnare il tag corretto (verbo, nome, aggeettivo) è il prerequisito per una lemmatizzazione di alta qualità
"la porta porta fortuna" la prima parola "porta" è un sostantivo la seconda un verbo. 
Uno stemmer le vedrebbe identiche.
Un lemmatizzatore pigro idem
Un lemmatizzatore intelligente guarda a destra ed a sinistra analizza il contesto ed assegna anche un etichetta grammaticale.

La buona notizia è che oggi abbiamo strumenti che fanno questo lavoro per noi

Integrazione con spaCy
Librerie moderne come spaCy eseguono automaticamente il POS taggin prima della lemmatizzazione per garantire la massima precisione possibile.
Vengono analizzati set di etichette universali (es. NOUN, VERB, ADJ) per uniformare il trattamente tra le ligue diverse.
Un errore nel POS tagging si propaga direttamente sul lemma, portando a una rappresentazione numerica errata nel modello neurale.

Probabilità Condizionata
Selezione del Tag Ottimale
Il sisetma che calcola il lemma, non tira a indovinare, calcola la probabilità condizionata dato una parola ed il suo contorno, minimizzando l'errore di interpretazione.
Possiamo esprimere la ricerca del lemma ottimo come la massimizzazione della probabilità del lemma 'l' data la parola 'w' e il contesto 'c'

Serve sempre tutto questo sforzo?
Lemmatizzazione vs Stemming
E' una sfida tra velocità e precisione
Scegliere lo strumento giusto per la pipeline
Sebbene entrambi i metodi mirino a ridurre la complessità del testo, le loro applicazioni differiscono drasticamente. Lo Stemming è rapido, economico e utile quando la velocità è prioritaria rispetto alla precisione lessicale. La Lemmatizzazione è più lenta, richiede più tempo, costa di più in termini computazionale, ma essenziale per compiti di compressione profonda.
A volte lo stemming risulta troppo aggressivo e/o produrre termini privi di senso rispetto alla pulizia della lemmattizzazione

Velocità vs Accuratezza
- Comlessita computazionale: lo stemming è istantaneo, la lemmatizzazione è più costosa per via del lookup nel dizionario
- Information Retrieval: lo stemming è spesso sufficiente per i motori di ricerca dove conta trovare la radice comune.
- NLU e Deep Learning: i modelli transformer beneficiano maggiormente della lemmatizzazione per mantenere relazioni semantiche chiare
- La perdita di informazione nello stemming 'E_s' è solitamente maggiore rispetto a quella della lemmatizzazione 'E_I'

L'errore informativo delle stemming è quasi sempre superiore, sascrifichiamo qualche secondo di calcolo per avere informazioni più precise.

Criteri si Scelta
- Analisi del Sentimento: la lemmatizzazione è preferibile perchè preserva sfumatura che lo stemming potrebbe cancellare raggrupando parole con valenze diverse
- Dimensione del dataset: con miliardi di documenti, lo stemming può accelerare drasticamente il preprocessing se la precisione millimetrica non è richiesta
- Supporto Linguistico: in lingue morfologicamente ricche come l'italiano lo stemming fallisce spesso, rendendo la lemmatizzazione quasi obbligatoria.

Over-Stemmig e Risoluzione
L'over-stemming accade quando parole diverse collassano sulla stessa radice. La lemmatizzazione risolve questo problema distinguendo i termini tramite l'appartenenza a classi grammaticali diverse.
La lemmatizzazione tende a preservare la distanza tra lemmi distinti 'l1' e 'l2' meglio dello stemming.
