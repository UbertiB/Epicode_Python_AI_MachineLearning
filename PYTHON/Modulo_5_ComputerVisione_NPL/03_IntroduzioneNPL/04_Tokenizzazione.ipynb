TOKENIZZAZIONE: CONCETTI E METODI

La tokenizzazione è uno dei prima passaggi del NLP: serve a spezzare una stringa di testo in unità più piccole chiamate token, che poi possono essere trasformate in numeri e date in input al modello.
La sequenza generale è:
testo -> tokenizzazione -> token -> token ID -> embedding -> modello
Quindi la tokenizzazione non capisce ancora il significato del testo: decide semplicemente come suddividerlo.
Prendiamo la frase: "il cliente ordine 100 pezzi"
una tokenizzazione semplice per parola potrebbe produrre
["il", "cliente" "ordina", "100", "pezzi"]
ogni elemento è un token

Il token non significa necessariamente parola.
Un token può essere:
- una parola intera - una parte di parola - un carattere - un segno di punteggiatura - in slcun tokenizer anche spazi o simnoli speciali

Questo dipende dal metodo utilizzato.
La scelta della granularità del token influenza drasticamente la capacità del modello di apprendere relazioni morfologiche e la sua efficienza computazionale in termini di memoria occupata dal vocabolario.

La tipologia di tokenizzazione determina il vocabolario
Se utiizzo i caratteri come token il vocabolario è piccolissimo e corrispondente ad un insieme finito di lettere. Mantenere un vocabolario piccolo permette di avere layer di embedding più leggeri migliorando la velocità di addestramento su hardware consumer.

La tokenizzazione per parola genera un vocabolario molto esteso e spesso carente di parole mai viste. Inoltre spesso fallisce nel riconoscere che 'andare' e 'andavamo' condividono la stessa radice, trattandole come entità totalmente diverse.

Più il vocabolario è vasto più la rete diventa lenta perchè deve imparare troppi parametri
La soluzione a questo problema è arrivata con l'idea di spezzare le parole in pezzi comune

L'era della Sub-word Tokenization
I modelli moderni come BERT e GPT utilizzano un approccio ibrido: la sub-word tokenization. Questa tecnica spezza le parole rare in sottounità comuni, mantenendo intere le parole frequenti. E' il punto di equilibrio perfetto tra flessibilità e compattezza.
Gli algoritmi imparano statisticamente quali pezzi di parole meritano di diventare token indipendenti basandosi sulla loro frequenza nel corpus.
E' flessibile, potente e non soffre mai di parole sconosciute.

BPE e WordPiece
Algoritmi di fusione e probabilità
-   Byte Pair Encoding (BPE): un algoritom iterativo che fonde le coppie di caratteri o sottostringhe più frequenti fino a raggiungere una dimensione del vocabolare prefissata.
-   WordPiece: simi a BPE, ma invece delle frequenza pura, utilizza un criterio della verosomiglianza per decidere quali unità fondere

- Gestione delle parole rare: perole mai viste vengono scomposte in pezzi couni che il modello sa già come gestire, prefissi, radici, e suffissi conosciuti (es. 'un-happi -ly')
- Indipendenza dalla Lingua: questi metodi funzionano bene anche su lingue agglutinati o senza chiari separatori di spazio

Ma come faccia a dire al modello che due o più token appartengono alla stessa parola?

Integrazione nei Transformere
Prima di applicare BPE, il testo viene diviso grossolanamente in parole. Questo evita che un token posso crossaare i confini di due parole diverse (pre-tokenizzazione)
L'uso di prefissi speciali (es. "##" per BERT o "G" in GPT serve a indicare alla rete se un token è l'inizio di una parola o una continuazione)
Questi marcatori sono vitali, permettono alla rete di ricostruire la struttra originale del testo.
In queste situazioni il vocabolario non è statico ma viene "addestrato sul dataset specifico, rendendo il tokenizer perta integrante dell'architettura del modello.
Se addestramo il vocabolario su testsi medici, il tokenizer imparerà a non dividere termin troppo complessi

E' il ricercatore (delle reti) che decide quando fermarsi, vogliamo 30.000 token? 50.000 più token abbiamo siamo precisi ma più memoria consumiamo.

Problemi comuni e Ambiguity
Il lato 'sporco' del testo reale
Teorizzare la tokenizzazione è semplice, ma applicarla a dati reali è complesso La punteggiatura, le abbreviazioni e i simboli speciali creano ambiguità che possono rompere la logica del modello se non gestite correttamente.
La punteggiatura crea un rumore enorme nei dati. 
- Apostrofi e Contrazioni: gestire caso come "l'aquila" o "don't" richiede di decidere se trattarli come uno o due token
- Abbreviazioni: distinguere un punti di fine frase da un punto di abbreviazione (es. Dott. vs fine.) Il punto dopo dott. è una fine frase? ed i : dell'orario? (10:30)?
- Simboli Speciali: la gestione di Emoji, URL e indirizzi mail che non dovrebbero essere spezzati arbitrariamente
- Normalizzazione Unicode: assicurarsi che caratteri visivamente uguali ma codificati diversamente vengano mappati sullo stesso token

Un tokenizer ingenuo spezzerebbe tutto, i modelli moderni invece creano logiche basate sul contesto
e normalizzazione.

Ma come è evoluta questa tencologia negli ultimi anni?

Evoluzione dei Tokenizer
Oggi usiamo strumenti come sentence paese che ignora completamente il concetto di parola e tratta il testo come flusso di byte, permette di trattare nello stesso modo italiano, cinese, codice python

La lunghezza della sequenza di token impatta sulla complessità dell'Attention Mechanism, che cresce quadraticamente rispetto al numero di elementi in input

La forma più intuitiva è la word tokenization, cioè tokenizzazione per parole.
E' semplice da capire ma genera un problema: quante parole possibili esistono?
postresivere avere: ordine, ordini, ordinare, ordinato, ordinando, riordinare
Se ogni forma fosse un tokern completamente diverso, il vocabolario diventebbe enorme.
E poi, cosa fai con una parola mai vista? es. iperindustrializzazione?
Se non è nel vocabolario con un tokenizer per parola avresti un problema di "Out Of Vocabulary" (OOV)
Per questo nei moderni modelli si usano molto i subword tokenizer.
Invece di "internazionalizzazione" come singolo token
potresti avere qualcosa tipo:
"intern", "azzional" "izzazione"
L'esempio preciso dipende dal tokenizer

Se il modello conosce pezzi come "zione","intern","auto","mente" può rappresentare anche parole mai vista esattamente in quella forma
Questo è il principio dietro molti tokenizer usati dai Transformer

La tokenizzazione è l'atto di dividere una frase 'umana' in unità che la rete posso iterpretare
Tra i metodi di tokenizzazione più importanti abbiamo BPE, WordPiece e Unigram

Dopo la tokenizzazione ogni token viene associato a un token ID
"il cliente ordina" potrebbe diventare [125,8421,3927]

Poi il modello prende l'ID e consulta una matrice di embedding
quindi cliente  che è id = [8421] vettorializzato potrebbe diventare 
[0.21,-0.44,0.78,..]
Ed è quello il vettore numerico che viene elaborato dalla rete

Poi ce il problema della lunghezza delle frasi
ogni frase, anche tokenizzata, ha una propria lunghezza
Ma quando lavori in batch devi spesso avere tensori della stessa dimensione
Quindi si usa il PADDING 0=[PAD]
Frase 1: [12,22]
Frase 2: [12,87,54,99,31,44]
la frase 1 postrebbe diventare [12,22,0,0,0,0]
il modello deve però sapere che [0] non sono parole vere, per questo spesso esiste una attention mask
esempio
token id:
[12,22,0,0,0,0]
attention mask
[1,1,0,0,0,0]

I modelli hanno una lunghezza massima di contesto. Se il testo supera la lunghezza, potresti doverlo tagliare
tokenizer(text,truncation=True, max_lenght=512)
se superi 512 token, taglia
Ed è importante dire token, non caratteri e non parole
Una frase di 500 parole può facilmente produrre più di 500 token
Con Huggin Face puoi vedere tutto queto concretamente

```python
import keras
import numpy as np
from transformers import AutoTokenizer

def demonstrate_tokenization_concepts(text):
    """
    Analisi comparativa dei metodi di tokenizzazione per il NLP moderno.
    """
    print(f"TESTO DI INPUT: '{text}'\n" + "="*50)

    # --- 1. WORD TOKENIZATION (Logica a Spazi) ---
    # Teoria: Scompone il testo basandosi sui delimitatori (spazi). 
    # Problema: 'magia,' viene visto come parola diversa da 'magia' a causa della virgola.
    word_tokens = text.lower().split()
    print(f"[1] WORD TOKENIZATION (Semplice)")
    print(f"    Token: {word_tokens}")
    print(f"    Problema: Nota come i segni di punteggiatura restano attaccati alle parole.\n")

    # --- 2. CHARACTER TOKENIZATION (Livello Atomico) ---
    # Teoria: Ogni carattere è un'entità. Utile per lingue senza spazi o modelli generativi puri.
    # Problema: Perdere la semantica (la lettera 'm' non ha il significato di 'matematica').
    char_tokens = list(text)
    char_ids = [ord(c) for c in char_tokens]
    print(f"[2] CHARACTER TOKENIZATION")
    print(f"    Primi 10 token: {char_tokens[:10]}...")
    print(f"    Mapping Numerico (Unicode): {char_ids[:10]}...")
    print(f"    Limiti: Sequenze troppo lunghe per frasi complesse.\n")

    # --- 3. SUB-WORD TOKENIZATION (Lo Standard dei Transformer) ---
    # Teoria: Algoritmi come WordPiece (BERT) o BPE (GPT) spezzano le parole rare.
    # Esempio: 'Intelligenza' potrebbe essere intera, ma 'iper-tecnologico' verrebbe spezzato.
    
    print(f"[3] SUB-WORD TOKENIZATION (BERT Base Cased)")
    # FIX: Utilizziamo l'identificativo corretto 'bert-base-cased'
    tokenizer = AutoTokenizer.from_pretrained("bert-base-cased")
    
    # Scomposizione in token
    tokens = tokenizer.tokenize(text)
    # Conversione in ID numerici (quelli che la rete neurale legge davvero)
    input_ids = tokenizer.encode(text, add_special_tokens=True)
    
    print(f"    Token BERT: {tokens}")
    print(f"    IDs (inclusi [CLS] e [SEP]): {input_ids}")
    
    # Dimostrazione Specifica: Ricostruzione (Decoding)
    decoded_text = tokenizer.decode(input_ids)
    print(f"    Decodifica: {decoded_text}")
    print(f"    Vantaggio: I token speciali (101 e 102) aiutano il modello a capire l'inizio e la fine della frase.\n")

    # --- 4. CONFRONTO OOV (Out-Of-Vocabulary) ---
    # Cosa succede se usiamo una parola inventata o molto rara?
    rare_word = "Cyber-pasticceria"
    print(f"[4] GESTIONE PAROLE RARE: '{rare_word}'")
    print(f"    BERT Token: {tokenizer.tokenize(rare_word)}")
    print(f"    Nota: La parola viene scomposta in radici conosciute invece di essere ignorata.")

# Esecuzione con una frase ricca di punteggiatura e termini composti
test_sentence = "L'intelligenza artificiale non è magia, è matematica."
demonstrate_tokenization_concepts(test_sentence)
```
