ESERCIZIO

sapere "leggere" come lavora un algoritmo di tokenizzazione è fondamentale per il debugging dei modelli. Il tuo compito è implementare una versione semplificata di un ciclo Byte Pair Encoding (BPE)

Data la parola "banana", scrivi una funzione che identifichi la coppia di caratteri adiacenti più frequente

Sostituisci tutte le occorrenze di quella coppi acon un nuovo token fittizio (es. "X")

Stampa la parole risultante e il nuovo "vocabolario" creato

Esempio atteso: sa la coppia puù frequente è "an", la parola diventerà "bXXa"

```python
from collections import Counter

def solve_bpe_exercise(word="banana"):
    """
    Implementazione di un ciclo semplificato di Byte Pair Encoding (BPE)
    1. Trova la coppia più frequente.
    2. Sostituisci con un nuovo token.
    3. Mostra il vocabolario risultante.
    """
    print(f"[5] ESERCIZIO BPE: Analisi della parola '{word}'")
    
    # Rappresentazione iniziale come lista di caratteri
    chars = list(word)
    
    # 1. Identificazione delle coppie adiacenti
    pairs = []
    for i in range(len(chars) - 1):
        pairs.append((chars[i], chars[i+1]))
    
    # Conteggio frequenze
    pair_counts = Counter(pairs)
    
    if not pair_counts:
        print("Impossibile trovare coppie.")
        return

    # Trova la coppia più frequente (se c'è un pareggio, Counter sceglie la prima incontrata)
    best_pair, frequency = pair_counts.most_common(1)[0]
    pair_str = "".join(best_pair)
    
    print(f"    Coppie trovate: {dict(pair_counts)}")
    print(f"    Coppia più frequente: '{pair_str}' (Appare {frequency} volte)")

    # 2. Sostituzione con il token fittizio "X"
    new_token = "X"
    # Creiamo la nuova stringa sostituendo la coppia
    # Usiamo replace per simulare la logica BPE
    result_word = word.replace(pair_str, new_token)
    
    # 3. Output Vocabolario
    # Inizialmente il vocabolario contiene i singoli caratteri
    vocab = set(word)
    # Aggiungiamo il nuovo token che rappresenta la coppia "fusa"
    vocab.add(f"{new_token} ({pair_str})")

    print(f"    Parola risultante: {result_word}")
    print(f"    Nuovo Vocabolario: {sorted(list(vocab))}")
    print("-" * 50)

if __name__ == "__main__":
    
    # Risoluzione Esercizio BPE
    solve_bpe_exercise("banana")
```
