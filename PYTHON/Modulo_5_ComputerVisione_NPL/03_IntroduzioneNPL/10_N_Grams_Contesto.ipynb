N-GRAMS E CONTESTO LOCALE

Il linguaggio non è solo un sacco di parole mescolate, l'ordine e la vicinanza, contano quanto i singoli termini.
Come passare dal singolo termine a delle sequenze strutturate.

Come cresce la complessità quanto uniano i puntini?
La Gerarchia degli N-grammi
Sal singolo token alle sequenze strutturate.

Fino ad ora abbiamo trattato le parole come entità isolate (Bag of Words), perdendo l'ordine originale. Gli N-grammi risolvono parzialmente questo problema raggruppando sequenze contigue di N elementi.
Un N-gramma è una finestra scorrevole appicata al testo che cattura frammenti di linguaggio. Questa tecnica permette di passare da una visione atomica a una visione sequenziale, fondamentale per i modelli linguistici probabilistici.
Ci permette di catturare frammenti di linguaggio nel loro originale, per permettere ad una macchina di non leggere delle singole parole ma di inizare a vedere la frase

Ma quante parole dobbiamo mettere dentro questa finestra?
Classificazione in base all'ampiezza della finestra

Gli N-grams sono  un modo semplice ma molto importante per rappresentare il contesto locale nel testo.
L'idea di base è questa: invece di guardare una parola isolata, guardo piccoli gruppi di parole consecutive.
Prendiamo la frase: "il cliente ordina 100 pezzi"
Se considero una parola per volta ho gli unigrammi
il, cliente, ordina, 100, pezzi
Quindi N=1 unigram
Se considero coppie consecutive, ho i bigrammi
"il cliente", "cliente ordina", "ordina 100", "100 pezzi"
Quindi N=2 bigram
Se considero gruppi di 3 parole ho i trigrammi
"il cliente ordina", "cliente ordina 100", "ordina 100 pezzi!
Quindi N=3 trigram
In generale un N-gram è una sequenza di N elementi consecutivi
Gli elementi possono essere parole, ma anche caratteri
Il motivo per cui gli N-gram sono interessanti è che introducono un po' di contesto.
Con Bag of Words
"cliente ordina prodotto" e "prodotto ordina cliente" contengono praticamente le stesse parole
Se utilizzi solo n-grams distingui bene l'ordine delle parole

E' da b-gramma in avanti che si inizia a capire una frase, più è grande n più storia catturiamo e meno sequenze totali avremo per ogni frase.

Sembrerebbe logico pertanto un N molto grande, ma c'è un prezzo da pagare

Limiti e Crescita Esponenziale
Maledizione della dimensionalità
Man mano che N aumenta, il numero di combinazioni possibili esplode in modo esponenziale, portando a problemi di memoria e scarsità di dati (sparsity), molte sequenze di 5 o 6 parole sono talmente rare che non le vedremo più nel nostro modello
Un N troppo piccolo, invece, non cattura la semantica complessa mentra un N grande rischia di memorizzare intere frasi senza generalizzare
Molte sequenze di parole possibili non appaiono mai nel set di addestramento rendendo difficile l'assegnazione di probabilità non nulle.

Rappresentazione Vettoriale Sequenziale
Matematica della sequenza
Invece di limitarci a contare solo le parole, iniziamo a contare le transizioni.
Immagina lo spazio vettoriale non più come un punto per ogni parola ma come un architettura che tiene conto di chi sta vicino a chi.
Questo permette di costruire spazi vettoriali che tengono conto della vicinanza fisica tra i termini

Questo ci porta ad una della applicazioni più pratiche, il riconoscimento delle entità

Il Contesto Locale e le Entità
Catturare il significato delle espressioni composte
Prendi l'espressione New York se la rete legge solo uni-grammi vede New e poi vede York, quindi new come nuovo e york come una contea inglese. Insieme però formano un'entità geografica precisa che non ha nulla a che fare con l'entità dei singoli pezzi. Gli n-grammi agiscono come collante  semantico. Trattando New York come un bi-gramma permettiamo all'algoritmo che deve considerare quei due token come un'unica idea indivisibile

Ma come fa il computer a capire quali parole devono incollarsi e quali no?

Significato e Collocazioni
Oltre l'interpretazione letterale delle singole parole
Usiamo misure statistiche si associazione. Una molto famosa è il PMI, ci dice quanto è probabile trovare due parole insieme rispetto a quanto è probabile trovarle vicine per puro caso, questo risolve anche i problemi di disambiguazione (la parola banca cambia completamente significato se è banca centrale o banca del fiume)
Il contesto locale non definisce solo le entità ma ci da un indizio fondamentale per capire quale dei significati di una parola stiamo usando in quel momento.

Ma come implementiamo questa unione all'interno del pre-processing?

Preprocessing con Bigrammi.
Oggi disponiamo di algoritmi statistici che scansionano milioni righe di testo e uniscono automaticamente i termini più frequenti, quindi automaticamente New e York saranno uniti nel token unico New_York (riconoscimento automatico).
Questo passaggio di pre-processing pulisce il dataset. 
Addestrare modelli su N-grammi permette di catturare vettori più precisi per concetti complessi che non hanno un corrispondente monolares.
Gli N-grammi non dovrebbero solitamente attraversare i confini delle frasi o i segni di punteggiatura forte per evitare associazione errate (attenzione alla punteggiatura).

L'Entropia del Contesto
Incertezza e predizione
In NLP l'entorpia misura l'incertezza. Più il contesto, quindi il nostro N è alto, più l'incertezza diminuisce. 
Catturare il contesto locale significa abbassare l'entropia, ridurre l'incertezza nella predizione della parola successiva, rendendo il computer molto più brava nel prevedere l'informazione.
Maggiore è il contesto (N più alto), minore è l'entropia della distribuzione di probabilità
L'entropia di una sorgente di informazione basata su N-grammi diminuisce all'aumentare di N, fino a un limite naturale imposto dalla lingua.

Ma come trasformiamo queste osservazioni in un vero modello probabilistico?

Frequenze e Probabilità
I motori statistici dietro la generazione del testo.
Un modello basato su N-grammi stima la probabilità di una parola basandosi esclusivamente sulle N-1 parole precedenti. Questa è la cosidetta Assunzione di Markov, che semplifica il calcolo matematico.
L'assunzione di Markov dice che per prevedere la prossima parola non serve conoscere tutta la storia del libro ma bastino solo le ultime parole viste. E' come avere una memoria da pesce rosso ma fovalizzata sull'informazione precedente.
Calcolando quante volte una sequenza appare rispetto al prefisso che la precede, possiamo prevedere con quale probabilità un certo termine seguire un altro in un dato contesto.
Contiamo quante volte appare la coppia Ne York e dividiamo per quante volte la parola New appare da sola. Questa è la probabilità condizionata di una parola data la precedente.

La probabilità di un'intera frase è il prodotto delle probabilità condizionate di ogni parola data la storia precedente.
Assumiamo che il futuro dipenda solo dal passato (gli n-1 token) anzichè dall'intera storia del documento (approsimazione di Markov).
Se un trigramma non è presente, il modello retrocede (back-off) verso il bigramam o l'unigramma per stimare comunque la probabilità

E' un sistema a cascata cercando di usare il contesto più ricco possibile prima di ripiegare su quello più semplice (monogramma)

C'è però  un problema moltiplicare migliaia di piccoli numeri decimali è pericoloso per un computer.
Le probabilità sono numeri tra 0 e 1, se ne moltiplichi migliaia i valori diventano tarlmente piccoli che il computer perde la precisione andando velocemente a errori di underflw numerico. Si preferisce quindi lavorare con la somma dei logaritmi
Trasformare il prodotto delle proabilità in una somma di log-probabilità garantisce stabilità ai calcoli del computer (è un truco algebrico per rendere i modelli molto più stabile).


```python
import keras
import nltk
from nltk.util import ngrams
from collections import Counter, defaultdict
import numpy as np

# Assicuriamoci di avere i dati per la tokenizzazione
nltk.download('punkt')

def build_ngram_statistical_model(corpus: str, n: int = 2):
    """
    Crea un modello statistico basato su N-grammi e calcola le probabilità condizionate.
    
    Teoria: Un modello di Markov di ordine N-1 assume che la probabilità di una parola 
    dipenda solo dagli N-1 token precedenti, riducendo la complessità computazionale.
    """
    
    # Tokenizzazione: trasformiamo il testo in una lista di atomi (unigrammi)
    tokens = nltk.word_tokenize(corpus.lower())
    
    # Generazione degli N-grammi
    # Teoria: usiamo una finestra scorrevole per catturare il contesto locale.
    # 
    generated_ngrams = list(ngrams(tokens, n))
    
    # Calcolo delle frequenze per il numeratore: Count(w_{n-1}, w_n)
    ngram_counts = Counter(generated_ngrams)
    
    # Calcolo delle frequenze per il denominatore (prefissi): Count(w_{n-1})
    # Se n=2, il prefisso è la singola parola precedente.
    prefix_counts = Counter([ng[:-1] for ng in generated_ngrams])
    
    # Calcolo delle probabilità condizionate: P(wn | wn-1) = Count(wn-1, wn) / Count(wn-1)
    # Utilizziamo un dizionario di dizionari per facilitare la predizione
    model = defaultdict(lambda: defaultdict(float))
    
    for ngram, count in ngram_counts.items():
        prefix = ngram[:-1]
        target_word = ngram[-1]
        model[prefix][target_word] = count / prefix_counts[prefix]
        
    return model, generated_ngrams

# --- ESEMPIO PRATICO: ENTITÀ COMPOSTE ---
corpus_test = """
New York è una città incredibile. Molte persone visitano New York ogni anno. 
New York offre opportunità uniche, ma vivere a New York è costoso.
"""

# Analizziamo i Bigrammi (N=2)
model, bigrams_list = build_ngram_statistical_model(corpus_test, n=2)

print(f"Esempio di Bigrammi estratti: {bigrams_list[:5]}")

# Verifichiamo la probabilità che 'york' segua 'new'
prefix_key = ('new',)
prob_york = model[prefix_key]['york']

print(f"\nTeoria delle Probabilità:")
print(f"P('york' | 'new') = {prob_york:.2f}")
# Se la probabilità è 1.0, 'New' e 'York' formano un'entità bi-grammatica perfetta nel corpus.
```
