Qualità dei dati
Bias ed Etica nei Dataset
La responsabilità del dato nell'addestramento
Un modello di intelligenza artificiale è specchio dei dati su cui viene istruito. Se un dataset di volti non è statisticamente rappresentative delle diversità umane, il modello manifesterà invetitabilmente dei pregiudizi algoritmici o bias.

Questi squilibri portano a disparità nelle prestazioni, dove soggetti appartenenti a minoranze etniche o di genere subiscono tassi di errore significativamente più alti, solleavndo gravi questioni di equità sociale.

Ma come si manifestano, concretamente, questi pregiudizi nel software?

Sistemi di Pregiudizio Algoritmico
Identificare e misurare la discriminazione
- Sottorappresentazione: la mancanza di campioni diversificati impedisce alla rete neurale di generalizzare correttamente le feature universali
- Disparità di Falsi Positivi: tassi di errore differenziati che causano identificazioni errate più frequenti in certi gruppi demografici
- Discriminazione Involontaria: un bias non intenzionale inserito dai ricercatori durante la fase di selezione o labelling dei dati
- Audit Algoritmico: la pratica di testare sistematicamente i modelli su subset bilanciati per quantificare lo scostamento delle prestazioni.
Se nella rete mancano donne o minoranza etniche, la rete non imparerà mai le caratteristiche distintive.

Metriche di Parità Statistica
Quantificare l'equità algoritmica
Per correggere il bias, i ricercatori utilizzano metriche matematiche che misurano le differenze di acccuratezza tra diversi gruppi demografici. L'obbiettivo è minimizzare la 'Disparate Impact' Ratio.
La matrice del kernel viene calcolato seguendo la distribuzione bidimensionale di Gauss. Maggiore è lo scarto quadratico medio, più forte è l'anonimizzazione.

Ma oltre all'etiche c'è una vera legge che ci osserva da vicino, il GDPR

GDPR e Dati Biometrici
Protezione giuridica dell'identità digitale.
Il General Data Protection Regulation (GDPR) classifica i dati biometrici come categorie particolari di dati personali, imponendo restrizioni rigorose sul loro trattamento, specialmente se finalizzati all'identificazione univoca.
Il passaggio fondamentale è comprendere che la pricavy non riguarda solo l'immagine grezza, ma anche gli embeddings che da essa derivano, poichè costituiscono una firma digitale univoca e non modificabile.
Per la legge un embedding di 128 feature è come l'impronta digitale, trattare dati biometrici senza seguire la norma è un rischio legale.

Principi di Conformità
Le regole d'ora per lo sviluppatore AI
- Minimizzazione dei dati: trattare solo i dati strettamente necessari per la finalità dichiarata, evitando la raccolta massiva.
- Purpose Limitation: i dati raccolti per uno scopo specifico non possono essere riutilizzati per finalità incompatibili senza nuovo consenso
- Privacy by Design: integrare le protezioni dei dati fin dalle fasi iniziali della progettazione del software
- Storage Limitation: definire periodi di conservazione certi e procedure di cancellazione automatica dei template biometrici.

Un trucco per la sicurezza è quella di separarsi dalle immagini il prima possibile

Gestione Sicura degli Embeddings
* Image-free Verification: e' best practice eliminare l'immagine sorgente immediatamente dopo aver estratto l'embedding. L'embedding è sufficiente per il matching e riduce il riscio in caso di data breach.
* Hasing Biometrico: tecniche avanzate permettono di trasformare gli embeddings tramite funzioni non invertibili, rendendo impossibile risalire alle feature facciali originali anche avendo accesso al database.
* Diritto all'Oblio: il sistema deve garantire all'utente la possibilità di richiedere la rimozione totale dei propri dati biometrici e di tutte le tracce di identificazione correlate.

E se volessimo mostrare un video senza violare la privacy di chi ci passa davanti

Anonimizzazione Automatica
Bilanciare monitoraggio e diritto alla privacy
Nello sviluppo di sistemi di monitoraggio o analisi video, spesso non è necessario conoscere l'identità dei soggetti. L'anonimizzazione in temo reale permette di estrarre metadata utili proteggendo l'identità visiva.
Vedremo come integrare le tecniche di detection studiate in precedenza con algoritmi di filtraggio per oscurare i volti prima della visualizzazione o dell'archiviazione del flusso video.

Tecniche di Offuscamento
Metodo computazionali per la protezione visiva.
- Gaussian Blur: applicazione di un filtro passa-basso sull'area del volto rilevata per rimuovere i dettagli identificativi
- Pixelazione: riduzione drastica della risoluzione spaziale all'interno della bounding box per creare l'effetto mosaico
- Mascheramento Solido: sostituzione dell'area del volto con un rettangolo opaco, metodo più drastico e sicuro.
- Blu dinamico: regolazione dell'intensità dell'oscuramento in base alla distanza del soggetto dalla camera o alla confidenza del detector.

Pipeline di Anonimizzazione
- Estrazione ROI: si isola la region of interest (ROI) corrispondnete al volto. Questa operazione di slicing numpy è fondamentale per non degradare l'intera immagine
- Applicazione Filtro: viene applicato un kernel di convoluzione. La dimensione del Kernel (ksize) determina quanto il volto risulterà irriconoscibile
- Reinserimento Frame: la ROI elaborata viene riposizionata nel frame originale. Questo processo deve essere ottimizzato per non causare cali di FPS nel loop video.

Ma cosa succede, matematicamente, dentro la sfumatura?

Il Kernel Gaussiano
La matematica dietro l'oscuramento
Il filtro Gaussiano utilizza una funzione di distribuzione normale per calcolare il peso dei pixel vicini. Un sigma elevato spalma l'informazione del pixel centrale su un'area vasta, eliminando le feature ad alta frequenza (i tratti del volto)
La matrici del kernel viene calcolata seguendo la distirbuzione bidimensionale di Gauss. Maggiore è lo scarto quadratico meido, più forte è l'anonimizzazione.


```python
import os
import cv2
import numpy as np
import face_recognition
import hashlib

# Configurazione 2026: Backend PyTorch per Keras 3
# Utilizzato per audit di fairness o analisi della qualità dell'immagine
os.environ["KERAS_BACKEND"] = "torch"
import keras

def anonymize_face(image, face_locations, kernel_size=(51, 51)):
    """
    Applica un filtro Gaussian Blur sulle coordinate dei volti rilevati.
    
    Teoria: La sfocatura distrugge le alte frequenze spaziali (dettagli),
    rendendo il dato non identificabile ai sensi del GDPR.
    """
    for top, right, bottom, left in face_locations:
        # Estrazione della Regione di Interesse (ROI)
        roi = image[top:bottom, left:right]
        
        # Applicazione della sfocatura se la ROI è valida
        if roi.size > 0:
            blurred_roi = cv2.GaussianBlur(roi, kernel_size, 0)
            # Reinserimento nel frame originale
            image[top:bottom, left:right] = blurred_roi
    
    return image

def secure_hash_embedding(embedding):
    """
    Genera un hash SHA-256 da un embedding biometrico.
    Assicura che la 'firma' del volto sia conservata in modo non invertibile.
    """
    return hashlib.sha256(embedding.tobytes()).hexdigest()

# --- INIZIALIZZAZIONE HARDWARE ---
video_capture = cv2.VideoCapture(0)

# Parametri di ottimizzazione
scale_factor = 0.25 
process_this_frame = True
face_locations = []

print("Avvio sistema di Privacy Real-time... (Premi 'q' per uscire)")

while True:
    # 1. Cattura frame
    ret, frame = video_capture.read()
    if not ret:
        break

    # 2. Elaborazione IA (ogni 2 frame per performance)
    if process_this_frame:
        # Ridimensionamento per accelerare la detection
        small_frame = cv2.resize(frame, (0, 0), fx=scale_factor, fy=scale_factor)
        rgb_small_frame = cv2.cvtColor(small_frame, cv2.COLOR_BGR2RGB)

        # Trova le coordinate dei volti
        # Nota: le coordinate sono relative al frame ridotto
        raw_face_locations = face_recognition.face_locations(rgb_small_frame)
        
        # Riporta le coordinate alla scala originale
        face_locations = []
        for top, right, bottom, left in raw_face_locations:
            face_locations.append((
                int(top / scale_factor),
                int(right / scale_factor),
                int(bottom / scale_factor),
                int(left / scale_factor)
            ))

        # Esempio di hashing (solo se viene rilevato un volto)
        # In un sistema reale, questo verrebbe salvato in un log di audit
        if face_locations:
            face_encodings = face_recognition.face_encodings(rgb_small_frame, raw_face_locations)
            for encoding in face_encodings:
                secure_id = secure_hash_embedding(encoding)
                # Stampiamo solo i primi 10 caratteri per brevità
                print(f"Rilevato ID Anonimo: {secure_id[:10]}...", end='\r')

    # Alterna il flag di elaborazione
    process_this_frame = not process_this_frame

    # 3. Applicazione Anonimizzazione (sempre attiva sul frame corrente)
    # Usiamo le ultime face_locations note per evitare sfarfallii
    frame = anonymize_face(frame, face_locations)

    # 4. Visualizzazione Risultato
    for (top, right, bottom, left) in face_locations:
        # Disegniamo un rettangolo sottile per indicare dove sta agendo il filtro
        cv2.rectangle(frame, (left, top), (right, bottom), (255, 255, 255), 1)
        cv2.putText(frame, "ANONIMIZZATO", (left, top - 10), 
                    cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255, 255, 255), 1)

    cv2.imshow('Privacy Shield 2026 - Real-time Anonymization', frame)

    # Uscita
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Pulizia
print("\nqChiusura sistema in corso...")
video_capture.release()
cv2.destroyAllWindows()
```
