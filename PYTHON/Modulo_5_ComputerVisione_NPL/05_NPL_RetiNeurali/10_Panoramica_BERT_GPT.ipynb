PANORAMICA SU BERT E GPT: COMPRENDERE E GENERARE IL LINGUAGGIO

BERT e GPT sono due famiglie fondamentali nate dall'architettura Transformer, ma usano il Transformer in modo diverso.

Differenza tra capire un testo e saper scrivere una storia. Sono due abilità distinte, quasi opposte.
Da un lato la comprensione profonda di BERT dall'altro la creatività generativa di GPT

BERT - soprattutto per COMPRENSIONE / rappresentazione del testo
GPT - soprattutto per GENERAZIONE autoregressiva del testo

Non significa che GPT 'non capisca' o che BERT non possa essere adatto ad altri compiti, indica soprattutto come sono costruiti ed addestrati

** BERT **
Bidirection Encoder Representations from Transformers
E' l'investigatore che legge in ogni direzione per non farsi sfuggire il contesto.

Immagina di leggere una frase ma una parola è macchiata dal caffè, per capire cosa c'è scritto non leggi solo le parole a sinistra, ma l'intera frase.
A differenza della RNN che camminava guardando solo avente, bert è come un drone che sorvola l'intera frase, catturando le relazioni tra parole indipendentemente dalla loro distanza.

Ma come si addestra un drone a capire un linguaggio senza dargli compiti espliciti?
Per addestrare BERT Google  ha inventato due giochi:
- Masked Language Model (MLM): BERT nasconde casualmente alcuni token (parole) e cerca di indovinarli usando il contesto circostante
- Next Sentence Prediction (NSP): il modello impara a capire se due frasi sono logicamente consecutive o meno. Gli mostriamo due frasi e gli chiediamo: "la seconda segue logicamente la prima o le ho messe a caso?"
Questi due compiti costringono l'architettura ad estrarre feature incredibilmente ricche.

BERT utilizza principalmente la parte Encoder del Transformer
testo -> tokenizzazione -> embedding + posizione -> transformer Encoder -> rappresentazioni testuali
La parola chiave è bidirectional
BERT, quando costruisce la rappresentazione di un token, può utilizzare il contesto sia a sinistra sia a destra
Per esempio.
"La banca ha approvato il mutuo"
per rappresentare "banca" può considerare anche "ha approvato il mutuo" e quindi capire il contesto che qui "banca" è un istituto finanziario
In: "La barca si è fermata sulla banca del fiume" il contesto porta verso un significato diverso
stessa parola + contesto diverso = rappresentazione contestuale diversa
Questa è una grande differenza rispetto a Word2Vec e GloVe, dove una parole ha normalmente lo stesso vettore indipendentemente dalla frase.

La Potenza della Bidirezionalità
La potenza di BERT sta anche nei dettagli, invece di usare le parole intere usa WordPiece, se non conosce la parole la spezza in pezzi che conosce e questo limita il problema delle parole sconosciute.
Poi c'è il token CLS, immaginate BERT come un team di esperti che analizza un documento, il token CLS è il loro capogruppo, alla fine della lettura il vettore associato a questo token, contiene il verdetto finale sull'intera frase, pronto per essere usato in una classificazione.

Come viene preaddestrato BERT?
uno dei meccanismi più importanti è il Masked Language Modeling
Prendiamo la frase: "il cliente ha ordinato 500 pezzi"
durante il pretraining una parte può diventare:
"il cliente ha [MASK] 500 pezzi"
e il modello deve prevedere "ordinato"
Per farlo usa contemporaneamente:
"il cliente ha" + "500 pezzi"
cioè usa il contesto sia a sinistra della parola da indovinare sia a destra.
Questo prta BERT a imparare rappresentazoini molto utili nel linguaggio.
Dopo il pretraining, BERT può essere fine-tuned per compiti specifici
Esempio:
BERT -> classificatore -> SPAM/HAM
BERT -> classificatore -> POSITIVO/NEGATIVO
BERT -> NER -> PERSONA/ORGANIZZAZIONE/LUOVO
Quindi BERT è stato utilizzato per diversi compiti:
- classificazione di testo
- sentiment analysis
- NER
- question answering estrattivo
- similarità / rappresentazione

** GPT **
Generative Pre-trained Transformer
Il narratore che guarda solo al passato per costruire il futuro.
E' il maestro della creazione, è un cantastorie seduto davanti alla macchina da scrivere, il suo compito nella vita è prevedere la prossima parola.

GPT utilizza invece un'architettura decoder-only basata su Transformer con attenzione casuale.
L'idea è:
Dato ciò che ho visto fino a questo momento, qual'è il prossimo token? GPT prevede la parola successiva data una sequenza di parole precedenti.
Utilizza maschere casuali per impedire al modello di 'sbirciare' nel futuro durante l'addestramento.
E' come se gli mettessimo dei paraocchi con la possibilità di vedere solo quanto già è passato.

Addestrando GPT su migliaia di pagine, emerge un fenomeno incredibile, lo Zero-shot-Learning cioè la capacità di risolvere task nuovi senza fine-tuning, solo tramite istruzioni nel prompt.
Senza esssere mai stato addestrato a tradurre, se gli dici: "traduci in inglese il gatto è sul tavolo" GPT lo fa, perchè? perchè durante il pretraining ha visto così tanti esempi di traduzioni umane che ha imparato lo schema. E' quello che chiamiamo Scaling Lows, più dati gli diamo più sembra acquisire logiche gratis, le prestazioni migliorano in modo prevedibile all'aumentare dei parametri e dei dati.

Esempio
"il cliente ha ordinato"
il modello potrebbe stimare: 
500     -   0.31
un      -   0.20
i       -   0.12
nuovi   -   0.08
...
e scegliere un token
Poi prende "il cliente ha ordinato 500" a predice il successivo:
"pezzi"
Questa è generazione autoregressiva
GPT non può, durante la predizione di un token, guardare i token futuri della stessa sequenza.
Usa una causal mask
Se abbiamo la frase: "il ciente ha ordinato 500 pezzi"
quando sta elaborando: "ordinato"
può vedere: "il", "cliente", "ha"
ma non può utilizzare "500", "pezzi"
come informazione futura per predire "ordinato"

La Fluidità di GPT deriva dal suo Causal Masking, nel meccanismo di self-attention, ogni token può parlare solo con quelli che lo precedono, preservando l'ordine temporale della generazione. Questo garantisce che la generazione sia coerente nel tempo.
E per quanto riguarda la creatività? Attraverso parametri come la temperatura, possiamo modulare quanto il modello debba essere deterministico o fantasioso. Se la temperatura è basso il modello sarà noioso e ripetitivo, se l'alziamo il modello inizia a fare collegamenti a volte originali.

Ogni parola è il miglior passo successivo possible in base a tutto ciò che è avvenuto prima, questo permette a GPT di scrivere interi paragrafi che sembrano scritti da una persona.

quindi:
BERT -> vede a sinistra e a destra nel testo disponibile
GPT -> vede solo i token precedenti per predire il successivo.

BERT
token 1 <-> token 2 <-> token 3 <-> token 4
GPT
token 1 -> token 2 -> token 3 -> token 4 -> prossimo token

C'è una differenza importante negli obbiettivi
BERT è stato progettato soprattutto per costruire rappresentazioni profonde del testo
GPT è stato progettato per modellare la probabilità della sequenza.
GPT ripetendo il meccanismo token -> token -> token può generare: frasi, paragrafi, codice, risposte, riassunti, traduzioni

Riassumento:
BERT -> ho tutto il testo davanti: cosa significa?
GPT -> ho ciò che è stato scritto finora, cosa viene dopo?

Per esempio, sentiment analysis
con BERT:
BERT Encodeer -> rappresentazione contestuale -> classifier -> POSITIVE
Qui non stai chiedendo a BERT di generare una frase. Gli chiedi di rappresentare il testo in modo utile per una classificazione
con GPT:
prompt: "classifica il sentiment: questo file è davvero fantastico". Risposta: GPT può generare Positivo
Qui il compito viene trasformato in generazione di testo
Quindi BERT genera una rappresentazione di classe
GPT una risposta testuale.
Ed è proprio questa seconda impostazione che ha portato i moderni LLM conversazionali

Naturalmente oggi i confini sono meno rigidi: un GPT moderno può svolgere classificazione, estrazione, traduzione, e molti altri compiti attraverso prompting o fine-tuning. Però l'architettura rimane un modello causale decoder-only

BERT usa il Transformer Encoder e costruisce rappresentazioni bidirezionali del testo, particolarmente utili per task di comprensione e classificazione.

GPT usa un Transformer Decoder causale e genera il linguaggio prevedendo autoregressivamente il token successivo sulla base dei token precedenti.

BERT:
"Guardo tutto il contesto per capire."

GPT:
"Guardo ciò che è già stato scritto per generare ciò che viene dopo."

ABBiamo quindi visto BERT e GPT, ma come facciamo ad utilizzarli per i nostri scopi senza spendere milioni in hardware?

L'Era del Pre-training e Fine-tuning
Il nuovo paradigma del Transfer Learning
Invece di addestrare modelli da zero per ogni task, oggi addestriamo modelli giganti su tutto il web e poi li adattiamo a compiti specifici.
Questo processo permette di trasferire la conoscenza linguistica generale (grammatica, fatti, logica) a problemi con pochi dati etichettati.
I modelli BERT e GPT hanno già una cultura genrale del dato, sanno cosa è un verbo, sanno chi è il presidente degli Stati Uniti, noi dobbiamo solo rifinirli per il nostro compito specifico.

WordFlow del Modello Moderno
Prima c'è il Massive Pre-training: addestriamo il modello su milioni dati non etichettati per imparare la struttura del linguaggio. Questa fase la fanno i giganti come google o open-ai
Poi arriviamo noi con il fine-tunning, prendiamo i modelli giganti pre-addestrati e gli diamo un dataset piccolo e prezioso, specifico per un task (es. analisi sentiment)
Servono molti meno esempi etichettati per ottenere prestazioni eccellenti.
Un modello generale (BERT o GPT) può diventare un esperto legale, esperto medico (gli passiamo 1000 cartele cliniche etichettate) o di programmazione

Ma come possiamo adattare un modello gigante senza romperlo?

Strategia di Adattamento
- Freesing dei pesi: possiamo bloccare i layer inferiori del modello (che hanno imparato la grammatica) e addestrare solo l'ultimo layer per task molto specifici.
- Catastrophic Forgetting: il rischio durante il fine-tuning è che il modello perda la conoscenza generale acquisita nel pre-trainig
- Adattatore (LoRA): tecniche moderne che aggiungono piccoli set di parametri addestrabili senza modificare i pesi originali del modello. Immagina di non voler riscrivere il libro di medicina ma di voler aggiungere solo dei post-it. Questo ci permette di avere dei modelli specializzati che rubano poco spazio in memoria e prevengono il catastrophic forgettig, cioè il rischio che il modello dimentichi come si parla mentre impari a diagnosticare.

Dinamica del Fine-tuning
Durante il fine-tuning, i parametri inizializzati dal pre-trainig non vengono buttati ma rifinit per minimizzare l'errore sul compito specifico.
Questa operazione sposta il modello dallo spazio della lingua generale allo spazio del dominio d'interesse.
Stiamo muovendo il modello da uno spazio di conoscenza universale verso una nicchia di competenza specifica.
