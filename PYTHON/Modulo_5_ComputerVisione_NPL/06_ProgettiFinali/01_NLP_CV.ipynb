** La Qualità del Dato **

La Checklist per la Qualità del Dato
Garbage In, Garbage Out: il dogma fondamentale
Prima di addestrare qualsiasi architettura complessa, è necessario validare il dataset. 
Un modello eccellente su dati poveri produrrà sempre risultati inaffidabili.
Analizziamo i punti critici sia per il Computer Vision (CV) che per il Natural Language Processing (NLP) concentrandoci sulla coerenza e sulla distribuzione

La Qualità del dato è uno dei fattori più importanti di tutto il Machine Learning e Deep Learning. Spesso conta più della scelta del modello.
Puoi avere un modello sofisticatissimo, ma se gli dai dati sbagliati, incoerenti o poco rappresentativi, imparerà male.
Non significa "più dati = meglio è". La parola chiave è la qualità

CORRETTEZZA
I modelli, durante il supervised learning confrontano la predizione con il target (loss) ed aggiornano i pesi. Se gli dai etichette sbagliate, impara relazioni sbagliate.
Quindi il primo concetto importante è la correttezza del dato.

COMPLETEZZA
supponiamo che tu voglia prevedere se un cliente pacherà in ritardo, e gli passi al modello i seguenti dati: cliente, fatturato, giorni medi di pagamento, scaduto precedente. Se il 60% dei "giorni medi di pagamento" sono null, hai un problema
Non significa che devi necessariamente eliminare tutte quelle righe, ma devi decidere consapevolmente come gestire i missing
Eliminazione? imputazione? media? corretto che manchi perchè ha significato? 
Il modello non deve ricevere dati mancanti senza che tu abbia deciso cosa significano.

CONSISTENZA
Immagina una colonna con: Italia, ITALIA, IT, Italy, ecc umanamente sono la stessa cosa, ma per un modello possono diventare categorie differenti
Se non normlizzi il significato, il dataset contiene informazioni incoerenti.
Quindi stesso significato -> stessa rappresentazione

RAPPRESENTATIVITA'
Supponiamo di voler classificare tra: mail italiane, mail inglesi, mail francesi, poi arriva una mail in tedesco, il modello non è stato preparato, per lui è un mondo che non esiste.
Una domanda da porci prima dell'addestramento è: il mio dataset rappresenta davvero ciò che il modello vedrà quadno sarà usato?

BILANCIAMENTO DELLE CLASSI
Immagina di dover controllare delle immagini mediche per individuare una malattia rara. Se gli passi il 99% di immagini senza malattie e solo 1% di immagini con malattia, se il modello predice sempre 'non malato', il modello ha un accuracy del 99% ma in realtà è un modello disastroso perchè non predice mai la malattia.
Per questo controlliamo: precision, recall, F1-score e confusion matrix
Per quantificare lo sbilanciamento, utilizziamo l'Imbalance Ratio (IR) che ci permette di decidere se applicare tecniche di oversampling o pesatura delle loss
IR= n. di esempi maggioritari /n. di esempi minoritari
Un valore IR elevato indica la necessità di intervenire sulla strategia di campionamento.

DUPLICATI
Se mostriamo gli stessi dati nel training e nel test il modello potrebbe avere un accuracy altissima ma non perchè ha imparato a generalizzare bene, ma perchè ha memorizzato

DATA LEAKAGE
Quando sporto i dati di training con informazioni future che non sarebbe disponibili in produzione.

RUMORE
Possono essere caratteri strani, in una mail replay precedenti, OCR sagliato, firme nella mail, immagini sfocate, immagini con riflessi, immagini con ombre, ecc.
Attenzione, non tutto il rumore deve essere limitato, se in produzione le immagini sono normalmente un pò sfuocate, addestrare solo su fotografie perfetto può essere peggio.
Il dataset deve rappresentare anche le normali imperfezioni del mondo reale.
Quindi: 
rumore artificiale inutile - da eliminare
rumore variabile reale - da conservare
Per questo si può usare la data augmentation

Un modello non impara la realtà: impara la realtà così come gliela rappresenta il dataset.

** SELEZIONE DEL MODELLO **

Scegliere l'architettura giusta per il contesto
Non sempre il modello più preciso è quello migliore. Dobbiamo considerare i vincoli hardware (GPU vs Edge device) e il tempo di rispostsa richiesto dall'utente

Le metriche di efficienza sono:
- Throughput: numero di redizioni che il sistema può gestire in un secondo 
- Memory Footprint: spazio occupato in RAM o VRAM fondamentale per dispositivi mobili o microcontrolli
- FLOPs: quantità di calcoli necessari, operazioni in virgola mobile necessarie per singola inferemza
- Quantizzazione: riduzione della precisione dei pesi (es. da FP32 a INT8) per accelelare il modello con perdita minima di accuratezza.

Se il modello è tropppo complesso abbiamo delle strategie di Ottimizzazione

Knowledge Distillation: addestrare un modello piccolo (studente) a imitare il comportamento di uno grande (insegnante)
Pruring: rimuovere le connessioni neurali meno significative per alleggerire la struttura computazionale
- Model Parallelism: divider il carico di lavoro su più unità di calcolo quando la memoria del singolo dispositivi è insufficiente.

La latenza non è casuale è un rapporto tra la complessità del modello e la potenza del nostro hardware
La latenza deve essere mantenuta sotto la soglia cui l'utente vuole aspettare

** Documentazione **
Documentare il progetto non significa solo commentare il codie, ma descrivre l'intero ciclo di vita, dai dati agli iperparametri.

Standard industriali
- Model Cards: schede tecnihce che riassumono architettura, dataset di addestramento e limitazione etihchette del modello
- Data Card: documentazione sulle origini del dato, metodi di campionamento e potenziali bias presenti
- Versioning: tracciare non solo il codice (Git) ma anche i pesi del modello e le versioni del dataset (DVC)
- Dependency Management: utilizzare file 'requirements.txt' o ambienti 'Conda' per garantire la portabilità

Documentare i casi in cui il modello fallisce sistematicamente per orientare lo sviluppo futuro.
Specificare chiaramento formati di input e output per facilitare l'integrazione con il frontend o altri servizi (API Documentation).

Tutto questo serve ad un unico grando scopo
Un progetto ben documentato deve avere una probabilità di essere duplicato da un altro ricercatore prossima al 100%
