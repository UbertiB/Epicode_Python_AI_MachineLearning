L'Ecosistema dei Vector Database
Gestione efficiente di grandi moli di rappresentazioni vettoriali.
Finora abbiamo giocato con gli embeddings tenendoli in semplici liste python, memorizzandoli in memoria RAM volatile. Ma immagina di dover gestire l'intera wikipeida, la RAM sarebbe satura in 2 secondi. In contesti di produzione, abbiamo bisogno di database specializzati in grado di gestire la persistenza, l'aggiornamento e la ricerca rapida su milioni di vettori ad alta dimensionalità
Un VectorDB non si limita a salvare i dati, implementa algoritmi di indicizzazione avanzati che permettono di trovare i vicini più prossimi in tempi sub-millisecondi, rendeno possibile la ricerca semantica in tempo reale su scale industriali.
I VectorDB non sono progettati per memorizzare nomi o numeri, ma per gestire la geometria dell'informazione. Se gli embedding sono le impronte digitali dei nostri testi, il vector db è la centrale operativa che permette di confrontare milioni di elementi in un secondo, garantendo che i dati siano sempre pronti e persistenti. I VectorDB rendono efficiente la ricerca semantica per significato, non solo per parole uguali.
Una ricerca semantica non cerca testi con le stesse parole, ma testi con significato vicino.
La documentazione Qdrant dice: il testo viene trasformato in vettori, e contenuti, semanticamente simili, finiscono vicini nello spazio vettoriale.

- Upsert e Persistenza: gestione del ciclo di vita dei vettori, garantendo che le modifiche ai documenti si riflettano nello spazio di ricerca.

La complessità della ricerca lineare
Perchè abbiamo bisogno di indici vettoriali
In una ricerca esautiva, dovremmo confrontare la query con ogni singolo documento nel database, portando a una complessità computazionale insostenibile all'aumentare dei dati.
Sia N il numero di documenti e D la dimensione dell'embedding. La ricerca lineare richiede O (N per D) operazioni, rendendo critico l'uso di indici spaziali.

Con gli indici vettoriali passiamo da una complessità che cresce linearmente ad una logaritmica.

Oltre la Parola Chiave
Il passaggio alla ricerca basata sull'intento semantico
La ricerca tradizionale si base sulla presenza di caratteri identici (es. BM25). Era una ricerca legata alla forma della parola e non al suo significato. 
Se cerchi 'automobile', non troverai documenti che parlano esclusivamente di 'macchina' o 'veicolo'. 
La ricerca semantica risolve questo problema analizzando la vicinanza concettuale nello spazio latente, creato dalla reti neurali. In questo spazio 'cane' e 'cucciolo' sono vicini di casa, anche se non condividono le stesse lettere.
Questo approccio permette di catturare sinonimi, contesti complessi e persino query multilingua, dove una domanda in italiano può recuperare documenti rilevanti in inglese perchè mappati in regione simili nello spazio vettoriale.
Questo è molto più vicino al cervello umano.

Ma come avvinene tecnicamente questa conversazione tra l'utente ed il database?

Query Encoding e Retrieval
Tutto parte dalla queu encoding, quando l'utente scrive una domanda, dobbiamo trasportarla nello stesso spazio in cui vivono i contenuti, usando lo stesso identico modello di embedding. Otteniamo così i primi k risultati, ovvero i primi k risultati più vicini, e poi spesso, con i sistemi più avanzati, facciamo un secondo passaggio chiamato Reranking, è come fare  prima una selezione di un concorso per individuare i finalisti, e poi esaminarli uno per uno con un giudice più severo e preciso e decidere il vincitore assoluto.
- Sincronia dei Modelli: perchè la ricerca funzioni, la query dell'utente deve essere codificata usando lo stesso modello di embedding utilizzato per popolare il database
- Top-k Retieval: il processo restituisce i primi K risultati con il punteggio di similarità più elevato, permettendo all'utente di definire una soglia di rilevanza.
- Reranking Semantico: spesso si usa una ricerca veloc per ottenere 100 candidati e un modello più pesante (Cross-Encoder) per riordinare i primi 10 con precisione estrema.

Qual'è l'operazione matematica che di dice che è il vincitore.
Il prodotto scalare tra due vettori misura quanto essi puntano nella stessa direzione. E' l'operazione fondamentale che sta alla base del calcolo della rilevanza semantica.
Maggiore è il valore del prodotto scalare tra il vettore query e il vettore documento, maggiore è la loro correlazione nel dominio dell'informazione.

Tuttavia il prodotto scalare ha un piccolo difetto che la similarità coseno corrette

Similarità Coseno per il Recupero
La metrica standard per confrontare la rilevanza dei documenti.
A differenza della distanza euclidea, che misura la distanza lineare tra due punti, la similarità coseno si concentra sull'angolo tra i vettori. In NLP questo è fondamentale perchè la magnitudo del vettore può essere influenzata dalla lunghezza del testo.
Implementando questa metrica per determinare quanto una query e un documneto siano allineati, indipendentemente dalla densità delle parole focalizzandoci puramente sulla loro direzione concettuale.

Immagina di dover trovare due navi in mare, non serve sapere quanto sono lunghe ma serve sapere se stanno seguendo la stessa rotta

Calcolo geometrico di questo calcolo.

Meccanica del Cosine Similarity
- Invarianza alla magnitudo: il punteggio non cambia se il vettore viene moltiplicato per una costante (utile per testi di lunghezze diverse). Se un concetto è espresso in 3 parole o in 30 la sua direzione nello spazio semantico è identico.
- Range normalizzato: il risultato varia tipicamente tra -1 e 1, dove 1 indica vettori identici e 0 vettori ortogonali
- Normalizzazione dei vettori: trasformazione dei vettori in vettori unitari per semplificare il calcolo del coseno a un semplice prodotto scalare.
- Analisi del cluster: utilizzo della similarità per raggruppare documenti con temi affini in regioni specifiche dello spazio.

Questo ci permette di creare dei cluster, zone nello spazio di documenti che parlano degli stessi documenti.

Ma come implementiamo tutto questo in modo che sia veloce?

Implementazione e Ottimizzazione
Utilizzeremo le operazioni matricali per calcolare la similartià tra una query e migliaia di documenti simultaneamente, massimizzando l'uso della cache della CPU (vettorizzazione con numpy)
Determinare un valore minimo di similarità (es. 0.75) permette di scartare risultati che sono semanticamente distanti dalla query originale (soglia di confidenza)
Pre-calcolare la norma dei vettori nel database (in modo preventivo, quando li salviamo nel database) accellera drasticamente la fase di ricerca riducendo le operazioni necessarie durante l'interrogazione.

Formula della Similarità Coseno
Rapporto tra prodotto scalare e norme
sotto il cofano stiamo dividendo il prodotot scalare con la grandezza dei vettori
Questa normalizzazione assicura che stiamo misurando solo l'angolo theta tra i vettori nello spazio multidimensionale. 

```python
#python -m pip install faiss-cpu
import os
import torch
import torch.nn.functional as F
from transformers import AutoTokenizer, AutoModel
import numpy as np
import faiss

# ==============================================================================
# CONFIGURAZIONE DELL'AMBIENTE DI ESECUZIONE
# ==============================================================================
# Disabilitiamo i messaggi di log ridondanti di TensorFlow per pulire l'output.
# Queste impostazioni assicurano che vengano mostrati solo errori critici.
os.environ["TF_ENABLE_ONEDNN_OPTS"] = "0"
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"

# ==============================================================================
# CLASSE: SimpleEmbedder
# ==============================================================================
# Questa classe funge da interfaccia tra i modelli pre-addestrati di Hugging Face
# e la nostra pipeline. Il suo scopo principale è convertire stringhe di testo
# in vettori numerici (embeddings) che catturano il significato semantico.
class SimpleEmbedder:
    def __init__(self, model_name):
        """
        Inizializza il tokenizer e il modello pre-addestrato.
        
        Args:
            model_name (str): Il percorso o l'ID del modello su Hugging Face Hub.
        """
        print(f"Caricamento del modello Hugging Face in corso: {model_name}...")
        
        # Il tokenizer converte il testo in ID numerici comprensibili dal modello.
        self.tokenizer = AutoTokenizer.from_pretrained(model_name)
        
        # Il modello (AutoModel) trasforma gli ID in rappresentazioni vettoriali complesse.
        self.model = AutoModel.from_pretrained(model_name)
        
        # Impostiamo il modello in modalità valutazione (eval) per disabilitare 
        # componenti come il Dropout, assicurando risultati deterministici.
        self.model.eval()

    def encode(self, texts):
        """
        Trasforma una lista di testi in una matrice di embeddings (NumPy array).
        Il processo segue tre fasi: tokenizzazione, inferenza e pooling.
        """
        # Fase 1: Tokenizzazione. 
        # Trasformiamo il testo in tensori PyTorch ('pt').
        # Il padding aggiunge zeri per rendere le frasi della stessa lunghezza nel batch.
        # Il truncation accorcia le frasi che superano la lunghezza massima del modello.
        inputs = self.tokenizer(texts, padding=True, truncation=True, return_tensors='pt')
        
        # Fase 2: Inferenza.
        # Utilizziamo torch.no_grad() perché non dobbiamo addestrare il modello,
        # riducendo drasticamente il consumo di memoria e aumentando la velocità.
        with torch.no_grad():
            outputs = self.model(**inputs)

        # Fase 3: Mean Pooling.
        # Il modello restituisce un vettore per ogni singolo token (parola/sottoparola).
        # Per ottenere un unico vettore che rappresenti l'intera frase, facciamo la media.
        
        # Estraiamo gli embeddings dell'ultimo layer nascosto.
        token_embeddings = outputs.last_hidden_state 
        
        # Utilizziamo la attention_mask per evitare di calcolare nella media i token di padding.
        attention_mask = inputs['attention_mask']
        
        # Espandiamo la maschera per matchare le dimensioni degli embeddings (batch, seq_len, hidden_dim).
        input_mask_expanded = attention_mask.unsqueeze(-1).expand(token_embeddings.size()).float()
        
        # Moltiplichiamo i vettori per la maschera (scartando il padding) e sommiamo lungo la sequenza.
        sum_embeddings = torch.sum(token_embeddings * input_mask_expanded, 1)
        
        # Contiamo quanti token validi (non padding) ci sono per ogni frase.
        sum_mask = torch.clamp(input_mask_expanded.sum(1), min=1e-9)
        
        # Eseguiamo la media dividendo la somma dei vettori per il numero di token reali.
        embeddings = sum_embeddings / sum_mask

        # Convertiamo il tensore PyTorch in un array NumPy per la massima compatibilità con FAISS.
        return embeddings.numpy()

# ==============================================================================
# FUNZIONE PRINCIPALE: Workflow di Ricerca Semantica
# ==============================================================================
def main():
    print("\n--- Inizio Procedura di Ricerca Semantica (Backend: PyTorch) ---\n")

    # 1. DEFINIZIONE DEL CORPUS
    # Rappresenta il nostro database testuale: i documenti in cui cercheremo informazioni.
    documents = [
        "La tua auto sportiva è molto veloce in autostrada.",    
        "Ho mangiato una mela rossa per merenda.",              
        "Il nuovo iPhone di Apple ha una fotocamera incredibile.", 
        "Il veicolo a quattro ruote ha bisogno di benzina.",     
        "Le azioni in borsa sono crollate ieri.",               
        "Una dieta ricca di frutta e verdura fa bene alla salute." 
    ]

    # 2. INIZIALIZZAZIONE DEL MODELLO E GENERAZIONE EMBEDDINGS
    # Istanziamo la classe helper specificando un modello ottimizzato per la similarità.
    embedder = SimpleEmbedder('sentence-transformers/all-MiniLM-L6-v2')

    print("Calcolo delle rappresentazioni vettoriali (embeddings) per i documenti...")
    doc_embeddings = embedder.encode(documents)

    # Per utilizzare la Similarità del Coseno con FAISS IndexFlatIP,
    # i vettori devono essere normalizzati (lunghezza unitaria).
    faiss.normalize_L2(doc_embeddings)
    
    # Identifichiamo la dimensione dello spazio vettoriale (es. 384 per questo modello).
    embedding_dim = doc_embeddings.shape[1]

    # 3. CREAZIONE DEL VECTOR DATABASE (FAISS)
    # IndexFlatIP utilizza il Prodotto Scalare (Inner Product). 
    # Se i vettori sono normalizzati, il Prodotto Scalare equivale alla Cosine Similarity.
    index = faiss.IndexFlatIP(embedding_dim)
    
    # Aggiungiamo i vettori dei documenti all'indice per permettere la ricerca rapida.
    index.add(doc_embeddings)
    print(f"Database vettoriale configurato con successo. Documenti indicizzati: {index.ntotal}\n")

    # 4. ESECUZIONE DELLA RICERCA SEMANTICA
    # L'utente inserisce una frase; il sistema cercherà i concetti affini, non solo le parole esatte.
    user_query = "Automobile rapida"
    k = 3 # Numero di risultati desiderati

    print(f"Testo cercato dall'utente: '{user_query}'")
    
    # Trasformiamo anche la query in un vettore utilizzando lo stesso processo dei documenti.
    query_vector = embedder.encode([user_query])
    faiss.normalize_L2(query_vector)

    # Eseguiamo la ricerca: FAISS confronta il query_vector con tutti i vettori nell'indice.
    # Restituisce i punteggi di similarità (distances) e le posizioni degli elementi (indices).
    distances, indices = index.search(query_vector, k)

    print(f"\nRisultati principali (Top-{k}):")
    print("-" * 60)
    for i in range(k):
        doc_id = indices[0][i]
        score = distances[0][i]
        # Mostriamo il rank, il punteggio di confidenza e il testo originale corrispondente.
        print(f"Posizione {i+1}: Similarità {score:.4f} | Documento: \"{documents[doc_id]}\"")
    print("-" * 60)

    # 5. VERIFICA DEI RISULTATI
    # Confrontiamo il calcolo di FAISS con un calcolo manuale tramite prodotto scalare NumPy.
    # Questo serve a dimostrare che la ricerca semantica è pura operazione matematica tra vettori.
    manual_score = np.dot(query_vector[0], doc_embeddings[indices[0][0]])
    print("\nVerifica tecnica dei calcoli:")
    print(f"   Score calcolato manualmente: {manual_score:.6f}")
    print(f"   Score restituito da FAISS:   {distances[0][0]:.6f}")

if __name__ == "__main__":
    main()
```
